# Task 1 — Your First Own Module

## 🐣 Concept: a module is just a file you point Python at

A module is any `.py` file containing functions and variables. To import it, the file must live in a folder listed on `sys.path` — the ordered list Python searches. The notebook-friendly trick: write the file into a temp folder (`tempfile.mkdtemp()`), then `sys.path.insert(0, tmp)` so `import salam` finds `salam.py` immediately. The file content is a plain string written with `open(..., "w")`; the `.py` extension is what makes it a module.

Once imported, `salam.BANNER` and `salam.greet("Sana")` are one dot away — exactly like `math.pi`. In the source string, `\n` is the newline character; it splits `BANNER` and `greet` into separate lines of real Python code.

3. `import salam`; print the banner and `greet("Sana")`.

# Task 3 — Guard Twins

## 🔑 Concept: `__name__` changes with how you launch the file

The same file can run two ways. As a script (`python party.py`) it gets `__name__ == "__main__"`; as an import (`import party`) it gets `__name__ == "party"`. An unguarded top-level print always fires — it is part of the module body — but a block under `if __name__ == "__main__":` fires *only* in script mode. That is the entire guard trick.

`importlib.invalidate_caches()` matters here: Python caches folder listings, so a file freshly written into an already-seen directory can be invisible to a new import. Invalidating the cache forces Python to re-scan.

### Your goal

3. Run the same file via `subprocess.run([sys.executable, party_path], ...)` — observe both prints.

```python
# ✅ SOLUTION — Task 3
import subprocess, importlib

party_code = '''def dance():
    return "dance-shuru! 💃"

print("baaraat-ati-hi 🎺")

if __name__ == "__main__":
    print("party-only-loud! 🎬")
'''

party_path = os.path.join(tmp, "party.py")
with open(party_path, "w") as f:
    f.write(party_code)

importlib.invalidate_caches()         # 🔄 new-file listing refresh!
print("=== IMPORT-side ===")
import party                          # baaraat-banner only!
print("dance:", party.dance())
print("=== SCRIPT-side ===")
res = subprocess.run([sys.executable, party_path], capture_output=True, text=True)
print(res.stdout.strip())
# Expected OUTPUT:
# === IMPORT-side ===
# baaraat-ati-hi 🎺
# dance: dance-shuru! 💃
# === SCRIPT-side ===
# baaraat-ati-hi 🎺
# party-only-loud! 🎬
```

# Task 4 — Grade-Utils Toolbelt

## 📚 Concept: helpers as a module

Three classic helper functions live in `grade_utils.py`:

- `average(marks)` → `round(sum(marks) / len(marks), 1)` — the mean rounded to one decimal (for `[50, 60, 70]` → `60.0`).
- `grade_letter(score)` → a cascade of `if score >= X` checks mapping 90→A, 75→B, 60→C, 40→D, else F.

As a module, any notebook can `from grade_utils import ...` and reuse them — the whole point of the lesson. Add docstrings: `gu.__doc__` and `gu.average.__doc__` should answer clearly.

```python
# ✅ SOLUTION — Task 4
gr_utils = '''"""grade_utils — exams-ka helper-module! 📚"""

def average(marks):
    """List-of-marks ka round-1 avg!"""
    return round(sum(marks) / len(marks), 1)

def grade_letter(score):
    """Score → exam-card letter!"""
    if score >= 90: return "A"
    if score >= 75: return "B"
    if score >= 60: return "C"
    if score >= 40: return "D"
    return "F"

def topper(students):
    """{'name': marks} me best!"""
    return max(students, key=students.get)
'''

with open(os.path.join(tmp, "grade_utils.py"), "w") as f:
    f.write(gr_utils)

importlib.invalidate_caches()         # 🔄 fresh-file dikhe!
import grade_utils as gu
print("avg   :", gu.average([50, 60, 70]))
print("grade :", gu.grade_letter(88))
print("topper:", gu.topper({"Sana": 92, "Amit": 85, "Priya": 96}))
print("doc   :", gu.__doc__)
# Expected OUTPUT:
# avg   : 60.0
# grade : B
# topper: Priya
# doc   : grade_utils — exams-ka helper-module! 📚
```

# Task 6 — Dotted + Relative-Note

## 🧭 Concept: package import styles and the root-entry rule

Two styles reach the same function:

- **Dotted import:** `from kedai.billing import tax` — full package path, then call `tax(2000)` bare.
- **Module import:** `from kedai import billing` — grab the submodule, then `billing.tax(2000)`.

Both return `100.0` for 2000. The deeper lesson is **relative imports**: submodules reach siblings with dots (`from .billing import tax`), but those dots only resolve when the file is loaded *as part of its package*. Running `python kedai/billing.py` directly fails with `ImportError: attempted relative import`, because there is no enclosing package — hence the golden rule: entry files live in the root folder; submodules are imported, never executed directly.

### Your goal

1. `from kedai.billing import tax` → print `tax(2000)`.

```python
# ✅ SOLUTION — Task 7 (Kotini-Toolkit BOSS 📋🔧)
BASE2 = tempfile.mkdtemp()
PKG2 = os.path.join(BASE2, "kotini_toolkit")
os.mkdir(PKG2)

INIT2 = '''"""kotini_toolkit — finance-team ka audit-pack! 📋🔧"""
print("kotini-toolkit banner: audit-ready! 🛡️")

from .audits import tax, savings_rate
from .reports import card
'''

AUDITS2 = '''"""audits — tax + savings helpers! 💰"""

def tax(amount, rate=0.05):
    """percent-rate ka tax round-2!"""
    return round(amount * rate, 2)

def savings_rate(savings, income):
    """income-share as-% round-1!"""
    return round(savings / income * 100, 1)
'''

REPORTS2 = '''"""reports — audit-card print-public! 📋"""

def card(name, income, savings, tax_amt):
    """4-line sync-align audit-card!"""
    lines = [
        "┌─ KOTINI-AUDIT-CARD ────────────────┐",
        f"│ name   : {name}",
        f"│ income : ₹{income:,}",
        f"│ savings: ₹{savings:,} ({round(savings / income * 100, 1)}%)",
        f"│ tax    : ₹{tax_amt:,}",
        "└─────────────────────────────────────┘",
    ]
    return "\\n".join(lines)
'''

with open(os.path.join(PKG2, "__init__.py"), "w") as f:
    f.write(INIT2)
with open(os.path.join(PKG2, "audits.py"), "w") as f:
    f.write(AUDITS2)
with open(os.path.join(PKG2, "reports.py"), "w") as f:
    f.write(REPORTS2)

sys.path.insert(0, BASE2)
import kotini_toolkit as kt

print("tax           :", kt.tax(100000))
print("savings-rate  :", kt.savings_rate(25000, 100000))
print("card-print 👇")
print(kt.card("Priya Sharma", 100000, 25000, 5000.0))
print("package-toolkit complete — import = ready-army! 🛡️")
# Expected OUTPUT:
# kotini-toolkit banner: audit-ready! 🛡️
# tax           : 5000.0
# savings-rate  : 25.0
# card-print 👇
# ┌─ KOTINI-AUDIT-CARD ────────────────┐
# │ name   : Priya Sharma
# │ income : ₹100,000
# │ savings: ₹25,000 (25.0%)
# │ tax    : ₹5,000.0
# └─────────────────────────────────────┘
# package-toolkit complete — import = ready-army! 🛡️
```
