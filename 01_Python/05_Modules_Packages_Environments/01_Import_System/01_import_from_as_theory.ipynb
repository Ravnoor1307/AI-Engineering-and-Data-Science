# 1. The Four Import Styles

## 1.1 Style 1 — `import module`: The Whole Counter, Prefix Calls

The first and safest style is the plain `import module`. Think of it as lifting the *entire counter* over to your seat: every tool that counter offers is now available, but every tool is labelled with its counter name. You call `math.sqrt(625)` rather than a bare `sqrt(625)`.

Why does this matter? Because labels prevent collisions. Python files reuse names freely — you might have a variable called `sum` or a function called `random`, anything. If every imported function came without a label, your variables and the library's names would fight for the same name, and whichever was defined later would silently win.

The code below imports `math` and immediately puts it to work:

- `math.sqrt(625)` — the square root; 625 is a perfect square, so `25.0`.
- `math.factorial(6)` — `6! = 6 * 5 * 4 * 3 * 2 * 1 = 720`.
- `round(math.pi, 4)` — pi shortened to four decimals: `3.1416`.
- `math.hypot(3, 4)` — the hypotenuse of a right triangle with legs 3 and 4; Pythagoras gives `5.0`.
- `math.gcd(48, 180)` — the greatest common divisor, `12`.
- `math.ceil(9.2)` and `math.floor(9.2)` — rounding up to `10` and down to `9`; they always move to the nearest whole numbers above and below.

Notice the pattern: `import math` creates a **namespace**, and you reach inside it with the dot operator. Later on, this habit — always going through a label — is the single best defensive habit in Python.

```python
# 📦 STYLE-1: `import module` — pura-counter lift, naam-se-call!
import math

print("sqrt(625)  :", math.sqrt(625))
print("factorial 6:", math.factorial(6))
print("pi round-4 :", round(math.pi, 4))
print("hypot(3,4) :", math.hypot(3, 4))
print("gcd(48,180):", math.gcd(48, 180))
print("ceil 9.2   :", math.ceil(9.2), "| floor 9.2:", math.floor(9.2))
# Expected OUTPUT:
# sqrt(625)  : 25.0
# factorial 6: 720
# pi round-4 : 3.1416
# hypot(3,4) : 5.0
# gcd(48,180): 12
# ceil 9.2   : 10 | floor 9.2: 9
```

## 1.2 Style 2 — `import module as alias`: Give It a Short Stamp

Typing `math.sqrt` all day is fine, but some module names are long. Python lets you attach a short nickname with `import module as alias`. After `import math as m`, the name `m` refers to the exact same module object — there is only ever one `math` loaded — and you call `m.hypot(5, 12)` exactly as you would call `math.hypot(5, 12)`.

The alias is your personal abbreviation, chosen by you, for you. It changes nothing about how Python resolves the import; it only changes the handle you use from that point on. Different files can even give the same module different aliases without conflict.

The same cell shows why `datetime` is a wonderful everyday module. Dates are objects, and objects know how to behave:

- `dt.date(2026, 9, 17)` builds a fixed date — no time, no timezone, just the calendar day.
- `.strftime("%A")` formats that date as the full weekday name, here "Thursday".
- Subtracting two dates gives a `timedelta`, and reading `.days` tells you how many calendar days lie between them — here `106` days from 17 September 2026 to 1 January 2027.
- Adding `dt.timedelta(weeks=2)` moves the date forward by fourteen days, landing on 1 October 2026.

This is the heart of library reuse: someone built the hard, fiddly calendar logic, and you get to say "add two weeks" in one readable line.

```python
# 🏷️ STYLE-2: `import module as alias` — short-stamp nickname!
import math as m
import datetime as dt

print("m.hypot    :", m.hypot(5, 12))
today = dt.date(2026, 9, 17)                       # fixed-date (deterministic!) 📅
print("today      :", today, "| weekday:", today.strftime("%A"))
newyear = dt.date(2027, 1, 1)
print("days-left  :", (newyear - today).days, "(NY-2027 tak!)")
print("+2 weeks   :", today + dt.timedelta(weeks=2))
# Expected OUTPUT:
# m.hypot    : 13.0
# today      : 2026-09-17 | weekday: Thursday
# days-left  : 106 (NY-2027 tak!)
# +2 weeks   : 2026-10-01
```

## 1.3 Style 3 — `from module import name`: Grab Only the Dish You Need

Sometimes you want just one or two tools, not a whole counter. `from math import hypot, factorial, comb` pulls only those three names into your file, and after that line they can be called directly — `comb(10, 5)` — with no prefix at all.

This is the most convenient style when used sparingly, which is exactly why the pairing below is so instructive. The cell imports `seed`, `randint`, `choice`, and `sample` from `random`, then demonstrates the single most important idea in random programming: **seeding**.

Computers cannot produce true randomness by themselves — they compute, and computing is deterministic. `random.randint(1, 6)` doesn't roll a physical die; it reads the next value from a long mathematical sequence that *looks* random. That sequence is controlled by a **seed**: a starting number.

- Set `seed(42)` and the sequence begins at a specific point.
- Ask for three dice rolls and you get `[6, 1, 1]`.
- Call `seed(42)` again, and the sequence restarts at exactly the same point.
- Ask for three dice rolls again and you get the *same* `[6, 1, 1]`.

This is the superpower hiding inside "randomness": with a fixed seed, an experiment becomes reproducible. A failing test — or a paper you want to defend — can be re-run with identical results. `seed()` plus the `random` family turns lucky guessing into controlled science.

Also note `comb(10, 5)` = 252: the number of ways to choose 5 items out of 10, a preview of the "how many possible papers?" idea in the task notebook.

```python
# 🥡 STYLE-3: `from module import name` — target-dish direct!
from math import hypot, factorial, comb
from random import seed, randint, choice, sample

print("booth-combo:", comb(10, 5), "ways")
seed(42)                                           # 🎲 same-seed = same-dice!
print("dice x3    :", [randint(1, 6) for _ in range(3)])
seed(42)                                           # re-seed: same-trick repeat!
print("redice x3  :", [randint(1, 6) for _ in range(3)], "← REPEAT ka proof!")
# Expected OUTPUT:
# booth-combo: 252 ways
# dice x3    : [6, 1, 1]
# redice x3  : [6, 1, 1] ← REPEAT ka proof!
```

## 1.4 Style 4 — `from module import *`: The Wildcard Trap

`from math import *` sounds incredibly convenient: "give me everything, no prefix, no fuss." It is — until it isn't. The star import takes *every* public name in the module and dumps it directly into your namespace. You don't get to choose; you get the whole counter thrown onto your plate.

The demonstration in this cell exposes the exact failure mode. Start with a plain user variable:

`ceil = 99` — your own, perfectly reasonable integer.

If, at that point, you had executed `from math import *`, Python's `math.ceil` function would have stomped over your variable. Conversely, if you already have your own `sum`, `min`, or `range`, they would stomp over the library's. Whichever assignment happens last wins, and nobody gets a warning. That silent, order-dependent overwriting is called **namespace pollution**.

The prefix style `math.ceil(9.2)` is immune: your `ceil` and math's `ceil` live in different namespaces, so both can exist peacefully. The quote in this cell sums it up: *prefix = the trustworthy road*.

The markdown cell right after this demo lists the three concrete harms — pollution, origin confusion, and linter blindness. When you import 200+ names you can no longer tell where any of them came from, and tools that catch typos and shadowing go quiet. The professional rule of thumb: prefer `import math` or an explicit `from math import ceil`; reserve the star for quick interactive experiments at your own risk.

```python
# ⚠️ STYLE-4: `from module import *` — sab-kuch NAKED-import ka big-bill!
# (demo WITHOUT-star: cool-concept explain; star-import pollution!)
print("ceil ke do-chehre — same-naam ke saamne:")
# Scenario: star-import ne `ceil` ko math.ceil se OVERWRITE!
import math
ceil = 99                                          # user-variable
print("user-ceil  :", ceil)
# agar `from math import *` kar diya hota:
# → math.ceil ne 'ceil' ko crush ho-jaata boom! (bhaari bill!)
print("math.ceil  :", math.ceil(9.2), "(prefix = bharosa raasta!)")
# Expected OUTPUT:
# ceil ke do-chehre — manas ke saamne:
# user-ceil  : 99
# math.ceil  : 10 (prefix = bharosa raasta!)
```

### ⚠️ The Harm of `import *` (3-part warning!)
1. **Namespace pollution:** 200+ names mix into your pocket — risk of your
   variables being hijacked (`sum`/`min` get taken over!)
2. **Origin confusion:** where did this function come from? (maintenance nightmare 😵)
3. **Lint pain:** IDEs/linters cannot track silent imports!
> 🎯 **Rule-card:** for script clarity use `import X` or `from X import …


# 2. Inspecting Modules

## 2.1 Peeking Inside with `dir()` and the Dunders

Modules can feel like black boxes until you learn to open them. Python gives you three instant inspection tools, all used in this cell:

- `type(math).__name__` — the class of the object, which prints `"module"`. This confirms that a module is just another kind of Python object.
- `math.__name__` — the module's own name, `"math"`. Every module knows who it is.
- `dir(math)` — the complete inventory of names inside the module. Filtering out names starting with an underscore (`_`) removes the private machinery and leaves the *public API*: the functions the library authors intend you to call.

On this machine, `dir(math)` reveals more than sixty public tools. The letters `acos`, `cbrt`, `comb`, `log`, `sin`, and `tan` are all there — the entire counter, listed for free. In an interactive console, running `dir(math)` yourself is often faster than searching the documentation, because the answer is right there in the interpreter.

The double-underscore names you see when you *don't* filter are called **dunders** (double underscores). Attributes like `__doc__`, `__file__`, and `__name__` are Python's metadata layer: they describe the module itself. Good libraries also give every function a `__doc__` string, which lets you request `help(math.sqrt)` and read a plain-English description. If you write your own modules with docstrings (the next lesson), you are giving your teammates exactly the same luxury — a library that documents itself.

```python
# 👀 Module peeping: `dir()` + module-name/type!
import math, random, datetime
print("math-module type:", type(math).__name__)
print("math.__name__   :", math.__name__)

peek = [n for n in dir(math) if not n.startswith("_")]
print("math's public-tools (first-10):", peek[:10])
print("total public  :", len(peek))
# Expected OUTPUT:
# math-module type: module
# math.__name__   : math
# math's public-tools (first-10): ['acos', 'acosh', 'asin', 'asinh', 'atan', 'atan2', 'atanh', 'cbrt', 'ceil', 'comb']
# total public  : (varies ~50+, signature version-stable head!)
```

# 3. The Shadow Trap — Never Name a File Like the Standard Library

## 3.1 When Your Own `math.py` Silently Replaces the Real One

Here is an interview classic that has burned thousands of beginners — and a few professionals. Imagine you create a small helper file and, in a hurry, name it `math.py`. Somewhere else in the same project you write `import math`, expecting the square-root machine you have used all along.

Python looks for `math` by scanning its import paths. The directory containing *your* file is almost always on that search path — frequently before the standard library. So `import math` binds to **your** `math.py`, not the real one. The standard library is effectively hidden behind your file.

This is the **shadow trap**: a local module that shadows (overwrites) a standard-library name. Now every call to `math.sqrt`, every alias, and every learning snippet that depends on the real `math` silently breaks — or worse, `from math import sqrt` imports your file's `sqrt`, which may not exist at all.

The fix is absurdly simple: choose a different filename. The hall of shame names the usual suspects — `math.py`, `random.py`, `datetime.py`, `json.py`, `os.py`, `types.py`, `email.py`, `test.py`. If your file is on that list (or could be mistaken for any stdlib module), rename it before you import anything.

The complete live proof appears in the task notebook: a temporary fake `math.py` is created, `sys.path` is pointed at it first, and `import math` really does pick up the fake. After cleanup, the genuine standard-library `math` comes back. Seeing it happen once is worth a thousand warnings.

## ⚠️ 5 Common Mistakes

| # | ❌ Mistake | ✅ Correct |
|---|---|---|
| 1 | `from math import *` again and again | prefix-style `import math` for safety |
| 2 | your filename = stdlib (`math.py`) shadow-trap | give your files a different name! 🪤 |
| 3 | `from random import *` + your own `sample()` clash | use only from-name-import (explicit!) |
| 4 | mixing string-dates in datetime.date() | parse the text → `datetime.strptime` |
| 5 | thinking seed = "always lucky" | seed is only a deterministic repeat (testing tool!) 🎲 |

> 🎯 **Toolbox-card:** math=numbers-pump ⛽ · random=seeded-luck 🎲 ·
> datetime=calendar-train 🚂 — install-free, docs-rich, import just this!


## Summary

A module is a reusable `.py` file, and the standard library is a collection of such modules that ships free with Python. You can import any of them in four ways: `import X` keeps a clean, prefixed namespace; `import X as y` gives a short alias; `from X import name` pulls in only the tools you need; and `from X import *` dumps in every name and risks silently overwriting your variables. We used `math` for number drills, `random` with a fixed `seed(42)` to prove deterministic replay, and `datetime` to subtract dates and add timedeltas. Inspecting a module with `dir()` and reading its dunders such as `__name__` and `__doc__` gives you a free tour of its public API. Finally, the shadow trap warns us never to name our own file `math.py` (or any stdlib name) — a mistake that silently hides the real library and breaks every dependent import.
