# 2. Task 2 — Alias + From-Import Mix

This task is a two-part drill on the difference between `import ... as` and `from ... import ...`.

### Concept: Two import styles, two mental models

`import math as m` binds the whole module to a short nickname. Now `m` refers to the same module object as `math`, and every function stays namespaced behind the prefix: `m.gcd(90, 216)`. `gcd` (greatest common divisor) is the largest integer that divides both 90 and 216 without remainder — here `18`. The alias only saves typing; it does not copy anything into your namespace.

`from math import ceil, floor` does something different: it copies those *two* names directly into your current namespace, so you call `ceil(2.5)` with no prefix. `ceil` (ceiling) rounds up to the nearest whole number → `3`; `floor` rounds down → `2`. Because `2.5` sits exactly between them, the pair shows the contrast neatly: same input, two different answers.

### Your goal

- `import math as m` → print `m.gcd(90, 216)` → `18`
- `from math import ceil, floor` → print `ceil(2.5)` → `3` and `floor(2.5)` → `2`

> **Hint:** an aliased module keeps its prefix (`m.`); a from-import gives the bare name directly.

# 3. Task 3 — Seeded-Dice Replay

The third task demonstrates one of `random`'s most useful features: reproducibility through the seed.

### Concept: Why `seed` makes "random" repeatable

Python's `random.randint` is not truly random — it is a deterministic *pseudo-random* generator. Given the same starting state, it produces exactly the same sequence of values. The `random.seed(42)` call sets that starting state. So if you seed with 42, generate three `randint(1, 10)` values, then seed with 42 again, the generator restarts from the same state and replays the identical three values. This is the backbone of debugging and reproducible experiments: a "random" draw you can repeat is a draw you can test.

### Your goal

- `random.seed(42)` then list-generate `random.randint(1, 10)` three times.
- Call `random.seed(42)` again and generate another three values.
- Print both lists and also the equality check, proving the same list came back.

> **Hint:** seed-reset → same sequence; print `first_deal == second_deal` as proof.

```python
# TODO
```

<details><summary>✅ Solution</summary>

[2, 1, 5] → re-seed → [2, 1, 5] — repeat-proof staged! 🔁
</details>

```python
# ✅ SOLUTION — Task 3
import random
random.seed(42)
first_deal = [random.randint(1, 10) for _ in range(3)]
print("deal-1      :", first_deal)
random.seed(42)                                   # replay-button! 🔁
second_deal = [random.randint(1, 10) for _ in range(3)]
print("deal-2      :", second_deal)
print("same-luck?  :", first_deal == second_deal)
# Expected OUTPUT:
# deal-1      : [2, 1, 5]
# deal-2      : [2, 1, 5]
# same-luck?  : True
```

# 4. Task 4 — Calendar-Train Dates

### Concept: Dates are objects you can do arithmetic on

The `datetime` module stores a calendar day in a `date` object. `dt.date(2026, 9, 17)` builds the exact day 17 September 2026. `.strftime("%A")` formats it as a full weekday name — `%A` means "full weekday", so you get `Thursday`. # 6. Task 6 — Shadow-Trap Proof (Temp-Folder Buddy)

The theory notebook warned you about naming a file `math.py`. This task proves the danger with a live experiment.

### Concept: How imports find their modules — and how they cache them

When you write `import math`, Python takes a two-step journey. First it checks `sys.modules` (the in-memory cache of everything already imported); if found, it reuses it and skips the disk entirely. Only if missing does it scan `sys.path` for a `math.py` (or a package folder), and it stops at the *first* match. That first-match rule is the trap: create a fake `math.py` in a temp folder, insert that folder at the front of `sys.path`, and `import math` resolves to YOUR fake file — the standard library stays hidden. The fake's variable `BANNER = "LOCAL-FAKE-MATH"` is the proof the fake won.

Notice the solution also pops `math` from `sys.modules` BEFORE inserting the path: earlier cells already imported the real `math`, so without evicting it the cache would win and the fake would never load. The cleanup runs in reverse: `sys.path.pop(0)` removes your folder from the search list, then `del sys.modules["math"]` evicts the fake module. The next `import math` reloads from disk, finds the genuine standard library, and `math.sqrt(9)` returns `3.0` again. Shadowing is real — and now you can prove it, and reverse it.

### Your goal

1. Temp folder → write fake `math.py` containing `BANNER`.
2. `sys.path.insert(0, tmp)` → `import math` → print `math.BANNER`.
3. Cleanup: `sys.path.pop(0)` + `del sys.modules["math"]` → re-import → `math.sqrt(9)` = `3.0`.

> **Hints:** use `tempfile.TemporaryDirectory()`; cleanup order matters — path-pop, then cache-del, then re-import.

```python
# TODO
```

<details><summary>✅ Solution</summary>

FAKE-math caught (LOCAL-FAKE-MATH 💀!) → cleanup → REAL sqrt=3.0!
Friend going forward: never shadow a file name again! 🪤✌️
</details>


```python
# ✅ SOLUTION — Task 6
import sys, os, tempfile

with tempfile.TemporaryDirectory() as tmp:
    with open(os.path.join(tmp, "math.py"), "w") as f:
        f.write('BANNER = "LOCAL-FAKE-MATH 💀"\n')

    sys.modules.pop("math", None)                    # cached-real hatao (pehle-cells ka!) 🧹
    sys.path.insert(0, tmp)
    import math                                        # 💀 abhi FAKE pakda!
    print("shadow-math :", math.BANNER)

    sys.path.pop(0)                                    # cleanup-step-1: path
    del sys.modules["math"]                            # cleanup-step-2: cache!
    import math                                        # 🛡️ ab REAL stdlib!
    print("real-sqrt9 :", math.sqrt(9))
    print("lesson      : apni file kabhi stdlib-naam mat rakho! 🪤")
# Expected OUTPUT:
# shadow-math : LOCAL-FAKE-MATH 💀
# real-sqrt9 : 3.0
# lesson      : apni file kabhi stdlib-naam mat rakho! 🪤
```
