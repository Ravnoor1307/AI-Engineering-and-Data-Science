## 1.1 Timestamps, DatetimeIndex, date_range, to_datetime

`pd.Timestamp('6th jan 2023 8:10')` builds a single moment in time — a **Timestamp** — automatically understanding strings like `'2023-1-5'`, `'5th January 2023'`, or `'2023, 1, 5'`. A `Timestamp` behaves like a superior `datetime` object: it exposes `.year`, `.month`, `.day`, and supports comparison and arithmetic.

A **DatetimeIndex** is a collection of timestamps. Placing one on a Series or DataFrame turns ordinary rows into time-stamped rows, which later enables slicing by date. `pd.date_range(start='2023-1-6', end='2023-1-31', freq='D')` manufactures a whole range of dates in one shot; the `freq` argument controls the spacing (`'D'` daily, `'B'` business days, `'M'` month-end, and so on). Finally, `pd.to_datetime()` converts a column of date-looking strings into real `datetime64` values, after which we can use the `.dt` accessor — for example `pd.to_datetime(s).dt.day_name()` prints the weekday name for every element.

These four tools are the foundation: a Timestamp is one point, a DatetimeIndex is many points as labels, `date_range` fabricates points, and `to_datetime` converts messy text into points. Run the code cell to see each one in action.

````python
# till now
# Timestamp
pd.Timestamp('6th jan 2023 8:10')
# DatetimeIndex -> df and series index
pd.DatetimeIndex([pd.Timestamp('6th jan 2023 8:10'),pd.Timestamp('7th jan 2023 8:10'),pd.Timestamp('8th jan 2023 8:10')])[0]
# date_range()
pd.date_range(start='2023-1-6',end='2023-1-31',freq='D')
# to_datetime()
s = pd.Series(['2023/1/6','2023/1/7','2023/1/7'])
pd.to_datetime(s).dt.day_name()
````

# 2. Timedelta — Duration Between Dates

The second big idea of the module: while a Timestamp answers *'when?'*, a **Timedelta** answers *'how long?'*. It is a duration, not a moment on the calendar.

### Timedelta Object

Represents a duration, the difference between two dates or times.


## 2.1 Constructing and using Timedelta objects

A Timedelta represents a span of time such as `2 days 10:35:00`. There are two natural ways to obtain one. First, **subtract one Timestamp from another**: `t2 - t1` yields the elapsed duration between the two moments, exactly like measuring a journey between a start and an end time. Second, you can build a standalone duration with `pd.Timedelta(days=2, hours=10, minutes=35)`, which is handy when you know the length in advance.

Timedeltas participate in arithmetic. Adding a Timedelta to a Timestamp (`pd.Timestamp('6th jan 2023') + pd.Timedelta(days=2, hours=10, minutes=35)`) shifts the moment forward and returns a new Timestamp; subtracting shifts it backwards. You can also subtract a Timedelta from an entire `DatetimeIndex`, for example a `date_range`, which slides every date by the same amount at once — a fully vectorised calendar operation.

These durations can also live inside data structures: subtract two datetime columns element-wise and you get a Timedelta column, which you will exploit in the next cell's delivery-time problem. When printed, a Timedelta reads naturally as `Timedelta('20 days 01:39:46')`, and you can later extract pieces of it with `.dt.days` or `.dt.total_seconds()`.

````python
# create using Timestamp objects
t1 = pd.Timestamp('6th Jan 2023 8:20:14')
t2 = pd.Timestamp('26th Jan 2023 10:00:00')

t2 - t1
````

# 5. Frequency and asfreq

The spacing between dates — the **frequency** — is a property of any time series, and Pandas provides tools to inspect it and force your data onto a regular grid.

````python
# frequency
google.index
````

## 5.1 Reading the frequency of a DatetimeIndex

Before you can convert or resample a time series, it helps to know the current frequency of the index. `google.index` is a `DatetimeIndex`; dumping it to the output shows each trading date. Notice that the dates are not evenly spaced — weekends and holidays are missing because the stock market is closed. Pandas represents this as a frequency string. A simple daily calendar would be `'D'`, a business-day calendar `'B'`, and here the gaps mean the index is **irregular** with respect to a pure calendar.

The `asfreq()` method then forces the index onto a fixed frequency grid. For example `google.asfreq('6H')` upsamples the daily data to 6-hour intervals, creating new timestamps that have no observed values. Because those new slots are empty, you must tell Pandas how to fill them: `method='bfill'` (back-fill) copies the *next* known value backwards, while `method='ffill'` would copy the *previous* value forward. `asfreq` only rearranges the layout of existing values — it never aggregates or smooths them. That is the key difference from `resample`, which groups values into periods and then applies a reduction like `mean()` or `sum()`. Think of `asfreq` as a ruler that forces your data onto a regular grid, and of the filling method as the rule that decides which neighbour fills each gap.

## 🔬 Deep Dive : Frequency & asfreq

```python
google.index            # check current frequency (eg 'D')
google.asfreq('6H')                     # upsample frequent Higher freq
google.asfreq('6H', method='bfill')     # fill missing by backfill
```

- freq = time spacing between observations
- asfreq — CONVERT index frequency (no aggregation)
- higher freq (D→6H) = upsample — gaps created → filling method needed
- lowest freq (6H→D) = down
- method: bfill (back fill — next), ffill (forward — previous)
- Creates regular grid for missing
- Compare same date across time enabling
- Aligns irregular data to uniform steps
- asfreq vs resample — asfreq layout only; resample does aggregate
- Padding = where values appear duplicated continuous assumption

## 5.2 asfreq — forcing data onto a regular grid

`asfreq('6H', method='bfill')` on the Google frame produces a new DatetimeIndex spaced every six hours — a much higher frequency than the original daily trading dates. This is **upsampling**: you manufacture new timestamps that fall between the observed ones. Of course no price was ever recorded at those intermediate instants, so the values at the new slots must be borrowed from neighbours. `method='bfill'` copies the next known value backwards to fill the gap (also called back-filling), while `method='ffill'` would copy the previous value forwards.

It is important to distinguish `asfreq` from what comes next. `asfreq` only *relabels* the index onto a fixed grid and optionally pads the holes; it never summarizes. Because it does not aggregate, you still see the raw closing prices repeated along each day. If instead you need daily → monthly summaries, or hourly → daily totals, you trade `asfreq` for `resample`, which genuinely groups rows into periods and reduces them. In a pipeline you will often use `asfreq` to build a clean regular calendar (fills required for joins, monthly reporting templates, and comparison grids) and `resample` to derive period-level aggregates.

````python
# asfreq
google.asfreq('6H',method='bfill')
````

# 6. Resampling — Changing Data Frequency

Resampling changes the frequency of the observations themselves, and unlike `asfreq` it always involves an aggregation step for coarser periods.

### Resampling

Resampling involves changing the frequency of your time series observations.

Two types of resampling are:

**Upsampling**: Where you increase the frequency of the samples, such as from minutes to seconds.<br>
**Downsampling**: Where you decrease the frequency of the samples, such as from days to months.

## 📖 Resampling

```python
yt['Subscribers'].resample('Y').mean()    # down — aggregate
google['Close'].resample('12H').interpolate(method='spline', order=2)  # up
```

- **Downsampling** — higher freq → lower (sum/mean)
- **Upsampling** — lower → higher (interpolate)
- `resample('Y')` daily → yearly
- `interpolate` — gaps smooth fill

## 🔬 Deep Dive : Resampling

```python
yt['Subscribers'].resample('Y').mean()      # DOWNSAMPLE → YEAR aggregate
google['Close'].resample('12H')             # UPSAMPLE → 12-hour grid
yt['Subscribers'].resample('M').sum()
# downsampling: plot smoother monthly/yearly trend
# upsampling: missing points via interpolate()
google['Close'].resample('12H').interpolate(method='spline', order=2)
```

- resample = groupby for time — CHANGE FREQUENCY + AGGREGATE
- Downsample (high→low freq): mean/sum/max per period (smooth trend)
- Upsample (low→high freq): new points → interpolate fills
  - linear, spline(order), ffill/bfill
- Rules: D/M/Y/Q/H; 'M' month-end, 'Y' year-end
- Aggregations per period like groupby
- Patterns: quarterly report, daily → weekly sales
- Rolling mean equivalent but period based
- Requires datetime index (sorted)
- Contrast: asfreq layout only; resample + aggregate

## 6.1 Downsampling and upsampling

Think of `resample` as **groupby for time**. It slices the series into consecutive periods (months, years, hours) and then applies a reduction to each slice. **Downsampling** moves from a higher frequency to a lower one — daily to yearly, say — and compresses many rows into one summary: `yt['Subscribers'].resample('Y').mean()` averages all the daily subscriber counts within each calendar year, and `.plot()` on the result overlays a smooth yearly trend on top of the raw daily curve so the long-term direction becomes visible. `'M'` (month-end) and `'Y'` (year-end) are the everyday rules; `'W'`, `'Q'`, and `'H'` extend the menu.

**Upsampling** is the reverse movement — daily to 12-hourly — and it creates more rows than exist, with no observed values at the new points. Those missing values cannot be summed away; they must be manufactured. `google['Close'].resample('12H').interpolate(method='spline', order=2)` fills the newly created gaps with a smooth spline curve fitted through the neighbouring prices. Ordinary `linear`, or a simple forward/backward fill, are the cheaper alternatives. In one sentence: downsampling aggregates *real* data into summaries, while upsampling invents plausible in-between points so that uneven or sparse series can share a common clock with other data.

````python
# Downsampling
yt['Subscribers'].plot()
yt['Subscribers'].resample('Y').mean().plot()
````

````python
# Upsampling
google['Close'].resample('12H').interpolate(method='spline',order=2).plot()
````

# 7. Rolling Window — Smoothing the Signal

Raw time-series data is often disgustingly noisy: a daily subscriber count jumps up and down for no good reason. Rolling windows are the classic cure.

### Rolling Window(Smoothing)
Time series data in original format can be quite volatile, especially on smaller aggregation levels. The concept of rolling, or moving averages is a useful technique for smoothing time series data. 

## 📖 Rolling window (smoothing)

```python
yt['Subscribers'].rolling(30).mean()   # 30-day avg
```

- Keep taking the mean over a window of data
- To smooth a volatile series
- Makes the trend visible


## 🔬 Deep Dive : Rolling window — smoothing

```python
yt['Subscribers'].rolling(30).mean()            # 30-day MOVING average
yt['Subscribers'].rolling(30).mean().plot(title='rolling')
yt['Subscribers'].rolling(30).max() / .std()
```

- rolling(n) — rolling WINDOW of last n observations then agg
- rolling mean = moving average — SMOOTHS noise, shows trend
- Volatile daily → clean mAvg curve
- window= n (rows) OR offset ('30D' time-based)
- agg: mean/max/std/min/sum
- Rolling std — volatility measure
- First n-1 → NaN (small window initials)
- The classic time-series smoothing tool
- vs resample: rolling = moving window; resample = buckets periods

## 7.1 Moving averages and exponential smoothing

`.rolling(30)` slides a window of the last 30 observations across the series, and `.mean()` computes the mean of the values currently inside the window at every step. The result is the 30-day **moving average**: each smoothed point replaces the volatile daily spike with the average of the trailing month, so the underlying trend emerges while transients get ironed out. The first 29 rows are `NaN` because the window has not yet collected enough history — a harmless and expected behaviour.

Because `rolling` returns a windowed object, any reduction works: `.max()`, `.min()`, `.std()` (a rolling volatility), `.sum()`, and more. Windows can also be time-based, e.g. `.rolling('30D')`, when you have irregularly spaced timestamps. The `ewm(30).mean()` line adds a second smoother — the **exponentially weighted moving average** — which weights the most recent observations more heavily than ancient ones, producing a tail that reacts faster and never waits a full window before it starts. Compare the two curves in the plot: rolling gives a clean moving average, `ewm` gives a quickly adapting, memory-weighted average. If `resample` buckets time into fixed periods, `rolling` is a window that *moves* one step at a time.

````python
# Rolling window
yt['Subscribers'].rolling(30).mean().plot(title='rolling')
yt['Subscribers'].ewm(30).mean().plot(title='ewm')
````

# 8. Shift — Creating Lag Features

Shifting does not change your measurement frequency at all; it moves the *values* relative to the rows, which is the foundation of feature engineering and of detecting change.

### Shifting

The shift() function is Pandas is used to, well, shift the entire series up or down by the desired number of periods.

## 📖 Shift

```python
yt['Subscribers'].shift(1)   # one step back
```

## 8.1 shift() and the gap-between-events trick

`shift(n)` slides the entire series up or down by `n` positions, keeping the index untouched. `yt['Subscribers'].shift(1)` places the *previous* day's value in the current row: row at date D now shows what was measured at D−1. Because nothing existed before the first row, that first slot receives `NaN`. A negative shift does the opposite — `shift(-1)` places the *next* day's value in the current row, and the last row becomes `NaN`.

Why care? A shifted column is a **lag feature**: yesterday's subscriber count is often a strong predictor of today's, so lagged columns feed forecasting models. The same machinery solves the login-gap problem. A login log is read with `header=None`, the useful columns are renamed `user_id` and `login_time`, and one user's rows are filtered out. After converting `login_time` to real datetimes, `user_df['shifted'] = user_df['login_time'].shift(1)` lines each login up with the login *before* it. Subtracting the two columns then measures the time *between* consecutive logins, and taking the mean gives the user's average idle gap — a rich behavioural metric built from three simple operations: parse, shift, subtract. Watch for the `SettingWithCopyWarning`; it is only a caution, not an error.
