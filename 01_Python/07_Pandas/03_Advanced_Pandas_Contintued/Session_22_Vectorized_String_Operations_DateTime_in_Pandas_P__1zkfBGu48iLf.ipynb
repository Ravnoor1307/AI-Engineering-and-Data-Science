# Datetime in Pandas — Timestamp, DatetimeIndex, and the .dt Accessor (Session 22)

Almost every real dataset has a date hiding in it: a purchase timestamp, a login time, a flight's departure moment, a day of COVID reporting. Teams quickly discover that treating those values as plain text is a trap — you cannot sort dates that are strings, subtract one moment from another, resample a year into months, or draw a meaningful time axis. This session builds your full time-handling vocabulary in Pandas, one object at a time.

You start with the **Timestamp** — Pandas' description of a single moment in time. Its superpower is parsing: `pd.Timestamp('2023/1/5')`, `pd.Timestamp('2023, 1, 5')`, `pd.Timestamp('2023')`, `pd.Timestamp('5th January 2023 9:21AM')`, and even a `datetime.datetime` object all become the same pluggable timestamp, with attributes like `.year`, `.month`, `.day`, `.hour`, `.minute`, and `.second` ready for the taking. Next you learn *why* Python's built-in `datetime` is not enough. Its API is pleasant, but pure-Python time objects are slow and bulky in bulk; NumPy's `datetime64` encodes dates as compact 64-bit integers and adds one day per unit — `date + np.arange(12)` hands back the next twelve days at once. Pandas takes the convenience of `datetime` and the speed of `datetime64` and merges them into `Timestamp`.

From single moments you move to collections. A **DatetimeIndex** is an array of timestamps good enough to sit on a Series or DataFrame, unlocking time-powered lookup. **`date_range`** generates whole calendars in one call — daily, business-day, weekly, hourly, month-end, year-end — controlled by `freq` codes and either a start/end or a period count. **`to_datetime`** converts existing string columns into proper datetimes (with `errors='coerce'` turning junk into `NaT` instead of crashing). Finally the **`.dt` accessor** extracts derived facts — month names, day names, quarter-start flags — which you exploit by grouping a real expense CSV by weekday and month and plotting the results. # 1. The Timestamp Object

A timestamp references a particular moment in time, e.g. Oct 24th, 2022 at 7:00pm. Everything in date analytics begins at this single point.

### Timestamp Object

Time stamps reference particular moments in time (e.g., Oct 24th, 2022 at 7:00pm)

### Creating Timestamp objects

type(pd.Timestamp('2023/1/5'))       # pandas._libs.tslibs.timestamps.Timestamp
pd.Timestamp('2023-1-5') / '2023, 1, 5' / '2023' / '5th January 2023' / '5th January 2023 9:21AM'

## 1.1 Constructing timestamps from almost any input

`type(pd.Timestamp('2023/1/5'))` returns `pandas._libs.tslibs.timestamps.Timestamp` — a dedicated class, not a plain `datetime`. The constructor is famously forgiving about formats: `'2023-1-5'` and `'2023, 1, 5'` both parse to `Timestamp('2023-01-05 00:00:00')`; `'2023'` alone still yields midnight of January 1st; the texty `'5th January 2023'` works; and `'5th January 2023 9:21AM'` carries the morning hour. The commented-out `'2023/1/5/9/21'` shows the rare input the parser refuses.

You can equally seed a Timestamp from a Python `datetime` object: `dt.datetime(2023, 1, 5, 9, 21, 56)` becomes a Timestamp with the full time preserved. The parsed object then answers attribute questions directly — `x.year` → 2023, `x.month` → 1, `x.day` → 5, and so on through `.hour`, `.minute`, `.second`.

Why choose Timestamp over `datetime`? Three reasons pile up at scale: *performance* (vectorised operations instead of Python loops), *uniform storage* (a compact 64-bit representation), and *ecosystem reach* (timestamps feed directly into DatetimeIndex, resampling, and plotting). Comparing timestamps, converting with `.to_pydatetime()`, and doing calendar arithmetic all come free. A single moment, parsed flexibly, answering component questions — that is the Timestamp.

# 2. Why Python's datetime Is Not Enough — numpy.datetime64

Python already ships a `datetime` module, so why does Pandas bother inventing its own time machinery? The honest answer is performance and memory: convenient loops are slow, but uniform arrays fly.

## 🔬 Deep Dive : numpy datetime64 & why vectorized time

```python
date = np.array('2015-07-04', dtype=np.datetime64)
date + np.arange(12)     # all 12 next days at once!

# uniform dtype → fast arithmetic
# python datetime list would need a loop
```

- np.datetime64 — numpy's time dtype (uniform, fast)
- Addition with ints = days shift vectorized
- Whole arrays shifted together — no python loop
- Reason: performance + uniform memory
- pandas builds on numpy → its own Timestamp layer
- pandas DatetimeIndex ops similarly vectorized
- Yearly/monthly/day arithmetic natural
- 64-bit int backing — compact memory
- This is performance foundation of DateTime in pandas

````python
# why separate objects to handle data and time when python already has datetime functionality?
````

- syntax wise datetime is very convenient
- But the performance takes a hit while working with huge data. List vs Numpy Array
- The weaknesses of Python's datetime format inspired the NumPy team to add a set of native time series data type to NumPy.
- The datetime64 dtype encodes dates as 64-bit integers, and thus allows arrays of dates to be represented very compactly.

## 2.1 datetime64 — dates as compact integers

The starting provocation is `date = np.array('2015-07-04', dtype=np.datetime64)`. With the `D` unit this single-cell array stores a date as one 64-bit integer — how many days since the epoch — which is why `datetime64` arrays represent dates so compactly and why arithmetic is so fast: adding `np.arange(12)` to the date performs a vectorised integer addition, returning the twelve following days in a single expression.

The bullet-point cells spell the argument out. Python's `datetime` is *syntax-friendly* but performance takes a hit on large data — a Python list of datetime objects stores pointers and objects, and any operation must call methods per element. NumPy's native `datetime64` dtype encodes dates uniformly, so whole arrays shift at C speed. Pandas then fuses the two worlds in its own `Timestamp` class: the ease of Python's `datetime` plus the efficient storage and vectorised interface of `numpy.datetime64`. From a group of these Timestamps, Pandas builds a **DatetimeIndex**, which the next section shows being used to index a Series or DataFrame. Remember the one-line mental model: `datetime` is ergonomic, `datetime64` is fast, `Timestamp` is both.

````python
import numpy as np
date = np.array('2015-07-04', dtype=np.datetime64)
date
````

````python
date + np.arange(12)
````

- Because of the uniform type in NumPy datetime64 arrays, this type of operation can be accomplished much more quickly than if we were working directly with Python's datetime objects, especially as arrays get large 

- Pandas Timestamp object combines the ease-of-use of python datetime with the efficient storage and vectorized interface of numpy.datetime64

- From a group of these Timestamp objects, Pandas can construct a DatetimeIndex that can be used to index data in a Series or DataFrame

# 3. DatetimeIndex — a Collection of Timestamps

Take the Timestamp idea and multiply it: a DatetimeIndex is simply an ordered collection of Pandas timestamps, ready to serve as row labels.

### DatetimeIndex Object

A collection of pandas timestamp

## 🔬 Deep Dive : DatetimeIndex

```python
pd.DatetimeIndex(['2023/1/1','2022/1/1','2021/1/1'])
pd.DatetimeIndex([dt.datetime(2023,1,1), dt.datetime(2020,1,1)])
pd.Series([1,2,3], index=pd.DatetimeIndex([pd.Timestamp(2023,1,1), ...]))
```

- DatetimeIndex = collection of Timestamps → used as INDEX
- Built from strings / datetime objects / Timestamps
- Series indexed by dates → time-indexed data
- Enables time-slicing, resampling, plotting with dates on x-axis
- can infer index from parseable column
- date_sorting matters (chronological indexing)
- `.index` on time-indexed df gives DatetimeIndex
- Advance slicing: yt.loc['2022-12'] partial-indexing works
- Time-based lookup instant after sort

## 3.1 Building and using a datetime index

Construct one from strings — `pd.DatetimeIndex(['2023/1/1','2022/1/1','2021/1/1'])` — from Python `datetime` objects, or from Timestamps, and the result is a `pandas.core.indexes.datetimes.DatetimeIndex`. Crucially the dates need not be sorted the moment you build them; Pandas keeps them in your order until you sort explicitly.

The payoff appears when the DatetimeIndex becomes the *index* of a Series: `pd.Series([1, 2, 3], index=dt_index)` labels each value with a date. A time-indexed Series or DataFrame is what unlocks the whole feature set used later in this module — slicing by year/month (`yt.loc['2022']`), resampling, and automatic date axes on plots. Because the index is internally ordered and hashable, lookups are fast and unambiguous, and the `.index` of any time-indexed frame is itself a DatetimeIndex.

Practical notes from these cells: pass a list of strings, datetime objects, or Timestamps interchangeably; store the index in a variable (`dt_index`) so you can reuse it; and remember the pattern for the real world — read a CSV, parse the date column with `to_datetime`, then `set_index` so time becomes the axis. Ordering and clean dtypes make every later time operation possible.

````python
# from strings
type(pd.DatetimeIndex(['2023/1/1','2022/1/1','2021/1/1']))
````

````python
# using python datetime object
pd.DatetimeIndex([dt.datetime(2023,1,1),dt.datetime(2022,1,1),dt.datetime(2021,1,1)])
````

````python
# using pd.timestamps
dt_index = pd.DatetimeIndex([pd.Timestamp(2023,1,1),pd.Timestamp(2022,1,1),pd.Timestamp(2021,1,1)])
````

````python
# using datatimeindex as series index

pd.Series([1,2,3],index=dt_index)
````

# 4. date_range — Generating Date Sequences

Rather than typing dates by hand, ask Pandas to produce a whole calendar for you with `pd.date_range(...)`.

### date_range function



## 🔬 Deep Dive : date_range — generate date sequences

```python
pd.date_range(start='2023/1/5', end='2023/2/28')
pd.date_range(start='2023/1/5', end='2023/2/28', freq='B')    # business days
pd.date_range(start='2023/1/5', end='2023/2/28', freq='H')    # hourly
pd.date_range(start='2023/1/5', end='2023/2/28', freq='MS')   # month start
pd.date_range(start='2023/1/5', periods=25)                   # by count
```

- freq codes:
  - D daily, B business, W weekly, H hourly
  - M month-end, MS month-start, A year-end
  - combos: '2D','3H','W-MON','BM'
- start/end OR periods — both ways
- Creates DatetimeIndex — template for data gaps
- Calendar-aware calendars (business days skip weekends)
- Filling experiment dates
- Common: daily series a month, sub-hourly
- resample/asfreq rely on freq logic
- ISO calendar/weekday support — freq detailed
- Creating dates without data — simulation, joins

## 4.1 Frequencies you should know

`pd.date_range(start='2023/1/5', end='2023/2/28', freq='3D')` walks from start to end in three-day steps, returning a DatetimeIndex. Change the `freq` and the rhythm of the walk changes completely:
- `'D'` — every calendar day; `'3D'` — every third day.
- `'B'` — business days only, skipping weekends (Jan 6→Jan 9 in the sample output).
- `'W-THU'` — once per week on the chosen weekday (Thursday in the example).
- `'6H'` — every six hours, stamping each day at 00:00, 06:00, 12:00, 18:00.
- `'M'` — month-end dates (2023-01-31, 2023-02-28); `'MS'` — month *starts*.
- `'A'` — year-end (2023-12-31, 2024-12-31), the annual rule `A-DEC`.

Instead of an end date you can specify `periods`: `pd.date_range(start='2023/1/5', periods=25, freq='M')` generates exactly 25 month-ends, useful when you want a template of fixed length regardless of calendar bounds.

Where do people use this? Generating experiment dates, expanding a sparse set of observed days into a full reference calendar, simulating time series, and constructing the index grids that `resample`/`asfreq` rely on. `freq='B'` already answers 'what dates are working days?' and `freq` strings compose with numbers, so `'2W-MON'` and `'BM'` stretch the vocabulary further.

````python
# generate daily dates in a given range
pd.date_range(start='2023/1/5',end='2023/2/28',freq='3D')
````

````python
# MS -> Month start
pd.date_range(start='2023/1/5',end='2023/2/28',freq='MS')
````

````python
# A -> Year end
pd.date_range(start='2023/1/5',end='2030/2/28',freq='A')
````

# 5. to_datetime — Converting String Columns

`date_range` *creates* dates; `to_datetime` *rescues* dates already sitting in your data as strings.

### to_datetime function

converts an existing objects to pandas timestamp/datetimeindex object

## 🔬 Deep Dive : to_datetime — string → Timestamp column

```python
s = pd.Series(['2023/1/1','2022/1/1','2021/1/1'])
pd.to_datetime(s)                          # parse whole column
pd.to_datetime(s, errors='coerce')         # bad dates → NaT (not crash)
```

- to_datetime — CONVERT existing series/df of strings to datetime
- errors='coerce' → unparseable become NaT (missing)
- errors='raise' default — crash on bad
- apply to df col: df['Date'] = pd.to_datetime(df['Date'])
- format= arg for custom formats (speed + correctness)
- yearfirst/dayfirst for ambiguous dates
- After conversion: dtypes datetime64[ns], info shows
- Then .dt accessor, plotting, resampling available
- Weekly pipeline: read CSV → to_datetime → index → analyze
- Datetime col = 80% of real datasets 'date' column

## 5.1 Parsing whole columns safely

`pd.to_datetime(s)` takes a Series of date-like strings — `['2023/1/1', '2022/1/1', '2021/1/1']` — and returns a datetime series, and because it is now a real datetime you can immediately hang the `.dt` accessor off it: `.dt.day_name()` prints `Sunday`, `Saturday`, `Friday`. No loop, no exceptions.

The killer feature for dirty data is `errors='coerce'`. Feed it `['2023/1/1', '2022/1/1', '2021/130/1']` — note the impossible day 130 — and instead of raising, Pandas marks that element as `NaT` (Pandas' missing timestamp), letting the rest of the column parse. The default `errors='raise'` would abort the program. This mirrors `pd.to_numeric(..., errors='coerce')`: one call to sanitise a messy column into a professional dtype.

The expense notebook then applies the real-world recipe: `df['Date'] = pd.to_datetime(df['Date'])` upgrades the `Date` column's dtype in place, and `df.info()` confirms `datetime64[ns]` in the Dtype column. From that moment everything becomes possible: `.dt` extractions (next section), resampling, plotting with time axes, and chronological sorting. If a dataset has a 'date' column, roughly 80% of the time its first cleaning step is exactly this line.

````python
# simple series example

s = pd.Series(['2023/1/1','2022/1/1','2021/1/1'])
pd.to_datetime(s).dt.day_name()
````

````python
# with errors
s = pd.Series(['2023/1/1','2022/1/1','2021/130/1'])
pd.to_datetime(s,errors='coerce').dt.month_name()
````

# 6. The .dt Accessor

A Series of datetimes has its own accessor, `.dt`, just as strings have `.str`. It broadcasts date-property extraction across every row.

### dt accessor

Accessor object for datetimelike properties of the Series values.

## 6.1 Extracting date parts and plotting time series

The first cell shows boolean flags straight off the accessor: `df['Date'].dt.is_quarter_start` returns a Series of `True`/`False` telling you which days open a quarter. The same family — `is_month_start`, `is_month_end`, `is_quarter_end`, plus `.weekday`, `.quarter`, `.hour` — turns the calendar into filterable conditions.

Plotting a raw time series is as simple as `plt.plot(df['Date'], df['INR'])`: matplotlib accepts the datetime column directly and spaces the x-axis chronologically. That alone beats the scramble of string axes.

The grouping cells then build the real dashboard. `df['day_name'] = df['Date'].dt.day_name()` labels every expense with its weekday name (`Monday`...), and `df.groupby('day_name')['INR'].mean().plot(kind='bar')` shows you the average spend per weekday — immediately revealing, say, that weekends cost more. Mirror it at monthly granularity: `df['month_name'] = df['Date'].dt.month_name()` followed by `df.groupby('month_name')['INR'].sum().plot(kind='bar')` summarises where the month's money goes. Filtering closes the loop: `df[df['Date'].dt.is_month_end]` returns only transactions recorded on the last day of a month — handy for payroll, subscription-lapse, or bill analysis. Pattern to internalise: parse → extract labels with `.dt` → `groupby` a derived label → plot the aggregate.

````python
df['Date'].dt.is_quarter_start
````

````python
# plot graph
import matplotlib.pyplot as plt
plt.plot(df['Date'],df['INR'])
````
