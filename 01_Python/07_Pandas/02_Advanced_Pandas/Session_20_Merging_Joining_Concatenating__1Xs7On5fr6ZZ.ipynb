The mental model to keep for the whole session: merge joins rows by keys, concat stacks tables, and the two combine into pipelines — merge when you need information from another column, concat when you need more rows or columns. # 2. concat — Stacking Tables

## 2.1 Pure assembly, no matching

```python
regs = pd.concat([nov, dec], ignore_index=True)
```

concat is the assembler: it stacks DataFrames and does NOT try to match anything. With `axis=0` (the default), the rows of dec sit underneath the rows of nov — registration month 1, then month 2 — making one combined `regs` frame of every enrolment of both months. `ignore_index=True` renumbers the result 0, 1, 2, ... instead of keeping two sets of repeated indices; without it, the concat result would repeat the original row numbers for each block, which breaks filtering and debugging almost instantly.

There is a legacy alias, `nov.append(dec, ignore_index=True)`, that behaves identically — and is deprecated. Modern pandas prefers pd.concat. Both appear in the notebook so that you recognise old code when you meet it.

The default concat is an OUTER union: if the two frames have different columns, the result contains ALL columns from both, and NaN fills the cells where one frame had no such column. concat is the tool for stacking month files, appending year folders, or widening across columns (axis=1) — but it never, ever joins on keys.

## 2.2 Deep dive — keys and horizontal stacking

concat has two extra superpowers worth seeing in the outputs.

```python
multi = pd.concat([nov, dec], keys=['Nov', 'Dec'])
multi.loc[('Dec', 4)]
```

The `keys=` argument labels each block with a name. The result `multi` becomes a MultiIndex DataFrame whose top level is the source ('Nov' or 'Dec') — so the origin of every row is recorded, not lost. `multi.loc[('Dec', 4)]` uses the tuple ('Dec', 4) to fetch the rows of the December block that came from original index 4: student 14, course 9. If you later need the source as a real column instead of an index level, `reset_index()` promotes it.

```python
pd.concat([nov, dec], axis=1)
```

`axis=1` stacks SIDEWAYS — columns, not rows. The result is a wide frame with student_id and course_id twice (once from nov, once from dec), aligned by original position. The printed headers repeat: `student_id course_id student_id course_id`. Different lengths or orderings produce NaN padding on the shorter frame, so horizontal concat is normally used only with compatible row counts.

The rule to remember: concat gives you MORE ROWS or MORE COLUMNS, never more information per row. For that you need merge — the next section.

# 3. merge — SQL-Style Joins

## 3.1 Joining rows on a key

merge is the relational JOIN: it combines two tables by matching a shared column, bringing columns from both frames into a single result.

```python
students.merge(regs, how='inner', on='student_id')
courses.merge(regs, how='left', on='course_id')
```

The grammar is `left_frame.merge(right_frame, how=..., on=...)`. `on` names the join key — the column that must exist in both frames. `how` chooses the join type:

- `inner`: only keys present in BOTH tables (the common core).
- `left`: every row of the LEFT frame, plus matches from the right; unmatched right columns become NaN.
- `right`: every row of the RIGHT frame, plus matches from the left.
- `outer`: the union of everything, NaN where the other side has no match.

So `students.merge(regs, how='inner', on='student_id')` joins every registration to its student record — the result adds name and partner columns to each registration row, keeping only students who enrolled. `courses.merge(regs, how='left', on='course_id')` keeps ALL courses — including those with zero registrations; registration columns fall out as NaN for the empty ones.

If the two keys have different names, use `left_on` and `right_on` instead of `on`. And mind many-to-many joins: if a key repeats in both tables, the result explodes into every combination (a cartesian product) — a classic source of over-counted rows.

> Mental hook: choose the base = left frame. 'Keep everything from courses' is a left join with courses on the left.

## 3.2 Deep dive — four joins, four shapes

Read the notebook's four merge outputs and their shapes tell you everything.

The INNER join, `students.merge(regs, how='inner', on='student_id')`, shows student_id, name, partner, course_id. Only the students who actually registered appear, each repeated once per course: Kailash Harjo and his registrations, then Esha Butala, and so on.

The LEFT join swaps perspective: `courses.merge(regs, how='left', on='course_id')` prints every course with its registrations. Python (2499) has many rows because many students registered for it; courses nobody took appear once with NaN in the student_id column — the left join is the classic 'keep the master list' pattern.

The RIGHT join is the mirror image. The notebook first ADDS students 26, 27, and 28 (Nitish, Ankit, Rahul) by concatenating a small frame onto `students`, then runs `students.merge(regs, how='right', on='student_id')`. Because the right frame regs drives the output, every registration appears, and the students who registered for nothing show up with NaN course_id. `regs.merge(students, how='left', on='student_id')` produces the identical result with the frames swapped — always possible, and a good sanity check that your join direction is right.

The OUTER join, `students.merge(regs, how='outer', on='student_id').tail(10)`, shows the union: the bottom rows carry some students with NaN course columns where the other side had no match.

Each join answers a different question. Inner: who is linked? Left/right: keep one side complete. Outer: keep everything, gaps and all.

# 4. Merge + GroupBy Patterns

## 4.1 Revenue from two tables

Merging is not the goal — answering questions is. The session's middle section now stitches merge onto the grouping skills you already have, in the classic pattern merge -> groupby -> sum.

```python
regs.merge(courses, how='inner', on='course_id')['price'].sum()
```

This is revenue in one line. First merge the registrations with the course table so each enrolment gains its course's price; then `['price'].sum()` totals the money earned — 154,247 in the notebook. The merge is what created the price column to begin with; no single input file contained it.

```python
temp_df = pd.concat([nov, dec], keys=['Nov', 'Dec']).reset_index()
temp_df.merge(courses, on='course_id').groupby('level_0')['price'].sum()
```

Month-by-month revenue merges the two tricks. The concat with keys=['Nov', 'Dec'] records the origin the moment the files are combined; reset_index turns the MultiIndex level into a plain column (named level_0); and the merge attaches prices before `groupby('level_0')['price'].sum()` totals November (89,175) and December (65,072) separately. The combine-first, merge-second, group-third order is deliberate — every step adds exactly the information the next step needs.

## 4.2 Deep dive — the join problem set

The next cells run a small test suite of join questions. Trace each one and the pattern becomes automatic.

```python
regs.merge(students, on='student_id').merge(courses, on='course_id')[['name', 'course_name', 'price']]
```

The three-table chain. Start from registrations, join students for the names, join courses for the names and prices, then keep the three report columns — the entire customer bill printed as one table. The key column survives every join so the chain can keep going.

```python
common_student_id = np.intersect1d(nov['student_id'], dec['student_id'])
students[students['student_id'].isin(common_student_id)]
```

'Who enrolled in both months?' is a set-intersection question that needs no merge at all: `np.intersect1d` returns the IDs present in both files, and `isin` filters the students table down to them.

```python
course_id_list = np.setdiff1d(courses['course_id'], regs['course_id'])
courses[courses['course_id'].isin(course_id_list)]
```

'Which courses got zero enrolments?' is the anti-join: `setdiff1d` (or the pandas version `~courses['course_id'].isin(regs['course_id'])`) finds IDs in courses but not in registrations — the answer: Numpy and C++.

```python
students.merge(students, how='inner', left_on='partner', right_on='student_id')[['name_x', 'name_y']]
```

The SELF-JOIN: merging students with itself pairs each student's partner ID with the partner's name — name_x is the student, name_y their partner. The merged columns carry the _x / _y suffixes, which is how pandas distinguishes the two 'name' columns.

The top-N cells finish the batch: group registrations by (student_id, name) and count for 'most enrolments', or join prices and sum for 'most money spent' — Chhavi Lachman tops both lists.

> Tip: revenue and ranking questions are almost always merge-then-group; membership questions (who/what is IN a set) are setdiff1d / intersect1d or isin.

# 5. merge — Syntax Variants

## 5.1 Two spellings, one join

merge can be called either on a DataFrame or on the pandas module, and the two are interchangeable.

```python
students.merge(regs, how='inner', on='student_id')
pd.merge(students, regs, how='inner', on='student_id')
```

`df.merge(other, ...)` and `pd.merge(df, other, ...)` compute the identical result; the method form reads more naturally in a chain, while the module form is useful when you are building the join from variables. Newer pandas also offers `df.join()` for index-based merges and the left_on / right_on spelling for mismatched key names, as the IPL case study will use.

The important vocabulary you must not lose: `how` (inner/left/right/outer), `on` (shared key), the `_x` / `_y` suffixes when both frames bring the same column name, and the cardinality warning — when keys repeat, rows multiply. Merging one student record against five registrations yields five rows for that student; that is expected for a many-to-one join, and it is exactly why the notebooks' top-N groupby counts registered users correctly.

Before moving on, re-run the alternate-syntax cell and confirm the output is byte-identical to the earlier inner join. Same tool, same result, two spellings — keep whichever you find clearer.

# 6. Case Study — IPL Merges

## 6.1 Two files become one

The session closes by merging two real cricket files whose key columns have DIFFERENT names — the situation `left_on` and `right_on` were made for.

```python
temp_df = delivery.merge(matches, left_on='match_id', right_on='id')
```

delivery's per-ball table identifies its game by `match_id`; matches' summary table calls the same thing `id`. Since the names differ, `on` cannot be used — the merge declares both: `left_on='match_id'` in the deliveries frame, `right_on='id'` in the matches frame. Every ball then gains its match's columns: season, city, venue, teams, winner.

The merged `temp_df` now answers questions neither file could answer alone. Stadium sixes ratio: count sixes per venue from the six-filtered merged frame (`six_df.groupby('venue')['venue'].count()`) and divide by the matches-per-venue counts (`matches['venue'].value_counts()`) — the ratio gives the six-happiest grounds, Holkar Stadium on top at 17.6 sixes per match. Per-season orange cap: `temp_df.groupby(['season', 'batsman'])['batsman_runs'].sum()` — one groupby, both files' information.

> Tip: always check that key dtypes match before merging — a string '12' and an int 12 will silently fail to join.

## 6.2 Deep dive — reading the merge pipeline

Walk the cells in the order the notebook runs them.

`matches` and `delivery` are printed raw first, so you can stare at the two key columns — `id` against `match_id` — and at the venue/season information you are about to gain per ball. Then the one-line merge creates `temp_df`, the 179k-row working frame; and `six_df = temp_df[temp_df['batsman_runs'] == 6]` keeps only the sixes.

The stadium ratio joins two countings: `num_sixes = six_df.groupby('venue')['venue'].count()` counts sixes per ground, `num_matches = matches['venue'].value_counts()` counts games per ground, and `(num_sixes / num_matches).sort_values(ascending=False).head(10)` ranks the ratio. Aligning the two Series works because both are indexed by venue — the same label-alignment magic you met with Series arithmetic. Holkar Cricket Stadium's 17.6 means its matches average more than 17 sixes, an extreme ground for hitting.

The final block computes the orange cap for every season:

```python
temp_df.groupby(['season', 'batsman'])['batsman_runs'].sum().reset_index() \
       .sort_values('batsman_runs', ascending=False) \
       .drop_duplicates(subset=['season'], keep='first') \
       .sort_values('season')
```

Group by (season, batsman) and sum the runs; reset_index flattens the MultiIndex; sort runs descending so each season's best is on top; drop_duplicates(season, keep='first') keeps exactly that best; and the final sort orders the winners by season — SE Marsh 616 in 2008, ML Hayden 572 in 2009, SR Tendulkar 618 in 2010, CH Gayle 2011-2012 with 608 and 733, and so on. A single pipeline, built from groupby + reset + drop_duplicates + sort, full of both files' data.
