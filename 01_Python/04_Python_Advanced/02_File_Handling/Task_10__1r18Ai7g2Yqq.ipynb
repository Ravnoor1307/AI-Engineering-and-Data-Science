## Task 1 — The Final Line of a File

Write a function `get_final_line(filename)` that returns the last line of the file. The function takes the filename, reads the file, and returns that final line.

**Approach** — you have two valid strategies:

1. *Read everything, then index.*
   ```python
   with open(filename, 'r') as f:
       lines = f.readlines()
   return lines[-1]
   ```
   Simple, but the whole file is loaded into memory.

2. *Stream and remember.*
   ```python
   last = ''
   with open(filename, 'r') as f:
       for line in f:
           last = line
   return last
   ```
   The variable `last` is overwritten on every iteration, so when the loop ends it holds the final line — and memory stays flat no matter how large the file grows.

**Watch out** — the returned line usually carries its trailing `'\n'`; decide whether the caller expects it kept or stripped. Also make sure the filename you test with actually exists, or the `open()` call raises `FileNotFoundError`. The real skill here is choosing the right reading tool for the job.

## Task 6 — String Length With Recursion

Return the length of a string without using `len()`: `'abcd'` must yield 4 and `'DataScience'` must yield 11.

**Approach**

- Base case: an empty string has length 0.
- Recursive step: one character, plus the length of the tail.

```python
def string_length(s):
    if s == '':
        return 0
    return 1 + string_length(s[1:])
```

**Hints**

- `s[1:]` is the string minus its first character, so every call shrinks `s` by one and the recursion always reaches the empty-string base case.
- Trace `'ab'`: `1 + length('b')` `→ 1 + 1 + length('')` `→ 1 + 1 + 0 = 2`.
- Forget the base case and you are stuck in infinite recursion — `RecursionError`.

## Task 7 — gcd() Without Loops

Write a function `gcd(a, b)` that returns the greatest common divisor of two numbers — recursion only, no loops. `gcd(16, 24)` should give 8.

**Approach** — the Euclidean algorithm is naturally recursive:

- Base case: if `b == 0`, the gcd is `a`.
- Otherwise the gcd of `a` and `b` equals the gcd of `b` and `a % b`.

```python
def gcd(a, b):
    if b == 0:
        return a
    return gcd(b, a % b)
```

**Hints**

- Trace `gcd(16, 24)`: `→ gcd(24, 16) → gcd(16, 8) → gcd(8, 0) → 8`. The numbers shrink fast, so the recursion stays shallow.
- `a % b` is the remainder — the heart of the algorithm.
- The task bans loops, not recursion: recursion is doing what a what-loop would otherwise do, but this is precisely its elegant niche.

*The edit distance between two strings is a measure of their similarity—the smaller the edit distance, the more similar the strings are with regard to the minimum number of insert, delete and substitute operations needed to transform one string into the other.*

Consider the strings `kitten` and `sitting`. The first string can be transformed
into the second string with the following operations:

This is the smallest number of operations that can be performed to transform kitten into sitting. As a result, the edit distance is `3`.


Write a recursive function that computes the edit distance between two strings.

Use the following algorithm:

```
Let s and t be the strings
    If the length of s is 0 then

    Else

        If the last character in s does not equal the last character in t then
            Set cost to 1
        Set d1 equal to the edit distance between all characters except the last one in s, and all characters in t, plus 1
```

## Task 8 — Edit Distance By Recursion

Edit distance is the minimum number of insert, delete and substitute operations that transforms one string into another. `kitten` → `sitting` needs 3 operations: substitute `k`→`s`, substitute `e`→`i`, insert `g`.

**Approach** — implement the given algorithm literally:

- If `len(s) == 0`, return `len(t)` (insert everything).
- If `len(t) == 0`, return `len(s)` (delete everything).
- Otherwise set `cost = 0` if the last characters match, else `cost = 1`.
- `d1 = edit_distance(s[:-1], t) + 1`   (delete the last char of `s`)
- `d2 = edit_distance(s, t[:-1]) + 1`   (insert the last char of `t`)
- `d3 = edit_distance(s[:-1], t[:-1]) + cost`  (substitute, or match for free)
- Return `min(d1, d2, d3)`.

**Hints**

- `s[:-1]` drops one character from the end; every level compares the two *last* characters.
- A matching pair costs 0, so the recursion naturally prefers keeping aligned equal characters.
- This naive version re-explores the same subproblems repeatedly — count the calls for `('kitten', 'sitting')` to appreciate why memoization exists.

###`Q-9:` Run-Length Encoding

Run-length encoding is a simple data compression technique that can be effective when repeated values occur at adjacent positions within a list. Compression is achieved by replacing groups of repeated values with one copy of the value, followed by the number of times that the value should be repeated. For example, the list 
```
["A", "A", "A", "A", "A", "A", "A", "A", "A", "A", "A", "A", "B", "B", "B", "B", "A", "A", "A", "A", "A", "A", "B"]
```
would be compressed as `["A", 12, "B", 4, "A", 6, "B", 1]`. 

Write a recursive function that implements the run-length compression technique
described above. Your function will take a list or a string as its only parameter. It should return the run-length compressed list as its only result. Include a main program that reads a string from the user, compresses it, and displays the run-length encoded result.

## Task 9 — Run-Length Encoding

Compress a list or string by replacing each run of identical adjacent values with one copy of the value followed by its run length. Twelve `A`s plus four `B`s plus six `A`s plus one `B` compress to `['A', 12, 'B', 4, 'A', 6, 'B', 1]`.

**Approach — recursive thinking**

- Base case: an empty input → empty output; a single character → `[char, 1]`.
- Recursive step: count how many times the first value repeats consecutively, emit `[value, run]`, and recurse on the remainder.

```python
def rle(data):
    if not data:
        return []
    run = 1
    while run < len(data) and data[run] == data[0]:
        run += 1
    return [data[0], run] + rle(data[run:])
```

**Hints**

- The opening run of 12 identical `'A'`s becomes `['A', 12]` — one entry, not twelve.
- The input can be a list *or* a string; indexing and slicing behave identically for both.
- Values are compared exactly, so `'A'` and `'a'` are distinct runs.
- A while loop inside the function is fine — the recursion is what drives the repeated run-splitting.

## Task 10 — Decimal To Binary Recursively

Convert a decimal integer into its binary representation using recursion.

**Approach**

- Base case: `n == 0` → the empty string (or `'0'` if you must print 0 itself).
- Recursive step: the final binary digit is `n % 2`; everything before it is `dec_to_bin(n // 2)`.

```python
def dec_to_bin(n):
    if n == 0:
        return ''
    return dec_to_bin(n // 2) + str(n % 2)
```

**Hints**

- Trace 5: `dec_to_bin(5)` `→ dec_to_bin(2) + '1'` `→ (dec_to_bin(1) + '0') + '1'` `→ (('' + '1') + '0') + '1'` `← '101'`, and 5 *is* 101 in binary.
- Repeated division by 2 is the same repeated-division method you use on paper; the recursion unwinds in the correct order because the string is built on the way back up.
- Guard the base case so `n` eventually reaches 0 — otherwise infinite recursion.
