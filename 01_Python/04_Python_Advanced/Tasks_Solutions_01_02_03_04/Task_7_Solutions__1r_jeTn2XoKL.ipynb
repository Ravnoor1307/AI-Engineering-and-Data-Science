### Q1 — a Rectangle with private helpers and one public `display`

The class stores its dimensions privately (`self.__length`, `self.__width`). Perimeter and area are written as *private* methods whose names start with two underscores — `__perimeter()` and `__area()` — because they are internal helpers that consumers should not call directly; the single public interface is `display()`, which prints all four facts:

```python
def display(self):
    print('length', self.__length)
    print('width', self.__width)
    print('perimeter', self.__perimeter())
    print('area', self.__area())
```

With `Rectangle(4, 5)`: length 4, width 5, perimeter `2*(4+5) = 18`, area `4*5 = 20`. Three points make this solution worth studying. First, **name mangling**: `self.__perimeter` inside the class actually resolves to `_Rectangle__perimeter`, so a stray `obj.__perimeter()` from outside would fail — the class controls who may use its internals. Second, private helper methods can be freely called from *within* the class, which is exactly why `display()` can compose them. Third, the constructor uses `l` and `b` only to initialise the private fields, and no public getter is exposed here — the only way to inspect the rectangle is through `display()`, the intended public face of the class.

````python
# write your code here

class Rectangle:

    def __init__(self,l,b):
        self.__length = l
        self.__width = b

    def __perimeter(self):
        return 2*(self.__length + self.__width)

    def __area(self):
        return self.__length * self.__width

    def display(self):
        print('length',self.__length)
        print('width',self.__width)
        print('perimeter',self.__perimeter())
        print('area',self.__area())

obj = Rectangle(4,5)
obj.display()
````

### Q2 — a Bank account: deposits, withdrawals, and hidden fees

The `Bank` class keeps `name`, `acc_no`, and `balance` all private. The public methods `deposit` and `withdrawl` are the *only* way to change the balance, which is the essence of encapsulation: outside code cannot just write `cust.__balance = 999999` (mangling prevents it), so all money movement is auditable through the class.

The withdrawal method is where the logic lives:

```python
def withdrawl(self, amount):
    if amount > self.__balance:
        print('Insufficient funds')
    else:
        self.__balance = self.__balance - amount
        reduction = self.__bankFees()
        self.__balance = self.__balance - reduction
```

It guards against overdraft, subtracts the requested amount, then applies the bank fee. The fee is computed by the private method `__bankFees()` returning `0.05 * self.__balance` — five percent of the balance *after* withdrawal. In the sample, `cust` starts at 500, deposits 500 (→ 1000), withdraws 100 (→ 900), is charged 5% of 900 = 45 (→ 855), giving the printed balance `855.0`. Notice the `₹` symbol in the display, which the output preserves. This question shows why fees should be a captive method: the calculation is buried inside the account logic and cannot be misapplied from outside.

````python
# write your code here

class Bank:

    def __init__(self,name,acc_no,balance):
        self.__name = name
        self.__acc_no = acc_no
        self.__balance = balance

    def display(self):
        print('Account Number :',self.__acc_no)
        print('Account Name :',self.__name)
        print('Account Balance :',self.__balance,'₹')

    def deposit(self,amount):
        self.__balance = self.__balance + amount

    def withdrawl(self,amount):
        if amount>self.__balance:
            print('Insufficient funds')
        else:
            self.__balance = self.__balance - amount
            reduction = self.__bankFees()
            self.__balance = self.__balance - reduction
            
    def __bankFees(self):
        return 0.05*self.__balance

cust = Bank('nitish',1000000,500)
cust.deposit(500)
cust.withdrawl(100)
cust.display()
````

### Q3 — a Computation utility class, pure functions in one place

This is the "calculator service" problem. The `Computation` class has a default constructor that does nothing (`pass`), because it needs no state — it is a class that exists purely to group a family of functions. The methods covered are:

- `factorial(n)` — multiply `1 · 2 · … · n` in a loop.
- `naturalSum(n)` — accumulate `1 + 2 + … + n` (note: the solution starts its accumulator at 1, which happens to give the correct sum only for these inputs, so recheck it).
- `testPrime(n)` — count divisors of `n`; it is prime iff exactly two divisors (1 and itself) are found.
- `testPrims(n, m)` — count divisors common to both numbers; exactly one shared divisor (1) means the two are **co-prime**.
- `tableMult(k)` and `allTables()` — print the 1–9 multiplication tables.
- `listDivisor(n)` and `listPrimeDivisor(n)` — collect all divisors, then only the prime ones (reusing `testPrime` internally).

The instantiation at the bottom calls several of them: `13` and `7` are co-prime, 18's divisors are `[1, 2, 3, 6, 9, 18]`, and its prime divisors are `[2, 3]`. Two design lessons stand out: private helper-like reuse (`listPrimeDivisor` calling `testPrime`) keeps logic in one place, and a class with no state is a tidy way to namespace many related functions.

````python
# write your code here
#Write your code here

class Computation:
    def __init__ (self):
        pass
# --- Factorial ------------
    def factorial(self, n):
        j = 1
        for i in range (1, n + 1):
            j = j * i
        return j
    
# --- Sum of the first n numbers ----
    def naturalSum(self, n):
        j = 1
        for i in range (1, n + 1):
            j = j + i
        return j
    
# --- Primality test of a number ------------
    def testPrime(self, n):
        j = 0
        for i in range (1, n + 1):
            if (n% i == 0):
                j = j + 1
        if (j == 2):
            return True
        else:
            return False

# --- Primality test of two integers ------------
    def testPrims(self, n, m):
        
        # initialize the number of commons divisors
        commonDiv = 0
        for i in range (1, n + 1):
            if (n% i == 0 and m% i == 0):
                commonDiv = commonDiv + 1
        if commonDiv == 1:
            print ("The numbers", n, "and", m, "are co-primes")
        else:
            print ("The numbers", n, "and", m, "are not co-primes")

#---Multiplication table-------------
    def tableMult(self, k):
        for i in range (1,10):
            print (i, "x", k, "=", i * k)

# --- All multiplication tables of the numbers 1, 2, .., 9
    def allTables(self):
        for k in range (1,10):
            print ("\nthe multiplication table of:", k, "is:")
            for i in range (1,10):
                print (i, "x", k, "=", i * k)

# ----- list of divisors of an integer
    def listDivisor(self, n):
        # initialization of the list of divisors
        lDiv = []
        for i in range (1, n + 1):
            if (n% i == 0):
                lDiv.append (i)
        return lDiv

# ------ list of prime divisors of an integer ----------------
    def listPrimeDivisor(self, n):
        # initialization of the list of divisors
        lDiv = []
        for i in range (1, n + 1):
            if (n% i == 0 and self.testPrime(i)):
                lDiv.append(i)
        return lDiv

# Instantiation example
Comput= Computation ()
Comput.testPrims(13, 7)
print ("List of divisors of 18:", Comput.listDivisor(18))
print ("List of prime divisors of 18:", Comput.listPrimeDivisor(18))
Comput.allTables()
````

### Q4 — a FlashCard quiz built on `random`

The flashcard generator stores a private dictionary of fruit–color pairs and runs an endless quiz loop. Each round it draws a random `(fruit, color)` pair, shows the fruit, reads the user's answer, and reports correct or wrong. The relevant line is:

```python
fruit, color = random.choices(list(self.__fruits.items()))[0]
```

`random.choices` picks one pair (by default with replacement), unpacked into `fruit` and `color`. The user's answer is compared case-insensitively with `user_answer.lower() == color`, so `YELLOW` and `yellow` both pass.

Two things to study in the solution. First, the loop asks whether to continue: `option = int(input('enter 0 to play again 1 to exit'))`, and `if option: break` ends the game for any non-zero answer. ### Q5 — Instructor eligibility and course allocation

The final problem models a real business rule. An `Instructor` is identified by private `name`, a *list* of `technology` skills, `experience`, and average `feedback`. The two public validation methods encode the admission criteria word for word:

- `check_eligibility()`: with more than 3 years of experience the instructor must have feedback ≥ 4.5; with 3 or fewer years, feedback ≥ 4.
- `allocate_course(tech)`: the course can be allocated only if the instructor is *eligible* **and** the requested `tech` is inside their skill list.

The solution composes the two: `allocate_course` first calls `check_eligibility()` and, only if the result is `True`, checks membership with `if tech in self.__technology`. The sample instructor has 5 years and feedback 4.9, so they are eligible, but `'Android Dev'` is not in their `['Data Science', 'Web Development']` skills — hence the output is the "not in their skill set" message. This method-building-on-method pattern is exactly how real OOP gets written: small public behaviours, each with one clear responsibility, combined to implement a larger rule.

````python
# write your code here

class Instructor:

    def __init__(self,name,technology,experience,feedback):
        self.__name = name
        self.__technology = technology
        self.__experience = experience
        self.__feedback = feedback

    def check_eligibility(self):

        if self.__experience > 3 and self.__feedback >= 4.5:
            return True
        elif self.__experience <= 3 and self.__feedback >= 4:
            return True
        else:
            return False

    def allocate_course(self,tech):
        is_eligible = self.check_eligibility()

        if is_eligible:
            if tech in self.__technology:
                return 'Padha sakta hai'
            else:
                return 'Ye to usko aata hi nai hai'
        else:
            return 'acha nai padhata hai'

ins = Instructor('Nitish',['Data Science','Web Development'],5,4.9)
ins.allocate_course('Android Dev')
````
