# 1. Class Relationships (Has-A / Is-A)

## 1.1 Aggregation — the Has-A relationship

Objects rarely live alone; more often one object *contains* another:

```python
class Customer:
    def __init__(self, name, gender, address):
        self.name = name
        self.gender = gender
        self.address = address    # Customer HAS-A Address
```

- **Aggregation (Has-A)** — a class holds an object of another class as one of its attributes.
- A `Customer` has an `Address`; an `Employee` has a `Department`; a `Car` has an `Engine`.
- The contained object usually exists on its own (an address makes sense without a customer) — this loose coupling makes aggregation flexible.
- You reach inside with the dot chain: `cust.address.city`.
- Class diagrams draw it with a hollow diamond on the container side.

The customer example coming up also shows methods delegating to the contained object: `edit_profile` updates the address by calling `self.address.edit_address(...)` — one object asking another to take care of itself.

### Aggregation(Has-A relationship)

## 1.2 Has-A vs Is-A — choose the right relationship

```python
# Aggregation  : Customer HAS-A Address
class Customer:
    def __init__(self, name, address):
        self.name = name
        self.address = address

# Inheritance  : Student IS-A User
class Student(User):
    pass
```

| Decision | Relationship | How | Meaning |
|---|---|---|---|
| Does one thing *contain* another? | Aggregation (Has-A) | nested attribute | `Customer` has `Address` |
| Is one thing a *kind of* another? | Inheritance (Is-A) | `class Child(Parent)` | `Student` is a `User` |

- Has-A → composition/containment; one object stores another.
- Is-A → specialization; the child inherits and extends the parent.
- Aggregation keeps classes independent and reusable; inheritance couples them through a hierarchy.
- Rule of thumb: prefer aggregation ("uses") unless there is a genuine "is a kind of" relationship.

This session explores both relationships, starting with the aggregation code cells below. In short: choose Has-A when one object *uses* another, and Is-A when one object is a more specific version of another. Getting that decision right keeps your hierarchy honest and your code easy to extend.

# 2. Inheritance — the Is-A Relationship

## 2.1 One class wearing another's clothes

**Inheritance** creates a parent-child relationship between classes:

```python
class User:                 # parent / base class
    def __init__(self):
        self.name = 'nitish'

class Student(User):        # child / derived class
    pass
```

- Syntax: `class Child(Parent):`.
- The child **receives** the parent's non-private attributes and methods — it can use them as its own.
- **Is-A**: a Student IS-A User.
- Benefits of inheritance:
  - **Code reuse** — you never rewrite the parent's logic.
  - **Extension** — the child adds its own methods and attributes on top.
  - **Overridability** — the child replaces parent methods, the seed of polymorphism.
  - **Hierarchy** — modeling the world as families (Vehicle → Car → ElectricCar).

The most common confusion: inherited members become *accessible*, not *mandatory* — the child decides what to use, override, or extend. Think of the child as the parent plus a few extra powers: it can use the inherited rooms (attributes and methods), owns new rooms of its own, and may remodel a kitchen (override a method) without tearing down the architecture.

## 2.2 Deep dive — method overriding

```python
class A:
    def __init__(self):
        self.var1 = 100
    def m1(self):
        print("Method in A")

class B(A):
    def m1(self):           # OVERRIDE — same name, own version
        print("Method in B")

b = B()
b.m1()       # "Method in B" — the child's version wins
```

- **Overriding** = the child redefines a parent method with the same name.
- Call resolution searches the child class first, then walks up; the child's version wins.
- To keep the parent logic too, call `super().m1()` from inside the child's version.
- Matching argument lists makes overrides clean and predictable.
- Real example: a `Phone` buys as a phone; a `SmartPhone` overrides `buy()` to announce a smartphone (as you will see in the code).
- Overriding is the foundation of polymorphism.
- Python does not do *overloading* by default — the last definition of a name wins. The chain of lookups explains the mechanism: Python checks the object's class first, then walks the parents in MRO order — an override behaves like a closer definition in a scoping chain, just at the class level.

## 2.3 Deep dive — what the child inherits

```python
class User:
    def __init__(self):
        self.name = 'nitish'
        self.gender = 'male'
    def login(self):
        print('login')

class Student(User):
    def __init__(self):          # own constructor → parent's is NOT auto-run
        self.rollno = 100
    def enroll(self):
        print('enroll into the course')
```

Three rules are visible here:

- **Constructor**: because `Student` defines its own `__init__`, calling `Student()` runs only the child's — `rollno` is set, but `name`/`gender` are NOT, unless the child calls `super().__init__()`.
- **Methods**: `login()` was inherited, and the child wrote `enroll()` — `s.login()` and `s.enroll()` both work.
- **Snapshot honesty**: the cell's saved output shows `nitish` because it was captured when `Student` had NO `__init__` of its own. Run the current source fresh and `print(s.name)` would raise `AttributeError` — a live reminder that *the constructor decides which attributes exist*.

That is exactly why the accompanying `Phone`/`SmartPhone` examples stress calling `super().__init__(...)` when a child wants both its own fields and the parent's.

## 2.4 Deep dive — constructor inheritance rules

| Thing | Inherited? |
|---|---|
| Constructor (`__init__`) | only if the child defines none; otherwise call `super().__init__()` |
| Non-private attributes/methods | ✅ |
| Private members (`__x`) | ❌ |
| Class/static variables | ✅ |

```python
class Phone:
    def __init__(self, price, brand, camera):
        self.__price = price
        self.brand = brand
        self.camera = camera

class SmartPhone(Phone):
    pass                        # inherits Phone.__init__ directly

s = SmartPhone(20000, "Apple", 13)   # Phone constructor runs
```

- Child with NO `__init__` → the parent's constructor is used automatically (via MRO).
- Child with its own `__init__` → the parent's is NOT run unless you call `super().__init__(...)`.
- Public attributes (`brand`, `camera`) come along; the private `__price` is mangled to `Phone` only, so the child cannot touch it — but a public getter like `show()` inside the parent CAN read it.
- Python resolves lookups through **MRO**, the Method Resolution Order.

# 3. Method Overriding

## 3.1 The child rewrites the parent

When a child defines a method with the same name as the parent's, it **overrides** it:

```python
class Phone:
    def buy(self):
        print("Buying a phone")

class SmartPhone(Phone):
    def buy(self):                       # override
        print("Buying a smartphone")

s = SmartPhone(20000, "Apple", 13)
s.buy()                                  # "Buying a smartphone"
```

- Overriding = same method name, the child supplies its own version.
- Python resolves the call at runtime by walking up the class chain (child first) — this is **dynamic dispatch**.
- The child's version fully replaces the parent's for that object.
- If the child still wants the parent's behavior too, it calls `super().buy()`.

You will see two further shapes in the code cells: a child that *adds a new method* (`display2`) without overriding — so the parent's `display1` still runs — and a `Phone`/`SmartPhone` pair where the child overrides `buy()` and prints its own message.

## Method overriding

```python
class A:
    def __init__(self):
        self.var1 = 100

    def display1(self, var1):
        print(var1)

class B(A):
    def display1(self, var1):    # same name naya — PARENT wala override
        print('This is overriding the parent method!', var1)
```

- Same method name in the child as in the parent → **overriding**
- At runtime the child's version runs (part of polymorphism)
- If you also want the parent's, call `super().method()`


````python
class A:
    def __init__(self):
        self.var1=100

    def display1(self,var1):
        print("class A :", self.var1)
class B(A):
  
    def display2(self,var1):
        print("class B :", self.var1)

obj=B()
obj.display1(200)
````

````python
# Method Overriding
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    def buy(self):
        print ("Buying a smartphone")

s=SmartPhone(20000, "Apple", 13)

s.buy()
````

# 4. The Super Keyword

## 4.1 Reaching back to the parent

`super()` is Python's reference *from the child up to the parent*:

```python
class SmartPhone(Phone):
    def __init__(self, price, brand, camera, os):
        super().__init__(price, brand, camera)   # parent constructor
        self.os = os
```

- `super().__init__(...)` invokes the parent's constructor explicitly.
- `super().method()` calls a parent method you overrode, e.g. `super().buy()`.
- Use it when the child wants BOTH the parent's setup AND its own extras.
- `super` can reach the parent's public and protected members, but not its private mangled names.
- It is meaningful only inside a child's methods — there is no sensible `super()` outside the class body.

The pattern to internalize: **override, then extend** — in the child's version, do your own work, then call `super().whatever()` to keep the parent's behavior flowing. A useful test: if you catch yourself copy-pasting the parent's `__init__` body into the child, stop — that is exactly the duplication `super()` exists to remove. Write the child's extras, and one line brings in everything the parent already does.

### Super Keyword

````python
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    def buy(self):
        print ("Buying a smartphone")
        # syntax to call parent ka buy method
        super().buy()

s=SmartPhone(20000, "Apple", 13)

s.buy()
````

        self.camera = camera

class SmartPhone(Phone):

- `super().__init__(...)` — calls the parent's constructor explicitly
- Use-case: the child wants both its own extra attributes and the parent's setup
- `super` can also access the parent's data (through methods)


## 4.2 Deep dive — super in the constructor chain

```python
class SmartPhone(Phone):
    def __init__(self, price, brand, camera, os, ram):
        print('Inside smartphone constructor')
        super().__init__(price, brand, camera)   # run the parent's setup
        self.os = os
        self.ram = ram
```

Call order for `SmartPhone(20000, "Samsung", 12, "Android", 2)`:

1. `SmartPhone.__init__` starts and prints 'Inside smartphone constructor'.
2. `super().__init__(...)` runs `Phone.__init__`, printing 'Inside phone constructor' and setting `__price`, `brand`, `camera`.
3. Control returns; `os` and `ram` are set on the child.
4. `s.os` → 'Android', `s.brand` → 'Samsung' — the public attributes are reachable.

- `super()` is MRO-aware: it resolves to the next class in the hierarchy.
- Use it in constructor chaining and when extending an overridden method.
- A `super()` call outside the class body has no meaning.
- Private parent data stays off-limits even through `super()` (name mangling binds it to the parent).

A small note: a cell that attempts `print(super().brand)` would raise `AttributeError` on a fresh run — `super()` resolves class-level lookups, while instance attributes like `brand` belong to `self` — so read instance data through `self`, and use `super()` for methods and constructor wiring.

````python
# using super outside the class
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    def buy(self):
        print ("Buying a smartphone")
        # syntax to call parent ka buy method
        super().buy()

s=SmartPhone(20000, "Apple", 13)

s.buy()
````

````python
# can super access parent ka data?
# using super outside the class
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    def buy(self):
        print ("Buying a smartphone")
        # syntax to call parent ka buy method
        print(super().brand)

s=SmartPhone(20000, "Apple", 13)

s.buy()
````

# 5. Types of Inheritance

## 5.1 The family shapes of classes

Inheritance comes in a few standard shapes:

| Type | Structure | Example |
|---|---|---|
| Single | 1 parent → 1 child | `class SmartPhone(Phone)` |
| Multilevel | grandparent → parent → child | `Product → Phone → SmartPhone` |
| Hierarchical | 1 parent → many children | `Phone → SmartPhone, FeaturePhone` |
| Multiple | many parents → 1 child | `class SmartPhone(Phone, Product)` |

```python
class SmartPhone(Phone): pass                    # single
class Phone(Product): pass                       # multilevel (with SmartPhone)
class FeaturePhone(Phone): pass                  # hierarchical
class SmartPhone(Phone, Product): pass           # multiple
```

- **Single** is the everyday case: one base, one child.
- **Multilevel** chains generations; the grandchild accumulates everything above it.
- **Hierarchical** shares one base between several siblings.
- **Multiple** imports behavior from several bases — powerful, but it introduces the **diamond problem**: when two parents share a common ancestor, whose method wins? Python's **MRO** (Method Resolution Order) answers deterministically, and `ClassName.__mro__` reveals the order.

### Types of Inheritance

- Single Inheritance
- Multilevel Inheritance
- Hierarchical Inheritance
- Multiple Inheritance(Diamond Problem)
- Hybrid Inheritance

| Type | Meaning |
|---|---|
| Single | 1 parent → 1 child |

## 5.2 Deep dive — MRO and the diamond problem

```python
class Phone:
    def buy(self): print("Buying a phone")
class Product:
    def buy(self): print("Product buy method")

class SmartPhone(Phone, Product):   # Phone listed FIRST
    pass

s = SmartPhone(20000, "Apple", 12)
s.buy()    # "Buying a phone" — from the FIRST listed parent
```

- Multiple inheritance `class C(A, B)` — the plain lookup order is left-to-right.
- **MRO** (C3 linearization) resolves conflicts consistently; the order in the class list matters.
- Diamond problem: with `A → B, C → D`, if `B` and `C` both override `A.m1`, which version runs on `D`? The MRO keeps the order in which the classes were declared.
- `D.mro()` prints the full resolution order — the debugging tool for surprises.
- In the running cells, `class SmartPhone(Phone, Product)` with duplicate `buy` prints `Phone`'s version because `Phone` is first.
- Keep inheritance trees shallow, and prefer composition over multiple inheritance in serious designs.

````python
# single inheritance
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    pass

SmartPhone(1000,"Apple","13px").buy()
````

## Deep Dive : Diamond problem

```python
class A:
    def m1(self): return 20
class B(A): ...
class C(A): ...
class D(B, C): ...        # which m1? from A via B or C?
```

- In multiple inheritance the same ancestor's method has 2 paths
- Conflict: which D.m1()?
- Python MRO (C3 linearization) solves it with a deterministic order
- `D.mro()` → check the order list
- `class D(B, C)` — B listed first → the B chain wins
- Ambiguity risk — the same base twice in the graph
- Fix: explicit MRO order, super() propagation
- Prefer composition over multiple inheritance (design advice)
- Interview classic — Python handles it quietly; other languages don't


````python
# the diamond problem
# https://stackoverflow.com/questions/56361048/what-is-the-diamond-problem-in-python-and-why-its-not-appear-in-python2
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class Product:
    def buy(self):
        print ("Product buy method")

# Method resolution order
class SmartPhone(Phone,Product):
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy()
````

````python
class A:

    def m1(self):
        return 20

class B(A):

    def m1(self):
        return 30

    def m2(self):
        return 40

class C(B):
  
    def m2(self):
        return 20
obj1=A()
obj2=B()
obj3=C()
print(obj1.m1() + obj3.m1()+ obj3.m2())
````

## Diamond problem

```python
class A:
    def m1(self):
        return 20

class B(A):
    def m1(self):
        val = super().m1() + 30
        return val

class C(B):
    def m1(self):
        val = self.m1() + 20   # warning: recursion risk!
        return val
```

- In multiple inheritance, when the same method exists in several parents → ambiguity
- Python's MRO (Method Resolution Order) decides which one runs
- `super()` looks up the parent's method according to the MRO
- Avoid recursion — `self.m1()` inside itself calls the same method again


````python
class A:

    def m1(self):
        return 20

class B(A):

    def m1(self):
        val=super().m1()+30
        return val

class C(B):
  
    def m1(self):
        val=self.m1()+20
        return val
obj=C()
print(obj.m1())
````

# 7. Abstraction

## 7.1 Contracts that force implementation

**Abstraction** hides the *how* and fixes the *what*. In OOP it is implemented with abstract base classes (`ABC`):

```python
from abc import ABC, abstractmethod

class BankApp(ABC):
    def database(self):            # concrete — inherited and shared
        print('connected')
    @abstractmethod
    def security(self):            # abstract — child MUST implement
        pass

class MobileApp(BankApp):
    def security(self):            # the required implementation
        print('secure login added')
```

- `@abstractmethod` marks a method children are forced to write.
- An `ABC` cannot be instantiated directly — `BankApp()` raises `TypeError`.
- A child stays abstract until it implements every abstract method.
- Concrete methods are inherited and shared normally.

| Idea | Effect |
|---|---|
| Abstract class | a template/contract; cannot create objects of it |
| Abstract method | the child must provide an implementation |
| Concrete method | inherited as-is |

Benefits: the parent *guarantees* an interface — consumers know `security()` and `display()` exist — while every child supplies its own details.
