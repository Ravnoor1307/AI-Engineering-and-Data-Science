## Welcome to Object-Oriented Programming

So far your Python programs have lived inside modules and functions. You have used `len()`, `append()` and `upper()`, but you have never built a *type of your own*. This session changes that. Object-Oriented Programming (OOP) is the idea that data and the operations on that data travel together as one unit — an **object**. Python is an object-oriented language: every value you touch, from a simple integer to a complicated list, is an object that belongs to some class. In this session you will learn to design your own classes, stamp out your own objects, and control how Python's operators behave on them.

4. `self` and instance attributes

## The mental model

Think of a **class** as a blueprint or cookie-cutter and an **object** as the cookie made from it. The blueprint declares which attributes every cookie should have (`flavor`, `size`) and which powers (methods) they all get. When you call the class, a fresh cookie is baked with its own private copy of each attribute. Two objects from the same class look alike on the surface but live in different memory — each carries its own state.

## Key definitions

- **Class:** a template/blueprint that bundles data (attributes) and behavior (methods).
- **Object:** a concrete, live instance built from a class.
- **Constructor:** the `__init__` method that automatically runs when an object is born.
- **Method:** a function defined inside a class and called with `object.method()`.
- **Attribute:** a piece of data owned by an object, written `object.attribute`.
- **Magic method:** a dunder (`__...__`) method Python calls behind the scenes for operators and built-ins.

By the end you will be able to create reusable, self-contained types just like Python's built-in `list` and `str`. Before you start, remember the shorthand: a **function** takes the object as an argument (`len(L)`), while a **method** is called with a dot inside the type (`L.append()`).

# 1. Function vs Method

## 1.1 Methods are attached, functions are standalone

Before you write your own classes you need to see the exact difference between a **function** and a **method**, because you will be creating methods all session.

- A **function** lives outside any type. You call it by passing the data in: `len(L)`, `max(L)`, `sorted(L)`.
- A **method** lives inside a type and is reached with a dot: `L.append(4)`, `L.sort()`, `s.upper()`.

The dot is the key clue: `object.method()`. When you write `L.upper()`, Python looks inside the list class, finds no `upper`, and raises `AttributeError: 'list' object has no attribute 'upper'`. Similarly `s.append('x')` fails because strings have no `append`.

| Call style | Example | Meaning |
|---|---|---|
| Function | `len(L)` | pass the object in as an argument |
| Method | `L.append(4)` | the object calls its own behavior |

This is the foundation of **abstraction**: data and the operations that make sense for that data live in the same place. A list knows how to append, sort and pop; a string knows how to upper, split and strip. Each type carries its own little toolbox.

## 1.2 Deep dive — everything is an object

```python
L = [1, 2, 3]
L.upper()       # AttributeError — 'list' object has no attribute 'upper'
s = 'hello'
s.append('x')   # AttributeError — 'str' object has no attribute 'append'
```

An object is a bundle of **state** (data) and **behavior** (methods). Every object in Python — an integer, a string, a list, a function — is an instance of some class, and it arrives with the methods of its own type. `type(L)` returns `<class 'list'>`, which tells you which toolbox the object carries.

- Methods are *attached* to an object: `object.method(arguments)`.
- Without attached methods you would have to write a separate function for every data type.
- OOP lets you define your own custom types so the "toolbox" can be exactly what your problem needs.

`L` is a list object: it can `append`, `sort`, and `reverse`, but it has no idea what `upper` means. `s` is a string object: it can `upper` and `split`, but it cannot `append`. The error is the class telling you clearly: "this operation does not exist in my toolbox." You will see both `AttributeError`s when you run the two cells below — that is intentional and instructive.

# 3. Constructors & Literals


- `str()` gives `''`, `list()` gives `[]`, `int()` gives `0`, `dict()` gives `{}`.

`type(L)` tells you which class an object belongs to — the constructor and the literal produce the same type. The constructor is essentially a tiny factory: call it, and a fresh object rolls off the line.

# 4. Naming Conventions

## 4.1 Reading code at a glance

Names carry meaning even before you read the code. Python's community follows **PEP 8**, and the most important rule for OOP is: a class name starts with a capital letter.

| What | Style | Example |
|---|---|---|
| Class | PascalCase | `HelloWorld`, `Atm` |
| Object/Variable | lowercase | `obj1`, `balance` |
| Function/Method | lowercase + underscore | `convert_to_decimal()` |
| Constant | ALL_CAPS | `MAX_SPEED` |

- **PascalCase**: every word capitalized, no spaces — `HelloWorld`.
- **snake_case**: lowercase words joined by underscores — `my_var`, `my_func`.
- Class names are nouns (`Person`, `Car`, `Atm`); methods are verbs (`get_balance`, `withdraw`).

This is coding etiquette: the moment you see `Atm` you know it is a class, and `atm1` tells you it is an instance. Consistency matters more than cleverness — a reader should never have to guess which thing is a type.

# 5. Making a Class + Constructor

## 5.1 The `class` keyword

To create a type of your own you write `class Atm:` and put everything the type owns — attributes and methods — inside its indented body.

```python
class Atm:
    def __init__(self):
        self.pin = ''
        self.balance = 0

    def menu(self):
        user_input = input("Hi how can I help you?")
```

Four new ideas appear here:

- `class Atm:` — the blueprint begins. `Atm` is now a user-defined type (PascalCase!).
- `__init__` — the **constructor**: it runs by itself the moment an object is created.
- `self` — a reference to the current object. `self.pin` means "the `pin` this particular object owns", so every object keeps its own copy of `pin` and `balance`.
- `menu()` — a method that can read and change the object's own attributes through `self`.

This is your first full class. The pattern is always `def __init__(self):` for setup, then ordinary methods that take `self` as their first parameter.

## Making a class + constructor

```python

```

- `class Atm:` — the blueprint starts
- `__init__` = **constructor** — called instantly as soon as an object is created
- `self` = reference to the current object — `self.pin` are the attributes that
  every object keeps its own copy of
- `menu()` — a method that can access its object's attributes


## 5.2 Deep dive — the constructor auto-runs

```python
class Temp:
    def __init__(self):
        print('hello')

obj = Temp()      # 'hello' prints — __init__ was auto-called
```

- Creating an object by calling the class — `Temp()` — makes the constructor run automatically.
- You never write `obj.__init__()` yourself.
- The constructor's job: initialize (set up) the new object's data.
- It is the "on-birth" hook — the first code that runs inside the freshly created object.
- A class can have only one `__init__` (no overloading); give flexibility with default arguments: `def __init__(self, x=5)`.
- Other languages write `new ClassName()`; Python uses the same call syntax as a plain function: `ClassName()`.

Run the `Temp` cell and watch `hello` print with no explicit call — that is the constructor running on its own. Add your own setup logic to `__init__` — setting attributes, printing, opening a file — and it will run at every birth, which is why `__init__` is the natural home for object setup routines.

# 6. Creating Objects & Identity

## 6.1 Two objects, two addresses

Every time you call a class, Python creates a brand-new object:

```python
obj1 = Atm()
obj2 = Atm()
print(id(obj1))   # unique memory address
print(id(obj2))   # DIFFERENT address
```

- `Atm()` allocates fresh memory for a new instance and runs the constructor again.
- `id(object)` returns that object's **unique identity** — effectively its memory address.
- `obj1` and `obj2` come from the same blueprint but live in different places, so each keeps its own `pin` and `balance`.
- Inside the constructor, `print(id(self))` shows the same address as `id(obj1)` — `self` really is the object being created.

Think of `id` as a fingerprint: two objects never share an identity. Checking `id()` is your first debugging tool for the question "is this a new object or the same one?". The upcoming cells show the `Employee` constructor assigning each incoming argument to `self` — watch how every subsequent method call reads the object's own copy of that data, which is the entire value proposition of the instance.

- `obj1` and `obj2` have different ids → different objects, different attributes
- Inside the constructor `print(id(self))` shows the same address that `id(obj1)` gives

# 7. Function vs Method — `len` vs `append`

## 7.1 Two call styles, two designs

`len` and `append` do different jobs and are called differently:

```python
L = [1, 2, 3]
len(L)        # function — pass the object in as an ARGUMENT
L.append(42)  # method — the object calls it with a dot
```

- `len(L)` — a function living OUTSIDE the list class; you hand it the list.
- `L.append()` — a method living INSIDE the list class; you ask the object itself.
- Similarly `str()`, `list()`, `sorted()` are functions; `s.upper()`, `L.sort()`, `L.pop()` are methods.

| Operation | Style | Lives |
|---|---|---|
| `len(L)` | function | outside the type |
| `L.append()` | method | inside the type |

Generic, shared utilities become functions; type-specific behavior becomes methods. Internally, `len(L)` actually delegates to the magic method `L.__len__()` — a preview of the power you gain in the magic-methods section.

## len vs append

```python
len(L)      # function -> you pass the list object as an ARGUMENT
L.append()  # method   -> the object itself calls the method
```

- `len(L)` — it is a function, outside the class; pass the list into it
- `L.append()` — it is a method, inside the class; called directly with `.`
- Similarly `str()`, `list()` — functions; `s.upper()`, `L.sort()` — methods


## 7.2 Deep dive — generic function vs type-specific method

```python
L = [1, 2, 3]
len(L)        # function — works on any sequence
L.append(42)  # method — belongs to the list class only
```

- **Function** (`len`, `min`, `max`, `sorted`) is defined OUTSIDE the object and accepts it as an argument.
- **Method** (`append`, `upper`, `sort`) is defined INSIDE the class and is attached to the object.
- `len` works on strings, lists, tuples — a generic utility.
- `append` works only on lists — type-specific behavior.
- Call style: `func(obj)` versus `obj.method()`.
- Internally, `len` calls the magic method `L.__len__()`.
- Design rule: shared generic operations belong to functions, type-unique ones belong to methods.

# 9. Magic Methods & Operator Overloading

## 9.1 Making `+` and `print` understand your class

Python reserves a set of "dunder" (double-underscore) methods for special operations. When you use an operator or a built-in, Python quietly translates it into a magic-method call.

```python
fr1 + fr2      # → fr1.__add__(fr2)
print(fr1)     # → uses __str__ / __repr__
len(L)         # → L.__len__()
```

| Magic method | Operator / built-in |
|---|---|
| `__str__` | `print()` and string conversion |
| `__add__` | `+` |
| `__sub__` | `-` |
| `__mul__` | `*` |
| `__truediv__` | `/` |

By defining these on your own class you **overload the operators** — you decide what `+` means for your objects. The `Fraction` class coming up defines `__str__` so `print(fr1)` shows `3/4`, and `__add__` so `fr1 + fr2` performs real fraction addition. Same operators, your own meaning — this is operator overloading, and it is exactly how Python lets a class feel like a built-in type to everyone using it.

## Magic methods & operator overloading

```python
class Fraction:
    def __init__(self, x, y):     # parameterized constructor
        self.num = x
        self.den = y

    def __str__(self):            # what is shown on print
        return '{}/{}'.format(self.num, self.den)

    def __add__(self, other):     # define + in our own way
        new_num = self.num * other.den + other.num * self.den
        new_den = self.den * other.den
        return Fraction(new_num, new_den)
```

- `__init__(self, x, y)` — constructor with arguments
- `__str__` — gives a readable output for `print(fr1)`
- `__add__` / `__sub__` / `__mul__` / `__truediv__` — operator overloading
- `fr1 + fr2` internally calls `fr1.__add__(fr2)`


## Deep Dive : Magic methods — operator overloading

```python
class Fraction:
    def __init__(self, x, y):   # constructor
        self.n, self.d = x, y
    def convert_to_decimal(self):
        return self.n / self.d
    def __add__(self, other):    # MAGIC method → defines +
        return Fraction(self.n*other.d + other.n*self.d, self.d*other.d)
    def __sub__(self, o):
        ...
    def __mul__(self, o):
        ...
    def __truediv__(self, o):
        ...

fr1 = Fraction(3,4)
fr2 = Fraction(1,2)
fr1 + fr2      # uses __add__

```

- **Magic methods** (dunder `__...__`) — python calls them internally for special operations
- `__add__` → `+` | `__sub__` → `-` | `__mul__` → `*` | `__truediv__` → `/`
- Operator overloading = defining operators on your own class
- `len(L)` internally `L.__len__()` | str → `__str__`
- `print(fr1)` → uses `__str__`/`__repr__`
- When you need it: custom math types (Fraction, Vector, Matrix)
- Formula for fraction addition: cross-multiply shortcut
- Magic methods = Python's "speech interface" for operators
- Favorite interview topic: name a few dunders


````python
class Fraction:

  # parameterized constructor
  def __init__(self,x,y):
    self.num = x
    self.den = y

  def __str__(self):
    return '{}/{}'.format(self.num,self.den)

  def __add__(self,other):
    new_num = self.num*other.den + other.num*self.den
    new_den = self.den*other.den

    return '{}/{}'.format(new_num,new_den)

  def __sub__(self,other):
    new_num = self.num*other.den - other.num*self.den
    new_den = self.den*other.den

    return '{}/{}'.format(new_num,new_den)

  def __mul__(self,other):
    new_num = self.num*other.num
    new_den = self.den*other.den

    return '{}/{}'.format(new_num,new_den)

  def __truediv__(self,other):
    new_num = self.num*other.den
    new_den = self.den*other.num

    return '{}/{}'.format(new_num,new_den)

  def convert_to_decimal(self):
    return self.num/self.den

    

  
````

````python
fr1 = Fraction(3,4)
fr2 = Fraction(1,2)
````

````python
fr1.convert_to_decimal()
# 3/4
````

````python
print(fr1 + fr2)
print(fr1 - fr2)
print(fr1 * fr2)
print(fr1 / fr2)
````

# 10. Every Type Has Its Own Operator Set

## 10.1 The same symbol, different meanings

Operators are not universal — every type decides which ones it supports.

```python
s1 = {1, 2, 3}
s2 = {3, 4, 5}
s1 + s2   # TypeError — sets do not support +
```

- `int` and `float` use `+` for addition; `list` and `str` use `+` for concatenation; `set` does not define `+` at all, so it raises `TypeError`.
- Defining the magic method enables the operator; omitting it means an error.
- For `Fraction` you defined `__add__`, so `+` worked; plain sets have no `__add__`, so `+` fails.

| Type | `+` behavior |
|---|---|
| `int` / `float` | numeric addition |
| `str` / `list` | concatenation |
| `set` | not defined → `TypeError` |
| `Fraction` (yours) | your `__add__` logic |

This is polymorphism in action — the same symbol, with the behavior decided by the class behind it.
