## Quick recap you will need

- Overriding: redefine a parent method with the same name; the child's version runs.
- `super()`: `super().__init__(...)` and `super().method()` reach the parent.
- Aggregation: `self.address = another_object` — object inside object.
- Abstraction: `from abc import ABC, abstractmethod` — children must implement `@abstractmethod`s.
- Subclassing built-ins: `class FlexibleDict(dict):` and override `__getitem__`.

## Task 1 · Concept — overriding a parent method

```python
class Vehicle:
    def __init__(self, seating_capacity):
        self.seating_capacity = seating_capacity
    def fare(self):
        return self.seating_capacity * 100

class Bus(Vehicle):
    def fare(self):
        base = super().fare()          # call the parent's fare
```

- Override `fare()` in `Bus`; use `super().fare()` for the base and add 10%: total fare + 10% of total fare.

## Task 3 · Concept — objects inside objects

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __str__(self):
        return f'({self.x}, {self.y})'

class Location:
    def __init__(self):
        self.Location = Point(3, 5)      # two owned Point objects
        self.Destination = Point(6, 2)
```

- `Point` is a tiny data holder with `x` and `y` and a readable `__str__`.
- `Location` owns TWO `Point` objects — aggregation (Has-A).
- Reflecting a point over the x-axis keeps `x` and flips the sign of `y`: `(x, -y)`.

**Hints**

- The problem asks for two `Point` objects named `Location` and `Destination` inside a `Location` class — you may rename for clarity, but keep a function that prints the reflected Destination.
- Reflect by building a new point: `Point(self.Destination.x, -self.Destination.y)` or `(x, -y)`.
- Verify: `Point(6, 2)` reflects to `(6, -2)`. Aggregation here is simple containment — `Location` does not inherit from `Point`, it owns two of them. That is Has-A versus Is-A made visible: both relationships are valid, but only containment fits this problem.

## Task 4 · Concept — abstract class forcing child areas

```python
from abc import ABC, abstractmethod

class Polygon(ABC):
    @abstractmethod
    def area(self):
        pass

class Rectangle(Polygon):
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth
    def area(self):                    # required!
        return self.length * self.breadth

class Triangle(Polygon):
    def __init__(self, base, height):
        self.base = base
        self.height = height
    def area(self):                    # required!
        return 0.5 * self.base * self.height
```

- `Polygon` is an `ABC` with an abstract `area()` — the children MUST implement it.
- You cannot instantiate `Polygon` directly; doing so raises `TypeError`.
- `Rectangle` and `Triangle` each supply their own formula.

**Hints**

- Collect each shape's dimensions at construction, then compute `area()` from them.
- Add a "get details" method per shape that prints its dimensions before the area.
- Instantiate both children and print their areas to confirm both work.

## Task 5 · Concept — inheritance for payment options

```python
class Bill:
    def __init__(self, amount, name):
        self.amount = amount
        self.name = name
    def pay(self):
        print(f'Payment of {self.amount} received for {self.name}')

class ChequePayment(Bill):
    def pay(self):
        super().pay()                    # base receipt
        print(f'Paying {self.amount} by Cheque ...')

class CashPayment(Bill):
    def pay(self):
        super().pay()
        print(f'Paying {self.amount} in Cash ...')
```

- `Bill` holds the shared data (`amount`, bill name).
- Children specialize the payment behavior: one for cheque, one for cash.
- Calling `super().pay()` reuses the shared receipt, then each child adds its payment channel — the override-and-extend pattern.

**Hints**

- Model the solution as a `Bill` base class with two subclasses (or a base plus one subclass per method); either is fine as long as the payment differs by method.
- Use `super()` so the shared receipt logic is not duplicated.
- Keep an amount and bill name so each payment prints a complete record.

## Task 6 · Concept — subclassing the built-in dict

```python
class FlexibleDict(dict):
    def __getitem__(self, key):
        if key in self:
            return super().__getitem__(key)
        # key was int but stored as str, or vice versa
        for k in self:
            if str(k) == key or (type(k) != type(key) and k == key):
                return super().__getitem__(k)
        raise KeyError(key)
```

- Normal dict lookup goes through `dict.__getitem__(key)` — the problem hints exactly at this.
- Override `__getitem__` so that when a normal lookup misses, you search once for a key whose string form matches.
- Everything else — assignment, iteration, other methods — is inherited from `dict`.

**Hints**

- The rule: `fd[1]` must find a stored `'1'`, and `fd['1']` must find a stored `1`.
- Compare `str(k) == key` or `k == str(key)` for type mismatches.
- After a failed normal lookup, iterate `self` once looking for a type-compatible key; raise `KeyError` if none exists.
- Sample check: `fd[1] = 100` then `fd['1']` → `100`; `fd['1'] = 100` then `fd[1]` → `100`.
