## Quick recap you will need

- Constructor: `def __init__(self, ...)` — runs automatically on object creation.
- Instance attributes: `self.name = ...` — each object keeps its own copy.
- Methods: `def method(self, ...)` — always receive the current object as `self`.
- Static helpers: `@staticmethod` — callable via the class, no `self` needed.
- Private fields: `self.__x` — accessed through getters/setters, an idea you will use again in the Instructor problem.

## Task 1 · Concept — a class stores data and computes

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def Perimeter(self):
        return 2 * (self.length + self.width)

    def Area(self):
        return self.length * self.width
```

- The constructor stores `length` and `width` as instance attributes.
- `Perimeter()` and `Area()` compute from `self`'s data — no extra arguments needed.
- `display()` prints each value on its own labelled line.

## Task 3 · Concept — a calculator as a class

```python
class Computation:
    def Factorial(self, n):
        return 1 if n <= 1 else n * self.Factorial(n - 1)

    def naturalSum(self, n):
        return n * (n + 1) // 2
```

A class whose methods are small math routines; each receives the number `n` as a parameter.

**Methods to build**

- `Factorial(n)` — product `1 * 2 * ... * n`.
- `naturalSum(n)` — sum `1 + 2 + ... + n`.
- `testPrime(n)` — `True`/`False` for primality.
- `testPrims(n1, n2)` — coprime test: the greatest common divisor must be `1` (use `math.gcd` or a loop).
- `tableMult(n)` — print `n * 1` up to `n * 10`.
- `allTablesMult()` — print the tables from 1 to 9.
- `listDiv(n)` — static method returning all divisors of `n` in a list `Ldiv`.
- `listDivPrim(n)` — static method returning only the prime divisors.

**Hints**

- `listDiv` is a `@staticmethod`: no `self`, called as `Computation.listDiv(10)`.
- Prime check: loop from 2 up to `n**0.5`; any clean division means "not prime".
- Divisors: every `i` in `1..n` where `n % i == 0`; a divisor is prime if `testPrime(i)` says so.
