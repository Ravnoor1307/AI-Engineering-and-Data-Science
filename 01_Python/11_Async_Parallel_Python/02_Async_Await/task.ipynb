## 👑 BOSS: 10-Assert Head-Waiter Badge (self-contained!) 🏆

```python
# 👑 BOSS CONSOLE
async def _chai(t):
    await asyncio.sleep(0.01)
    return f"chai-{t}"

assert asyncio.iscoroutine(_chai(1)),                   "① tokra ≠ chai — coroutine-object check!"
assert not asyncio.iscoroutine("chai-1"),               "② string tokra nahi hai!"

async def _b3():
    return await asyncio.gather(_chai(1), _chai(2), _chai(3))
assert run(_b3()) == ["chai-1", "chai-2", "chai-3"],    "③ gather order-preserved!"

async def _job(i, s):
    await asyncio.sleep(s); return i
async def _race():
    waits = [0.3, 0.1, 0.2]
    results = await asyncio.gather(*(_job(i, waits[i]) for i in range(3)))
    finish = []
    for c in asyncio.as_completed([_job(i, waits[i]) for i in range(3)]):
        finish.append(await c)
    return results, finish
_res, _fin = run(_race())
assert _res == [0, 1, 2],                               "④ results submit-order!"
assert _fin == [1, 2, 0],                               "⑤ finish = chhoti-pehle [1,2,0]!"

random.seed(11)
_marks = []
async def _fm(name):
    w_ = round(random.uniform(0.1, 0.4), 2); m_ = random.randint(60, 95)
    _marks.append((name, m_))
async def _storm():
    for n in ["Asha", "Bhuvan", "Chirag", "Diya", "Eshan"]:
        await _fm(n)
run(_storm())
assert _marks == [("Asha", 95), ("Bhuvan", 89), ("Chirag", 72), ("Diya", 92), ("Eshan", 71)], "⑥ seed-squad exact!"
assert round(sum(m for _, m in _marks) / 5, 1) == 83.8, "⑦ avg 83.8!"

async def _slow(): await asyncio.sleep(0.5); return "real"
_flag = False
try:
    run(asyncio.wait_for(_slow(), timeout=0.1))
except TimeoutError:
    _flag = True
assert _flag,                                           "⑧ shield fired (TimeoutError)!"
_t0 = time.perf_counter(); run(asyncio.gather(*(_job(0, 0.1) for _ in range(3))))
assert time.perf_counter() - _t0 < 0.3,                 "⑨ gather < sequential (0.3s margin)!"
assert asyncio.iscoroutinefunction(_chai),              "⑩ iscoroutinefunction True!"
print("🏆 10/10 HEAD-WAITER BADGE — async/await tumhari tray me! 🦸🎫")
```

## ✅ SOLUTIONS 🔓

<details><summary><b>S1-S4</b></summary>

```python
# S1-S4 ✅
async def serve_chai(t):
    await asyncio.sleep(0.01)
    return f"chai-{t}"
ISCORO = asyncio.iscoroutine(serve_chai(1))
print("tokra hai?", ISCORO)                              # True — call karne se sirf tokra!

TASTE = run(serve_chai(5))
print("chai taiyaar:", TASTE)                            # chai-5

TRAY = run(asyncio.gather(serve_chai(1), serve_chai(2), serve_chai(3)))
print("tray:", TRAY, "| order?", TRAY == ["chai-1", "chai-2", "chai-3"])   # True!

async def job(i, s):
    await asyncio.sleep(s); return i
async def finish_order():
    waits = [0.3, 0.1, 0.2]
    out = []
    for c in asyncio.as_completed([job(i, waits[i]) for i in range(3)]):
        out.append(await c)
    return out
FINISH_ORD = run(finish_order())
print("finish-order:", FINISH_ORD)                       # [1, 2, 0] — chhoti pehle!
```

<details><summary><b>S5-S7</b></summary>

```python
# S5-S7 ✅
random.seed(11)
async def fetch_marks(name):
    w_ = round(random.uniform(0.1, 0.4), 2)
    m_ = random.randint(60, 95)
    return (name, m_)
SQUAD = run(asyncio.gather(*(fetch_marks(n) for n in ["Asha", "Bhuvan", "Chirag", "Diya", "Eshan"])))
print("squad:", SQUAD)                                   # seed-exact pairs!

TIMED_OUT = False
async def slow_api(): await asyncio.sleep(0.5); return "real-data"
try:
    run(asyncio.wait_for(slow_api(), timeout=0.25))
    SHIELD = "real-data"
except TimeoutError:
    TIMED_OUT = True; SHIELD = "CACHED"
print("shield fired?", TIMED_OUT, "| user ko:", SHIELD)  # True / CACHED

t0 = time.perf_counter()
run(asyncio.gather(*(job(0, 0.1) for i in range(3))))
EL = time.perf_counter() - t0
print(f"3×0.1 gather: {EL:.2f}s (sequential 0.30 se kam? {EL < 0.3})")
# Takeaway: gather = slowest-ek-tak; as_completed = finishing-feed; wait_for = helmet! 🛡️
```
