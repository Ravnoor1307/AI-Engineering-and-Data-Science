**Topic 14.02 | Production_Ready_Mindset — Part 2 of 4**

🌍 **Hook:** You built a 1000-line single-slab — one giant script holding
CSV too, validation too, plotting too, billing too! Then the ELECTRICIAN's job
came up → you had to demolish the whole house!! 😱 Pro-builders don't ship a single piece:
kitchen-team, electric-team, plumber-team — each one builds its own ROOM!
That's what we call in code: **MODULE** (one .py = one craftsman-team!) 🧩

📖 **Definitions:**
- **module** = one `.py` file — its own functions/constants
- **package** = a folder + `__init__.py` (the team's tower!)
- **import-scene** = use another team's work: `import team_stats` 🪢
- **`if __name__ == "__main__":`** = demo when run AS a script; SILENT when
  imported — the guard-master! (no side-effects on module load) 🛡️

🔍 **WHY modular?** smaller files → testable → zero team-collisions →
reuse (marks_stats will work in every project!)

## 🐣 Level 0 — The `__name__` Mirror 🪞 (the guard-master LIVE!)

```python
# notebook==script-direct — __name__ hamesha "__main__"!
print("yeh file chal rahi as:", __name__)

# module jagah pe dikhein: report ka madhna daan sirf direct-chalane pe — guard:
def demo_only():
    return "main-guard ke ANDAR wala kaam (import pe SILENT!) 🛡️"

if __name__ == "__main__":
    print("guard-inside:", demo_only())   # import se chala toh SKIP!
# Expected OUTPUT:
# yeh file chal rahi as: __main__
# guard-inside: main-guard ke ANDAR wala kaam (import pe SILENT!) 🛡️
```

## 🐥 Level 1 — Your Own Module: Write + Import (LIVE!) 📦

Let's make a file `marktools_module.py` (real file work!): load/clean fns —
then import it and RUN it — module-mechanics LIVE! (plus cleanup!)

```python
from pathlib import Path
import importlib, sys

# 1) module likho (asali-file, permanent Lalit‑pur pattern!)
Path("marktools_module.py").write_text('''
"""marktools — parse/format helpers (demo-module!) 🧩"""

def parse_mark_line(line: str):
    """"Asha: 95" → ("Asha", 95); kharab → None"""
    if ":" not in line:
        return None
    name, mark = line.split(":")
    return name.strip(), int(mark)

def pretty_rows(rows) -> str:
    return ", ".join(f"{n}={m}" for n, m in rows)

if __name__ == "__main__":
    print("SELF-DEMO:", pretty_rows([parse_mark_line("Asha: 95")]))
''')

# 2) import karo (cwd path pe hai — sys.path me "." hote hi!)
sys.path.insert(0, ".")
importlib.invalidate_caches()   # fresh-written files ETIC-recognize ete!
mt = importlib.import_module("marktools_module")

# 3) chalao!
r1 = mt.parse_mark_line("Asha: 95")
r2 = mt.parse_mark_line("bekaar-line")
print("parse-good:", r1, "| parse-bad:", r2)
print("pretty     :", mt.pretty_rows([r1, ("BK", 75)]))

# 4) cleanup (demo-files leave-mat-karo!)
for f in ("marktools_module.py",):
    Path(f).unlink()
import shutil; shutil.rmtree("__pycache__", ignore_errors=True)
print("cleanup ✓ (module kaam kar gaya, files saaf!)")
# Expected OUTPUT:
# parse-good: ('Asha', 95) | parse-bad: None
# pretty     : Asha=95, BK=75
# cleanup ✓ (module kaam kar gaya, files saaf!)
```

## 🐔 Level 2 — Package-Tower + the Import-Cycle Trap 🗼🪢

**Package** = a folder + `__init__.py`: several door-keeper modules of one team.
**CYCLE-TRAP:** `a` imports `b`, `b` imports `a` → an import-lock! 🔒
**CURE:** move the shared thing into a THIRD module (c-team!) — a linear flow!

```python
Path("shoptower").mkdir(exist_ok=True)
Path("shoptower/__init__.py").write_text('"""shoptower package — godown-team!"""\nfrom .pricing import apply_discount\n')
Path("shoptower/pricing.py").write_text('''
def apply_discount(total: float, rate: float = 0.9) -> float:
    return round(total * rate, 2)

if __name__ == "__main__":
    print("SELF:", apply_discount(100))
''')

import shoptower.pricing as spr
from shoptower import apply_discount             # __init__ ne EXPOSE kiya! ✨
print("dot-import :", spr.apply_discount(200))
print("pkg-expose :", apply_discount(200), "(tallest-ka-direct!)")

shutil.rmtree("shoptower", ignore_errors=True); shutil.rmtree("__pycache__", ignore_errors=True)
print("tower demolish ✓ prototype-teach kaam complete!")
# Expected OUTPUT:
# dot-import : 180.0
# pkg-expose : 180.0 (tallest-ka-direct!)
# tower demolish ✓ prototype-teach kaam complete!
```

## ⚠️ 5 Classic Mistakes (Modular)
| # | ❌ Mistake | ✅ Correct |
|---|---|---|
| 1 | import-time side-effects (code runs!) | `__main__`-guard 🛡️ |
| 2 | import-cycle (a↔b deadlock!) 🔒 | shared thing → third-module (c-team!) |
| 3 | importing everything via `from x import *` | explicit-select (clean namespace!) |
| 4 | mixing teams in one file (CSV+plot+billing) | one-team-one-file 🧩 |
| 5 | hardcoding absolute-paths `C:\data\...` | relative-paths/config (portable!) |
