🛩️ 01 — Logging & Monitoring Mindset: The Black-Box Mentality!

**Topic 14.03 | Production_Ready_Mindset — Part 3 of 4**

🌍 **Hook:** A flight crashes — the first investigation looks at the BLACK-BOX:
"what did the plane SAY in its last 1 hour?!" On a production-server too, your
app-crash is the same thing: the boss asks "when did it come, what happened?" — only logs can tell!
`print` = writing a note on paper and throwing it in the wind ✉️💨 — a logger = the black-box!

📖 **MINDSET-First (you learned the syntax in §11, now the THINKING):**
| Thought | Direct meaning |
|---|---|
| WHICH level? | DEBUG=developer-diary · INFO=work-flowing 🟢 · WARNING=heavy-breathing! 🟡 · ERROR=fell! 🔴 · CRITICAL=fire-alarm! 🚨 |
| WHAT to log? | events + decisions + numbers (rows→parse→avg!); clues for future fixes |
| WHAT NOT to log? | PASSWORDS/tokens/cards — always mask/no-store ⚠️ |
| Monitoring = health-checks + dashboards ON TOP of logs (the alert-pyramid!) |

🔍 **WHY the mindset?** library-level syntax takes 1 hour to learn; the real-born-thinking
of logging (what, how much, when to step up) — that's what makes 10x engineers!

## 🐣 Level 0 — Black-Box Setup Live: Timestamps + Levels + FILE 📦

(the syntax §11-05 buddy-pair — now with a standard-shipping vibe: time-name-level!)

````python
import logging
from logging.handlers import RotatingFileHandler

log = logging.getLogger("chaihut-etl ✈️")
log.setLevel(logging.DEBUG)
for h in log.handlers[:]: log.removeHandler(h)          # rebuild-safety!

handler = RotatingFileHandler("blackbox_demo.log", maxBytes=220, backupCount=2)
handler.setFormatter(logging.Formatter("%(asctime)s | %(levelname)s | %(name)s | %(message)s"))
log.addHandler(handler)

log.info("ETL shuru 🟢")
log.debug("parse-attempt batch-1 (dev-detail!)")
log.warning("s2 line me ERR status ∘ strange? 🟡")
log.error("s9 parse FAIL — column-count mismatch 🔴")
log.critical("avg=0 aaya — poori pipeline rok di! 🚨")
log.info("config-file loaded ✓")
log.info("rows=417 processed-sync")
log.info("ETL khatam 🏁")

print("blackbox lines:", open("blackbox_demo.log").read())
# Expected OUTPUT (style — timestamps tumhare run-wale!):
# YYYY-MM-DD hh:mm:ss | INFO | chaihut-etl ✈️ | ETL shuru 🟢
# ...
# (8+ formatted lines dikhni chahiye — STATIC-style, level-wise emoji-labels!)
````

## 🐥 Level 1 — The Magic of ROTATION: Overflow = Self-Archive! 🌀

Once the file fills past maxBytes=220 → `backupCount=2` auto-archives it!
In production this fights the file-growing-to-death problem! Watch the overflow live:

````python
for i in range(40):
    log.info(f"bulk-line #{i} — spam-overflow rotation dikhane!")

files = sorted(f for f in __import__("os").listdir(".") if f.startswith("blackbox_demo"))
print("rotated-files:", files)
cur = open("blackbox_demo.log").read().strip().splitlines()[-1]
print("last-line:", cur[-40:])
for f in files: __import__("os").remove(f)          # cleanup black-box!
print("cleanup ✓")
# Expected OUTPUT:
# rotated-files: ['blackbox_demo.log', 'blackbox_demo.log.1', 'blackbox_demo.log.2']
# last-line: ... bulk-line #39 — spam-overflow ...
# cleanup ✓
````

## 🐔 Level 2 — The Monitoring-Pyramid: ON TOP of Logs (concepts!) 🗼

```
          🚨 ALERTS (page/mail: "ERROR>10/min chal rahi!" — HUMAN jogao!)
        ╱
      📊 DASHBOARDS (metrics: rows/min, error-rate, latency — NUMBERS-ka-satellite!)
╱
  📦 LOGS (har-event ka raw-dera — black-box!)
```
- **Heartbeat** = an info-line every 15-min: "I am alive 💓" (silence = a death-signal!)
- **Count-metric** = counts mined by parsing logs ("parse-fail% > 2" → alert!)
- DS-pipelines: compare rows-in vs rows-out — hidden LOSSES detect data-bugs!

## ⚠️ 5 Classic Mistakes (Logging-Mindset)
| # | ❌ Mistake | ✅ Correct |
|---|---|---|
| 1 | `print` in production (vanishes in the air) | logger + handlers (file/rotation!) |
| 2 | directly logging PASSWORD/tokens POST-payload ⚠️ | mask/ids-only, audit-friendly! |
| 3 | making EVERYTHING DEBUG-level (unreadable soup!) | mix the levels: diary/context/alarm-order |
| 4 | one file growing to GBs (open-crash!) | RotatingFile(maxBytes, backupCount) 🌀 |
| 5 | write the logs but never look at them | heartbeat+alerts (monitoring ON TOP of logs!) |
