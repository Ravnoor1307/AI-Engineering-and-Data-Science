📖 **The 5-Step Navigator Strategy (page-turning-friendship!):**
1. **README/entry FIRST** — where does the story begin? (`main.py/cli.py/README`!)
2. **TREE-VIEW** — folder names reveal the grouping of things (io/ viz/ calc/?)
3. **Trace ONE REQUEST** — one full journey (entry → helpers → return!)
4. **Tests = living-docs** 📗 — read the tests to understand the API's real-behavior!
5. **SMALL changes + verify** — the first edit is a small rehearsal 🎭

In this notebook we build a mini **mystery-repo** (3 files!) — and apply that very strategy
to it LIVE! 🕵️

## 🐣 Step 0 — Materialize a Mystery-Repo 📁 (the teaching spot!)

```python
from pathlib import Path
import shutil

repo = Path("mystery_repo"); repo.mkdir(exist_ok=True)
(repo / "README.md").write_text("# Myst-StackBot v1\nLooks like a small CLI for marks-bands!\n")
(repo / "cli.py").write_text('''from core import band_report

def main():
    rows = [("Asha", 95), ("BK", 62), ("CD", 41)]
    print(band_report(rows))

if __name__ == "__main__":
    main()
''')
(repo / "core.py").write_text('''from utils import classify

def band_report(rows):
    bags = {"A": [], "B": [], "C": [], "F": []}
    for name, mark in rows:
        band = classify(mark)
        bags[band].append(name)
    lines = [f"{b}Band: {', '.join(bags[b]) or '-'}" for b in "ABCF"]
    return "\\n".join(lines)
''')
(repo / "utils.py").write_text('''def classify(m):
    if m >= 85: return "A"
    if m >= 50: return "B"
    if m >= 35: return "C"
    return "F"
''')
for p in sorted(repo.iterdir()):
    print("📄", p.name, f"({p.stat().st_size}b)")
# Expected OUTPUT:
# 📄 README.md (52b)
# 📄 cli.py (168b)
# 📄 core.py (290b)
# 📄 utils.py (120b)
```

## 🐥 Step 1+2 — README-Glimpse + TREE + ENTRY-Find 🔍

How do you find the entry point? LINE-GREP for the `__main__`-guard (the craftsman's pattern!)

### Exceptions
Handle errors gracefully with try/except/else/finally. Avoid bare except; catch specific exceptions.

```python
print("README peeks:", (repo / "README.md").read_text().splitlines()[1])

def find_entry_guards(root: Path):
    hits = []
    for py in sorted(root.glob("*.py")):
        for i, line in enumerate(py.read_text().splitlines(), 1):
            if '__name__' in line:
                hits.append(f"{py.name}:{i}: {line.strip()}")
    return hits

print("\n__main__-guards scan:", find_entry_guards(repo))
print("→ ENTRY = cli.py (fit line 6-8 guard!) 🎯")
# Expected OUTPUT:
# README peeks: Looks like a small CLI for marks-bands!
#
# __main__-guards scan: ['cli.py:6: if __name__ == "__main__":']
# → ENTRY = cli.py (fit line 6-8 guard!) 🎯
```

## 🐔 Step 3 — One-REQUEST Trace + Run (draw the map, verify it!) 🗺️

### Functional tools
map/filter apply functions; reduce aggregates. Often paired with lambdas.

```python
# Imports web: cli → core → utils (grep 'from/import' AND RUN-finally!)
chain = []
for py in ["cli.py", "core.py", "utils.py"]:
    first = (repo / py).read_text().splitlines()[0]
    chain.append(f"{py}: {first}")
print("import-chain scan:")
for c in chain: print("  ", c)

import importlib.util, sys
sys.path.insert(0, "mystery_repo")
importlib.invalidate_caches()
import cli
print("\n--- execute entry (entry=run-path verified!) ---")
rows = [("Asha", 95), ("BK", 62), ("CD", 41)]
from core import band_report
print(band_report(rows))

# now map done → clean up (demo repo tum kaam-parciyal-ghosts-chhodenge!)
shutil.rmtree("mystery_repo", ignore_errors=True)
shutil.rmtree("__pycache__", ignore_errors=True); sys.path.remove("mystery_repo")
print("\nmap-complete ✓ mystery solved → cleaned! 🕵️")
# Expected OUTPUT:
# import-chain scan:
#    cli.py: from core import band_report
#    core.py: from utils import classify
#    utils.py: def classify(m):
#
# --- execute entry (entry=run-path verified!) ---
# ABand: Asha
# BBand: BK
# CBand: CD
# FBand: -
#
# map-complete ✓ mystery solved → cleaned! 🕵️
```

## ⚠️ 5 Classic Mistakes (Reading-Code)
| # | ❌ Mistake | ✅ Correct |
