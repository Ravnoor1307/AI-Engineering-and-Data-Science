🧼 01 — Clean/Readable Code: The 6 Laws of Straight-Forward Style! ✍️

**Topic 14.01 | Production_Ready_Mindset — Part 1 of 4**

🌍 **Hook:** You wrote code that today's-genius understood; 3 MONTHS later you read it
yourself and say "which MADMAN wrote this?!" 😂 Code is not just for the compiler — it's for
HUMANS! In interviews/production/quick-fixes the most VALUABLE skill = READABLE code!

📖 **The 6 Laws (rules-of-thumb!):**
1. **Names = the story** — say WHOSE, WHAT (a, x1 ❌ → `student_avg` ✅)
2. **Small functions** — one function, one answer (a ~20-line ceiling!)
3. **DRY** — copy-paste = the clone-army of future bugs 🪳
4. **Magic-numbers = ghosts** — 0.9?? put it in constants!
5. **Comments = WHY, code = WHAT** — `i += 1  # i badhata` 😑 worthless!
6. **Dead-code = delete** — git-history keeps the memory — scrap it! 🗑️

🔍 **WHY:** bug-hunting 10x faster, adding features 3x faster, and the team loves you! ♥

## 🐣 Level 0 — The Name's Story-Test 🗣️

```python
# ❌ GANDA (padhke bhi matlab zero):
def calc(l):
    t = 0
    for v in l:
        t += v * 0.9          # 0.9?? fest-discount?! 😵 mystery-ghost!
    return t / len(l)

# ✅ CLEAN (seedha-paani, kahani-bolta!):
DISCOUNT_RATE = 0.9            # shop-wide festive-discount 🎉 (ghost = CAPTURED!)

def average_discounted_price(prices: list[float]) -> float:
    """har price pe discount laga ke average 🧾"""
    return sum(p * DISCOUNT_RATE for p in prices) / len(prices)

prices = [100, 200, 350]
print("ganda-answer :", calc(prices))
print("clean-answer :", average_discounted_price(prices), "(same math, NAYI story!)")
# Expected OUTPUT:
# ganda-answer : 195.0
# clean-answer : 195.0 (same math, NAYI story!)
```

## 🐥 Level 1 — Small Functions = Small Blame! 🔨
Getting "load+parse+filter+report" all inside one function = chaos-storm! Break it apart:

```python
# ❌ MEGA-FN (1 darwaza, 5 kaam):
def mega(filepath):
    raw = open(filepath).read()
    rows = []
    for line in raw.splitlines():
        if line.startswith("#") or not line.strip():
            continue
        name, mark = line.split(":")
        if int(mark) < 0:
            continue
        rows.append((name, int(mark)))
    avg = sum(m for _, m in rows) / len(rows)
    return f"students={len(rows)} avg={avg:.1f}"

# ✅ 3 CHHOTE (har-ek ek-jawaab, har-ek TESTABLE!):
def load_marks_file(filepath: str) -> list[tuple[str, int]]:
    """file → (name, mark) rows — comments/blank/negative OUT!"""
    rows = []
    for line in open(filepath):
        if line.startswith("#") or not line.strip():
            continue
        name, mark = line.split(":")
        if int(mark) >= 0:
            rows.append((name.strip(), int(mark)))
    return rows

def class_average(rows: list[tuple[str, int]]) -> float:
    return round(sum(m for _, m in rows) / len(rows), 1) if rows else 0.0

def class_report(rows) -> str:
    return f"students={len(rows)} avg={class_average(rows)}"

rows = load_marks_file("students_marks.txt")
print("rows-load:", rows)
print("chhotte-se-bani report:", class_report(rows))
# Expected OUTPUT:
# rows-load: [('Asha', 95), ('BK', 75), ('Chirag', 100), ('Diya', 77)]
# chhotte-se-bani report: students=4 avg=86.8
```

## ⚠️ 5 Classic Mistakes (Clean-Code)
| # | ❌ Mistake | ✅ Correct |
|---|---|---|
| 1 | `calc/process/handle` zero-meaning names | verb+noun: `average_discounted_price` |
| 2 | comments ECHO (worthless `i+=1 # +1`) | write WHY, let code say WHAT |
| 3 | copy-paste the same-logic everywhere | make a fn, stay DRY 🔁 |
| 4 | `if status == 3:` (3???) | constants: `PAID = 3` 🪙 |
| 5 | keeping dead-code ON-hold (future-plans!) | git-history remembers — DELETE! 🗑️ |
