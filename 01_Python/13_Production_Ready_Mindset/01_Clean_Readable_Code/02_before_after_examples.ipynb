## 🐣 Makeover-1: Sensor-Scam Separator 📡 (BEFORE)

```python
# ❌ BEFORE — anonymous-field guessing game (l[1]? l[3]?? kon? kya?!)
def old_parse(lines):
    out = []
    for l in lines:
        p = l.split("|")
        if len(p) >= 3 and p[2] == "OK":
            out.append((p[0], float(p[1])))
    tot = 0
    for _, v in out:
        tot += v
    return out, round(tot / len(out), 1) if out else 0

RAW = ["s1|42.5|OK", "s2|17.3|ERR", "s3|60.0|OK", "note|x|OK?"]
res, avg = old_parse(RAW)
print("before-out:", res, "| avg:", avg)
# Expected OUTPUT:
# before-out: [('s1', 42.5), ('s3', 60.0)] | avg: 51.2
```

## 🐥 Makeover-1: AFTER ✅ (names! constants! docstrings! small fns!)

```python
STATUS_OK = "OK"            # kabhi "OKAY"/"ok" nahi — ek-hi shabd! 🎯

def parse_sensor_line(line: str):
    """"s1|42.5|OK" → ("s1", 42.5)  |  wahi-line status OK nahi → None"""
    parts = line.split("|")
    if len(parts) < 3 or parts[2] != STATUS_OK:
        return None
    return parts[0], float(parts[1])

def average_of(keepers: list) -> float:
    return round(sum(v for _, v in keepers) / len(keepers), 1) if keepers else 0.0

def sensor_sift(lines: list[str]):
    keepers = [k for k in (parse_sensor_line(l) for l in lines) if k]
    return keepers, average_of(keepers)

res2, avg2 = sensor_sift(RAW)
print("after-out :", res2, "| avg:", avg2)
assert (res, avg) == (res2, avg2)     # SAME outputs, NAYI readability! 🎉
print("✅ same-answer proof — parlour seva success!")
# Expected OUTPUT:
# after-out : [('s1', 42.5), ('s3', 60.0)] | avg: 51.2
# ✅ same-answer proof — parlour seva success!
```

## 🐔 Makeover-2: The Warehouse-Bill Goblin 🧾 (BEFORE → AFTER)

```python
# ❌ BEFORE — nested-if monster + magic percents + same trip-lines thrice!
def old_bill(cart, member):
    tot = 0
    for n, p, q in cart:
        tot += p * q
    if member:
        tot = tot - tot * 0.1
        if tot > 500:
            tot -= 50
    else:
        if tot > 1000:
            tot = tot - tot * 0.05
    return round(tot, 2)

# ✅ AFTER — named-rates + small-pieces (sab combination clear!)
MEMBER_DISCOUNT = 0.10        # members ka flat-10%
BULK_THRESHOLD  = 500         # iske upar → ₹50 flat-off (members only)
BULK_FLAT_OFF   = 50
GUEST_TAX_RULE   = (1_000, 0.05)   # guests: 1000+ pe 5% 😐

def subtotal(cart) -> float:
    return sum(price * qty for _, price, qty in cart)

def apply_deals(total: float, is_member: bool) -> float:
    if is_member:
        total *= (1 - MEMBER_DISCOUNT)
        if total > BULK_THRESHOLD:
            total -= BULK_FLAT_OFF
    else:
        (cut, rate) = GUEST_TAX_RULE
        if total > cut:
            total *= (1 - rate)
    return round(total, 2)

def final_bill(cart, is_member: bool) -> float:
    return apply_deals(subtotal(cart), is_member)

cart = [("chai", 12, 10), ("bun", 20, 12), ("paneer", 150, 5)]
print("member-before:", old_bill(cart, True), "| guest-before:", old_bill(cart, False))
print("member-after :", final_bill(cart, True), "| guest-after :", final_bill(cart, False))
assert old_bill(cart, True) == final_bill(cart, True)
assert old_bill(cart, False) == final_bill(cart, False)
print("✅ both tables — SAME cheez, padhna AISAAN!")
# Expected OUTPUT:
# member-before: 949.0 | guest-before: 1054.5
# member-after : 949.0 | guest-after : 1054.5
# ✅ both tables — SAME cheez, padhna AISAAN!
```
