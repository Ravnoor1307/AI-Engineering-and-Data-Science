## 1. What is Open Source Software?

Open Source Software (OSS) is software whose source code is made available under a license that allows people to use, inspect, modify, and redistribute it, subject to the conditions of that license.

Let's break this definition down.

- Software: A program or application that performs a task.
- Source code: Human-readable instructions written by developers in a programming language such as Python, C++, or Java.
- Open source: The source code is available under a license that grants specific permissions to users.

Open-source software — a recipe with sharing permissions

You can inspect the recipe and, if its license permits, adapt it, improve it, and share your version with others.

The analogy has one important limitation: open-source software is not necessarily free of charge, and its license may impose conditions on how you use or redistribute it.

### What does source code look like?

Here is a tiny Python program that calculates the average of three numbers.

```
marks = [80, 90, 70]average = sum(marks) / len(marks)print("Average marks:", average)
```

If this program is published under an appropriate open-source license, other developers can inspect how it works, modify it, and potentially reuse it according to that license.

For example, you could change it to calculate the average marks of an entire class, add error handling, or turn it into part of a larger student-performance analysis tool.

That ability to study and build on existing software is particularly valuable in Data Science and AI Engineering.

Modern technology is built on software created by many people and organizations working together.

Consider what happens when you train a machine learning model in Python:

Many widely used tools in this workflow are open source. Instead of implementing every mathematical operation, data-processing function, and visualization system yourself, you can build on software maintained by existing communities.

This gives you several advantages:

1. Faster learning: You can study real implementations and documentation.
2. Lower barriers to entry: Many powerful tools can be used without paying a software license fee.

5. Transparency: You can inspect code to understand its behavior, although code availability alone does not guarantee that it is correct or secure.
6. Career development: You can learn from public repositories and contribute to projects that demonstrate your engineering skills.

## 3. Open Source vs. Closed-Source Software

Not every software tool gives you the same rights. Understanding this distinction helps you choose tools for projects, internships, research, and commercial applications.

| Feature                | Open-source software                   | Closed-source software                           |
| ---------------------- | -------------------------------------- | ------------------------------------------------ |
| Source code            | Available under an open-source license | Generally not publicly available                 |
| Inspect implementation | Generally permitted                    | Usually restricted                               |
| Modify the code        | Permitted under license conditions     | Usually restricted                               |
| Redistribute copies    | Permitted under license conditions     | Usually restricted                               |
| Cost                   | May be free or paid                    | May be free or paid                              |
| Examples               | Python, pandas, scikit-learn           | Proprietary applications such as Microsoft Excel |

These are general patterns, not absolute rules for every product or service.

### An important distinction: free software vs. open source

The word free can mean two different things:

- Free of charge: You can use something without paying.
- Freedom to use, study, modify, and share: Your rights are granted by its license.

For example, a proprietary application might offer a free version without revealing its source code. Conversely, a company might charge money for a product built using open-source software.

Therefore, free software is not automatically open source, and open-source software is not automatically free of charge.

## 4. How does Open Source work?

Open-source development is a collaborative process. Let's understand it using a fictional project called `student-performance-analyzer`.

1. A developer creates a project

   A developer writes code to analyze student marks and publishes the project in a Git repository.
2. The project receives an open-source license

   The license explains what other people may do with the code and what conditions apply.
3. Other developers use it

   Someone uses the project to analyze attendance and academic performance.
4. A contributor proposes an improvement

   Another developer adds visualizations or fixes a bug and submits a change for review.
5. The project improves over time

   Maintainers review contributions, test changes, document features, and publish releases.

### Where does GitHub fit in?

GitHub is a platform for hosting Git repositories and collaborating on software projects. It is widely used by open-source communities, but GitHub itself is not the definition of open source.

Understand these four terms:

- Git: A version-control system that tracks changes in files.
- GitHub: A platform for hosting repositories and collaborating.
- Repository: A project space containing code, documentation, and often its change history.
- License: The legal terms that determine how others may use, modify, and distribute the project.

A project can be hosted on GitHub and still be closed source if its license does not grant open-source permissions.

Similarly, a public repository without a license does not automatically grant everyone the right to reuse its code. Copyright restrictions generally remain in place unless permissions are granted.

### Common open-source collaboration terms

| Term         | Meaning                                                                               |
| ------------ | ------------------------------------------------------------------------------------- |
| Fork         | Your own copy of a repository on a hosting platform, commonly used to propose changes |
| Clone        | Download a repository and its Git history to your computer                            |
| Commit       | A recorded set of changes                                                             |
| Branch       | An independent line of development                                                    |
| Pull request | A proposal to merge changes into another branch or repository                         |
| Issue        | A place to report bugs, request features, or discuss work                             |
| Maintainer   | Someone responsible for managing a project and its contributions                      |

Career connection: Understanding these concepts helps you collaborate with other developers and contribute to projects beyond your own coursework.

## 6. Open Source in AI Engineering

Now let's move from Data Science to AI Engineering.

Data Science focuses on understanding data, discovering patterns, and generating insights. AI Engineering focuses on building, integrating, testing, deploying, and maintaining AI-powered systems.

There is significant overlap between these fields, and open-source tools are important in both.

### The AI Engineering pipeline

Imagine that you want to build an AI assistant that answers questions about your college's notes and documents.

1\. Collect and prepare documents

Extract text from PDFs, organize files, and prepare the content.

Examples: Python, PyMuPDF

2\. Create embeddings

Convert text into numerical representations that capture aspects of meaning.

Examples: sentence-transformers, Transformers

3\. Retrieve relevant information

Find document passages related to the user's question.

Examples: FAISS, Chroma

4\. Generate an answer

Send the question and retrieved context to a language model.

Examples: Transformers, a locally hosted model, or a hosted model API

5\. Deploy and monitor the application

Build an interface, expose an API, test quality, and monitor performance.

Examples: FastAPI, Docker, Streamlit

This architecture is an example of Retrieval-Augmented Generation (RAG). It combines document retrieval with a language model so that the system can use relevant information from your documents when answering questions.

Notice that you do not have to build every component from scratch. Open-source libraries can provide much of the underlying functionality.

## 7. Is an open-source AI model the same as open-source software?

No. This is one of the most important distinctions in AI Engineering.

A machine learning model is not just a program. It can also involve learned parameters, training data, training code, evaluation methods, and documentation.

Consider a language model.

Model implementation

The software that defines how the model operates

Model weights

Numerical parameters learned during training

Training information

Data, training procedures, and related documentation

These are distinct components, and access to one does not automatically mean access to all the others.

### Four concepts you should distinguish

| Concept              | What it means                                                                           |
| -------------------- | --------------------------------------------------------------------------------------- |
| Open-source software | Software whose license grants recognized open-source rights to its source code          |
| Open weights         | A model's learned numerical parameters are made available                               |
| Open training code   | The code used to train a model is available                                             |
| Open training data   | The data used to train a model is made available, subject to its terms and restrictions |

A model may have downloadable weights without providing its complete training dataset or training pipeline. Some models also impose additional restrictions on usage.

Consequently, downloadable model weights do not automatically make a model fully open source.

The Open Source Initiative has also published a definition specifically addressing AI systems, recognizing that openness in AI involves more than releasing weights alone.

### Why does this matter to you?

Suppose you are developing an AI chatbot for a business.

You need to ask:

- Can I download and run the model?
- Can I modify the implementation?
- Does the license permit commercial use?
- Are there restrictions on redistribution or serving the model?
- Can I use the model with my intended data?
- Do I have enough information to evaluate its limitations and risks?

A model that works well for experimentation might not be appropriate for a commercial product.

## 8. Open-source licenses: What you must understand

Open-source licenses are legal permissions that determine how software can be used, modified, and redistributed.

Here are three important licenses you'll frequently encounter.

| License            | General characteristics                                                                                                                                 |
| ------------------ | ------------------------------------------------------------------------------------------------------------------------------------------------------- |
| MIT License        | Permissive; generally allows use, modification, distribution, and commercial use, subject to conditions such as retaining copyright and license notices |
| Apache License 2.0 | Permissive; includes express patent-related provisions and requirements such as preserving relevant notices                                             |
| GNU GPL            | A family of copyleft licenses; distributing covered derivative works can require providing source code under the applicable GPL terms                   |

These descriptions are simplified. The precise obligations depend on the license, the software, and how you use or distribute it.

### Practical example

Imagine you find a useful open-source Python library for your AI project.

Before using it, you should:

1. Find its `LICENSE` file or official licensing information.
2. Identify the actual license and its version.
3. Check whether your intended use is permitted.
4. Follow requirements for attribution, notices, source-code availability, or other obligations where applicable.
5. Review the licenses of model weights, datasets, and bundled dependencies separately.

Do not assume that everything in a public GitHub repository has the same license.

### What if you build your own GitHub project?

For example, you create a student-performance prediction system and want other developers to use it.

- If you choose an appropriate open-source license, you grant permissions described by that license.
- If you do not provide a license, making your repository public does not automatically give everyone permission to reuse its code.

Teacher's tip: Read the licenses of tools you use, and include an appropriate `LICENSE` file when you intentionally publish your own project as open source.

## 10. Limitations and risks of Open Source

Open source is powerful, but it is not a guarantee of quality, safety, or suitability.

| Challenge                | Why it matters                                                       | What you should do                                                       |
| ------------------------ | -------------------------------------------------------------------- | ------------------------------------------------------------------------ |
| Security vulnerabilities | A library may contain exploitable bugs                               | Use trusted sources, update dependencies, and review security advisories |
| Outdated packages        | Old dependencies may stop working or receive no fixes                | Check maintenance status and compatibility                               |
| Dependency conflicts     | Different libraries may require incompatible versions                | Use virtual environments and lock dependencies                           |
| Limited support          | Some projects have no guaranteed support team                        | Review documentation, community activity, and support options            |
| Licensing issues         | Components can impose different obligations                          | Review licenses before reuse or distribution                             |
| Hidden costs             | Training, GPUs, storage, and cloud infrastructure may be expensive   | Estimate total operating costs                                           |
| Model bias and errors    | An AI model can produce inaccurate or unfair results                 | Evaluate it on representative data and monitor performance               |
| Data privacy             | Data sent to an external service may be processed under its policies | Check data-handling terms and avoid exposing confidential information    |

### An important security lesson

Suppose a tutorial asks you to install a package using:

```
pip install some-package
```

That command installs a package from the configured package index. It does not guarantee that the package is trustworthy.

Before installing unfamiliar dependencies, verify the package name, source, maintainer, release history, and relevant security information.

A similarly named package could be malicious. This is one reason developers use dependency scanning and controlled environments.

### Is open source always better than proprietary software?

No.

Open source is often a strong choice when you need flexibility, transparency, customization, and community-developed tools.

Proprietary tools can be preferable when you need a particular managed service, commercial support agreement, specialized functionality, or integration with an existing enterprise system.

Many real-world organizations use a combination of both.

7. Publish your project on GitHub, following the licenses and privacy requirements that apply.

For a responsible prediction system, avoid claiming that the model can reliably predict a student's future from a small or biased dataset. Treat predictions as estimates, evaluate errors, and avoid using them as the sole basis for consequential decisions about students.

### Why this is a valuable project

You would be learning more than machine learning. You would also practice data cleaning, experimentation, evaluation, software organization, documentation, and version control.

That combination is what makes a portfolio project more meaningful than a notebook containing only a few model-training commands.

## 15. Final revision notes

Remember these key ideas:

- Open Source Software: Software distributed under a license that grants rights to access its source code and use, modify, and redistribute it under specified conditions.
- Data Science: Uses tools such as Python, NumPy, pandas, Matplotlib, Jupyter, and scikit-learn to analyze data and build traditional ML models.
- AI Engineering: Uses tools such as PyTorch, Transformers, FastAPI, and Docker to develop and deploy AI applications.
- Git and GitHub: Help you manage changes, collaborate, and showcase projects.
- Open-source licensing: Determines what you are permitted to do with software and what obligations apply.
- Open AI models: Model weights, software code, training code, and training data are distinct components; availability of one does not guarantee openness of all the others.
- Responsible use: Evaluate security, quality, privacy, bias, cost, and licensing before using a tool in a real project.

The most important lesson is this: Open source is not just a way to obtain software. It is a way to learn from existing work, collaborate with developers, and build increasingly sophisticated systems without having to start from zero.
