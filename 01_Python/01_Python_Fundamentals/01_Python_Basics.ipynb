```python
# Implicit and explicit type conversion

integer_value = 10
float_value = 2.5
mixed_result = integer_value + float_value

number_text = "125"
price_text = "99.50"
converted_number = int(number_text)
converted_price = float(price_text)

print(mixed_result, type(mixed_result))
print(converted_number, type(converted_number))
print(converted_price, type(converted_price))
print(str(converted_number), type(str(converted_number)))
```

```python
#  Truth-value conversion edge cases

print(bool(0))
print(bool(1))
print(bool(""))
print(bool("False"))
print(bool([]))
print(bool([0]))
print(int(3.9))
```
