## Conditional Statements : 

```python
# If Statement
age = 20
if age >= 18:
    print("Adult")
print("Thus line runs after the if statement.")
```

```python
# If-Alse Statement
number = 7
if number % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
# If-Elif-Else Chain
score = 86

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "Needs improvement"

print(f"Grade: {grade}")
```

```python
# Order matters in If-Elif Chain
score = 95

# This order is correct: test the most restrictive/highest threshold first.
if score >= 90:
    label = "A"
elif score >= 80:
    label = "B"
else:
    label = "Other"

print(label)
```

```python
# Nested Conditional
age = 21
has_valid_id = True

if age >= 18:
    if has_valid_id:
        print("Entry permitted")
    else:
        print("Valid ID required")
else:
    print("Minimum age not met")
```

```python
# Boolean Logic : and, or, not
age = 22
has_ticket = True
is_guest_listed = False
```

```python
can_enter = age >= 18 and (has_ticket or is_guest_listed)
```

```python
# Short-Circuit Evaluation
user_name = "Ravnoor"

# The second expression is evaluated only when the first part is truthy.
if user_name and user_name.strip():
    print("A non-empty name was supplied")
else:
    print("Name is missing")
```

```python
# Safe fallback: the right-hand side is used when the left-hand side is falsy.
provided_label = ""
label = provided_label or "Untitled"
print(label)
```

```python
# truthly and Falsy values

values = [0, 1, 0.0, "", "False", [], [0], {}, {"x": 1}, None]
for value in values:
    print(repr(value), "->", bool(value))
```

```python
# Explicit membership and range checks
marks = 74

if 0 <= marks <= 100:
    print("Marks are in the valid range")
else:
    print("Invalid marks")
```

```python
# match-case: 

http_status = 404

match http_status:
    case 200:
        message = "OK"
    case 400:
        message = "Bad request"
    case 404:
        message = "Not found"
    case 500:
        message = "Server error"
    case _:
        message = "Unknown status"

print(message)
```

```python
# match-case: 

status = 403

match status:
    case 200 | 201:
        print("Successful response")
    case 401 | 403:
        print("Authentication or permission issue")
    case _:
        print("Other response")
```

```python
# match-case: CAPTURE AND SEQUENCE PATTERNS

point = (0, 8)

match point:
    case (0, 0):
        print("Origin")
    case (0, y):
        print(f"On the y-axis at {y}")
    case (x, 0):
        print(f"On the x-axis at {x}")
    case (x, y):
        print(f"Point: x={x}, y={y}")
    case _:
        print("Not a two-item point")
```

```python
# match-case: MAPPING PATTERN

request = {"method": "GET", "path": "/students", "authenticated": True}

match request:
    case {"method": "GET", "path": path, "authenticated": True}:
        print(f"Read request for {path}")
    case {"method": "POST", "path": path}:
        print(f"Create request for {path}")
    case _:
        print("Unsupported or incomplete request")
```

```python
# match-case: GUARD CONDITIONS

point = (4, 4)

match point:
    case (x, y) if x == y:
        print("Point lies on the line y = x")
    case (x, y) if x == 0 or y == 0:
        print("Point lies on an axis")
    case (x, y):
        print("General point")
    case _:
        print("Invalid point")
```

```python
# LOOKAHEAD AND LOOKBEHIND ASSERTIONS

import re

text = "USD 45, EUR 30, USD 120"

# Positive lookbehind: digits preceded by the literal text USD.
print(re.findall(r"(?<=USD )\d+", text))

# Positive lookahead: digits followed by the literal text USD.
print(re.findall(r"\d+(?= USD)", "45 USD and 30 EUR"))

# Negative lookahead: words not followed by a digit.
print(re.findall(r"\b[A-Za-z]+\b(?!\d)", "item1 item two3 three"))
```

```python
# BASIC FORMAT VALIDATION EXAMPLES

import re

email_pattern = re.compile(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}")
phone_pattern = re.compile(r"\+?[0-9][0-9 -]{7,14}[0-9]")
date_shape_pattern = re.compile(r"\d{4}-\d{2}-\d{2}")

print(bool(email_pattern.fullmatch("student@example.com")))
print(bool(phone_pattern.fullmatch("+91 98765 43210")))
print(bool(date_shape_pattern.fullmatch("2026-11-20")))

# Regex shape validation does not prove that a date exists or an email is deliverable.
```

```python
# DATE VALIDATION: REGEX SHAPE + datetime CALENDAR CHECK

import re
from datetime import datetime

date_text = "2026-02-30"
if re.fullmatch(r"\d{4}-\d{2}-\d{2}", date_text):
    try:
        parsed_date = datetime.strptime(date_text, "%Y-%m-%d").date()
        print("Valid date:", parsed_date)
    except ValueError:
        print("Date-shaped text, but not a real calendar date")
else:
    print("Wrong date format")
```

```python
# REGEX FOR LOG PARSING

import re

log_pattern = re.compile(
    r"(?P<timestamp>\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2})\s+"
    r"(?P<level>INFO|WARNING|ERROR)\s+"
    r"(?P<message>.*)"
)

line = "2026-11-20 09:15:12 WARNING Disk usage reached 82%"
match = log_pattern.fullmatch(line)
if match:
    print(match.groupdict())
else:
    print("Malformed log line")
```

```python
#  MINI IMPLEMENTATION - SUPPORT LOG ANALYZER
# Concepts: multiline strings, splitlines, strip, regex, for, if, match-case,
# continue, dictionaries, counting, and formatted output.

import re

log_text = """2026-11-20 09:12:01 INFO User login successful user=rav
2026-11-20 09:15:12 WARNING Disk usage reached 82% path=/data
2026-11-20 09:16:44 ERROR Failed to connect host=api.example.com
2026-11-20 09:17:30 INFO Contact support@example.com for help
INVALID LINE WITHOUT TIMESTAMP
"""

log_pattern = re.compile(
    r"(?P<timestamp>\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2})\s+"
    r"(?P<level>INFO|WARNING|ERROR)\s+"
    r"(?P<message>.*)"
)
email_pattern = re.compile(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}")

counts = {"INFO": 0, "WARNING": 0, "ERROR": 0}
malformed_lines = 0
emails_found = []

for raw_line in log_text.splitlines():
    line = raw_line.strip()
    if not line:
        continue

    parsed = log_pattern.fullmatch(line)
    if parsed is None:
        malformed_lines += 1
        print("Malformed:", line)
        continue

    details = parsed.groupdict()
    level = details["level"]
    counts[level] += 1

    match level:
        case "INFO":
            category = "Informational"
        case "WARNING":
            category = "Needs attention"
        case "ERROR":
            category = "Failure"
        case _:
            category = "Other"

    print(f"[{category}] {details['timestamp']} - {details['message']}")
    emails_found.extend(email_pattern.findall(details["message"]))

print("\nSUMMARY")
for level, count in counts.items():
    print(f"{level}: {count}")
print("Malformed lines:", malformed_lines)
print("Emails:", emails_found)
```
