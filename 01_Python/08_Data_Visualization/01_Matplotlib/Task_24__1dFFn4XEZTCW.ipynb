Two real datasets are used — a power-station capacity dataset and a stock-market dataset — plus several mathematical surfaces.

Keep these concepts ready while you work:

- **Subplots**: `plt.subplots(1, 2, figsize=...)` returns a figure and an array of axes; plot on `ax[0]` and `ax[1]` separately, and give *each* subplot its own title, xlabel and ylabel.
- **Reference lines**: `ax.axhline(value, c='red')` and `ax.axvline(value, c='black')` draw full-width lines — perfect for marking average values.
- **Color as a third dimension**: pass a numeric array or category codes to the `c=` argument and pick a `cmap`. For text labels (Power Station names), convert them to numbers with `sklearn.preprocessing.LabelEncoder`.
- **3D figures**: create `plt.figure()` then `ax = plt.subplot(projection='3d')`; use `ax.scatter3D(x, y, z, c=...)`, `ax.plot_surface(xx, yy, z, cmap='viridis')`, `ax.contour(...)` and `ax.contourf(...)`.
- **Surfaces**: build 1D `x`, `y` with `np.linspace`, expand with `np.meshgrid`, then compute `z` from a formula.
- **Pandas plotting**: `df.plot(kind='line', y='Close', use_index=True, ax=ax)` draws lines straight from a DataFrame, and `.plot(kind='scatter', x=..., y=..., c='Symbol', colormap='viridis')` colors points by a categorical column.
