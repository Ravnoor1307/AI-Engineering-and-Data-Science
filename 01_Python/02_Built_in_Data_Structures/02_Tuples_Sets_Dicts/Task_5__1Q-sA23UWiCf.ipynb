# Tuple


### Section 1 — The Tuple toolbox

Tuples are immutable, ordered records. For these five problems you will mostly *read* and *group* tuples, never modify them. Keep these tools in mind:

- **`zip()`** pairs two iterables position by position — perfect for joining tuples that share their first element (Problem 1).
- **Looping over a tuple** works exactly like a list: `for item in t:` or index-based `t[i]`. You cannot `append`, but you can build a **new** tuple or list from the old one.
- **Slicing and indexing** (`t[i]`, `t[i:j]`) are your readers; `t[-1]` reaches the last item.
- **`tuple(...)` / `list(...)`** convert between container types — useful when a problem asks for the output as a list of tuples (Problem 1) or as a tuple of results (Problem 2, Problem 4).
- **Identity checks** — two tuples are "same" when each element equals the element at the same index (Problem 3). `zip` both tuples and compare pair by pair, or use the direct `==`.
- **`type(x)`** distinguishes a tuple from a list or a set — the tool for counting container types (Problem 4).
- **Input collection** into tuples (Problem 5): read the user's answer with `input()`, store each record as a `(name, higher_ed, skill, grad_year)` tuple inside a list, then compare the fields against the job requirement.

Notice the recurring shape: *build a new structure by filtering or inspecting an old one*. Because tuples cannot change, every solution constructs the result fresh.

# Set

### Section 2 — The Set toolbox

Sets are unordered, unique, hashable-element collections with O(1) membership. These five problems are all about *commonality* and *uniqueness*, which is precisely what sets are for:

- **`set(iterable)`** builds a unique collection from any list or string — the single tool behind "common elements" (Problem 1), "unique vowels" (Problem 2), and "binary string" (Problem 3).
- **`&` (intersection)** keeps items present in several sets: `set(a) & set(b) & set(c)` finds the elements common to three lists at once (Problem 1).
- **`len(set_1 & set_2)`** counts how many distinct items are shared — used to count unique vowels across a case-sensitive vowel set (Problem 2). Remember the problem says "Lowercase and uppercase vowels will be taken as different", so the vowel set must contain both cases.
- **`len(set(s))`** tells you how many *distinct* characters a string uses — the entire test for a "binary string" is `len(set(s)) == 2` (Problem 3).
- **`s.update(iterable)`** pours many items into a set — sweep a list of arrays into one set to build their union (Problem 4).
- **List comprehension with `in`** can emulate intersection without set operators — the requirement of Problem 5 ("only use list comprehension") is exactly `[item for item in lst1 if item in lst2]`.
- When order clashes with sets, convert back with `list(...)`/`sorted(...)` for the required output.

The mental model for all five: **convert the input to a set, intersect/union/count, then present the result in the required form.**

# Dictionary

### Section 3 — The Dictionary toolbox

Dictionaries store key→value pairs with O(1) keyed lookup. These five problems exercise reading, iterating, grouping, and building dicts:

- **Iterating a dict** yields its **keys** by default: `for k in d:` — and `d[k]` (or the value in a `d.items()` loop) reaches the value. Problem 1 ("key with maximum unique values") is a scan: for each key compute `len(set(d[k]))`, remember the key with the largest count.
- **`str.split()`** splits a sentence into words — then, for each word, check membership in a replacement dict with `if word in repl_dict` and map it (Problem 2). Words not found pass through unchanged.
- **`range(0, n, 2)`** steps by 2 — the pattern used to re-pair a flat list into key/value pairs when building a list of dictionaries (Problem 3). Each step picks `test_list[i]` as the key and `test_list[i+1]` as the value, stamped with the shared `key_list`.
- **`for k, v in list_of_pairs:`** unpacks a list of tuples — the direct way to build a dict where each first item becomes a key and each second item a value (Problem 4). Note the expected output wraps each value in a list: `d[name] = [value]`.
- **`sorted(d)`** sorts a dict's **keys**; `sorted(d[k])` sorts a key's value list. Iterating the sorted keys and rebuilding a fresh dict produces the sorted dictionary of Problem 5.

Recurring idea: a dict is either **built** (from pairs, from lists) or **scanned** (keys and their values), and most problems are one or the other. Decide which before writing code.
