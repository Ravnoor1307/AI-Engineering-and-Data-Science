3. **A working-simplicity-first attitude** — start with the simplest correct solution, run it on the example, then optimise only if asked. Many questions repeat across interviews, so internalising the *approach pattern* matters more than memorising exact code.

## 📖 1. kth largest/smallest

```python
L = [12, 23, 1, 4, 56, 34, 22, 3]
L.sort()            # ascending
L[k - 1]            # kth smallest
L[-k]               # kth largest
```

- `sort()` first, then index
- kth smallest = index k-1
- kth largest = index -k
- O(n log n) sort based — fine for interviews

````python
# 1. Find the kth largest/smallest item from a list

L = [12,23,1,4,56,34,22,3]
k=3

L.sort(reverse=True)
print(L[k-1])
````

## 📖 2. Check array sorted

```python
L = [1, 2, 3, 4, 5]
flag = True
for i in range(len(L) - 1):
    if L[i] > L[i + 1]:
        flag = False
        break
```

- One loop, adjacent comparison
- `L[i] > L[i+1]` — means descending
- Break early = efficient

````python
# 2. Check if an array is sorted
L = [1,2,3,4,5]

flag = True

for i in range(0,len(L)-1):
  if L[i] > L[i+1]:
    flag = False

if flag:
  print('sorted')
else:
  print('not sorted')
````

## 📖 3. Min / Max

```python
L = [21, 1, 34, 23, 54, 11, 10]
max_val = L[0]
for x in L:
    if x > max_val: max_val = x
```

- Compare-and-update in one pass
- `min()`, `max()` are also builtins
- Manual loop preferred in interviews

````python
# 3. Find Min/Max in a given array

L = [21,1,34,23,54,11,10]

max_val = L[0]

for i in L:
  if i > max_val:
    max_val = i

print(max_val)
````

## 📖 4. First element occurred k times

```python
d = {}            # count map
for i in L:
    d[i] = d.get(i, 0) + 1
    if d[i] == k:
        print(i)   # the first element to reach count k
        break
```

- Count the frequency of each element with a dictionary.
- `d.get(i, 0)` returns the current count (0 if the key is new), so `d[i] = d.get(i, 0) + 1` is the safe one-line counting idiom.
- The moment an element's count reaches `k`, that element is the answer — print it and `break` immediately.
- Because you scan left to right and stop at the first hit, you naturally get the *first* element to occur `k` times.
- Time is O(n) with a single pass over the list; the dict gives O(1) increments.


````python
# 4. Find the first element to occur k times in an array

L = [1,1,2,3,4,4,5,5]

k = 2

d = {}

for i in L:
  if i in d:
    d[i] = d[i] + 1
  else:
    d[i] = 1

for i in d:
  if d[i] == k:
    print(i)
    break
````

## 📖 5. Duplicates

```python
d = {}
for i in L:
    if i in d:
        print(i)      # duplicate
    else:
        d[i] = 1
```

- Track with a seen set/dict
- A second appearance → duplicate
- O(n) time, O(n) space

````python
# 5. Find duplicates in an array
L = [1,1,2,3,4,4,5,5]

d = {}

for i in L:
  if i in d:
    d[i] = d[i] + 1
  else:
    d[i] = 1

for i in d:
  if d[i] > 1:
    print(i)


````

## 📖 6. Left rotate by d

```python
L = [1, 2, 3, 4, 5]; d = 2
for i in range(d):
    L.append(L.pop(0))     # pop front → push end
```

- Each rotation: remove the first element and place it at the end
- `L[1:] + L[:1]` — slice trick also works
- d rotations

````python
# 6. Rotate array to left d items

L = [1,2,3,4,5]
rotate = 2

for i in range(rotate):
  temp = L[0]
  for j in range(0,len(L)-1):
    L[j] = L[j+1]
  L[len(L)-1] = temp

print(L)
````

## 📖 7. Intersection of sorted arrays

```python
while i < len(a) and j < len(b):
    if a[i] == b[j]:
        print(a[i]); i += 1; j += 1
    elif a[i] < b[j]:
        i += 1
    else:
        j += 1
```

- Two pointers approach
- Both sorted → compare and advance
- O(n+m) — more efficient than sets

````python
a = [1,2,3,4,5,8]
b = [3,6,7,8]

i=j=0

while i<len(a) and j<len(b):
  if a[i] == b[j]:
    print(a[i])
    i+=1
    j+=1
  elif a[i] > b[j]:
    j+=1
  else:
    i+=1
````

## 📖 8. Subarray with given sum

```python
d = {0: -1}
curr_sum = 0
for i, x in enumerate(L):
    curr_sum += x
    if curr_sum - S in d:
        print(d[curr_sum - S] + 1, i)
    d[curr_sum] = i
```

- Prefix-sum map: if curr-S appeared earlier, the sum is found
- Sliding window or prefix-sum
- O(n)

````python
# 8. Find continous subarray with a given sum(given non-negative numbers)
# return the starting and ending index of the subarray
# return 1st subarray in case of multiple

L = [1,22,13,7,9,11,10]
S = 16

for i in range(0,len(L)):
  subarray = []
  for j in range(i,len(L)):
    subarray.append(L[j])
    if sum(subarray) == S:
      print(subarray)
````

## 📖 9. Left smaller / right greater

```python
for i in range(1, len(L) - 1):
    if max(L[:i]) < L[i] < min(L[i+1:]):
        print(L[i])
```

- Check: left max < element < right min
- Precompute max_left/min_right arrays
- Naive slicing O(n²), precompute O(n)

````python
# 9. Find element with left side smaller/right side greater in an array

L = [3,1,2,5,8,7,9]

for i in range(1,len(L)-1):
  flag = True

  for j in range(0,i):
    if L[j] > L[i]:
      flag = False

  for k in range(i+1,len(L)):
    if L[k] < L[i]:
      flag = False

  if flag:
    print(L[i])

````

````python
L = [3,1,2,5,8,7,9]

for i in range(1,len(L)-1):
  if max(L[:i]) < L[i] < min(L[i+1:]):
    print(L[i])
````

````python
L = [3,1,2,5,8,7,9]

max_arr = []
min_arr = []

max_val = L[0]
min_val = L[-1]

for i in L:
  if i>max_val:
    max_val = i
  max_arr.append(max_val)

for i in range(len(L)-1,-1,-1):
  if L[i] < min_val:
    min_val = L[i]
  
  min_arr.insert(0,min_val)


for i in range(1,len(L)-1):
  if max_arr[i-1] < L[i] < min_arr[i+1]:
    print(L[i])
````

## 📖 10. Max subarray (Kadane)

```python
curr_sum = 0
best_sum = float('-inf')   # or the first element
for x in L:
    curr_sum = max(x, curr_sum + x)   # restart the subarray, or continue it
    best_sum = max(best_sum, curr_sum)
```

- Kadane's algorithm keeps a *running* sum `curr_sum`. For each element, the running subarray either **continues** (add the element to the current subarray) or **restarts** (the element alone is better). That decision is exactly `max(x, curr_sum + x)`.
- `best_sum` remembers the best running sum ever seen — the maximum subarray sum.
- Initialize `best_sum` to a very small value (`float('-inf')`) or `L[0]` so negative-only arrays still work.
- Complexity is O(n) in a single pass — the classic efficient answer.
- Advanced variants also track the *indices* of the best subarray; the last code cell in this section shows that version, tracking sequences instead of just the sum.


````python
# 10. Maximum sum subarray
L = [-2,4,7,-1,6,-11,14,3,-1,-6]

d = {}

for i in range(0,len(L)):
  subarray = []
  for j in range(i,len(L)):
    subarray.append(L[j])
    d[sum(subarray)] = subarray

max_val = max(d.keys())

for i in d:
  if i == max_val:
    print(d[i])
````

````python

L = [-2,4,7,-1,6,-11,14,3,-1,-6]


curr_sum = 0
curr_seq = []
best_sum = L[0]
best_seq = []

for i in L:
  if i + curr_sum > i:
    curr_sum = curr_sum + i
    curr_seq.append(i)
  else:
    curr_sum = i
    curr_seq.clear()
    curr_seq.append(i)

  if curr_sum > best_sum:
    best_sum = curr_sum
    best_seq = curr_seq

print(best_sum,best_seq)


````

### Approach notes — the practice set (problems 11–15)

These five problems are the "do it yourself" set. Each one either uses a pattern you have already seen or a small idea you should recognise:

- **Problem 11 — Sort 0s/1s:** the simplest correct approach is *count and rewrite*: count how many zeros exist, write that many `0` at the front and the rest `1`. If you want the O(n) two-pointer alternative, move zeros to the front while swapping every `1` to the back (the "Dutch national flag" idea with two buckets).
- **Problem 12 — Move all negative numbers to the end:** two-pointer swap — a left pointer that scans for `+` values and a right pointer that scans for `-` values; swap them when found, then advance both. One pass, in place.
- **Problem 13 — Maximum product subarray:** track the *maximum* and *minimum* running product (because a negative × negative can become the largest). At each element the new max is `max(x, max_so_far*x, min_so_far*x)`, and likewise for the min; keep the best max seen.
- **Problem 14 — Union of two arrays:** the two classic routes: `list(set(a) | set(b))` when order does not matter (O(n)), or merge the two sorted arrays with two pointers when order must be preserved.
- **Problem 15 — Pythagorean triplets:** sort the array, then for each `c` (from the largest end) use two pointers `a` and `b` on the remaining left part and check `a*a + b*b == c*c`, moving pointers according to the comparison — the same two-pointer search you used for intersection, applied to squares.
