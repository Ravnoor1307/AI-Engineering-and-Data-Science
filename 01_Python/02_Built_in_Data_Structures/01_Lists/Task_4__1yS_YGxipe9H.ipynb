### `Problem 13:` Write a list comprehension to print the following matrix

[[0, 1, 2], [3, 4, 5], [6, 7, 8]]

### Approach hint — Problem 13

Build the specific matrix `[[0,1,2],[3,4,5],[6,7,8]]` with one **list comprehension** — no hard-coding the rows.

- Row `0` is `[0,1,2]`, row `1` is `[3,4,5]`, row `2` is `[6,7,8]`. In general, row `r` starts at `3*r` and continues for three numbers.
- Outer comprehension produces the rows, inner comprehension produces the cells: for each row index `i` in `range(3)`, each column value is `j + 3*i` for `j` in `range(3)`.
- So the expression `[[j + 3*i for j in range(0,3)] for i in range(0,3)]` yields exactly the required matrix.
- Read it as: "for each of the three rows, generate the three column values for that row." The inner `for` completes a row; the outer `for` stacks the rows.
- This is the canonical *nested list comprehension for a grid* — row expression outside, cell expression inside.
