### Group 1 — Problems 1 to 5 : list & string functions

These five problems are short and focused; each needs one clear idea from Session 6.

- **Uniqueness (P1):** to return a list of unique elements you can either keep an *order-preserving loop* — append `x` only `if x not in result` — or use `list(set(L))`, which loses order. The sample output `[1,2,3,4,5]` keeps the original order, so prefer the `not in` loop for this one.
- **String splitting + sorting (P2):** `seq.split('-')` turns the hyphen-separated string into a list of words; `sorted(...)` orders them alphabetically; `'-'.join(...)` rebuilds the hyphen-separated string. The function receives the string and returns the transformed string.
- **Case counting (P3):** loop over the characters with `c.isupper()` and `c.islower()` to count upper vs lower letters; ignore characters that are neither (spaces, punctuation). Return both counts.
- **Even filter (P4):** keep only the items where `x % 2 == 0`. A `for` loop with an accumulator list, or `[x for x in L if x % 2 == 0]`, both match the sample.
- **Perfect number (P5):** a number `n` is perfect when the sum of its *proper* divisors (all divisors from 1 up to but excluding `n`) equals `n`. So loop `for i in range(1, n):` and sum the `i` where `n % i == 0`, then compare with `n`.

Group reflex: for problems that say "returns", use `return` — not `print` — so the caller gets the value.

- **Most frequent word (P7):** tokenise with `s.split()`, then count occurrences by building a dict: `d[word] = d.get(word, 0) + 1`. Find the maximum count with `max(d.values())`, then scan the dict for the first word with that count and report `'word -> count'`.
- **Histogram of bin size 10 (P8):** values `1–10` fall in bin `11–20`? No — bin `1–10`, then `11–20`, `21–30`, ... Here the sample calls them `11-20`, `21-30`, `31-40`, `41-50` while counting values greater than the lower edge (e.g. `13` counts in `11-20`). Compute `floor(min/10)*10` and `ceil(max/10)*10` to bound the bin range, step `10`, and count items `i+1 <= v <= i+10` into a dict keyed by `'i+1-i+10'`.

### Group 3 — Problems 11 to 15 : map, filter, lambda, reduce

These five problems must be solved with Session 6's higher-order functions — no ordinary loops.

- **Add three lists (P11):** `map(lambda x, y, z: x + y + z, L1, L2, L3)` maps with *three* iterables — the lambda receives one element from each. Wrap in `list(...)` to realise the result.
- **Base raised to index (P12):** you need pairs `(number, index)`, so map over `list1` and `range(len(list1))` together: `list(map(lambda x, i: x ** i, list1, range(len(list1))))`.
- **Filter vowels (P13):** `filter` keeps characters for which the predicate `True if ch.lower() in 'aeiou' else False` holds. Lowercasing first makes the test case-insensitive; wrap the filter with `list(...)` and note that the characters come out exactly as they appeared (case preserved).
- **Flatten a 2D list with reduce (P14):** `functools.reduce(lambda x, y: x + y, matrix)` repeatedly concatenates rows: `([1,2,3] + [3,6,7]) + [7,5,4]`. Import `functools` and wrap with `list(...)`. It works because `+` concatenates lists.
- **Filter then map employees (P15):** compose left-to-right: `map(lambda e: e['fname'] + ' ' + e['lname'], filter(lambda e: e['grade'] == 'highly-skilled', employees))`. The `filter` keeps the highly-skilled records, the `map` extracts first+last names. Read any `filter(...)` result before `map` consumes it by wrapping the filter in `list(...)`.

Group reflex: `map` transforms every item, `filter` selects a subset, `reduce` collapses many items into one — and lambdas supply the small functions to each of them.
