###`Q1:` Join Tuples if similar initial element
While working with Python tuples, we can have a problem in which we need to perform concatenation of records from the similarity of initial element. This problem can have applications in data domains such as Data Science.

For eg.
```
Input  : test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)] 
Output : [(5, 6, 7, 8), (6, 10), (7, 13)] 
```

````python
# write your code here
test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]

unique = []

for i in test_list:
  unique.append(i[0])
unique = set(unique)

result = []
for i in unique:
  result.append([i])
  for j in test_list:
    if j[0] == i:
      result[-1].append(j[1])

print(list(map(tuple,result)))


````

###`Q2:` Multiply Adjacent elements (both side) and take sum of right and lest side multiplication result. 


For eg.
```
The original tuple : (1, 5, 7, 8, 10)
Resultant tuple after multiplication : 

(1*5, 1*5+5*7, 7*5 + 7*8, 8*7 + 8*10, 10*8) -> (5, 40, 91, 136, 80)

output-(5, 40, 91, 136, 80)
```

````python
# write your code here
t = (1, 5, 7, 8, 10)

L = []

L.append(t[0]*t[1])

for i in range(1,len(t)-1):
  L.append(t[i]*t[i-1] + t[i]*t[i+1])

L.append(t[-1]*t[-2])

print(tuple(L))
````

###`Q3`: Check is tuples are same or not?
Two tuples would be same if both tuples have same element at same index
```
t1 = (1,2,3,0)
t2 = (0,1,2,3)

t1 and t2 are not same
```

````python
# write your code here
t1 = (1,2,3,0)
t2 = (1,2,3,0)

flag = True
for i,j in zip(t1,t2):
  if i == j:
    continue
  else:
    flag = False
    break
if flag:
  print('same')
else:
  print('not same')
````

###`Q4`: Count no of tuples, list and set from a list

# Set

### Section 2 — Set solutions at a glance

The set section shows how one idea — uniqueness — powers five different problems:

- **Q1 (common elements in three lists):** convert all three lists to sets and chain intersections: `list((s1 & s2) & s3)`. `set(a)` removes duplicates first, and `&` keeps only shared elements. For the sample, `[80, 20]`.
- **Q2 (unique vowels, case-sensitive):** build the vowel set for *both* cases — `set('aeiouAEIOU')` — because "lowercase and uppercase vowels are different". Intersect it with `set(sentence)` (the set of distinct characters) and report the length: `len(s & vowels)` → `6`.
- **Q3 (binary string = exactly two unique characters):** a string is "binary" precisely when `len(set(s)) == 2` — the two-character requirement is exactly a uniqueness requirement. Print `'binary'`/`'not binary'`.
- **Q4 (union of n arrays):** sweep every number into one growing set with `s.update(i)` per sublist, then print it. Sets absorb duplicates automatically, so the union falls out.
- **Q5 (intersection, comprehension-only):** reproduce intersection without operators, as demanded: `[item for item in lst1 if item in lst2]` — the comprehension keeps exactly the elements of `lst1` that also appear in `lst2`.

The through-line: both "common elements" and "unique count" are one `&` or one `len(set(...))` away; duplicates only exist to be removed.


###`Q1:` Write a program to find set of common elements in three lists using sets.
```
Input : ar1 = [1, 5, 10, 20, 40, 80]
        ar2 = [6, 7, 20, 80, 100]
        ar3 = [3, 4, 15, 20, 30, 70, 80, 120]

Output : [80, 20]
```

````python
# write your code here
ar1 = [1, 5, 10, 20, 40, 80]
ar2 = [6, 7, 20, 80, 100]
ar3 = [3, 4, 15, 20, 30, 70, 80, 120]

s1 = set(ar1)
s2 = set(ar2)
s3 = set(ar3)

result = list((s1 & s2) & s3)
print(result)
````

###`Q2:` Write a program to count unique number of vowels using sets in a given string. Lowercase and upercase vowels will be taken as different.

`Input:`
```
Str1 = "hands-on data science mentorship progrAm with live classes at affordable fee only on CampusX"
```
`Output:`
```
No of unique vowels-6
```

````python
# write your code here
vowels = set('aeiouAEIOU')

s = set("hands-on data science mentorship progrAm with live classes at affordable fee only on CampusX")

print('No of unique vowels-',len(s & vowels))
````

# Dictionary

### Section 3 — Dictionary solutions at a glance

The dictionary section exercises keyed scans and rebuilds:

- **Q1 (key with the most unique values):** scan every key, and for each compute `len(set(d[key]))`; remember the key with the largest unique-value count (`max_key`). `('Best', unique 3)` beats `('CampusX', unique 1)` and `('is', unique 2)`, so the printed key is `Best`.
- **Q2 (replace words via lookup):** split the sentence with `test_str.split()`, and for each word return `repl_dict[word]` if present else the word itself; join the parts back with `' '.join(...)`. "Words not in the dict pass through" is the `else` branch.
- **Q3 (flat list → list of dicts):** step through the flat values two at a time with `range(0, n, 2)`, and for each pair build `{key_list[0]: test_list[i], key_list[1]: test_list[i+1]}`. Result: `[{'name': 'CampusX', 'id': 10}]`.
- **Q4 (list of tuples → dict):** unpack each pair with `for i, j in L:` and write `d[i] = [j]` — note the value is wrapped in a list because the expected output is `{'A': [1], ...}`.
- **Q5 (sort keys and value lists):** iterate `for i in sorted(d)` to walk keys in order, and rebuild `res[i] = sorted(d[i])` so each value list is sorted too. `{'a': [4, 19], 'b': [10, 12], 'c': [3]}`.

The recurring reflex: a dict problem is either *scanning* keys and values, or *building* a fresh dict — Q1/Q2 scan, Q3/Q4/Q5 build.


### `Q1`: Key with maximum unique values

Given a dictionary with values list, extract key whose value has most unique values.

**Example 1:**

Input:

```bash
test_dict = {"CampusX" : [5, 7, 9, 4, 0], "is" : [6, 7, 4, 3, 3], "Best" : [9, 9, 6, 5, 5]}
```

Output:
```bash
CampusX
```

**Example 2:**

Input:
```bash
test_dict = {"CampusX" : [5, 7, 7, 7, 7], "is" : [6, 7, 7, 7], "Best" : [9, 9, 6, 5, 5]}
```

Output:
```bash
Best
```

````python
# write your code here
test_dict = {"CampusX" : [5, 7, 7, 7, 7], "is" : [6, 7, 7, 7], "Best" : [9, 9, 6, 5, 5]}

max_val = 0
max_key = ''
for i in test_dict:
  if max_val < len(set(test_dict[i])):
    max_val = len(set(test_dict[i]))
    max_key = i

print(max_key)
````
