###`Problem 1:` Combine two lists index-wise(columns wise)

Write a program to add two lists index-wise. Create a new list that contains the 0th index item from both the list, then the 1st index item, and so on till the last element. any leftover items will get added at the end of the new list.

`Given List:`
```
list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]
```

`Output:`
```
[['M','y'], ['na', me'], ['i', 's'], ['Kh', 'an']]
```

### Solution walkthrough — Problem 1

The reference combines the two lists index-wise with `zip` and a comprehension:

```python
list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]
[[i, j] for (i, j) in zip(list1, list2)]
```

- `zip(list1, list2)` steps both lists in parallel, yielding `('M','y')`, then `('na','me')`, and so on.
- The comprehension builds a two-item list from each pair: `['M','y']`, `['na','me']`, ...
- Output: `[['M','y'], ['na','me'], ['i','s'], ['Kh','an']]` — matches the expected output (note the problem's sample has a display typo, `me'`).
- If the lists were unequal in length, `zip` would stop at the shorter one; leftover items would need to be appended afterwards — not needed here because both lists have four items.
- Alternative loop version: append `[list1[i], list2[i]]` for each index. The comprehension is the Pythonic one-liner.


````python
# Write code here
list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]

[[i,j] for (i,j) in zip(list1,list2)]
````

###`Problem 4:` Running Sum on list
Write a program to print a list after performing running sum on it.

i.e:

`Input:`
```
list1 = [1,2,3,4,5,6]
```
`Output:`
```
[1,3,6,10,15,21]
```

### Solution walkthrough — Problem 4

The reference computes the running (cumulative) sum and collects every step:

```python
list1 = [1,2,3,4,5,6]
result = []
total = 0
for i in list1:
    total = total + i
    result.append(total)
print(result)
```

- `total` accumulates the running sum; after `i` it holds the sum of everything seen so far including `i`.
- Appending `total` each iteration produces `[1, 1+2, 1+2+3, ...]` = `[1, 3, 6, 10, 15, 21]`.
- Output printed is exactly `[1, 3, 6, 10, 15, 21]` — the expected running-sum sequence.
- Key point: you store the *accumulated* value, not the current item — the one-line difference between "running sum" and "copy".
- Note: naming the accumulator `sum` would shadow the built-in `sum()`; the reference wisely avoids that.


````python
# Write code here
list1 = [1,2,3,4,5,6]

result = []
sum = 0

for i in list1:
  sum = sum + i
  result.append(sum)

print(result)
````

###`Problem 5:` You are given a list of integers. You are asked to make a list by running through elements of the list by adding all elements greater and itself.

i.e. Say given list is `[2,4,6,10,1]`
resultant list will be `[22,20,10,23]`.

For 1st element `2` ->> these are greater `(4+6+10)` values and `2` itself so on adding becomes `22`.

For 2nd element `4` ->> greater elements are `(6, 10)` and `4` itself, so on adding `20`

like wise for all other elememts.

`[2,4,6,10,1]`-->`[22,20,16,10,23]`




### Solution walkthrough — Problem 5

The reference uses a nested loop: for each element, sum that element plus every element greater than or equal to it.

```python
L = [2,4,6,10,1]
result = []
for i in L:
    total = 0
    for j in L:
        if i <= j:
            total = total + j
    result.append(total)
print(result)
```

- Outer loop fixes the "current" element `i`.
- Inner loop scans the **whole** list and adds every `j` with `j >= i` — hence "greater and itself".
- For `2` → `2+4+6+10 = 22`, for `4` → `4+6+10 = 20`, for `6` → `6+10 = 16`, for `10` → `10`, for `1` → `1+2+4+6+10 = 23`.
- Output: `[22, 20, 16, 10, 23]` — matches the problem's expected result.
- Complexity is O(n²) because every element re-scans the list; that is acceptable at this size and the problem is about the *condition*, not optimisation.


````python
# Write code here
L = [2,4,6,10,1]
result = []

for i in L:
  sum = 0
  for j in L:
    if i <= j:
      sum = sum + j

  result.append(sum)

print(result)
````

###`Problem 6:` Find list of common unique items from two list. and show in increasing order

`Input`

```
num1 = [23,45,67,78,89,34]
num2 = [34,89,55,56,39,67]
```

`Output:`
```
[34, 67, 89]
```


### Solution walkthrough — Problem 6

The reference finds common items while guarding against duplicates, preserving order:

```python
num1 = [23,45,67,78,89,34,67]
num2 = [34,89,55,56,39,67,67]
common = []
for i in num1:
    if i in num2 and i not in common:
        common.append(i)
print(common)
```

- `i in num2` — the element appears in both lists.
- `i not in common` — the deduplication guard; without it `67` and `89` would repeat where inputs repeat.
- Output for the given inputs (which themselves contain duplicates, `67` twice in `num1`): `[67, 89, 34]` — unique commons in first-seen order.
- The problem statement's sample `[34, 67, 89]` shows the sorted order for its (duplicate-free) input; the reference is not sorted but is *correct for its own input*. Sorting with `sorted(common)` would reconcile them.
- Set shortcut: `list(set(num1) & set(num2))` gives the same commons; add `sorted()` when the expected output is ascending.


````python
# Write code here
num1 = [23,45,67,78,89,34,67]
num2 = [34,89,55,56,39,67,67]

common = []

for i in num1:
  if i in num2:
    if i not in common:
      common.append(i)

print(common)
````

### `Problem 9:` Convert Character Matrix to single String using string comprehension.

**Example 1:** 

Input:
```bash
[['c', 'a', 'm', 'p', 'u', 'x'], ['i', 's'], ['b', 'e', 's', 't'], ['c', 'h', 'a', 'n', 'n', 'e', 'l']]
```

Output:
```bash
campux is best channel
```

### Solution walkthrough — Problem 9

The reference joins the character matrix into a sentence with two nested joins:

```python
L = [['c','a','m','p','u','x'], ['i','s'], ['b','e','s','t'], ['c','h','a','n','n','e','l']]
print(" ".join(["".join(i) for i in L]))
```

- `"".join(i)` glues each row of characters together with **no separator**: `['c','a','m','p','u','x']` → `'campux'`.
- The comprehension `["".join(i) for i in L]` maps every row to its word.
- `" ".join([...])` joins the words with single spaces.
- Output: `campux is best channel` — matching the problem exactly.
- This is the nested-comprehension habit made concrete: transform every piece *first* (inner join), assemble the collection *second* (outer join). Mixing the two produces `c a m p u x i s ...`.


````python
# Write code here
L = [['c', 'a', 'm', 'p', 'u', 'x'], ['i', 's'], ['b', 'e', 's', 't'], ['c', 'h', 'a', 'n', 'n', 'e', 'l']]

print(" ".join(["".join(i) for i in L]))
````

### `Problem 10:` Add Space between Potential Words.

**Example:**

```bash
['campusxIs', 'bestFor', 'dataScientist']
```

Output:
```bash
['campusx Is', 'best For', 'data Scientist']
```

### Solution walkthrough — Problem 10

The reference splits each camel-case word into per-word chunks by detecting uppercase boundaries, then joins the chunks.

```python
test_list = ["campusxIs", "bestFor", "dataScientist"]
res = []
for i in test_list:
    temp = [[]]
    for char in i:
        if char.isupper():
            temp.append([])
        temp[-1].append(char)
    temp_string = ""
    for item in temp:
        temp_string = temp_string + "".join(item) + " "
    res.append(temp_string[0:-1])
print(res)
```

- `temp = [[]]` starts one empty word-chunk.
- For every character, `char.isupper()` detects the start of a new word and opens a fresh chunk (`temp.append([])`); the character then joins whichever chunk is current (`temp[-1]`).
- `"".join(item)` rebuilds each word from its characters; the loop spaces them out, and `temp_string[0:-1]` trims the trailing space.
- For `'campusxIs'` → chunks `['campusx', ['I','s']]` → `'campusx Is'`.
- The reference also prints the intermediate character-chunk lists (`campusx: [['c','a',...], ['I','s']]`), so you can literally see the word splits.
- Output: `['campusx Is', 'best For', 'data Scientist']` — the expected spacing.


````python
# Code here
# initializing list
test_list = ["campusxIs", "bestFor", "dataScientist"]
 
res = []

for i in test_list:
    temp = [[]]
    for char in i:
        
        if char.isupper():
            temp.append([])
        
        temp[-1].append(char)
    
    temp_string = ""
    for item in temp:
        temp_string = temp_string + "".join(item) + " "
    res.append(temp_string[0:-1])

print(res)
````

### `Problem 11:` Write a program that can perform union operation on 2 lists

**Example:**

Input:

```bash
[1,2,3,4,5,1]
[2,3,5,7,8]
```

Output:
```bash
[1,2,3,4,5,7,8]
```

### Solution walkthrough — Problem 11

The reference builds the union by adding items from both lists, skipping duplicates:

```python
L1 = [1,2,3,4,5,1]
L2 = [2,3,5,7,8]
union = []
for i in L1:
    if i not in union:
        union.append(i)
for j in L2:
    if j not in union:
        union.append(j)
print(union)
```

- First loop copies `L1` into `union`, dropping `L1`'s duplicate `1` (only the first occurrence is added).
- Second loop adds every `L2` item that isn't already present — `2,3,5` are skipped, `7,8` are added.
- Output: `[1, 2, 3, 4, 5, 7, 8]` — the union, in first-seen order.
- The `not in union` check is the deduplication core of a manual union.
- Shorter alternative: `list(set(L1) | set(L2))` loses order; `list(dict.fromkeys(L1 + L2))` keeps order. The loop is the most explicit.


````python
# Write code here
L1 = [1,2,3,4,5,1]
L2 = [2,3,5,7,8]

union = []

for i in L1:
  if i not in union:
    union.append(i)

for j in L2:
  if j not in union:
    union.append(j)

print(union)
````

### `Problem 12:` Write a program that can find the max number of each row of a matrix

**Example:**

Input:

```bash
[[1,2,3],[4,5,6],[7,8,9]]
```

Output:
```bash
[3,6,9]
```

### Solution walkthrough — Problem 12

The reference finds the maximum of each row and collects the maxima:

```python
L = [[1,2,3],[4,5,6],[7,8,9]]
result = []
for i in L:
    result.append(max(i))
print(result)
```

- The outer loop visits each row `i` (each is a list).
- `max(i)` computes the largest value inside that row.
- Append to `result`: `[3, 6, 9]` — printed, matching the expected output.
- Compositional version: `[max(row) for row in L]`.
- The subtlety avoided: `max(L)` would compare *whole rows* lexicographically (and yield `[7,8,9]` as a list), not the per-row maximum. Applying `max` to each inner list is the correct granularity.


````python
# Write code here
L = [[1,2,3],[4,5,6],[7,8,9]]
result = []
for i in L:
  result.append(max(i))

print(result)
````

### `Problem 13:` Write a list comprehension to print the following matrix

[[0, 1, 2], [3, 4, 5], [6, 7, 8]]

### Solution walkthrough — Problem 13

The reference builds the matrix row by row with a nested comprehension:

```python
[[j + 3*i for j in range(0, 3)] for i in range(0, 3)]
```

- Outer comprehension runs `i = 0, 1, 2` — the row index.
- Inner comprehension runs `j = 0, 1, 2` — the column value, offset by `3*i`.
- Row 0: `[0+0, 1+0, 2+0]` = `[0,1,2]`; row 1: `[0+3, 1+3, 2+3]` = `[3,4,5]`; row 2: `[6,7,8]`.
- Output: `[[0, 1, 2], [3, 4, 5], [6, 7, 8]]` — exactly the requested matrix, no hard-coding.
- Reading order: the inner `for` completes one row, the outer `for` stacks the rows — the standard shape for any generated grid.


````python
# Write code here
[[j + 3*i for j in range(0,3)] for i in range(0,3)]
````

### Solution walkthrough — Problem 14

The reference transposes the matrix with a nested comprehension:

```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]
[[row[i] for row in matrix] for i in range(len(matrix))]
```

- Inner comprehension: `[row[i] for row in matrix]` walks *down the column* — it pulls the `i`-th element out of every row. For `i=0` → `[1, 4, 7]`.
- Outer comprehension runs `i` over the column indices `0..2`, building one transposed row each.
- Output: `[[1, 4, 7], [2, 5, 8], [3, 6, 9]]` — rows and columns swapped.
- The trick to remember: transposing means the *inner* loop iterates the rows of the *original* matrix while the outer loop selects the column. `zip(*matrix)` does the same in one built-in call, but this comprehension shows the mechanics the problem is testing.


````python
# Write code here
matrix = [
[1,2,3],
[4,5,6],
[7,8,9]
]

[[row[i] for row in matrix]for i in range(len(matrix))]
````

### `Problem 15:` Write a list comprehension that can flatten a nested list

matrix = [<br>

          [4,5,6],<br>
          [7,8,9]<br>]<br>

Output:<br>
[1, 2, 3, 4, 5, 6, 7, 8, 9]

### Solution walkthrough — Problem 15

The reference flattens the matrix with the two-`for` comprehension:

```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]
[item for row in matrix for item in row]
```

- The two `for` clauses read exactly like nested loops: first `for row in matrix`, then `for item in row`.
- The collected `item` is the innermost value, so the result is every cell in order: `[1,2,3,4,5,6,7,8,9]`.
- Output printed matches the expected flattened list.
- This is *the* flatten idiom — worth memorising over a `sum(matrix, [])` or a double `append` loop.
- It also generalises: more levels of nesting just add more `for ... in ...` clauses at the end.


````python
# Write code here
matrix = [
[1,2,3],
[4,5,6],
[7,8,9]
]


[item for row in matrix for item in row]
````
