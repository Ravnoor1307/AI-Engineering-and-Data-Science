## 🐣 Step 0 — Rehearsal Setup + MINI-ZOMATO 🧺

```python
import logging
logging.getLogger("streamlit").setLevel(logging.ERROR)
from streamlit.testing.v1 import AppTest

MINI = """
import streamlit as st
import pandas as pd

st.title("Zomato-Dash 🍽️⭐")
df = pd.DataFrame({
    "name":     ["Karim", "Dimnaa", "BBQ Nation", "Sagar", "Udupi", "Haveli", "RollsCo", "Chaihut"],
    "city":     ["Delhi", "Mumbai", "Delhi",      "Mumbai","Delhi", "Delhi",  "Mumbai",  "Delhi"],
    "rating":   [4.2,     4.6,      4.8,          3.9,     4.1,     3.4,      4.5,       4.9],
    "votes":    [1200,    5400,     8700,         900,     3100,    400,      2200,      180],
    "cost_two": [500,     1200,     1900,         450,     700,     950,      400,       150],
})
"""
print("mini-zomato: 8 rests / 2 cities / rating 3.4–4.9 ✓")
# Expected OUTPUT:
# mini-zomato: 8 rests / 2 cities / rating 3.4–4.9 ✓
```

## 🐥 Step 1 — KPI-Tiles + City-Filter 🎚️

```python
APP = MINI + """
city = st.selectbox("Sheher chuno", ["All", "Delhi", "Mumbai"])
minr = st.slider("Min rating", 3.0, 5.0, 3.5)

show = df.copy()
if city != "All":
    show = show[show.city == city]
show = show[show.rating >= minr]

st.metric("Restaurants", len(show))
st.metric("Avg Rating", round(show.rating.mean(), 2) if len(show) else "-")
st.metric("Avg Cost-2", int(show.cost_two.mean()) if len(show) else 0)
st.markdown(f"KPI|n={len(show)}|avg={round(show.rating.mean(),2)}|cost={int(show.cost_two.mean())}")
"""
at = AppTest.from_string(APP)
at.run()
print("default:", [m.value for m in at.markdown], "| metrics:", [(m.label, m.value) for m in at.metric])
at.selectbox[0].set_value("Delhi"); at.run()
print("delhi:  ", [m.value for m in at.markdown])
at.slider[0].set_value(4.2); at.run()
print("del 4.2:", [m.value for m in at.markdown])
# Expected OUTPUT:
# default: ['KPI|n=7|avg=4.43|cost=757'] | metrics: [('Restaurants', '7'), ('Avg Rating', '4.43'), ('Avg Cost-2', '757')]
# delhi:   ['KPI|n=4|avg=4.5|cost=812']
# del 4.2: ['KPI|n=3|avg=4.63|cost=850']
```

## 🐔 Step 2 — Top-Votes Table + Cheap-Picks (button-trigger!) 🏆💰

```python
APP2 = MINI + """
top3 = df.nlargest(3, "votes")[["name", "votes"]]
st.dataframe(top3)
st.markdown("TOP3=" + ",".join(top3.name))

if st.button("Sasta batao 💰"):
    cheap = df[df.cost_two <= 500].sort_values("cost_two")
    st.markdown("CHEAP=" + ",".join(cheap.name))
"""
at2 = AppTest.from_string(APP2)
at2.run()
print("top3-line:", [m.value for m in at2.markdown])
at2.button[0].click(); at2.run()
print("after-click:", [m.value for m in at2.markdown])
# Expected OUTPUT:
# top3-line: ['TOP3=BBQ Nation,Dimnaa,Udupi']
# after-click: ['TOP3=BBQ Nation,Dimnaa,Udupi', 'CHEAP=Chaihut,RollsCo,Sagar,Karim']
```
