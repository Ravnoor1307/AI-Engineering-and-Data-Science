```python
# 🧺 INTAKE — mini-zomato df (app-string banaane ke liye df-code!)
import logging
logging.getLogger("streamlit").setLevel(logging.ERROR)
from streamlit.testing.v1 import AppTest

DF_CODE = '''
import streamlit as st
import pandas as pd
df = pd.DataFrame({
    "name":     ["Karim", "Dimnaa", "BBQ Nation", "Sagar", "Udupi", "Haveli", "RollsCo", "Chaihut"],
    "city":     ["Delhi", "Mumbai", "Delhi",      "Mumbai","Delhi", "Delhi",  "Mumbai",  "Delhi"],
    "rating":   [4.2,     4.6,      4.8,          3.9,     4.1,     3.4,      4.5,       4.9],
    "votes":    [1200,    5400,     8700,         900,     3100,    400,      2200,      180],
    "cost_two": [500,     1200,     1900,         450,     700,     950,      400,       150],
})
'''
print("shield-intake ready ✓ (8 rests, 2 cities)")
# Expected OUTPUT:
# shield-intake ready ✓ (8 rests, 2 cities)
```

## 👑 BOSS: 10-Assert Dash-Don Badge (AppTest police-show!) 🏆

```python
# 👑 BOSS CONSOLE — canonical Zomato-Shield (apna app banao, in asserts pe test!)
BOSS_APP = DF_CODE + '''
st.title("Zomato-Shield 🍽️🛡️")
st.markdown("8rest|2city|mini-df")

city = st.selectbox("Sheher", ["All", "Delhi", "Mumbai"])
minr = st.slider("Min rating", 3.0, 5.0, 3.5)

show = df.copy()
if city != "All":
    show = show[show.city == city]
show = show[show.rating >= minr]

st.metric("Restaurants", len(show))
st.metric("Avg Rating", round(show.rating.mean(), 2) if len(show) else "-")
st.markdown(f"RESULT|city={city}|n={len(show)}|avg={round(show.rating.mean(),2)}")

top3 = df.nlargest(3, "votes")[["name", "votes"]]
st.markdown("TOP3=" + ",".join(top3.name))

high = show[show.rating >= 4.5]
st.markdown(f"N_HIGH={len(high)}")

if st.button("Sasta batao"):
    cheap = df[df.cost_two <= 500].sort_values("cost_two")
    st.markdown("CHEAP=" + ",".join(cheap.name))
'''
at = AppTest.from_string(BOSS_APP)
at.run()
assert at.title[0].value == "Zomato-Shield 🍽️🛡️",            "① title exact!"
assert (at.metric[0].label, at.metric[0].value) == ("Restaurants", "7"), "② default n=7 (3.5 cut)!"
assert (at.metric[1].label, at.metric[1].value) == ("Avg Rating", "4.43"), "③ avg 4.43!"
assert [m.value for m in at.markdown][:3] == ["8rest|2city|mini-df",
    "RESULT|city=All|n=7|avg=4.43", "TOP3=BBQ Nation,Dimnaa,Udupi"],  "④ opening mds!"
assert [m.value for m in at.markdown][3] == "N_HIGH=4",        "⑤ high-rating 4 (BBQ 4.8, Dimnaa 4.6, RollsCo 4.5, Chaihut 4.9 ≥4.5!)"
at.selectbox[0].set_value("Delhi"); at.run()
assert [m.value for m in at.markdown][1] == "RESULT|city=Delhi|n=4|avg=4.5", "⑥ delhi n=4 avg=4.5!"
at.slider[0].set_value(4.2); at.run()
assert [m.value for m in at.markdown][1] == "RESULT|city=Delhi|n=3|avg=4.63", "⑦ delhi-4.2 n=3!"
assert [m.value for m in at.markdown][3] == "N_HIGH=2",        "⑧ del-high 2 (BBQ 4.8 + Chaihut 4.9!)"
at.button[0].click(); at.run()
assert [m.value for m in at.markdown][4] == "CHEAP=Chaihut,RollsCo,Sagar,Karim", "⑨ anon-cheap line!"
at.selectbox[0].set_value("Mumbai"); at.slider[0].set_value(5.0); at.run()
assert [m.value for m in at.markdown][1] == "RESULT|city=Mumbai|n=0|avg=nan", "⑩ empty-set still-runs (RERUN-proof!)!"
print("🏆 10/10 DASH-DON BADGE — streamlit tumhari coaching pe rehearse hua! 📊🎯")
```

## ✅ SOLUTIONS 🔓

<details><summary><b>S1-S4</b></summary>

```python
# S1-S4 ✅
app1 = DF_CODE + '''
st.title("Zomato-Shield 🍽️🛡️")
show = df
st.metric("Restaurants", len(show))
st.metric("Avg Rating", round(show.rating.mean(), 2))
st.markdown(f"AUDIT|n={len(show)}|avg={round(show.rating.mean(),2)}")
'''
at1 = AppTest.from_string(app1); at1.run()
print("① title:", at1.title[0].value)
print("② metrics:", [(m.label, m.value) for m in at1.metric])
print("   audit-line:", [m.value for m in at1.markdown])

app2 = DF_CODE + '''
city = st.selectbox("Sheher", ["All", "Delhi", "Mumbai"])
show = df if city == "All" else df[df.city == city]
st.markdown(f"CITY|{city}|n={len(show)}")
'''
at2 = AppTest.from_string(app2); at2.run()
at2.selectbox[0].set_value("Delhi"); at2.run()
print("③ city-filter:", [m.value for m in at2.markdown], "(Delhi=5 rows — raw df!)")

app3 = DF_CODE + '''
city = st.selectbox("Sheher", ["All", "Delhi", "Mumbai"])
minr = st.slider("Min rating", 3.0, 5.0, 3.5)
show = df.copy()
if city != "All":
    show = show[show.city == city]
show = show[show.rating >= minr]
st.markdown(f"RESULT|city={city}|n={len(show)}")
'''
at3 = AppTest.from_string(app3); at3.run()
print("④ default-slider:", [m.value for m in at3.markdown], "(All cut 3.5 → 7!)")
```

<details><summary><b>S5-S7</b></summary>

```python
# S5-S7 ✅
app4 = DF_CODE + '''
top3 = df.nlargest(3, "votes")[["name", "votes"]]
st.markdown("TOP3=" + ",".join(top3.name))
if st.button("Sasta batao"):
    cheap = df[df.cost_two <= 500].sort_values("cost_two")
    st.markdown("CHEAP=" + ",".join(cheap.name))
'''
at4 = AppTest.from_string(app4); at4.run()
print("⑤ TOP3:", [m.value for m in at4.markdown])
at4.button[0].click(); at4.run()
print("⑥ click-ke-baad:", [m.value for m in at4.markdown])

app5 = DF_CODE + '''
city = st.selectbox("Sheher", ["All", "Delhi", "Mumbai"])
minr = st.slider("Min rating", 3.0, 5.0, 4.5)
show = df.copy()
if city != "All":
    show = show[show.city == city]
show = show[show.rating >= minr]
st.markdown(f"N_HIGH={len(show)}")
'''
at5 = AppTest.from_string(app5); at5.run()
at5.selectbox[0].set_value("Delhi"); at5.run()
print("⑦ Delhi+4.5:", [m.value for m in at5.markdown], "(BBQ+Chaihut=2!)")
# Takeaway: anchors (markdown-tags) + metrics = dashboard ko robot-proof banao! 🤖
```
