📖 **Definitions (fix these 3 in your mind!):**
- **st** = the streamlit module — `st.title/write/dataframe/metric` = putting
  "plates" on the page (in top-to-bottom order!)
- **RERUN-MODEL 🔄** = on every widget-touch the WHOLE script runs again
  (top to bottom!) — no socket, no callback — just the loop-of-truth!
- **AppTest 🎭** = a notebook-friendly rehearsal: run an app-string, grab the
  elements, assert the values (the real server is `streamlit run` on your machine!)

🔍 **WHY is it GOLD for DS/AI?** pandas → SELECTBOX → chart = a demo-into-app in 10 minutes!


## 🐣 Level 0 — First Rehearsal: Title + Metric + Data-Plate 🪗

```python
import logging
logging.getLogger("streamlit").setLevel(logging.ERROR)    # rehearsal-shor band!

from streamlit.testing.v1 import AppTest

APP0 = """
import streamlit as st
st.title("Chaihut Analytics ☕")
st.metric("Aaj ki Sale", "₹4,570")
st.markdown("swagat hai! COUNTER=42")
"""
at = AppTest.from_string(APP0)
at.run()
print("title-value :", at.title[0].value)
print("metric      :", (at.metric[0].label, at.metric[0].value))
print("markdown    :", [m.value for m in at.markdown])
# Expected OUTPUT:
# title-value : Chaihut Analytics ☕
# metric      : ('Aaj ki Sale', '₹4,570')
# markdown    : ['swagat hai! COUNTER=42']
```


## 🐥 Level 1 — Widgets = Live Filters (select / slider) 🎚️

Understand the RERUN-spell: a widget-change → the SCRIPT runs again from top-to-bottom!
That's why you write the filter-code directly — `df[df.city == city]` — in AppTest
`set_value(...)`; `.run()` = "spin+refresh"!


```python
APP1 = """
import streamlit as st
import pandas as pd

st.title("Zomato Explorer 🍽️")
df = pd.DataFrame({
    "rest": ["Karim","Dimnaa","BBQ","Sagar","Udupi"],
    "city": ["Delhi","Mumbai","Delhi","Mumbai","Delhi"],
    "rating": [4.2, 4.6, 4.8, 3.9, 4.1],
})
city = st.selectbox("City", ["All", "Delhi", "Mumbai"])
minr = st.slider("Min rating", 3.5, 5.0, 4.0)

show = df.copy()
if city != "All":
    show = show[show.city == city]
show = show[show.rating >= minr]
st.metric("Restaurants", len(show))
st.markdown(f"RESULT_COUNT={len(show)}")
"""
at2 = AppTest.from_string(APP1)
at2.run()
print("default →", [m.value for m in at2.markdown], "| metric:", at2.metric[0].value)
at2.selectbox[0].set_value("Delhi"); at2.run()
print("delhi   →", [m.value for m in at2.markdown], "| metric:", at2.metric[0].value)
at2.selectbox[0].set_value("Mumbai"); at2.slider[0].set_value(4.5); at2.run()
print("mum 4.5→", [m.value for m in at2.markdown])
# Expected OUTPUT:
# default → ['RESULT_COUNT=4'] | metric: 4
# delhi   → ['RESULT_COUNT=3'] | metric: 3
# mum 4.5→ ['RESULT_COUNT=1']
```


## 🐔 Level 2 — The Full RERUN Picture: Button + Marker-Print 🔄

Inside an AppScript, `print(...)` GLOWS AGAIN on every run! A button-click =
one rerun — watch the sequence (this is streamlit's core-mantra — once you
understand it, even complex apps become easy!)


```python
APP2 = """
import streamlit as st
print("RERUN-HUA!")
if st.button("Badhaoo"):
    st.markdown("CLICK_MILA")
"""
at3 = AppTest.from_string(APP2)
at3.run()                                   # run #1
at3.run()                                   # run #2 (bina-change!)
at3.button[0].click(); at3.run()            # click + rerun
print("markdowns:", [m.value for m in at3.markdown])
print("⚡ script `print` har-run pe = RERUN-MODEL saboot-kaand!")
# Expected OUTPUT:
# RERUN-HUA!
# RERUN-HUA!
# RERUN-HUA!
# markdowns: ['CLICK_MILA']
# ⚡ script `print` har-run pe = RERUN-MODEL saboot-kaand!
```

## ⚠️ 5 Classic Mistakes (Streamlit)
| # | ❌ Mistake | ✅ Correct |
|---|---|---|
| 1 | thinking widgets magically go live in isolation | understand that every change = full-RERUN 🔄 |
| 2 | asserting on st.dataframe in AppTest (arrow-bytes!) | assert on deterministic markdown/metric |
| 3 | running `python app.py` instead of `streamlit run` | ONLY streamlit run (the magic-server!) |
| 4 | forgetting that loop-variables refresh every rerun (state-reset!) | `st.session_state` (an advanced note!) |
| 5 | reloading heavy data on every run (slow app!) | the `@st.cache_data` decorator (a hint!) |
