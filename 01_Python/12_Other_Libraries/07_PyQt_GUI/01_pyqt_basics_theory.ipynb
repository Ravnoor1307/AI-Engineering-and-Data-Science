📖 **Definitions (office-speak!):**
- `QApplication` = the office's STATION-SERVER (there's just ONE — all windows
  hang off it!) 🖥️
- `QWidget/QLabel/QLineEdit/QPushButton` = the pieces-on-desk 🧩
- **signal↔slot** = a doorbell (signal) + whoever-opens-the-door (the slot/function)!
  `button.clicked.connect(fn)` = hook up the wire! 🔔
- **event-loop** = the station-server's heartbeat 💓 (`app.exec()` — even without
  running the loop in headless, signals still work!) 🎛️

💡 **Names-note:** `PySide6` = the official Qt-for-Python; `PyQt6` = the same-API
other-discussion (just change the import-line!). We'll use PySide6. Even in

style! 🎭


## 🐣 Level 0 — Station-Server + First Doorbell 🔔 (LIVE, display-free!)

```python
import os, sys
os.environ.setdefault("QT_QPA_PLATFORM", "offscreen")    # display-free workstation!
from PySide6.QtWidgets import QApplication, QWidget, QLabel, QPushButton, QVBoxLayout

qapp = QApplication.instance() or QApplication(sys.argv)   # ek-hi server; again-call safe!

desk = QWidget(); lay = QVBoxLayout(desk)
board = QLabel("shuruat")
bell  = QPushButton("Bajao!")
lay.addWidget(board); lay.addWidget(bell)

hits = []
bell.clicked.connect(lambda: (hits.append("ting!"), board.setText(f"bell #{len(hits)} 🔔")))

bell.click(); bell.click(); bell.click()
print("board-text:", board.text())
print("hits-log  :", hits)
# Expected OUTPUT:
# board-text: bell #3 🔔
# hits-log  : ['ting!', 'ting!', 'ting!']
```

## 🐥 Level 1 — QLineEdit: Form-Line + Greet-Reply 📝

```python
from PySide6.QtWidgets import QLineEdit

desk2 = QWidget(); lay2 = QVBoxLayout(desk2)
name_edt = QLineEdit(); naam = QLabel("?"); go = QPushButton("Namaste-do!")
lay2.addWidget(name_edt); lay2.addWidget(go); lay2.addWidget(naam)

go.clicked.connect(lambda: naam.setText(f"Hello {name_edt.text()}! 👋"))

name_edt.setText("Asha");  go.click()
print("1:", naam.text())
name_edt.setText("");      go.click()
print("2:", naam.text())    # khali-name watchdog absent — mistakes-section!
# Expected OUTPUT:
# 1: Hello Asha! 👋
# 2: Hello ! 👋
```


## 🐔 Level 2 — Employee-Mini: Spin-Salary + Board-Refresh 🧑‍💼

QSpinBox = a numeric spinner; QTextEdit = a multiline-board. The registry is plain
python; the GUI is a thin-layer! (the thin-GUI mantra from 13.06 — the SAME-crime here too!)


```python
from PySide6.QtWidgets import QSpinBox, QTextEdit

REG = []

def reg_add(name: str, salary: int):
    REG.append({"name": name.strip(), "salary": salary})
    board3.setPlainText("\n".join(f"{i+1}. {r['name']} — ₹{r['salary']}" for i, r in enumerate(REG)))
    count_lbl.setText(f"staff={len(REG)} total=₹{sum(r['salary'] for r in REG)}")

desk3 = QWidget(); lay3 = QVBoxLayout(desk3)
name_edt3 = QLineEdit(); sal_edt = QSpinBox(); sal_edt.setRange(0, 10_000_000)
add_btn = QPushButton("Hire!"); count_lbl = QLabel("staff=0"); board3 = QTextEdit()
for w_ in (name_edt3, sal_edt, add_btn, count_lbl, board3): lay3.addWidget(w_)

add_btn.clicked.connect(lambda: reg_add(name_edt3.text(), sal_edt.value()))

name_edt3.setText("Asha");   sal_edt.setValue(45000); add_btn.click()
name_edt3.setText("BK");     sal_edt.setValue(38000); add_btn.click()
name_edt3.setText("Chirag"); sal_edt.setValue(52000); add_btn.click()
print(count_lbl.text())
print("--- board ---")
print(board3.toPlainText())
# Expected OUTPUT:
# staff=3 total=₹135000
# --- board ---
# 1. Asha — ₹45000
# 2. BK — ₹38000
# 3. Chirag — ₹52000
```

## ⚠️ 5 Classic Mistakes (PyQt/PySide)
| # | ❌ Mistake | ✅ Correct |
|---|---|---|
| 1 | connect(fn()) — the bracket-crime again! | the NAME `connect(fn)` |
| 2 | calling QApplication([...]) twice (an ironic crash) | `instance() or QApplication` |
| 3 | not managing the Entry/filled text/value | clear the inputs after add (pro-UX!) |
| 4 | an empty-name staff member being hired (see Level-1!) | guard: `if not name: return` |
| 5 | expecting windows in a headless-console | QT_QPA_PLATFORM=offscreen works in both! |
