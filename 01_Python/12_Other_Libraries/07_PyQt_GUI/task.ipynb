## 👑 BOSS: 10-Assert Dock-Pilot Badge (widgets LIVE offscreen!) 🏆

```python
# 👑 BOSS CONSOLE — workstation jama, phir click-assault!
_desk = QWidget(); _lay = QVBoxLayout(_desk)
_name = QLineEdit(); _sal = QSpinBox(); _sal.setRange(0, 10_000_000)
_hire = QPushButton("Hire"); _fire = QPushButton("Fire-Top"); _find = QLineEdit()
_count = QLabel("staff=0"); _board = QTextEdit(); _warn = QLabel("")
for w in (_name, _sal, _hire, _fire, _find, _count, _board, _warn): _lay.addWidget(w)

_REG = []
def _refresh():
    _count.setText(f"staff={len(_REG)} total=₹{sum(r['salary'] for r in _REG)}")
    _board.setPlainText("\n".join(f"{r['name']}({r['salary']//1000}k)" for r in _REG))

def _do_hire():
    n, s = _name.text().strip(), _sal.value()
    if not n:
        _warn.setText("⚠️ naam toh likho!"); return
    _warn.setText("")
    _REG.append({"name": n, "salary": s}); _refresh()

def _do_fire():
    if not _REG:
        _warn.setText("⚠️ khali!"); return
    top = max(_REG, key=lambda r: r["salary"])
    _REG.remove(top); _warn.setText(f"🔥 {top['name']} chala gaya"); _refresh()

_hire.clicked.connect(_do_hire); _fire.clicked.connect(_do_fire)
_name.returnPressed.connect(lambda: _board.setPlainText(str(_find.text())))

_name.setText(""); _sal.setValue(50000); _hire.click()
assert _warn.text() == "⚠️ naam toh likho!" and len(_REG) == 0,         "① watchdog roka-naam!"
_name.setText("Asha"); _sal.setValue(45000); _hire.click()
_name.setText("BK");   _sal.setValue(38000); _hire.click()
_name.setText("Diya"); _sal.setValue(52000); _hire.click()
assert _count.text() == "staff=3 total=₹135000",                        "② count+total label!"
assert "Asha(45k)" in _board.toPlainText(),                             "③ board-render 45k!"
_fire.click()
assert [r["name"] for r in _REG] == ["Asha", "BK"],                     "④ fire-top= Diya(52k) gaya!"
assert _warn.text() == "🔥 Diya chala gaya",                            "⑤ fire-banner!"
assert _count.text() == "staff=2 total=₹83000",                         "⑥ post-fire total!"
_fire.click(); _fire.click()
assert len(_REG) == 0,                                                  "⑦ sab fire!"
_fire.click()
assert _warn.text() == "⚠️ khali!",                                     "⑧ empty-fire guard!"
_name.setText("Q"); _sal.setValue(10000); _hire.click()
_name.setText("Qs"); _sal.setValue(20000); _hire.click()
matches = [r["name"] for r in _REG if "q" in r["name"].lower()]
assert matches == ["Q", "Qs"],                                          "⑨ search-substring case-ins!"
totals = sum(r["salary"] for r in _REG)
assert totals == 30000 and len(_REG) == 2,                              "⑩ final-registry math!"
print("🏆 10/10 DOCK-PILOT BADGE — workstation tumhare clicks pe hilty! 🎛️👑")
```

## ✅ SOLUTIONS 🔓

<details><summary><b>S1-S4</b></summary>

```python
# S1-S4 ✅
desk = QWidget(); lay = QVBoxLayout(desk)
lbl = QLabel("..."); btn = QPushButton("Dock!"); greet_in = QLineEdit()
go = QPushButton("GO"); greet_lbl = QLabel("...")
nm = QLineEdit(); sp = QSpinBox(); sp.setRange(0, 10_000_000)
hire = QPushButton("Hire"); stf = QLabel("staff=0"); warn = QLabel("")
board = QTextEdit()
for w in (lbl, btn, greet_in, go, greet_lbl, nm, sp, hire, stf, warn, board): lay.addWidget(w)

clicks = []
btn.clicked.connect(lambda: (clicks.append(1), lbl.setText(f"Dock khul gaya! #{len(clicks)}")))
go.clicked.connect(lambda: greet_lbl.setText(f"Hello {greet_in.text()}!"))
REG = []
def do_hire():
    n, s = nm.text().strip(), sp.value()
    if not n:
        warn.setText("⚠️ naam toh likho!"); return
    REG.append({"name": n, "salary": s}); warn.setText("")
    stf.setText(f"staff={len(REG)} total=₹{sum(r['salary'] for r in REG)}")
hire.clicked.connect(do_hire)

btn.click(); btn.click()
print("①", lbl.text(), "| count:", len(clicks))
greet_in.setText("Rani"); go.click()
print("②", greet_lbl.text())
nm.setText("Asha"); sp.setValue(45000); hire.click()
nm.setText("BK");   sp.setValue(38000); hire.click()
print("③", stf.text())
nm.setText(""); hire.click()
print("④", warn.text(), "| safe-len:", len(REG))
```

<details><summary><b>S5-S7</b></summary>

```python
# S5-S7 ✅
find = QLineEdit("a")
matches = [r["name"] for r in REG if find.text().lower() in r["name"].lower()]
print("⑤ search 'a':", matches)

def fire_top():
    if not REG:
        warn.setText("⚠️ khali!"); return None
    top = max(REG, key=lambda r: r["salary"])
    REG.remove(top); warn.setText(f"🔥 {top['name']} chala gaya")
    stf.setText(f"staff={len(REG)} total=₹{sum(r['salary'] for r in REG)}")
    return top["name"]

fired = fire_top()
print("⑥ fired:", fired, "| now:", stf.text())

def team_lines():
    return [f"{r['name']}({r['salary']//1000}k)" for r in REG]
board.setPlainText("\n".join(team_lines()))
print("⑦", board.toPlainText())
# Takeaway: click-assault + guards + formatted-board = kaam-ka-deskdock! 🧑‍💼
```
