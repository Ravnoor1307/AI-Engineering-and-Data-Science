## 👑 BOSS: 10-Assert Room-Chief Badge (direct-call only!) 🏆

```python
# 👑 BOSS CONSOLE — full control-room def, phir seedha phone-calls!
import plotly.graph_objects as go
_bDF = SALES.copy()

def _show_for(reg):
    return _bDF if reg == "All" else _bDF[_bDF.region == reg]

# NB-rehearsal: callbacks ko SEEDHA functions bana ke test — browser-independent! ☎️
def upd(region, metric):
    show = _show_for(region)
    if len(show) == 0:
        fig = go.Figure().update_layout(title="NONE")
        return fig, "NO-DATA", "NO-BEST"
    fig = px.bar(show, x="month", y=metric, title=f"{metric.upper()} — {region}")
    kpi = f"TOTAL={int(show[metric].sum())}"
    bm = show.groupby("month")[metric].sum().idxmax()
    return fig, kpi, f"BEST={bm}@{int(show.groupby('month')[metric].sum().max())}"

def cmp2(metric):
    north = _bDF[_bDF.region == "North"]; south = _bDF[_bDF.region == "South"]
    import plotly.graph_objects as go
    fig = go.Figure()
    fig.add_bar(x=list(north.month), y=list(north[metric]), name="North")
    fig.add_bar(x=list(south.month), y=list(south[metric]), name="South")
    return fig

fig, kpi, best = upd("North", "sales")
assert [int(v) for v in fig.data[0].y] == [100, 140],              "① north-sales y!"
assert fig.layout.title.text == "SALES — North",                   "② title exact!"
assert kpi == "TOTAL=240",                                         "③ TOTAL=240!"
figE, kpiE, _ = upd("East", "profit")
assert [int(v) for v in figE.data[0].y] == [21, 18] and kpiE == "TOTAL=39", "④ east-profit!"
figA, kpiA, bestA = upd("All", "sales")
assert kpiA == "TOTAL=580",                                        "⑤ all-sales 580!"
assert bestA == "BEST=Feb@320",                                    "⑥ Feb@320 best-month!"
_, kpiP, bestP = upd("All", "profit")
assert kpiP == "TOTAL=174" and bestP == "BEST=Feb@96",             "⑦ profit 174 / Feb@96!"
_, _, bestN = upd("North", "sales")
assert bestN == "BEST=Feb@140",                                    "⑧ north-best Feb@140!"
figC = cmp2("sales")
assert [t.name for t in figC.data] == ["North", "South"],          "⑨ compare traces!"
figW, kpiW, bestW = upd("West", "sales")
assert (kpiW, bestW, figW.layout.title.text) == ("NO-DATA", "NO-BEST", "NONE"), "⑩ empty-guard!"
print("🏆 10/10 ROOM-CHIEF BADGE — callbacks tumhare phone-pe uthate! 📈🎛️")
```

## ✅ SOLUTIONS 🔓

<details><summary><b>S1-S4</b></summary>

```python
# S1-S4 ✅
app.layout = html.Div([
    html.H1("Sales-Shield 📈🛡️"),
    dcc.Dropdown(["All", "North", "South", "East"], "All", id="reg"),
    dcc.RadioItems(["sales", "profit"], "sales", id="metric"),
    dcc.Graph(id="chart"),
    html.Div(id="kpi"),
])
print("① tree:", [(type(ch).__name__, getattr(ch, "id", None)) for ch in app.layout.children])

def update_dash(region, metric):
    show = SALES if region == "All" else SALES[SALES.region == region]
    fig = px.bar(show, x="month", y=metric, title=f"{metric.upper()} — {region}")
    kpi = f"TOTAL={int(show[metric].sum())}"
    return fig, kpi

fig2, kpi2 = update_dash("North", "sales")
print("②", fig2.layout.title.text, "|", kpi2, "| y:", [int(v) for v in fig2.data[0].y])
fig3, kpi3 = update_dash("All", "profit")
print("③", kpi3, "| y:", [int(v) for v in fig3.data[0].y])

def best_month(region, metric):
    show = SALES if region == "All" else SALES[SALES.region == region]
    g = show.groupby("month")[metric].sum()
    return f"BEST={g.idxmax()}@{int(g.max())}"
print("④", best_month("All", "sales"), "|", best_month("North", "sales"))
```

<details><summary><b>S5-S7</b></summary>

```python
# S5-S7 ✅
import plotly.graph_objects as go
fig5, k5 = update_dash("East", "sales")
print("⑤ exact-title:", repr(fig5.layout.title.text))     # 'SALES — East'

def cmp2(metric="sales"):
    fig = go.Figure()
    for reg in ["North", "South"]:
        sub = SALES[SALES.region == reg]
        fig.add_bar(x=list(sub.month), y=list(sub[metric]), name=reg)
    return fig
fc = cmp2("profit")
print("⑥ traces:", [t.name for t in fc.data], "| profit-south:", [int(v) for v in fc.data[1].y])

def guarded(region, metric):
    show = SALES if region == "All" else SALES[SALES.region == region]
    if len(show) == 0:
        return go.Figure().update_layout(title="NONE"), "NO-DATA"
    return px.bar(show, x="month", y=metric), "OK"
fw, kw = guarded("West", "sales")
print("⑦ west:", kw, "| title:", fw.layout.title.text)
# Takeaway: directly-testable callbacks = dash bhi notebook-friendly! ☎️📈
```
