**Topic 13.05 | Other_Libraries — Part 5 of 8**

🌍 **Hook:** Streamlit = VLOG (a new frame on every touch!). Dash = CONTROL-ROOM:
a system that does NOT fully re-render on every knob — only the targeted wire
rings whose remote-connection exists. For real-world production time, Dash =
SERIOUS dashboards ◉ (React! Flask! Plotly! all inside!)

📖 **Definitions (the remote-control ones!):**
- **layout** = the TREE of `html.Div/Dropdown/Graph` — the page's xray-skeleton 🩻
- **callback** = `@app.callback(Output("chart","figure"), Input("reg","value"))` —
  "when reg spins, change chart's figure" (wirewire!)
- **figure** = the plotly-graph's dict (x/y/title) — the callback's MAIN parcel!

🔍 Streamlit vs Dash (a no-memorizing table):
| question | streamlit | dash |
|---|---|---|
| run-model | full-script rerun 🔄 | targeted-updates 🎯 |
| best-for | quick demos ⚡ | multi-knob control-rooms ◉ |
| test-pattern? | AppTest | callbacks called DIRECTLY (this notebook!) ☎️ |


## 🐣 Level 0 — First Layout-Tree + Xray-Skeleton 🩻

```python
from dash import Dash, html, dcc
import pandas as pd
import plotly.express as px

app = Dash("sales-room")

app.layout = html.Div([
    html.H1("Sales-Room 📈"),
    dcc.Dropdown(["All", "North", "South", "East"], "All", id="reg"),
    dcc.RadioItems(["sales", "profit"], "sales", id="metric"),
    dcc.Graph(id="chart"),
    html.Div(id="kpi"),
])

print("layout root:", type(app.layout).__name__, "| children:", len(app.layout.children))
print("component map:", [(type(ch).__name__, getattr(ch, "id", None)) for ch in app.layout.children])
# Expected OUTPUT:
# layout root: Div | children: 5
# component map: [('H1', None), ('Dropdown', 'reg'), ('RadioItems', 'metric'), ('Graph', 'chart'), ('Div', 'kpi')]
```


## 🐥 Level 1 — The Callback's Remote ☎️ (a direct-call rehearsal!)

In the real browser the wire triggers on its own; for learning, we call the
CALLBACK-FUNCTION directly in the notebook — testable by design!


```python
from dash import Input, Output

SALES = pd.DataFrame({
    "region": ["North","North","South","South","East","East"],
    "month":  ["Jan","Feb","Jan","Feb","Jan","Feb"],
    "sales":  [100,140,90,120,70,60],
    "profit": [30,42,27,36,21,18],
})

@app.callback(Output("chart", "figure"), Output("kpi", "children"),
              Input("reg", "value"), Input("metric", "value"))
def update_dash(region, metric):
    show = SALES if region == "All" else SALES[SALES.region == region]
    fig = px.bar(show, x="month", y=metric, title=f"{metric.upper()} — {region}")
    kpi = f"TOTAL={int(show[metric].sum())}"
    return fig, kpi

fig, kpi = update_dash("North", "sales")
print("title:", fig.layout.title.text)
print("x:", list(fig.data[0].x), "| y:", [int(v) for v in fig.data[0].y], "| kpi:", kpi)
fig2, kpi2 = update_dash("East", "profit")
print("east-y:", [int(v) for v in fig2.data[0].y], "| kpi:", kpi2)
# Expected OUTPUT:
# title: SALES — North
# x: ['Jan', 'Feb'] | y: [100, 140] | kpi: TOTAL=240
# east-y: [21, 18] | kpi: TOTAL=39
```


## 🐔 Level 2 — All-Region Merge + Layout Hacks 🍱

All = the full df — 6 bars; a profit-side-switch — the same callback, two avatars!
(figure-attr: `.data[0].y` = clean numpy; since `to_dict()` packs binary-packed
data, attribute-access is the real key here!) 🗝️


```python
figA, kpiA = update_dash("All", "sales")
figP, kpiP = update_dash("All", "profit")
print("ALL sales y:", [int(v) for v in figA.data[0].y], "|", kpiA)
print("ALL profit y:", [int(v) for v in figP.data[0].y], "|", kpiP)
print("fig-type:", figA.data[0].type, "| traces:", len(figA.data), "| y-label:", figA.layout.yaxis.title.text)
print("💡 layouts hack: dcc/Tabs, Row/Col via html.Div(style={...}) — same-tree-idea!")
# Expected OUTPUT:
# ALL sales y: [100, 140, 90, 120, 70, 60] | TOTAL=580
# ALL profit y: [30, 42, 27, 36, 21, 18] | TOTAL=174
# fig-type: bar | traces: 1 | y-label: sales
```

## ⚠️ 5 Classic Mistakes (Dash)
| # | ❌ Mistake | ✅ Correct |
|---|---|---|
| 1 | forgetting dcc.Graph(id=...) (the callback needs a target!) | an id on every component |
| 2 | mixing up the callback function's RETURN-order | Output-list order ≠ jumble! |
| 3 | expecting a State-type in the Input-list | Input=trigger, State=read-only 🔧 |
| 4 | asserting on the figure's EXPECT-ed dict-data | `.data[0]/.layout` attribute-access 🗝️ |
| 5 | app.run(debug=True) inside the notebook | layouts + callback-call rehearsal 🎭 |
