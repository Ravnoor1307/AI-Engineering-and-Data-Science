## 👑 BOSS: 10-Assert Toolkit-Chief Badge (logic-only, dual-env safe!) 🏆

```python
# 👑 BOSS CONSOLE — thin-GUI toolkit poora, phir value-assault!
def _parse(text: str) -> list[int]:
    out = []
    for tok in text.split(","):
        tok = tok.strip()
        try:
            out.append(int(tok))
        except ValueError:
            pass                # "abc" type tokens IGNORE (harsh-guard!)
    return out

def _avg(marks) -> float:
    return round(sum(marks) / len(marks), 1) if marks else 0.0

def _grade(avg: float) -> str:
    return ("A" if avg >= 85 else "B" if avg >= 70 else "C" if avg >= 50
            else "D" if avg >= 35 else "F")

def _report(name: str, text: str) -> str:
    m = _parse(text); a = _avg(m)
    return f"{name}: avg={a} grade={_grade(a)}"

class _Board:
    def __init__(self):
        self.rows = []
    def add(self, name, raw):
        self.rows.append({"name": name, "avg": _avg(_parse(raw))})
    def top(self, n=1):
        return sorted(self.rows, key=lambda r: -r["avg"])[:n]
    def summary(self):
        return [f"{r['name']}: avg={r['avg']} grade={_grade(r['avg'])}" for r in self.rows]

b = _Board()
for n, s in zip(NAMES, SAMPLES):
    b.add(n, s)

assert _parse("80, 90, 100") == [80, 90, 100],                    "① parse clean!"
assert _parse("") == [] and _parse("80, abc, 90") == [80, 90],    "② empty+harsh-guard!"
assert _avg([80, 90, 100]) == 90.0 and _avg([]) == 0.0,           "③ avg exact+safe!"
assert (_grade(90.0), _grade(84.9), _grade(67.3), _grade(36.0), _grade(20)) == ("A", "B", "C", "D", "F") and _grade(50) == "C", "④ grade-bands boundaries!"
assert _report("Asha", "80,90,100") == "Asha: avg=90.0 grade=A",  "⑤ report-string!"
assert b.rows[0] == {"name": "Asha", "avg": 90.0},                "⑥ board-rows struct!"
assert [r["name"] for r in b.top(2)] == ["Diya", "Asha"],         "⑦ top2 = Diya(91.7)→Asha(90)!"
assert len(b.summary()) == 5,                                     "⑧ summary-5!"

def could_show_gui():
    try:
        r = tk.Tk(); r.destroy(); return True
    except Exception:
        return False
gc = could_show_gui()
assert isinstance(gc, bool),                                      "⑨ detector bool (env-safe — dono-world!)"

assert round(_avg(_parse(SAMPLES[-1])), 1) == 52.5,               "⑩ Eshan avg 52.5!"
print(f"🏆 10/10 TOOLKIT-CHIEF BADGE — thin-GUI logic tumhari badge! 🖥️👑  (this-box can_show_gui={gc})")
```

## ✅ SOLUTIONS 🔓

<details><summary><b>S1-S4</b></summary>

```python
# S1-S4 ✅
def parse_marks(text: str) -> list[int]:
    return [int(x.strip()) for x in text.split(",") if x.strip()]

def average(marks):
    return sum(marks) / len(marks) if marks else 0.0

def grade_of(avg: float) -> str:
    return ("A" if avg >= 85 else "B" if avg >= 70 else "C" if avg >= 50
            else "D" if avg >= 35 else "F")

def report_text(name, marks_str):
    m = parse_marks(marks_str); avg = round(average(m), 1)
    return f"{name}: marks={m} avg={avg} grade={grade_of(avg)}"

print("①", parse_marks("80, 90, 100"), "| empty:", parse_marks(""))
print("②", average([60, 70, 72]), grade_of(round(average([60, 70, 72]), 1)))
print("③", report_text("Asha", SAMPLES[0]))

board = sorted((("".join(n), round(average(parse_marks(s)), 1)) for n, s in zip(NAMES, SAMPLES)),
               key=lambda t: -t[1])
print("④ leaderboard:", board)
```

<details><summary><b>S5-S7</b></summary>

```python
# S5-S7 ✅
def parse_harsh(text):
    out = []
    for tok in text.split(","):
        try:
            out.append(int(tok.strip()))
        except ValueError:
            pass
    return out
print("⑤ harsh:", parse_harsh("80, abc, 90, xyz, 70"))

def can_show_gui():
    try:
        r = tk.Tk(); r.destroy(); return True
    except Exception:
        return False
print("⑥ can_show_gui:", can_show_gui(), "(headless pe False — laptop pe True!)")

class MarksBoard:
    def __init__(self):
        self.rows = []
    def add(self, name, raw):
        m = parse_harsh(raw)
        self.rows.append({"name": name, "avg": round(average(m), 1)})
    def top(self, n=1):
        return sorted(self.rows, key=lambda r: -r["avg"])[:n]
    def summary(self):
        return [f"{r['name']}: avg={r['avg']}" for r in self.rows]

mb = MarksBoard()
for n, s in zip(NAMES, SAMPLES):
    mb.add(n, s)
print("⑦", mb.top(2), "| lines:", mb.summary()[:2])
# Takeaway: logic = reusable EVERYWHERE (GUI ✗ NB ✓ tests ✓) — fat-logic niyat! 🧠
```
