🌍 **Hook:** Desktop "button-work" — press a button, type in a text-box, windows
open-close — python's STANDARD tkinter toolkit (zero-install, usable by everyone!).
But here's a PLOT-TWIST: a GUI needs a DISPLAY (screen/terminal-X) — on a
headless notebook server ... let's watch it live!

📖 **Definitions:**
- `tk.Tk()` = the root-window (the box's SHOP-front!) 🪟
- widgets: `Label` (board), `Entry` (form-line), `Button` (switch) 🧩
- `command=fn` = the function behind the button (the signal-concept!)
- `mainloop()` = the GUI's heartbeat-loop (the box ALWAYS alive!) 💓

🔍 **TK-secret:** on notebook-servers `Tk()` gives **TclError: no display** —
run the GUI on your own machine! But the LOGIC INSIDE the GUI (calculations) —
THAT is always testable. That's why pro-devs keep the GUI and the LOGIC
separate — the "thin-GUI, fat-logic" mantra! 🧠


## 🐣 Level 0 — The Headless Drama LIVE: Where's the Display?! 🫣

````python
import tkinter as tk

try:
    root = tk.Tk()          # yahan dat-screen: WINDOW khul jaati!
    print("window ban gayi!", root)
    root.destroy()
except Exception as e:
    print(f"❌ {type(e).__name__}: {e}")
    print("💡 HEADLESS-signal detected — GUI apne LAPTOP pe chalegi; yahan LOGIC seekhenge!")
# Expected OUTPUT (notebook-server pe):
# ❌ TclError: no display name and no $DISPLAY environment variable
# 💡 HEADLESS-signal detected — GUI apne LAPTOP pe chalegi; yahan LOGIC seekhenge!
````


## 🐥 Level 1 — The Marks-Calculator's PURE-LOGIC (tested without the GUI!) 🧮

Whatever calculations need to go into the GUI's button, keep them in PLAIN
functions. → test them without a display, (a rehearsal-room logic-check!) 🎭


````python
def parse_marks(text: str) -> list[int]:
    """"80,90,70" → [80, 90, 70] (GUI Entry ka text-in raw string!)"""
    return [int(x.strip()) for x in text.split(",") if x.strip()]

def average(marks: list[int]) -> float:
    return sum(marks) / len(marks) if marks else 0.0

def grade_of(avg: float) -> str:
    return "A" if avg >= 85 else "B" if avg >= 70 else "C" if avg >= 50 else "D" if avg >= 35 else "F"

def report_text(name: str, marks_str: str) -> str:
    m = parse_marks(marks_str)
    avg = round(average(m), 1)
    return f"{name}: marks={m} avg={avg} grade={grade_of(avg)}"

print(report_text("Asha", "80,90,100"))
print(report_text("BK",   "60,70,72"))
print(report_text("CD",   "33,40,35"))
# Expected OUTPUT:
# Asha: marks=[80, 90, 100] avg=90.0 grade=A
# BK: marks=[60, 70, 72] avg=67.3 grade=C
# CD: marks=[33, 40, 35] avg=36.0 grade=D
````


## 🐔 Level 2 — Full GUI-PRINT-Sample (run it on your laptop!) 📝

Below is the code that opens the window on YOUR machine — here we only print it
(headless-honesty!):
```python
import tkinter as tk

def parse_marks(text):
    return [int(x.strip()) for x in text.split(",") if x.strip()]

def on_calc():
    m = parse_marks(entry.get())
    avg = sum(m) / len(m) if m else 0
    gr = "A" if avg >= 85 else "B" if avg >= 70 else "C" if avg >= 50 else "D" if avg >= 35 else "F"
    result.config(text=f"avg={avg:.1f} grade={gr}")

root = tk.Tk()
root.title("Marks-Calc 🖥️")
tk.Label(root, text="Marks comma-se daalo:").pack()
entry = tk.Entry(root, width=30); entry.pack()
tk.Button(root, text="Compute!", command=on_calc).pack()
result = tk.Label(root, text="?"); result.pack()
root.mainloop()     # 💓 heartbeat — click karke LIVE window aayegi!
```
🔍 **Anatomy-note:** `pack()` = top-to-bottom stacking; don't put brackets
after `command=` (give the function-NAME, not a call!); the logic = the same
functions we tested in Level-1 — that's why the GUI is "thin"! 🧠

## ⚠️ 5 Classic Mistakes (Tkinter)
| # | ❌ Mistake | ✅ Correct |
|---|---|---|
| 1 | `command=on_calc()` (an instant-call!) | give the NAME `command=on_calc` |
| 2 | expecting Tk() on a headless-server | laptop / the try-except note 🪟 |
| 3 | stuffing the calculation code into the button | pure-functions, thin-GUI 🧠 |
| 4 | pack/grid both together (trouble!) | ONE geometry-manager |
| 5 | forgetting mainloop (the window vanishes) | `root.mainloop()` always |
