## 👑 BOSS: 10-Assert Demo-Don Badge (brain → counter proof!) 🏆

```python
# 👑 BOSS CONSOLE — poora studio, seedha + launch assaul!
def _predict(study, sleep):
    raw = 10 * study + 2 * sleep + 5
    return min(100.0, max(0.0, raw))

def _ui(study, sleep):
    s = _predict(study, sleep)
    g = grade_of(s)
    return f"score={s:.0f} grade={g}", {g: 0.9, "down-step": 0.1}

assert _predict(8, 7) == 99.0,                                  "① 80+14+5=99!"
assert _predict(10, 8) == 100.0,                                "② CLIP 121→100 🛡️!"
assert _predict(0, 0) == 5.0,                                   "③ floor 5.0!"
t8, l8 = _ui(8, 7)
assert t8 == "score=99 grade=A" and l8 == {"A": 0.9, "down-step": 0.1}, "④ text+label pack!"
t_mid, _ = _ui(6, 6); t_low, _ = _ui(0, 4)
assert t_mid == "score=77 grade=B" and t_low == "score=13 grade=F", "⑤ bands B/F exact!"

demo = gr.Interface(
    fn=_ui,
    inputs=[gr.Slider(0, 12, value=6, step=1, label="study"),
            gr.Slider(4, 10, value=7, step=1, label="sleep")],
    outputs=[gr.Textbox(), gr.Label()],
    title="Marks-Studio 🎓",
    examples=[[8, 7], [6, 6]],
)
assert demo.title == "Marks-Studio 🎓",                          "⑥ title exact!"
assert [type(c).__name__ for c in demo.input_components] == ["Slider", "Slider"], "⑦ 2 sliders!"
assert [type(c).__name__ for c in demo.output_components] == ["Textbox", "Label"], "⑧ text+label out!"

demo.launch(prevent_thread_lock=True, server_name="127.0.0.1", server_port=7874, quiet=True)
_r = httpx.get("http://127.0.0.1:7874/", timeout=10)
assert _r.status_code == 200 and "gradio" in _r.text.lower(),   "⑨ counter-OPEN 200!"
demo.close()
assert "7874" in demo.local_url,                          "⑩ local_url port-sure + closed-hygiene!"
print("🏆 10/10 DEMO-DON BADGE — StudioNoor ka demo-day blockbuster! 🎓👑")
print("   (Section 13 COMPLETE — function → counter ka safar SIKH liya! ♻️🔥)")
```

## ✅ SOLUTIONS 🔓

<details><summary><b>S1-S4</b></summary>

```python
# S1-S4 ✅
def predict_raw(study, sleep):
    return 10 * study + 2 * sleep + 5

def predict_score(study, sleep):
    return min(100.0, max(0.0, predict_raw(study, sleep)))

def ui_predict(study, sleep):
    s = predict_score(study, sleep)
    g = grade_of(s)
    return f"score={s:.0f} grade={g}", {g: 0.9, "down-step": 0.1}

print("① raw(8,7):", predict_raw(8, 7), "| raw(10,8):", predict_raw(10, 8))
print("② clip(10,8):", predict_score(10, 8), "| clip(-5,0):", predict_score(-5, 0))
print("③ ui(6,6):", ui_predict(6, 6))

demo = gr.Interface(
    fn=ui_predict,
    inputs=[gr.Slider(0, 12, value=6, step=1, label="study"),
            gr.Slider(4, 10, value=7, step=1, label="sleep")],
    outputs=[gr.Textbox(), gr.Label()],
    title="Marks-Studio 🎓",
)
print("④ parts:", [type(c).__name__ for c in demo.input_components],
      "→", [type(c).__name__ for c in demo.output_components])
```

<details><summary><b>S5-S7</b></summary>

```python
# S5-S7 ✅
print("⑤ title:", demo.title)

demo.launch(prevent_thread_lock=True, server_name="127.0.0.1", server_port=7874, quiet=True)
r = httpx.get("http://127.0.0.1:7874/", timeout=10)
print("⑥ OPEN:", r.status_code, "| on?", "gradio" in r.text.lower(), "| url:", demo.local_url)
demo.close()

demo2 = gr.Interface(fn=ui_predict,
                     inputs=[gr.Slider(0, 12, value=6), gr.Slider(4, 10, value=7)],
                     outputs=[gr.Textbox(), gr.Label()],
                     title="Marks-Studio 🎓",
                     examples=[[8, 7], [6, 6]])
print("⑦ examples registered (cards):", len(demo2.examples), "| first:", demo2.examples[0])
# Takeaway: brain pure + clamp + frame + hygiene-close = demo-repeatable! 🎓♻️
```
