2. `ui_fn(...)` = pack the model's output (text + a LABEL dict!)
3. `gr.Interface(ui_fn, sliders, [text, Label])` = the counter
4. fn-level tests → launch → GET-200 → **close** (hygiene!)



## 🐣 Step 1 — Brain: Handcrafted Linear Model 🧠

(in the real ML section, `model.predict()` will come here — the SAME-FRAME!)


```python
import warnings; warnings.filterwarnings("ignore")

def predict_score(study_hrs: float, sleep_hrs: float) -> float:
    """linear-korma: padhai dominant, neend-thodi; clip 0..100 🎯"""
    raw = 10 * study_hrs + 2 * sleep_hrs + 5
    return min(100.0, max(0.0, raw))

def grade_of(avg: float) -> str:          # 13.06 flashback — same bands! ♻️
    return "A" if avg >= 85 else "B" if avg >= 70 else "C" if avg >= 50 else "D" if avg >= 35 else "F"

def ui_predict(study_hrs: float, sleep_hrs: float) -> tuple[str, dict]:
    score = predict_score(study_hrs, sleep_hrs)
    g = grade_of(score)
    label = {g: 0.9, "down-step": 0.1}          # simple-confidence demo 📇
    return f"score={score:.0f} grade={g}", label

for s, sl in [(8, 7), (6, 6), (0, 4), (10, 8)]:
    txt, lab = ui_predict(s, sl)
    print(f"study={s} sleep={sl} → {txt} | label={lab}")
# Expected OUTPUT:
# study=8 sleep=7 → score=99 grade=A | label={'A': 0.9, 'down-step': 0.1}
# study=6 sleep=6 → score=77 grade=B | label={'B': 0.9, 'down-step': 0.1}
# study=0 sleep=4 → score=13 grade=F | label={'F': 0.9, 'down-step': 0.1}
# study=10 sleep=8 → score=100 grade=A | label={'A': 0.9, 'down-step': 0.1}   (CLIP: raw 121 → 100!)
```

## 🐥 Step 2 — Counter-Frame: Sliders-Style Interface 🎛️

```python
import gradio as gr

demo = gr.Interface(
    fn=ui_predict,
    inputs=[
        gr.Slider(0, 12, value=6, step=1, label="📚 Study hours/day"),
        gr.Slider(4, 10, value=7, step=1, label="😴 Sleep hours"),
    ],
    outputs=[
        gr.Textbox(label="Prediction"),
        gr.Label(label="Grade Confidence"),
    ],
    title="Marks-Predictor 🎓 (mini-korma edition)",
    description="Padhai ghumao → forecast pao! (ML ke baad model-swap! ♻️)",
)
print("floats →", [type(c).__name__ for c in demo.input_components])
print("outputs →", [type(c).__name__ for c in demo.output_components])
print("title:", demo.title)
# Expected OUTPUT:
# floats → ['Slider', 'Slider']
# outputs → ['Textbox', 'Label']
# title: Marks-Predictor 🎓 (mini-korma edition)
```

## 🐔 Step 3 — OPEN → GET-200 → CLOSE (hygiene-loop!) 🚪

```python
import httpx
demo.launch(prevent_thread_lock=True, server_name="127.0.0.1", server_port=7873, quiet=True)
r = httpx.get("http://127.0.0.1:7873/", timeout=10)
print("marks-counter:", r.status_code, "| gradio-page?", "gradio" in r.text.lower())
demo.close()
print("closed ✓ — hygiene-meter top!")
# Expected OUTPUT:
# marks-counter: 200 | gradio-page? True
# closed ✓ — hygiene-meter top!
```
