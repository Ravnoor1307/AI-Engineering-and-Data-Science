## What is being tested

This task drops you into the clinical-trial dataset from the session and asks for the *cleaning judgement* rather than a specific function: identify the messy columns, sort the defects by severity, and fix them in a defensible order — then report what you changed and why. In other words, the skill being examined is exactly the session's core rhythm: **assess before you act, define a list, and make every repair traceable to a real defect.**

## How to approach it

**Replay the session's loop.** Load the data (`info()` + `head()` first), audit missing values, duplicates and invalid categories, and keep a written list of what you find before changing a single cell. **Fix by priority.** The column that breaks analysis (the invalid severity label, the dose that reads as text because of stray rows) outranks cosmetic tidy-ups — and every change gets verified after it is made with the same checks that found the problem. **Document.** A cleaning deliverable without a description of the defects and the decisions is a deliverable without evidence. Think in before → after and you are already thinking like a professional.

The dataset is the same four clinical-trial files you cleaned in the session, so you are not guessing in the dark — you are replaying a known mess under timing. That is the trick of the assessment: the *moves* are memorised, the *order* and the *documentation* are what get tested. Work the session's rhythm in exactly that order — `info()` and `head()` on arrival, a systematic missing-and-duplicate-and-category audit, a written defect list, repairs triaged by severity, and a re-audit after every repair until the frame is quiet. Deliver in the session's shape too: cleaned frames plus a short report that names each defect, the decision, and the before-and-after numbers. A report that reads like evidence is a report that reads like a professional cleaned it.
