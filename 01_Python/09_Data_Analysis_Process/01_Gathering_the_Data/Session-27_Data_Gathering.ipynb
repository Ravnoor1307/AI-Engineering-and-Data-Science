## Where every analysis really begins

## Read with suspicion

Data comes in with assumptions baked in — a missing cell that is actually the string `NA`, a column of numbers with a stray rupee sign, a date read as text. The single habit this session most wants to install is: **never trust a read until `info()`, `head()` and `dtypes` confirm it**. The parameters you study here are precisely the tools that rebuild your control over incoming data.


## What is Data Analysis?

**Data Analysis** is the systematic process of
**inspecting, cleaning, transforming, and modelling** data with the goal of
**discovering useful information, drawing conclusions, and supporting decision-making**.

It is not just "making charts". A real analysis answers a **business or scientific question**
and produces an **actionable recommendation**.

### Why does it matter?
| Reason | Explanation |
|--------|-------------|
| Evidence over opinion | Decisions based on data instead of guesswork |
| Find hidden patterns | Relationships that are invisible in raw data |
| Predict the future | Forecasting sales, churn, risk, demand |
| Measure performance | KPIs like revenue, profit, conversion, retention |
| Tell a story | Communicate findings so others act on them |

### Data → Information → Insight → Decision
```text
Raw Data   (unstructured, messy numbers)
   ↓  (clean, organise)
Information (structured, meaningful values)
   ↓  (analyze, compare)
Insight    (what it means, why it matters)
   ↓  (act)
Decision   (business action / recommendation)
```

##  Step 2 — Data Cleaning

**Definition:** Finding and fixing data-quality problems so your analysis is trustworthy.

### Common problems you will fix
| Problem | Example | Tool |
|---------|---------|------|
| Missing values | `NaN`, blank cells, "N/A" | `dropna()`, `fillna()` |
| Duplicates | same customer twice | `drop_duplicates()` |

### The cleaning mantra
> **Inspect first → decide a rule → apply it → verify.**  
> Always *look* at the data with `.info()`, `.isna().sum()`, `.duplicated().sum()` and
> `.describe()` **before** you change anything.





##  Step 3 — Data Transformation

**Definition:** Converting the clean data into a shape that is ready for analysis and modelling.

### Typical transformations
1. **Feature engineering** — create new, useful columns
   ```python
   df["Revenue"] = df["Price"] * df["Quantity"]
   df["AOV"]    = df["Revenue"] / df["Orders"]
   ```
2. **Encoding categoricals** — text → numbers
   - Label encoding (`pd.factorize`), One-hot encoding (`pd.get_dummies`)
3. **Scaling / normalising** — bring features to a similar range
   - Min-Max, Z-score (Standardisation)
4. **Aggregation** — summarize groups
   ```python
   df.groupby("Category")["Revenue"].sum()
   ```
5. **Reshaping** — pivot / melt / stack
   ```python
   df.pivot_table(index="month", columns="region", values="revenue")
   ```
6. **Joining / merging** datasets together (`merge`, `join`, `concat`)

**Deliverable:** a tidy, analysis-ready table.





##  Step 4 — Data Exploration (EDA)

- What are the main statistics? (`.describe()` — mean, median, std, min, max)
- How are values distributed? (histograms, boxplots)
- What correlates with what? (correlation matrix, heatmap)
- Which groups behave differently? (groupby comparisons)
- Are there patterns or anomalies?

### Typical EDA commands

```python
df.info()           # dtypes + non-null counts
df.describe()       # numeric stats
df.isna().sum()     # missing values

```

### 3. Opening a CSV file from a URL

Plain files are not the only source — datasets routinely live at a public URL, and pandas can read them directly. The classic trick shown here is the two-step: `requests.get(url)` downloads the file's content over HTTP, and `.text` gives that content as one long string; wrapping it in `io.StringIO(...)` hands pandas a *file-like object* it can parse exactly as if the file were local.

The indirection matters because it buys **control**: you can inspect the status code, add headers, or handle authentication before the bytes ever touch pandas — things a raw URL read cannot do. Once the DataFrame exists, everything is identical to a local file: a `.head()` to check, a `.shape` to size, and the analysis can begin. Remote data is just local data with a download step — and the download step is the part you now understand rather than trust.


````python
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
````

### 4. The `sep` parameter

CSV stands for comma-separated values, but hardly any real dataset uses only commas. `sep` tells pandas which character actually separates the columns: `sep='\t'` for tab-separated `.tsv` exports, `sep=';'` for European-style CSVs where commas are decimal points, and `sep='|'` for pipe-separated logs. If you ever see a whole column read into a single string, or a frame that has exactly one column, the first suspect is a wrong separator.

Two notes make this practical. `sep='\t'` is the escaped tab character, and whitespace-heavy files answer to `sep=r'\s+'` — a regex meaning any run of whitespace. And remember the reverse direction: `df.to_csv(path, sep=';')` writes with your chosen separator, so code that reads unusual files can also produce them. Know the separator, and a misparse becomes a two-second fix instead of a mystery.

One extra path: some exports smuggle quoted fields that contain the separator itself. If such a line misbehaves, that is exactly what the `quoting` and `escapechar` knobs exist for — recognise the symptom (one column, wrong length) and reach for the parameter, not a hack.



````python
pd.read_csv('Datasets/movie_titles_metadata.tsv')
````

````python
pd.read_csv('Datasets/movie_titles_metadata.tsv',sep='\t',names=['sno','name','release_year','rating','votes','genres'])
````

### 5. The `index_col` parameter

A DataFrame always carries an index — the labels printed in the unshaded first column. By default pandas numbers the rows `0, 1, 2, ...`, but when the file already contains a natural key (an `id`, a `date`, an `order_no`), `index_col` promotes that column to become the index instead of leaving it an ordinary column and inventing an integer index.

The benefit is bookkeeping: look-ups like `df.loc[key]`, joins on the index, and time-series alignment all become cheaper and clearer when the index is meaningful. Pass `index_col=0` for the first column, `index_col='order_id'` for a named column, or a list like `index_col=[0, 1]` to build a hierarchical MultiIndex from several columns. Everyday work rarely needs it; the moment your data has a stable identifier, this parameter turns a label from data into structure.

One mental check keeps it honest: the index is not a column, so `reset_index()` is the door back. If a downstream step expects a plain column, call `df = df.reset_index()` to reclaim that key before you merge or export.



````python
pd.read_csv('Datasets/aug_train.csv',index_col='enrollee_id')
````

### 6. The `header` parameter

When pandas reads a file, it uses the first row as the column names — unless you tell it otherwise. `header=None` says there is no header row, generate integer column names instead, and `header=4` says skip the first four rows of preamble and use row five as the header. Both appear constantly in the wild, because real files begin with comment lines, title lines or blank rows before the actual table starts.

Use `header=None` for computer-to-computer exports that carry no names; use an explicit integer when the true header sits a few rows down. The symmetric writer counterpart is `df.to_csv(path, header=False)`. One extra tip: if the file has junk rows *after* the table, that is `skipfooter`'s role — but the moment you need `header=`, also check `comment='#'` to drop lines beginning with a marker. Read the file's shape twice, and the header parameter will never surprise you again.


````python
pd.read_csv('Datasets/test (1).csv')
````

````python
pd.read_csv('Datasets/test (1).csv',header=1)
````

### 7. The `usecols` parameter

Large files are rarely fully useful: you may need only a handful of columns and loading the rest wastes memory and clutters every `head()`. `usecols=['name', 'age', 'salary']` reads only the columns you list — named by label, or by position with `usecols=[0, 2, 4]`.

Do this at *read time*, not after the fact: reading a 200-column file and then dropping 190 columns still loads all 200 into memory, while `usecols` keeps the whole parse lean from the start. It also does double duty as documentation — the short list is a concise statement of which fields the analysis actually depends on. When a file is wide or when you loop over many files, making `usecols` part of every read is a habit that pays in speed and in clarity.

One symmetry worth noticing: `usecols` and `dtype` solve overlapping problems — the first drops columns you never wanted, the second pins the types of the ones you keep. Together they make a read both lean and honest, and both are decisions you can copy from one file to the next.



````python
pd.read_csv('Datasets/aug_train.csv')
````

````python
pd.read_csv('Datasets/aug_train.csv',usecols=['enrollee_id','gender','education_level'])
````

### 8. The `squeeze` parameter

A single-column file read by pandas becomes a one-column DataFrame — rows with a column. With `squeeze=True`, a file that resolves to exactly one column is returned as a **Series** instead: a 1-D labelled array that behaves like a vector, with `.iloc[i]`, `.mean()`, `.value_counts()` and plotting semantics that suit a bare sequence.

The catch is in the name: `squeeze` only activates when the result genuinely is one-column (or one-row); multi-column files return a normal DataFrame untouched. So this is a *shorthand*, not a coercion — use it when you genuinely want the data as a Series and find the extra column wrapper noise. If you need a Series regardless of shape, `df.iloc[:, 0]` is the deterministic alternative. Know both, keep them separate, and the squeeze surprise (the Series you expected, the DataFrame you got) stops happening.

When in doubt, verify the return with `type(df)` — a two-character habit that settles the Series-versus-DataFrame question instantly and forever.



````python
pd.read_csv('Datasets/aug_train.csv',usecols=['gender'])
````

````python
pd.read_csv('Datasets/aug_train.csv',usecols=['gender']).squeeze()
````

### 9. The `skiprows` / `nrows` parameters

Two knobs control *how much* of the file pandas sees. `nrows=1000` reads only the first 1,000 data rows — the standard move for a first exploration of a huge file: fast, and enough to see column structure, dtypes and sample values before committing to a full read. `skiprows=5` discards the first five lines entirely, which is how you skip banner comments and other preamble that would otherwise corrupt the header.

The two combine naturally: `skiprows=1, nrows=500` means drop the junk line, then peek at the next five hundred rows. The parameter stays honest about scale too — `nrows` does not sample randomly, it reads the *head* of the file, so use it for orienteering, and reach for actual sampling later when you need a representative slice. Reading a preview first is the cheapest data-quality decision there is.

One caution: `skiprows` skips before the header is detected, so count carefully when the first data line also holds column names — a miscounted skip yields a header that is really a row.



````python
pd.read_csv('Datasets/aug_train.csv',nrows=100)
````

### 10. The `encoding` parameter

Text files store characters as bytes, and the mapping from bytes back to characters is the file's **encoding**. Modern files are usually UTF-8; older Windows exports often use latin-1/cp1252. When the bytes and the assumed encoding disagree you get `UnicodeDecodeError`, or worse, a file that reads clean but prints mojibake where accented letters should be. The fix is declaring the truth: `encoding='utf-8'`, `encoding='latin-1'`, or `encoding='utf-16'` for some exotic exports.

The robust workflow has three moves. First, try the read; second, when it fails, look at the *actual* bytes and match them; third, pin the correct encoding in the read so every later run is deterministic. Also remember the writer side: `pd.to_csv(path, encoding='utf-8')` and read it back with the same value. Encodings are invisible patience — they cost nothing when they work and cost an hour when they silently do not.

One practical shortcut: when the error message names a codec it cannot decode, the hint usually reveals the culprit — read the first raw bytes and match the encoding to them.



````python
pd.read_csv('Datasets/zomato.csv')
````

````python
pd.read_csv('Datasets/zomato.csv',encoding='latin-1')
````

### 11. The `on_bad_lines` parameter

No data file arrives perfect: a line with too few fields, a stray comma, a quote that never closes. In newer pandas these are handled deliberately via `on_bad_lines`: `'error'` (the default) raises and stops the read; `'warn'` prints a warning but continues past the offending line; `'skip'` quietly drops bad lines; and a callable lets you implement custom handling per malformed line.

The important shift is that a parse error is **information**: whichever mode you choose, the message should prompt a look at the raw file rather than a shrug. In practice, `'warn'` balances visibility with progress for exploratory reads, while rigorous pipelines end by keeping `'error'` so bad rows can never slip in silently. Pair it with `comment='#'` and deliberate `dtype` choices, and you have turned the least cooperative text files into something pandas can be trusted with.

One naming note: very old pandas called this `error_bad_lines`; if a Stack Overflow answer you copy uses that name, you are reading pre-1.3 material and should translate the modern spelling.



````python
pd.read_csv('Datasets/BX-Books.csv', sep=';', encoding="latin-1",on_bad_lines='skip')
````

### 12. The `dtype` parameter

pandas infers a column's type from its contents — and that inference is wrong disturbingly often. Mixed or dirty values push numeric columns into `object`; padded IDs like `0012` get their leading zeros stripped; downstream `.astype()` calls are then a correction cascade. `dtype={'id': str, 'price': 'float32'}` short-circuits all of it by declaring the type at read time.

The parameter earns its keep three ways: as **memory control** (`'float32'` instead of `'float64'`, or explicit `'category'`), as **correctness** (keeping a real identifier a string), and as **stability** (the file's columns keep their intended types regardless of which rows exist today). The honest discipline is to read the file, inspect `df.dtypes`, and then decide which columns the inference *should not* be trusted with — those get pinned in `dtype`. A column whose type you declare is a column you have understood.

One follow-up instinct: after a `dtype`-pinned read, run `df.dtypes` anyway and confirm the declaration took — the file may quietly disagree, and the check keeps the two in sync.



````python
pd.read_csv('Datasets/aug_train.csv').info()
````

````python
pd.read_csv('Datasets/aug_train.csv',dtype={'target':int}).info()
````

### 13. Handling dates

Dates arrive as strings, and strings cannot be sorted, filtered by day, or plotted on a proper time axis. `parse_dates=['date']` (or `parse_dates=['a', 'b']` for several) tells pandas to convert the listed columns to `datetime64` at read time. The payoff is immediate: `.dt.year`, `.dt.month`, `.dt.day` accessors, year-over-year compares, resampling, and time-aware plotting all become available.

Two refinements matter. When the date is split into separate year/month/day columns, pass a list of lists: `parse_dates=[['year', 'month', 'day']]` and pandas combines them while reading. And when the string format is unusual, `date_format` or a post-hoc `pd.to_datetime(col, format=...)` gives you full control. A column formally typed as a date is the difference between a chart that knows trends and a table that only thinks it does.

One habit to install now: make the format explicit whenever the values are at all unusual. `pd.to_datetime(col, format='%d-%m-%Y')` fails loudly on a wrong guess, whereas silent inference can misread `01-02-2026` as February first or January second. Ambiguous dates decoded wrong are a classic silent bug — decide the order, document it in the code, and the analysis never has to wonder.



````python
pd.read_csv('Datasets/IPL Matches 2008-2020.csv').info()
````

````python
pd.read_csv('Datasets/IPL Matches 2008-2020.csv',parse_dates=['date']).info()
````

### 14. The `converters` parameter

The `converters` parameter applies a function to a column *while the cell is being parsed*, before pandas even finishes reading: `converters={'price': lambda x: float(x.replace('INR', ''))}` strips currency symbols inside the read. It is per-column, runs once per value, and is the cleanest place to fix a column whose raw text is systematically dirty.

Compare it with the alternatives: `dtype` handles types but not cleaning, and a post-read swap performs cleaning but runs after the file is already in memory. Converters win when the transformation is simple, column-specific and known up front — parse failures mid-read stay isolated to that one column's function instead of bubbling up later. Reserve them for small, pure functions; anything complex deserves a named function, and anything requiring context across rows belongs after the read, in the DataFrame itself.

A clean mental division: converters are per-value honest, per-column narrow, and cheap to test in isolation — which makes them the right first reflex for a column like `price` that misbehaves the same way on every row.



````python
pd.read_csv('Datasets/IPL Matches 2008-2020.csv',converters={'team1':rename})
````

### 15. The `na_values` parameter

Real files mark missing data in dozens of ways — `NA`, `N/A`, `null`, `-`, `?`, `Not Available`, even a stray `nan` as text. By default pandas recognises empty cells plus a few common tokens, and everything else survives as a confusing *string* that looks like data. `na_values=['NA', 'N/A', 'null', '-', '?']` teaches pandas exactly which tokens mean *missing* for *this* file, so they arrive as `NaN` where missing-value logic (`dropna`, `fillna`, `isnull`) actually works.

The counterpart is `keep_default_na=False`, which switches off the built-in set so you can fully own the list — useful when a file legitimately contains the string `NA` as a real value, or when your missing list is deliberately minimal. Read the raw file once, list every junk token you can find, and pin them in `na_values`: missing data that arrives typed as missing is missing data you can finally handle.

One symmetry to keep in mind: `na_values` governs *reading*, while `fillna` and `dropna` govern *using* — decide the tokens once at the door, then lean on the standard missing-value toolkit everywhere after.



````python
pd.read_csv('Datasets/aug_train.csv')
````

````python
pd.read_csv('Datasets/aug_train.csv',na_values=['Male',])
````

### 16. Loading a huge dataset in chunks

When a file is too big for memory, do not read it at all — **stream it**. `chunksize=10000` returns an iterator of DataFrames; with `for chunk in reader:` you process ten thousand rows at a time and never hold the whole file. The classic pattern aggregates each chunk (`chunk.groupby(...).sum()`) and accumulates the partial results, which is how you compute totals and summaries over a dataset that would otherwise be unreadable.

Two design rules keep chunked code honest: every per-chunk operation must reduce (each chunk returns a *summary*, not the raw rows), and the accumulation must be idempotent so re-running the loop produces the same answer. Chunking trades a little code for a huge drop in memory pressure, and it is the pattern behind every big-file workflow from here to distributed data. Same pandas, same answers — just taken in bites instead of one gulp.

### Handling the 403 — server-side blocks

A `403 Forbidden` response is the server's polite refusal: the request was received and understood, but access is not granted. ### Handling errors with try / except
```python
try:
    r = requests.get(url, timeout=10)
    r.raise_for_status()
    soup = BeautifulSoup(r.text, "html.parser")
    ...extract...
except requests.exceptions.RequestException as e:
    print("Skipping page:", e)
```
If one page fails, you **skip it and continue** instead of crashing the whole job.



````python
import requests
from bs4 import BeautifulSoup

def scrape_page(url):
    try:
        r = requests.get(url, timeout=10)
        r.raise_for_status()
        return BeautifulSoup(r.text, "html.parser")
    except requests.exceptions.RequestException as e:
        print("Failed:", url, "->", e)
        return None

# soup = scrape_page("https://example.com")    # replace with a real URL
print("Robust scraper function defined.")
````

## B24 · Web Scraping Dynamic Sites with Selenium

Some sites render content with **JavaScript** — the HTML you download with `requests`
is empty. `Selenium` controls a **real browser** (Chrome/Firefox), executes the JS,
and lets you read the *rendered* page.

```bash
pip install selenium
# also need a matching chromedriver (or use Selenium Manager)
```

### Basic workflow
1. Create a `WebDriver` (Chrome options: headless mode, no sandbox)
2. `driver.get(url)` — load the page (JS runs)
3. Wait for elements (explicit waits, **not** `time.sleep`)
4. Find elements (`By.ID`, `.CLASS_NAME`, `.CSS_SELECTOR`...)
5. Read / click / scroll
6. `driver.quit()` to close it



### Selenium locators (`By`)

Selenium finds elements by explicit locator type, and the import `from selenium.webdriver.common.by import By` gives you the whole menu. `By.ID` for ids, `By.CLASS_NAME` / `By.CSS_SELECTOR` for classes, `By.NAME` for named fields, `By.XPATH` for structural paths, and `By.TAG_NAME` for plain tags like `links = driver.find_elements(By.TAG_NAME, 'a')`. Every watched element answers to at least one of these — the skill is choosing the one that is (a) stable and (b) pinpoints exactly what you want.

Rank them in practice: **CSS selectors first** (readable, robust to markup shifts), **XPath** when you must climb the tree or match on text, **id/name** when the page author provides them (they are uniquely designed for this), and raw tag names only for quick sweeps. The wrong locator is the classic flaky-test culprit: fine on Monday, broken on Tuesday, for no visible reason. Pick locators as carefully as you pick column names.

One last tip: when a selector stops matching, verify it in a fresh browser session before editing code — a stale page or a closed driver causes more false alarms than the site's markup does.



### Waiting strategies — explicit waits

JavaScript changes a page *after* it loads, and a scraper that reads too early sees empty containers. The robust answer is **explicit waits**: `wait = WebDriverWait(driver, 10)` and then `wait.until(EC.presence_of_element_located((By.ID, 'result')))` — the driver actively polls until the condition holds, or raises after ten seconds. That is fundamentally safer than a fixed `time.sleep`, because it waits for the *right thing* (the element existing) rather than a guessed amount of time.

Choose the condition to match what you actually need: `presence_of_element_located` for it being in the DOM, `visibility_of_element_located` for it being visible and clickable, `element_to_be_clickable` before `.click()`. Explicit waits localise the flake to the one interaction that needs it; implicit waits and blind sleeps are the currency of frustration. Browser automation is only as stable as its waits — engineer them deliberately.

One tuning note: set the timeout slightly longer than the slowest legitimate load, and combine the explicit wait with a guard — `wait.until(...)` then one defensive check — so a genuinely dead page raises a clear message instead of a vague one.
