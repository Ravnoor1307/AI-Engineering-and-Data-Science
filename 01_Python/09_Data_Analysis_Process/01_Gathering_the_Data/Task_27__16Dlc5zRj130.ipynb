## Problem 2 — the API pagination loop

**The pattern to reuse:** query the RapidAPI endpoint, read `totalCount` from the JSON response, compute how many pages of results exist, then loop fetching page by page, sleeping each iteration to stay inside rate limits. Each page's records get concatenated into one growing DataFrame — that is the whole problem in miniature.

**Be gentle and be thorough.** `time.sleep(2)` between calls is not a nicety; it is what keeps your key alive. And after the loop, verify closure: `len(df)` should equal `totalCount`, not a guess — print both and compare. Finally, writing the result into MySQL with the DataFrame's `to_sql` closes the circle: gathered data now lives in a queryable database. The line between a script that fetches data and a pipeline that *owns* data is exactly this verification step.

Two failure modes to head off: a page that stops returning results mid-loop (guard with a page limit) and a rate-limit reply inside a 2xx status (check the response body's message field, not just the code). Both are common, both are cheap to catch.
