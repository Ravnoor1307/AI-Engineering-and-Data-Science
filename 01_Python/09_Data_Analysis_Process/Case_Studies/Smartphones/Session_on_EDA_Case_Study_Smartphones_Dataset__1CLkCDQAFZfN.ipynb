# 📊 EDA Case Study — Smartphones

**Topic:** 09 Data Analysis Process · **Level:** 🟠 Intermediate · **Type:** CASE STUDY

## What this notebook is

This is the EDA companion to the smartphone cleaning case study: the same dataset, now already cleaned (by its own `v2` export), walking through the full exploratory loop — **univariate** distributions for every column, **bivariate** price-versus-feature plots, and **correlations** that compress the whole frame into numbers. The running question throughout is the one that makes EDA paying work: *what makes a smartphone cost what it costs?*

## The flow you will see

The notebook moves in four acts. **Setup** — display options and the cleaned CSV loaded with a glance at `shape` / `info()`. **Univariate** — each column interrogated with a plot and a count: price distribution, rating distribution, brand counts, and one-by-one looks at the engineered flags (`has_5g`, `has_nfc`, `has_ir_blaster`) and the numeric features (battery, RAM, screen, camera). **Bivariate** — the money questions: `barplot` of price against brand and cores, and **scatterplots** of price against processor speed and screen size, which is where the strongest direction reveals itself. **Correlations** — `df.corr()['price']` ranks every numeric column by how strongly it moves price; a `KNNImputer` fills the missing ratings so the correlation matrix is honest; and `get_dummies` turns the categorical columns into binary flags so even brand and OS take part.

## What to notice

Two habits carry the whole notebook. **Every chart answers a question** — if you cannot state the question, the plot is decoration. And **evidence precedes conclusions**: the correlation table names `has_nfc`, storage and screen size as price drivers; the notebook's job is to make you *see* that from the plots before the numbers confirm it. By the end you should be able to open any cell, say what it asks, and point to where the answer lives.

## Scatter — the continuous relationship

Two barplots just compared price across brands and core counts; the scatterplots that follow ask the *continuous* question — does price climb smoothly with processor speed, and does it climb with screen size? `sns.scatterplot(data=df, x='processor_speed', y='price')` draws every row as a point, so you can see both the trend and the dispersion around it.

Read a scatter like a detective. A clear upward drift means the variable genuinely prices into the phone; a cloud with no shape means it does not. Also look at the *edges*: the points hugging the top are the flagship phones (OnePlus, Samsung Ultra), and whether they cluster at the high end of the speed axis tells you if premium speed is a real driver or just a label that travels with premium phones.

The takeaway beats bar comparisons for continuous variables: a bar gives you an average, but a scatter gives you the relationship and its noise. When the two agree — bar shows higher medians, scatter shows an upward drift — you have a bivariate finding worth writing down.
