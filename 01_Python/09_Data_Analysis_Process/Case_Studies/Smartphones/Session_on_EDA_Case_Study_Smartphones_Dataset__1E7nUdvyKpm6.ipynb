# 📊 EDA Case Study — Smartphones (v3 / v4 / v5)

**Topic:** 09 Data Analysis Process · **Level:** 🟠 Intermediate · **Type:** CASE STUDY

## What this notebook is

This is the *second* EDA pass over the smartphone dataset — a continuation session that loads an already-cleaned `v3` file and keeps refining it: fixing leftover tag errors, engineering two more flags (`fast_charging_available`, `extended_memory`), re-exporting as `v4` and then `v5` (where `num_cores` finally becomes numeric), and finishing with KNN imputation, correlations, and dummies for modelling. The running theme is that **EDA is iterative** — each pass cleans and re-questioned the data, and each export freezes a better version.

## The flow you will see

The notebook moves through four phases. **Load and audit** — `v3` arrives at 980 rows and 24 columns, `info()` shows the remaining dodgy dtypes (`num_cores` still text), and `isnull().sum()` names the missing values (101 empty ratings). **Fix and enrich** — the tag errors are corrected in place, and the two engineered flags are built with `str.contains` and string parsing, giving each row a consistent answer to "does it fast-charge?" and "how far can memory expand?". **Export and retype** — `v4` is written, then `v5`, with `num_cores` converted from `Octa Core`-style text to its numeric equivalent. **Model-ready analysis** — `KNNImputer` fills the 101 missing ratings, `.corr()['price']` and `['rating']` quantify the drivers before and after imputation, `get_dummies(drop_first=True)` turns brand/OS into binary flags, and the final CSV ships as a ready modelling input.

## What to notice

Watch the version-control discipline: every meaningful change is checkpointed to a numbered `smartphone_cleaned_vN.csv`, so the notebook can always be re-run delta by delta. And notice the compare hand: correlations are computed *before* imputation and *after*, merged side by side, so you can see exactly what the KNN fill changed. That habit — always measure the effect of a step, not just its existence — is the professional differentiator this session teaches.

## 📖 Drop the redundancies, then correlate

The engineered columns have made several raw columns redundant, so this cell drops what is no longer needed: `df.drop(columns=['processor_name', 'extended_memory'], inplace=True)` removes the source columns whose useful content has already been captured by processed versions (`processor_brand`, `num_cores`, `extended_upto`). A column you parse into better columns and never look at again is dead weight — it clutters every `head()` and every future correlation.

With the frame lightened, the analysis asks its central question: `df.corr()['rating']` ranks every numeric column by how strongly it moves *rating* rather than price — this session's target variable. A strong positive coefficient means the feature travels with customer satisfaction; a near-zero one means it does not.

The sequencing is the lesson: **clean before you correlate**. Correlations computed on messy text columns or un-imputed gaps are correlations you cannot trust. Drop the redundant, normalise the remaining types, and only then read the coefficients.

## 📖 KNN impute, then correlate again

The 101 missing ratings are real data loss, so this cell fills them with **KNN imputation**: `KNNImputer(n_neighbors=3, weights='distance')` looks at each incomplete row, finds its three nearest neighbours in the numeric feature space, and guesses the rating from their (distance-weighted) average. `weights='distance'` matters — the closer a neighbour is, the more it influences the fill, which keeps the imputed values faithful to the local structure of the data.

Then the notebook does the honest thing: it correlates **before and after** the imputation. `x` holds the imputed correlations with price, `y` the raw ones, and `x.merge(y, on='index')` prints both side by side in one table. If the imputed coefficients have shifted — say `price` vs `rating` creeping from 0.28 to 0.34 — that shift is *evidence* that the missing rows were non-random, and the imputation recovered signal the naive table was hiding.

The habit to keep: never present a filled-in column without showing what the fill changed. An imputed frame is a claim, and a before-after compare is its receipt.


````python
from sklearn.impute import KNNImputer

knn = KNNImputer(n_neighbors=3,weights='distance')

return_array = knn.fit_transform(x_df)
````

````python
x = pd.DataFrame(return_array, columns=x_df.columns).corr()['price'].reset_index()
````
