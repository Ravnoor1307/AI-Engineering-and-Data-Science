## What you are about to do

This case study drops you into a genuinely messy **smartphones dataset** and lets you watch — cell by cell — how a professional cleans it. The notebook is structured as a single loop repeated until it is quiet: **assess** the data for quality and tidiness issues, **clean** the specific defects found, **restructure** columns that hide structure, and **export** a tidy file. The sheet lists over twenty quality issues and seven tidiness issues up front, so from the very first `info()` you know precisely which messes to hunt for.

## The assessment-first mindset

The opening cells do the detective work: `df.info()` and `df.describe()` give the shape and the ranges, then a long bullet list names the offenders — model names written inconsistently (OPPO / oppo), prices stored as strings with a rupee symbol and commas, columns whose values leaked into their neighbours (card text in the camera column, OS text in the card column), and a long catalogue of specific row numbers where a value is shifted, missing, or wrong. The important trick to internalise: **the assessment comes before any cleaning code, and the cleaning code always names the defect it is answering.**

## What to pay attention to

Watch the three deep moves the notebook uses again and again. First, **targeted repair over wholesale removal** — rows are dropped only when they are genuinely junk, and most broken rows are *fixed* by moving or re-parsing values. Second, **surgical pandas**: `df.copy()` keeps the original safe, `.str` accessors vectorise every text fix, `shift(1, axis=1)` realigns shifted columns in one call, and `value_counts()` after every repair proves the column is clean. Third, **typed, analysis-ready output** — every cleaned column ends as a sensible dtype (int, float, boolean) before the notebook exports `smartphone_cleaned_v2.csv`. By the end you should be able to say, for any cell, which defect it fixes and why that fix is safe.

## 📖 Working copies and price cleaning

This cell sets the tone for the whole case study. `df1 = df.copy()` creates a working copy so the original `df` stays untouched as an audit trail — every mutation from now on hits `df1`, and any doubt sends you back to the pristine original. The "original is sacred, fixes go to a copy" habit is the first rule of serious cleaning.

The price fix is a classic three-part pattern: **parse → clean → cast**. `str.replace('₹', '')` removes the rupee symbol, `str.replace(',', '')` strips the thousand separators, and `astype('int')` turns the remaining string into a real number. All of it runs through the vectorised `.str` accessor, so it is fast and loop-free.

Then comes a detail that marks careful work: `reset_index()` followed by `+ 2` renumbers the rows to match the *raw file's* line numbers. That tiny offset means any fix you make later can be cross-referenced against the original CSV exactly — when a school of analysts shares one file, row numbers are the common language.

Two habits on display are worth stealing. Second, **prefer repair to removal** — the notebook drops a handful of genuinely bad rows, but the *majority* of broken rows are fixed by relocating values, because data that can be saved should be saved. Wholesale row-dropping is the grader's anti-pattern: cheap, lazy, and quietly throws away salvageable information.

Notice also the sequencing: categories are worked one at a time — drop for one defect, verify, then the next. Cleaning that proceeds in auditable slices is cleaning you can defend afterwards.

## The shift realignment pattern

Some of this dataset's rows are not wrong — they are merely *one place to the left*. When a phone's text was assembled a cell short, everything after that gap slid right by one column, which is exactly what the row indices captured during assessment. This cell shows the pandas idiom that repairs it in one shot.

`temp_df = df1[df1['index'].isin(battery_rows)]` isolates the broken rows. `temp_df.iloc[:, 7:].shift(1, axis=1)` takes the affected tail columns and moves every value one column *right*, opening a blank in the first slot where the missing value will live. Then `df1.loc[temp_df.index, temp_df.columns[7:]] = x` writes the shifted block back into place — `.loc` aligning rows and columns so only the intended subset changes.

The key insight: **the values exist, they are just in the wrong position**. Realigning is not guesswork, and it is vectorised — one block operation repairs the whole batch. Once you have seen `shift(1, axis=1)` used this way, you will recognise the same move in later cells for camera and processor columns.

## value_counts — the misalignment detector

The cheapest way to find a wrong value is to ask the column to list what it actually contains: `df['os'].value_counts()` prints every distinct OS string with its count. Clean columns look tidy — a handful of `Android v12`, `Android v11`, `Android v9.0 (Pie)` buckets. Suspicious columns leak a comic offender in full view: a lone `Bluetooth` sitting among the operating systems.

That is precisely what happens before this cell: the `card` repairs finish with a clean count, and the next audit of `os` reveals stale tokens that belong nowhere near an OS column. The misfits are then set to `np.nan` and re-counted, proving the repair.

Hold onto the reflex: **`value_counts()` is the first tool for spotting misalignment, and `np.nan` is the honest way to clear a wrong value you cannot fix.** A categorical column that cannot be enumerated cleanly in one printout is a categorical column you have not finished cleaning.

## 📖 Camera count extraction

The camera column stores the *number* of cameras as English words — `Quad`, `Triple`, `Dual` — followed by megapixel details, with front and rear counts separated by an ampersand. Regex and `split` alone are too blunt, so this cell introduces a small **helper function**: `camera_extractor` maps each keyword to a digit (`Quad` → `'4'`, `Triple` → `'3'`, `Dual` → `'2'`, otherwise `'1'`), and `.apply()` runs it against every value.

Notice the two structural decisions. Rear and front counts are separated first with `split('&')`, so each lane is extracted independently. And `fillna('Missing')` fills the missing rear-camera rows with a sentinel the extractor understands — missingness handled at parse time, not discovered later.

The lesson is that **some cleaning needs a naming function, not a lookup**: a custom, testable function that encodes the domain logic (what the words mean) beats a monster one-liner, and `.apply()` keeps it vectorised over the whole column in one pass.


````python
def camera_extractor(text):

  if 'Quad' in text:
    return '4'
  elif 'Triple' in text:
    return '3'
  elif 'Dual' in text:
    return '2'
  elif 'Missing' in text:
    return 'Missing'
  else:
    return '1'
````

## Lambda-powered feature extraction

`extended_memory` is carved out of the `card` column with a single wall of pandas: `df1['card'].apply(lambda x: '0' if 'Not' in x else x.split('upto')).str.get(-1).str.strip().str.replace('Memory Card Supported', 'Not Specified')`.

Read it right to left. The lambda decides the sentinel first: if 'Not' appears in the card text, the phone has no expandable storage and yields `'0'`; otherwise the value is split on 'upto' and `.get(-1)` takes the *maximum* capacity mentioned. Then `.str.strip()` cleans whitespace and a final `.str.replace()` settles one known anomaly. All of it runs through the vectorised `.str` accessor, so the whole column transforms in one statement.

The wider lesson is **extraction is a pipeline of tiny steps, not one clever regex**: branch on a sentinel, split to a usable part, strip, replace. Name each step's role and the whole chain stays readable, repeatable, and reviewable. The pattern also scales to the remaining card and network fields: once proven on one column, repeat the same build for the rest and verify each result with the same `value_counts` sweep you already trust.


````python
df1['extended_memory'] = df1['card'].apply(lambda x:'0' if 'Not' in x else x.split('upto')).str.get(-1).str.strip().str.replace('Memory Card Supported','Not Specified')
````
