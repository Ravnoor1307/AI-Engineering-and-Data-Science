# Task 29 — EDA, the assessment

**Topic:** 09 Data Analysis Process · **Level:** 🟠 Intermediate · **Type:** TASK

## What is being tested

This task hands you an **insurance claims dataset** from Kaggle and one instruction: perform all the steps of EDA, document your conclusions, and then perform feature engineering. What is actually being examined is not the functions you can name but the *discipline* you bring to them — can you audit the columns, ask each column about its distribution, slice the data by the outcome you care about, and write down what you learned in a way another analyst could follow?

## How to approach it

**Do it in the session's order.** Load and glance first — `shape`, `head()`, `info()`, `describe()` — then audit missingness and dtypes. Chart each numeric column's distribution and each categorical column's counts. Then move to the bivariate questions the dataset is designed to answer: insurance charges against age, BMI, smoking status, region, and number of dependants, using the grouping, cross-tabulation, or scatter that fits the variable type. **Document as you go.** Every conclusion needs evidence — a number from the output, a plot, a comparison against a baseline. A notebook full of plots with no annotation is as weak as a notebook full of text with no plots.

**Finish with feature engineering, deliberately.** Treat it as the proof that the EDA worked: encode the flags the analysis surfaced (smoker, BMI band, age band, dependant count, region) into columns a modelling step can actually consume. Then audit the result the way the session taught you — re-run `info()` and `describe()` on the engineered frame, check for introduced missingness, and confirm every new column is traceable to a finding from the exploration. The task is complete when the notebook reads as one continuous argument, not a scrapbook of cells.

## Reading the task like a project brief

The problem statement looks like two lines, but treat it as a contract with three deliverables: the EDA itself, the conclusions, and the feature engineering. Plan the pipeline on paper before touching the notebook — what does the raw file give you, and what does the finished notebook have to prove?

**Lock a workflow in your head.** Glance (`shape`, `head`, `info`, `describe`) → audit (missingness per column, dtypes that surprise you) → univariate (distributions) → bivariate (relationships with `charges`) → conclude (write the finding next to each output) → engineer (new columns that encode the findings). If you keep that order, the notebook practically writes itself.

**Two habits score marks.** First, always state a baseline before a breakdown — the overall mean charges before the smoker-versus-non-smoker split, so the comparison has context. Second, back every claim with its evidence — when you write *obese smokers pay the most*, point at the grouped table that says so. The grader can re-run every cell; make the reasoning as inspectable as the code.
