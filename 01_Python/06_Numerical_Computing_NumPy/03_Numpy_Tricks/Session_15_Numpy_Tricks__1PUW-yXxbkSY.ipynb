### np.append

`np.append` adds values to the end of an array. It accepts the array, the values to add, and an optional `axis`. Without an axis the array is flattened first and the new values are simply tacked on, so the result has one more element. With `axis=1` the function appends a whole column provided the row counts match, and with `axis=0` it appends rows. Important: `np.append` always **returns a new array** — it never modifies the original — and because it copies data it is comparatively slow in loops, so prefer building a list and converting once when appending many times.

## 🔬 Deep Dive : np.append

```python
np.append(a, 200)                    # add element at end (copy!)
np.append(b, np.random.random((b.shape[0],1)), axis=1)  # add column
```

- Append — **returns new array** (costly; copy)
- axis=None → flatten + append
- single element append
- axis=0 rows append (must shape match)
- axis=1 add column
- Original unchanged
- For repeated appends prefer lists then np.array
- dims alignment required
- int/float coercion to common dtype
- panel note: append can be slow bulk — plan

````python
# code
np.append(a,200)
````

### np.concatenate

`np.concatenate` joins a sequence of arrays along an existing axis. It takes a tuple of arrays and an `axis` parameter, and every array must match in every dimension except the one being joined. `axis=0` stacks along rows, `axis=1` along columns; `hstack` and `vstack` are convenient wrappers for the same operation. The result is always a new combined array. This is the workhorse for assembling matrices out of pieces and the NumPy counterpart of pandas `concat`.

````python
# code
c = np.arange(6).reshape(2,3)
d = np.arange(6,12).reshape(2,3)

print(c)
print(d)
````

```python
np.concatenate((c,d), axis=0)   # stack rows → (4,3)
np.concatenate((c,d), axis=1)   # stack cols → (2,6)
```

- concatenate((a,b,...), axis) — join along axis
- axis=0 → rows (samples), axis=1 → cols (features)
- Must match along other axes
- List/tuple of arrays
- vs hstack/vstack convenient wrappers

### np.expand_dims

`np.expand_dims` inserts a new axis of size 1 at a chosen position. Shape `(15,)` becomes `(1, 15)` with `axis=0` or `(15, 1)` with `axis=1` — same data, new shape. That extra dimension is what makes broadcasting work: a row vector needs a leading 1, a column vector needs a trailing 1. You will reach for this whenever a function expects `(1, n)` or `(n, 1)` input, such as feeding one sample into a model or subtracting a column mean. `a[:, None]` and `a[None, :]` are equivalent shorthand.

### np.where

`np.where` is NumPy's vectorised conditional. Called with one argument — just a condition — it returns the indices where the condition is True. Called with three arguments — a condition, a value/array for True, and a value/array for False — it returns a new array built by choosing between the two at every position. This turns an elementwise if-else written by hand into a single vectorised call, which is dramatically faster on large arrays. It is the closest NumPy thing to a functional 'replace where'.

````python
a
````

## 🔬 Deep Dive : np.where — conditional select

```python
a
np.where(a > 50)          # INDICES of True positions (tuple)
np.where(a > 50, 0, a)    # replace: cond true → 0 else a
np.where(a % 2 == 0, 0, a)  # even→0, odd keep
```

- Two modes:
  1. `np.where(cond)` → indices (tuple of arrays)
  2. `np.where(cond, x, y)` → array: True→x, False→y
- Index mode: `a[np.where(a>50)[0]]` ≈ mask
- Fancy: positional finding
- Vectorized if/else — no loop!
- replace values based on condition
- Nested: np.select / boolean combo
- pandas .where similar
- Critical utility for conditional ops

````python
# find all indices with value greater than 50
np.where(a>50)
````

````python
# replace all values > 50 with 0
np.where(a>50,0,a)
````

````python
np.where(a%2 == 0,0,a)
````

### np.argmax and np.argmin

`np.argmax` returns the index of the maximum value, and `np.argmin` the index of the minimum — not the value itself. Flat (no axis) returns a single position; with `axis=0` it reports, for each column, which row holds that column's maximum; with `axis=1` it reports, for each row, which column holds that row's maximum. Retrieving the actual value is just `a[np.argmax(a)]` (or `a.max()`). Deep learning leans on this constantly: the predicted class is `np.argmax` over the probability vector. On ties, the first occurrence wins.

Cumulative sums are everywhere — cumulative revenue, running scores, and survival-style analyses — and cumulative products model successive multiplication, such as probability chains.

### np.percentile

`np.percentile` finds the value below which a given percentage of the data falls. The 50th percentile is the median, the 25th and 75th are the quartiles, and the gap between them is the interquartile range. You can request several at once by passing a list like `[25, 50, 75]`. Percentiles are robust summaries because a single weird outlier barely moves them, unlike the mean — which is why box plots and outlier sieves are built from them. `np.median(a)` is a handy alias for the 50th percentile.

### np.corrcoef

`np.corrcoef` computes the Pearson correlation matrix for one or more arrays supplied as rows. For two series `salary` and `experience` it returns a 2x2 matrix: the diagonal is 1.0 (every variable correlates perfectly with itself) and the off-diagonal is their correlation coefficient r. A coefficient near +1 means 'move together', near -1 means 'move opposite', and near 0 means 'no linear relationship'. Because correlations feed feature selection, multicollinearity checks, and basic EDA, this little function is a staple of any modelling workflow.

## 🔬 Deep Dive : np.corrcoef — correlation

```python
salary = np.array([20000,40000,25000,35000,60000])
experience = np.array([...])
np.corrcoef(salary, experience)
# → correlation matrix (1 diag, off-diag r)
```

- Pearson correlation coefficient matrix
- r ∈ [-1,1]: strength + direction
- r ≈ 1 positive strong, ≈ -1 negative, ≈ 0 none
- Interpretation: salary increases w/ experience?
- Matrix: [[var1_var1, var1_var2], [var2_var1, var2_var2]]
- r = cov(x,y)/(stdx*stdy)
- Data insight: feature relationships
- Pre ML: correlation screening
- Spearman: scipy alternative
- np.dot normalization manual possible

````python
salary = np.array([20000,40000,25000,35000,60000])
experience = np.array([1,3,2,4,2])

np.corrcoef(salary,experience)
````

### The closing eight — swapaxes, uniform, count_nonzero, tile, repeat, allclose

The final stubs round out the collection.

- `np.swapaxes(a, i, j)` exchanges two axes of an array; for 2D input it is the same as transposing, but it works on any pair of axes in higher dimensions.
- `np.random.uniform(low, high, size)` draws continuous random values evenly spread between `low` and `high` — useful when you need a custom range instead of `[0, 1)`.
- `np.count_nonzero(a)` counts how many elements are non-zero, or equivalently how many are True in a boolean mask — the fast way to tally matches.
- `np.tile(a, reps)` repeats the **whole array** to build a bigger block.
- `np.repeat(a, n)` repeats **each element** n times before moving on — the distinction from `tile` is worth memorising.
- `np.allclose(a, b)` returns True when two arrays are equal within a small tolerance; because floats rarely match exactly after computation, `allclose` is the community-standard comparison for test assertions.
