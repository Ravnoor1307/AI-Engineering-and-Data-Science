## 9.1 Hint — Q9: Percentile identities

Start with an array in [0, 1], round it to three decimals with `np.around(arr, decimals=3)`, then verify four facts:

- `np.percentile(arr, 0)` equals `arr.min()` — the smallest value is at percentile zero by definition.
- `np.percentile(arr, 100)` equals `arr.max()`.
- `np.percentile(arr, 50)` is the median.
