Welcome to Numerical Computing with NumPy! This session is where your Python skills start turning into true data-science skills. So far you have worked with Python lists, which are flexible but slow when dealing with tens of thousands of numbers. NumPy — short for Numerical Python — is the foundational library for scientific computing. It provides the **ndarray** (n-dimensional array), a homogeneous, fixed-size grid of numbers that lives in one contiguous block of memory. Because of that layout, operations on whole arrays run in compiled C code, often tens of times faster than an equivalent Python loop.

Why does this matter? Nearly every serious Python library used for machine learning and data analysis — pandas, scikit-learn, TensorFlow, PyTorch — is built on top of NumPy's array model. When you load images, tables, or sensor data, the numbers arrive as NumPy arrays. When you train a neural network, the weights are stored and updated inside arrays. Mastering NumPy is therefore not optional for a data scientist; it is the shared language of the field.

A mental model that will serve you well: think of a NumPy array as a **grid of numbers with a fixed shape**. Operations act on the whole grid at once, and the shape — `(rows, columns)` or `(depth, rows, columns)` — defines how you step through the data. Everything else in NumPy is built from that single idea. Let's begin.

### Numpy Arrays Vs Python Sequences

- NumPy arrays have a fixed size at creation, unlike Python lists (which can grow dynamically). Changing the size of an ndarray will create a new array and delete the original.

- The elements in a NumPy array are all required to be of the same data type, and thus will be the same size in memory.

- NumPy arrays facilitate advanced mathematical and other types of operations on large numbers of data. Typically, such operations are executed more efficiently and with less code than is possible using Python’s built-in sequences.

- A growing plethora of scientific and mathematical Python-based packages are using NumPy arrays; though these typically support Python-sequence input, they convert such input to NumPy arrays prior to processing, and they often output NumPy arrays.

## ones / zeros / random / linspace / identity

```python
np.ones((3, 4))                # a 3x4 array filled with 1s
np.zeros((3, 4))               # a 3x4 array filled with 0s
np.random.random((3, 4))       # uniform random floats in [0, 1)
np.linspace(-10, 10, 10, dtype=int)   # 10 evenly spaced integers
np.identity(3)                 # 3x3 identity matrix
```

- `ones(shape)` and `zeros(shape)` accept a shape tuple and fill every cell with 1 or 0 — perfect for placeholder matrices, masks, and bias initialisation.
- `random.random(shape)` produces uniform random floats between 0 and 1; `np.random.rand` is a similar helper.
- `linspace(start, stop, n)` returns exactly `n` points evenly spaced between `start` and `stop`, both included. Use `linspace` when you know how many points you want; use `arange` when you know the step size.
- `identity(n)` builds the n x n identity matrix with 1s on the main diagonal.
- All of these builders accept either a single integer (1D) or a shape tuple (N-D).

## Why these builders matter

A good NumPy habit is to allocate arrays with a one-line builder instead of filling a Python list with a loop.

- `ones` and `zeros` are the standard way to create blank arrays before filling them. In machine learning you will initialise weight matrices, masks, and bias vectors to zeros all the time.
- `random.random` samples uniformly from the interval `[0, 1)`. When you need a normal (Gaussian) distribution instead, reach for `np.random.randn`; a uniform sample produces a flat histogram, while a normal sample produces the familiar bell curve.
- `linspace` is count-based: `np.linspace(-10, 10, 10)` always gives exactly 10 values, `-10` through `10`. `arange` is step-based and end-exclusive: `np.arange(-10, 11, 2)` gives `-10, -8, ..., 10`. Choosing the right one depends on whether you think in terms of the number of points or the step size.
- `identity(n)` is the matrix that leaves every vector unchanged under multiplication: `I @ v == v`. It is the multiplicative identity for matrices, the same way 1 is for numbers.
- Passing `dtype=int` to `linspace` truncates the evenly spaced floats to integers, which is exactly what the code cell below demonstrates.

### Changing Datatype

## Changing datatype

```python
array.astype(np.int32)   # returns a NEW array in the new type
```

- `astype(type)` converts an array to a new `dtype` and returns a new array; the original remains untouched.
- Float to int truncates towards zero, so 3.9 becomes 3.
- Int to float widens the type so fractional values can be represented.
- Why bother? Lower-precision types such as `int32` or `float32` use half the memory of their 64-bit counterparts. Downcasting large datasets saves memory, and aligning the `dtype` of model inputs avoids cryptic errors later.

## 🔬 Deep Dive : Changing datatype — astype

```python

```

- astype — dtype change; original untouched (immutable-ish)
- float64 → int32: truncates decimals
- int → float: precision up
- memory: float64 takes double int32
- usage: downcast dataset for efficiency
- object → numeric conversions reserved
- errors: 'raise'/'ignore'
- monetary precision: keep float64
- ML: model input dtype alignment
- Check result: .dtype / .nbytes

### Iterating

## Iterating arrays

```python
for i in a1:            # 1D — yields each element
for i in a2:            # 2D — yields each row
for i in a3:            # 3D — yields each 2D block
for i in np.nditer(a3): # yields every scalar element
```

- Iterating an ndarray with a plain `for` loop walks the **outermost axis**.
- A 2D array yields rows; a 3D array yields 2D slabs; only a 1D array yields scalars.
- `np.nditer(a)` flattens the traversal so you visit every element one by one.
- In real data pipelines, prefer vectorised functions over loops — but know how iteration works so you can inspect arrays during debugging.

## 🔬 Deep Dive : Iterating arrays

- Default iteration traverses **first axis**
- 1D → items, 2D → rows, 3D → 2D blocks
- `np.nditer` — flatten to elements (np.ravel too)
- Vectorized preferred: `a1.sum()` not sum(a1)
- flatten: `.flat`, `.ravel()`, `.flatten()`
- Read: axis-wise structure
- Iteration pythonic loops slow — avoid big ones
- use apply/vectorized/functions instead
- nditer useful debug & custom traversal

### Reshaping

## reshape / transpose / ravel

```python
np.transpose(a2)   # swap rows and columns
a2.T               # shorthand for the same
a3.ravel()         # flatten into a 1D view
```

- `reshape` — change the shape while keeping the total number of elements fixed. The product of the new dimensions must match the old ones.
- `transpose` / `.T` — reverse or permute the axes (for 2D that means rows to columns).
- `ravel()` — collapse everything into one flat 1D array; it is a view when possible, so use `flatten()` if you want a guaranteed copy.
- After reshaping, all indexing and slicing is done on the new shape.

### Stacking

## Stacking — concatenate

```python
np.hstack((a4, a5))   # side by side: columns are joined
np.vstack((a4, a5))   # one below the other: rows are joined
```

- `hstack` joins arrays along the column direction — the number of rows must match, and the result has more columns.
- `vstack` joins arrays along the row direction — the number of columns must match, and the result has more rows.
- Shape consistency matters: mismatched shapes raise a `ValueError`.
- Stacking is how you combine new features (hstack) or new observations (vstack) in a data pipeline.

## 🔬 Deep Dive : Stacking — concatenate

```python
a4 = np.arange(12).reshape(3,4)
a5 = np.arange(12,24).reshape(3,4)

np.hstack((a4,a5))   # side-by-side (concat cols) → (3,8)
np.vstack((a4,a5))   # top-bottom (concat rows) → (6,4)
np.concatenate((a4,a5), axis=0)   # general
np.dstack(...)       # depth
```

- hstack — horizontal: same rows, columns add (along axis 1)
- vstack — vertical: rows add (axis 0)
- concatenate(axis=) — general combine
- dimensions must align on non-joining axis
- stacking appends dataset columns/rows (feature join)
- arrays for joining — datasets more observations/features
- setup: a4, a5 shape (3,4) both
- Misalignment → ValueError
- For lists of arrays: `np.stack([a,b,c])` new axis

````python
# horizontal stacking
a4 = np.arange(12).reshape(3,4)
a5 = np.arange(12,24).reshape(3,4)
a5
````
