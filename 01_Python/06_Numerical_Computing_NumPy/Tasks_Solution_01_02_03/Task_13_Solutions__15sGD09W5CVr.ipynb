### `Q-3`Write a function to create a 2d array with 1 on the border and 0 inside. Take 2-D array shape as (a,b) as parameter to function.

```
[[1,1,1,1],
[1,0,0,1],
[1,0,0,1],

```

## 3.1 Solution notes — Q3: Border array

Start from `np.ones((a, b))` so every cell is 1. The interior is selected by the slice `x[1:-1, 1:-1]` — all rows from index 1 to the second-to-last, and all columns likewise — and assigning 0 to that region hollows out the middle. The border survives untouched because the slice never includes row/column 0 or the last one. For a (4,5) input the function returns a 4x5 ring, matching the spec's example shape. Note how a corner case like `a=1` or `b=1` would behave: the slice selects nothing, so the 'interior' is empty — an edge worth remembering.

````python
#Code here
def borderArray(a,b):
    x = np.ones((a,b))
    x[1:-1, 1:-1] = 0    # Except the border, changing all other elements to 0
    return x
borderArray(4,5)
````

## 7.1 Solution notes — Q7: Distances in integer type

The array `a = np.random.randint(1, 100, (10, 2))` holds ten 2D points. Subtracting `p = np.array([2, 3])` broadcasts the single point against every row, squaring `(a-p)**2` stays elementwise, and `np.sum(..., axis=1)` collapses the coordinate axis to one distance per point. `np.sqrt` finishes the Euclidean metric and `.astype(int)` satisfies the 'integer distance' requirement by truncating. ## 11.1 Solution notes — Q11: Softmax with validation

The function guards two requirements before any math: `type(arr) != np.ndarray` raises `TypeError('Requires Numpy Array')`, and `arr.ndim > 1` raises `TypeError('Requires 1D Array')`. Then it computes `s = np.sum(np.exp(arr))` and returns `np.exp(arr) / s` — a vectorised softmax. The example outputs show how dramatically exponentials separate the scores: the largest input dominates the probability mass while the rest collapse toward vanishing values.

````python
#Code here
def softmax(arr):
    if type(arr) != np.ndarray:
        raise TypeError("Requires Numpy Array")
    elif arr.ndim > 1:
        raise TypeError("Requires 1D Array")
    s = np.sum(np.exp(arr))
    return np.exp(arr)/s
softmax(np.array([86.03331084, 37.7285648,  48.64908087, 87.16563062, 38.40852563, 37.20006318]))
````

## 12.1 Solution notes — Q12: Variadic vertical stack

`def vertical_stack(*args)` collects any number of arguments into a tuple. A loop validates that each is an ndarray, raising `TypeError` otherwise; then `np.vstack(args)` joins them along axis 0. The demo builds `a` from `np.arange(10).reshape(2, -1)` (auto-inferred 5 columns), `b` from `np.repeat(1, 10).reshape(2, -1)`, and stacks two and then three matrices, mixing integer and float dtypes (promoted to float). The variadic signature plus validation is the reusable pattern.

````python
#Coder here
def vertical_stack(*args):
    for i in args:
        if type(i) != np.ndarray:
            raise TypeError("Requires Numpy Array")
    return np.vstack(args)

a = np.arange(10).reshape(2, -1)
print("a=",a)
b = np.repeat(1, 10).reshape(2, -1)
print("b=",b)
print(vertical_stack(a,b))
c = np.random.random((2,5))
print("c=", c)
vertical_stack(a,b,c)
````

## 13.1 Solution notes — Q13: Date array

This is the most intricate function in the task. `np.arange(start, end, dtype='datetime64[D]')` generates daily dates but is end-exclusive, so the solution first advances the supplied end date by one day. It parses the end string, increments the day, and — when that overflows the month — rolls into the next month using a hardcoded dictionary of days-per-month. The adjusted end is then fed to `arange`, making the range inclusive. The three examples all return the full expected date lists. This is a great demonstration of mixing string handling with datetime64 arithmetic.

````python
#Code here

def date_array(start: str, end: str):
    if type(start) != str or type(end) != str:
        raise TypeError
        
    total_days_of_month = {"01": 31, "02": 28, "03": 31, "04":30, "05": 31, "06":30, "07":31, "08":31, "09":30, "10": 31, "11":30, "12":31}

    end = end.split("-")
    end_last = int(end[-1]) + 1
    
    # If the next day of end falls in the next month, account for that
    if total_days_of_month[end[-2]] < end_last:
        days_diff = end_last - total_days_of_month[end[-2]]
        end[-1] = f'0{days_diff}' if days_diff< 10 else f'{days_diff}'
        next_month = int(end[-2]) + 1
        end[-2] = f'0{next_month}' if next_month< 10 else f'{next_month}'
    else:
        end[-1] = f'0{end_last}' if end_last < 10 else f'{end_last}'
    end = "-".join(end)
    return np.arange(start, end, dtype="datetime64[D]")    # Use arange() to generate all dates between start and end 
date_array(start = '2020-11-25', end = '2020-11-30')
````

### `Q-19:` Write a function which will accept 2 arguments.
First: A 1D numpy array arr

Second: An integer n {Please make sure n<=len(arr)}

Output: The output should be the nth largest item out of the array
```
# Example1 : arr=(12,34,40,7,1,0) and n=3, the output should be 12
# Example2 : arr=(12,34,40,7,1,0) and n=1, the output should be 40
```

## 19.1 Solution notes — Q19: nth largest

The function sorts the array in place with `arr.sort()`, validates `n` against the length (raising `IndexError`), and returns `arr[-n]` — the n-th from the end, since ascending sort puts the largest values last. For `arr = [12, 34, 40, 7, 1, 0]` and `n=2`, the sorted values are `[0, 1, 7, 12, 34, 40]` and `arr[-2]` is 34. If preserving the caller's original order mattered, `np.sort(arr)` would return a sorted copy instead of mutating.

````python
#Code here
def nthmax(arr, n):
    if n>len(arr):
        raise IndexError("n is way out of limit")
    arr.sort()
    return arr[-n]
nthmax(np.array([12,34,40,7,1,0]),2)
````

## 20.1 Solution notes — Q20: Repeat and tile pattern

`np.repeat(a, 3)` repeats each element consecutively, producing `[1, 1, 1, 2, 2, 2, 3, 3, 3]`; `np.tile(a, 3)` repeats the whole array, producing `[1, 2, 3, 1, 2, 3, 1, 2, 3]`. `np.hstack` joins them into the exact target sequence — 18 elements with no hardcoding, purely composed from the input array. This contrast between per-element repetition (`repeat`) and whole-array tiling (`tile`) is a classic wording lesson.

````python
#code here
a = np.array([1,2,3])
np.hstack([np.repeat(a, 3), np.tile(a, 3)])
````
