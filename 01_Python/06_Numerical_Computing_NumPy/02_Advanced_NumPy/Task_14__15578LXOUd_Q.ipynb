## 6.1 Hint — Q6: The Cauchy matrix

The Cauchy matrix is defined by `C[i, j] = 1 / (x_i - y_j)`. The whole construction is one broadcast: reshape x into a column with `x.reshape(-1, 1)`, then write `1 / (x - y)`. The column broadcasts against the 1D y, forming a complete grid of differences whose reciprocals fill the matrix — shape (4,3) for the given vectors.
