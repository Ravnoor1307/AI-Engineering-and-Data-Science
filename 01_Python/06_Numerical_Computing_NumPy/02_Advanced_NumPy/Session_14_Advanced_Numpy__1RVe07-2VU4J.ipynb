# Advanced NumPy — Session 14

Session 13 gave you the building blocks: creating arrays, reading their shape, slicing, and reshaping. This session upgrades you from a NumPy user to a NumPy thinker. The four big ideas you will meet are speed, power indexing, broadcasting, and formulas.

First, **speed**. A Python loop touches each element through the interpreter, which is slow. A NumPy operation hands a whole array to compiled C code that walks the memory directly, so the same addition can be 50-100 times faster — and leaner in memory, because every element is stored cleanly next to its neighbours. This is why real machine learning loops are almost invisible inside NumPy calls.

Second, **power indexing**. Plain slicing picks blocks. Fancy indexing picks any scattered positions with an index array, and boolean masking picks exactly the values that satisfy a condition. These two techniques replace entire pages of filter logic with a single line and are the closest thing data scientists have to a SQL WHERE clause.

Third, **broadcasting**. When two arrays have different shapes, NumPy does not panic — it stretches the smaller one across the larger, provided the dimensions are compatible. Mastering the two broadcasting rules lets you write code for the general case and stop reshaping by hand.

Fourth, **formulas**. Machine learning is a collection of vectorised equations. The sigmoid, mean-squared-error, and binary cross-entropy functions are each a one-liner once you combine `np.exp`, `np.mean`, and elementwise arithmetic. Seeing a neural network as a chain of array formulas turns scary math into plain NumPy.

You will also handle missing data with `np.nan`, visualise formulas with matplotlib, and build coordinate grids with `meshgrid` for surface plots. Every idea here is a tool you will reuse in pandas and in every model you train. Let's begin with why NumPy wins on speed.

## Numpy vs Python lists — speed

```python
import numpy as np
import time

# Python list addition (loop)
c = [x + y for x, y in zip(a, b)]

# Numpy vectorized
a = np.arange(10000000)
c = a + b
```

- NumPy is **C/CPU optimized** — ~50-100x faster than a pure loop
- This code is measured with `time.time()`
- A vectorized operation runs in a single pass, no loop involved
- In data science, this is exactly what makes the difference on large arrays


## 🔬 Deep Dive : Speed — vectorization wins

```python
# list approach
a = [i for i in range(10000000)]
b = [i for i in range(10000000)]
# manual loop... slow

# numpy
import numpy as np
a = np.arange(10000000)
b = np.arange(10000000)
# a + b one call — C speed
```

- Vectorized ops run at C speed (SIMD + no Python loop)
- Same operation: numpy ~50-100x faster than Python list loop
- Reason: tight C loop + contiguous memory
- Python per-element creates interpreter overhead
- Approx 3.26s vs 0.06s (~54x)
- CPU: numpy uses optimized BLAS/LAPACK in many ops
- Rule: never element-loop when vector op exists
- This scale: 10M element addition instant
- ML data millions row — numpy is the backbone

````python
# speed
# list
a = [i for i in range(10000000)]
b = [i for i in range(10000000,20000000)]

c = []
import time 

start = time.time()
for i in range(len(a)):
  c.append(a[i] + b[i])
print(time.time()-start)
````

````python
# numpy
import numpy as np
a = np.arange(10000000)
b = np.arange(10000000,20000000)

start = time.time()
c = a + b
print(time.time()-start)
````

- int8 → 1 byte; float32 4; float64 8
- Contiguous = cache-friendly fast
- numpy memory ~= size * itemsize only
- Efficiency: store 10M ints → MB not GB
- Convenience: homogeneous type assumption
- ML models train on (N, features) numpy — compact

## Fancy indexing

```python
a = np.arange(24).reshape(6, 4)
a[:, [0, 2, 3]]    # select columns 0, 2 and 3 with a list
```

- Normal slicing selects a *range* of positions (`1:3`).
- **Fancy indexing** selects *arbitrary* positions using a list or array of indices such as `[0, 2, 3]`.
- The index list can be in any order, does not need to be sorted, and may repeat values.
- Unlike a slice, the result of fancy indexing is a **copy**, not a view — later changes to it will not affect the original array.

## 🔬 Deep Dive : Fancy indexing — index arrays

```python
a = np.arange(24).reshape(6,4)
a[:,[0,2,3]]      # select specific COLUMNS (0,2,3) — copy
a[[1,2,0],:]      # specific rows
```

- Fancy indexing: index/pick arbitrary positions with INDEX ARRAYS
- Result is a **COPY** (unlike slicing views)
- List/array of indices → gather elements
- Row/col: `a[rows, cols]` tag cross
- Boolean masks alongside (next)
- Use: reorder rows, subsample, feature select
- `a[[0,1],[1,2]]` → (a[0,1], a[1,2]) pair-wise
- Different from slices — fancy = copy semantics
- DataFrame `.iloc` same concept later
- Powerful data extraction tool

## Boolean / mask indexing

```python
a = np.random.randint(1, 100, 24).reshape(6, 4)

a[a > 50]                    # keep every element greater than 50
a[a % 2 == 0]                # keep even numbers
a[(a > 50) & (a % 2 == 0)]   # combine several conditions
a[~(a % 7 == 0)]             # keep elements not divisible by 7
```

- A comparison produces a boolean array of the same shape — the **mask**.
- Placing the mask inside brackets, `a[mask]`, returns only the elements where the mask is True.
- Combine conditions with `&` (and), `|` (or), and `~` (not) — never Python's `and`/`or`, which do not broadcast over arrays.
- Surround every condition in parentheses when combining them.
- This is NumPy's equivalent of a SQL WHERE clause and your go-to tool for filtering data.

## Deep Dive : Boolean / mask indexing

```python
a = np.random.randint(1,100,24).reshape(6,4)
a > 50                          # bool mask same shape
a[a > 50]                       # values where True (1D result)
a[a % 2 == 0]                   # evens
a[(a > 50) & (a % 2 == 0)]      # combine conditions with &
a[~(a % 7 == 0)]                # not divisible by 7
```

- Boolean mask = same-shape bool array
- `a[mask]` → elements where mask True (1D flat result)
- Conditions produce mask: `a>50`, `a%2==0`
- Combine: `&` (and), `|` (or), `~` (not) — NOT python and/or
- Parentheses required around conditions
- Masking = most common filtering tool
- Write: `a[(a>50) & (a<90)]` ranged
- Sets membership by mask (np.isin later)
- Result always 1D flattened selection
- Critical: & not and — numpy error otherwise


````python
# Boolean Indexing
a = np.random.randint(1,100,24).reshape(6,4)
a
````

### Broadcasting

The term broadcasting describes how NumPy treats arrays with different shapes during arithmetic operations.

The smaller array is “broadcast” across the larger array so that they have compatible shapes.

## Broadcasting — same shape

```python
a = np.arange(6).reshape(2, 3)     # (2,3)
b = np.arange(6, 12).reshape(2, 3) # (2,3)
a + b   # elementwise direct
```

- Same shape → direct elementwise, no magic involved
- Broadcasting comes into play when the shapes DIFFER


## Deep Dive : Broadcasting — same shape

```python
a = np.arange(6).reshape(2,3)
b = np.arange(6,12).reshape(2,3)
a + b        # elementwise — shapes match directly
```

- Broadcasting = elementwise ops between compatible shapes
- Same shape → direct elementwise, no copy
- Compares shape dim-by-dim
- Rule 1: same dims → fine
- a + b → c[i,j] = a[i,j]+b[i,j]
- Vectorized — fast
- Diff shape needs broadcast rule (covered later)
- Matrices: shapes equal or 1 / missing
- Alignment: trailing dims first
- Works for + - * / ** etc.
- numba-vectorized intuitive


````python
# same shape
a = np.arange(6).reshape(2,3)
b = np.arange(6,12).reshape(2,3)

print(a)
print(b)

print(a+b)
````

## Broadcasting — different shapes

```python
a = np.arange(6).reshape(2, 3)     # (2, 3)
b = np.arange(3).reshape(1, 3)     # (1, 3)
a + b   # b is stretched to (2, 3) and added elementwise
```

- When shapes differ, NumPy **stretches** the smaller array so both end up compatible.
- Here `b` has shape `(1, 3)`; its single row is replicated until `b` behaves like `(2, 3)` before the addition.
- No loops are written — the operation stays fully vectorised.
- Stretching only works when every dimension of the two arrays matches or is 1; the formal rules follow below.

## 🔬 Deep Dive : Broadcasting — different shapes

```python
a = np.arange(6).reshape(2,3)   # (2,3)
b = np.arange(3).reshape(3)     # (3,)
a + b        # b stretches along rows → each row + b
```

- a (2,3) + b (3,) → b broadcast to (2,3)
- Trailing dimensions align; dims of size 1 stretch
- Rule 2: dims either equal or one of them is 1
- Rules summary:
  1. align trailing dims
  2. minus-minus: equal | 1 stretches | missing treated 1
  3. result shape = max dims
- Examples:
  - (3,4) + (4,) ✓
  - (3,4) + (1,4) ✓
  - (3,1)+(1,4) → (3,4) ✓
  - (3,4)+(4,3) ✗ mismatch
  - (4,4)+(2,2) ✗
- Scalar broadcast: (2,3)+5 → all
- Concept: no explicit loops — implicit replication
- Widely used w/ mean subtract, scaling, neural nets bias

````python
# diff shape
a = np.arange(6).reshape(2,3)
b = np.arange(3).reshape(1,3)

print(a)
print(b)

print(a+b)
````

#### Broadcasting Rules

**1. Make the two arrays have the same number of dimensions.**<br>
- If the numbers of dimensions of the two arrays are different, add new dimensions with size 1 to the head of the array with the smaller dimension.<br>

**2. Make each dimension of the two arrays the same size.**<br>
- If the sizes of each dimension of the two arrays do not match, dimensions with size 1 are stretched to the size of the other array.
- If there is a dimension whose size is not 1 in either of the two arrays, it cannot be broadcasted, and an error is raised.

<img src = "https://jakevdp.github.io/PythonDataScienceHandbook/figures/02.05-broadcasting.png">

## Broadcasting rules quick reference

```python
a = np.arange(12).reshape(4, 3)   # (4,3)
b = np.arange(3)                  # (3,) → (1,3) → (4,3)
a + b

a = np.arange(12).reshape(3, 4)   # (3,4)
b = np.arange(3)                  # (3,) → can't (4 vs 3) → ERROR
```

2 rules:
1. Right-align the shapes; every dim must be `1` or equal
2. Dims of size `1` get stretched

| Shape | Compatible? |
|---|---|
| (4,3) + (3,) | ✅ |
| (3,4) + (3,) | ❌ |
| (1,3) + (3,1) | ✅ (→ (3,3)) |
| (4,4) + (2,2) | ❌ |

### Working with missing values

## Missing values — np.nan

```python
a = np.array([1, 2, 3, 4, np.nan, 6])
a[~np.isnan(a)]     # keep only the valid (non-NaN) values
```

- `np.nan` (Not a Number) is NumPy's marker for missing data in float arrays.
- `np.isnan(a)` builds a mask that is True exactly where values are missing.
- Negating with `~` flips the mask, and `a[mask]` then yields a clean array without the missing entries.
- This is the standard data-cleaning step you will run before feeding an array into a model.

## 🔬 Deep Dive : Missing values — np.nan

```python
a = np.array([1,2,3,4,np.nan,5,6])
a[~np.isnan(a)]      # drop nan → array([1,2,3,4,5,6])
```

- **np.nan** — Not a Number (missing marker)
- nan propagates: any op with nan → nan (careful)
- Detect: `np.isnan(a)` → bool mask
- Remove/filter: `a[~np.isnan(a)]`
- Functions with nan: np.nanmean, np.nansum, etc.
- nan replaces None in numeric arrays (float only)
- int arrays can't hold nan (dtype object needed)
- Fill: np.nan_to_num(x, nan=0)
- DataFrame uses NaN too (pandas) — cornerstone
- Check any: `np.isnan(a).any()`

````python
# Working with missing values -> np.nan
a = np.array([1,2,3,4,np.nan,6])
a
````

### Meshgrids

## Meshgrid — the formula for building 2D grids

```python
X, Y = np.meshgrid(x, y)
```

- `meshgrid` — builds a coordinate grid from 2 vectors
- X has repeating rows, Y has repeating columns
- The base for 3D surface plots / contour plots
- `plt.contour`, `ax.plot_surface` work on this grid


## Deep Dive : Meshgrid — 2D grid

```python
X, Y = np.meshgrid(np.arange(0,10), np.arange(0,10))
# X: x repeated rows | Y: y repeated cols
# grid coordinates for 2D surface plots
```

- meshgrid — 1D x, y → 2D coordinate grids
- X[i,j] = x values repeated rows
- Y[i,j] = y values repeated columns
- Mesh = every (xi, yi) pair — grid of coordinates
- Use: 2D functions `Z = f(X,Y)` surfaces, contour, contourf
- Also: vectorized eval on all grid points
- Interpolation, heatmap plotting
- pair with plt.contourf/plot_surface
- Evaluate formula at every grid cell without X-loop
- Numpy grid = scientific computation essential
