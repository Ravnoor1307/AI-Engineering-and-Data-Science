## 1. Setup — Loading the Catalogue

The setup reuses the catalogue loader and builds `name2price`, a dictionary mapping each product name to its price. Every later task that needs a price reads from this dictionary.

The loader tries two file paths and raises `FileNotFoundError` if neither works. With the lookup table ready, the tasks can focus purely on signatures and unpacking rather than file handling.

```python
# ⚙️ SETUP — dono datasets ek saath!
import csv
from functools import reduce

def _load(fname, sublist):
    for path in [f"../../00_Resources/datasets/{fname}", f"00_Python/00_Resources/datasets/{fname}"]:
        try:
            with open(path, newline="", encoding="utf-8") as f:
                return list(csv.DictReader(f))
        except FileNotFoundError:
            continue
    raise FileNotFoundError(f"{fname} nahi mila!")

students = _load("students_marks.csv", None)
products = _load("products_catalog.csv", None)
SUBS = ["math", "science", "english", "hindi", "computer"]
names  = [s["name"] for s in students]
avgs   = [sum(int(s[c]) for c in SUBS) / 5 for s in students]
pnames = [p["product_name"] for p in products]
prices = [float(p["price"]) for p in products]
stocks = [int(p["stock"]) for p in products]
print(f"Ready! {len(students)} students + {len(products)} products — machines ON 🏭")
# Expected OUTPUT:
# Ready! 50 students + 30 products — machines ON 🏭
```

```python
# ✅ SOLUTION — Task 7 (Report Card BOSS 🎓)
print("  🎓 CLASS OF 2024 — REPORT CARD")
print("  " + "─" * 34)

# 1) Leaderboard top-5 (zip + sorted + enumerate!)
board = sorted(zip(avgs, names), reverse=True)
for rank, (avg, nm) in enumerate(board[:5], start=1):
    print(f"  #{rank} │ {nm:<12} │ avg {avg:>5.2f}")
print("  " + "─" * 34)

# 2) Distinction club (≥80) — filter! (zip(avgs, names) — na[0]=avg, na[1]=name!)
dist = sorted(filter(lambda na: na[0] >= 80, zip(avgs, names)), reverse=True)
print("  🎖️ Distinction:", [f"{n} ({a:.1f})" for a, n in dist])

# 3) Class stats
class_avg = reduce(lambda a, b: a + b, avgs) / len(avgs)
above = sum(1 for a in avgs if a > class_avg)
print(f"  📊 class avg : {class_avg:.2f} | above-avg : {above}/{len(avgs)} students")

# 4) Subject topper (computer) — max with key!
cs = [int(s["computer"]) for s in students]
topper = max(zip(cs, names))
print(f"  💻 Computer topper: {topper[1]} ({topper[0]}/100)")

# 5) Sign-off
print("  — Card generated via 4-machines assembly 🏭")
# Expected OUTPUT:
#   🎓 CLASS OF 2024 — REPORT CARD
#   ──────────────────────────────────
#   #1 │ Manish_29    │ avg 83.60
#   #2 │ Ananya_20    │ avg  80.20
#   #3 │ Arjun_11     │ avg  79.40
#   #4 │ Sneha_24     │ avg  78.60
#   #5 │ Priya_44     │ avg  78.00
#   ──────────────────────────────────
#   🎖️ Distinction: ['Manish_29 (83.6)', 'Ananya_20 (80.2)']
#   📊 class avg : 65.69 | above-avg : 27/50 students
#   💻 Computer topper: Kavya_22 (100/100)
#   — Card generated via 4-machines assembly 🏭
```

## Summary

- Positional arguments match by order, keyword arguments by name, and duplicates raise `TypeError`.
- `*args` packs any number of positionals into a tuple, including the empty tuple.
- `**kwargs` packs named extras into a dictionary and pairs with `.get` for safe defaults.
- Call-site stars do the opposite: `f(*list)` and `f(**dict)` expand stored data into arguments.
- A bare `*` creates a keyword-only zone that rejects positional calls.
- The boss task combined all of these into a smart bill that handles discounts, GST, member bonuses, unknown flags, and invalid items.
