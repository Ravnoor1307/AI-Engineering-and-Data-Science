## Arithmetic Operators

**What it is:** the normal math operators — `+` (add), `-` (subtract), `*` (multiply), `/` (true divide → float), `//` (floor divide → integer), `%` (remainder/modulus), `**` (power).

```python
print(5 + 6)     # 11   addition
print(10 - 3)    # 7    subtraction
print(5 / 2)     # 2.5  true division
print(5 // 2)    # 2    floor division
print(5 % 2)     # 1    remainder
print(5 ** 2)    # 25   power
```

**Why it matters:** nearly every numeric pipeline — scores, deltas, normalization, splittling digits — is built on these five symbols plus power.

**Run the code cell below** and match each line of output to its operator.

````python
# Arithmetric Operators
print(5+6)

print(5-6)

print(5*6)

print(5/2)

print(5//2)

print(5%2)

print(5**2)
````

## Relational (Comparison) Operators

**What it is:** operators that compare two values and give a **boolean** (`True` / `False`): `>`, `<`, `>=`, `<=`, `==` (equal), `!=` (not equal).

```python
print(4 > 5)     # False
print(4 < 5)     # True
print(4 >= 4)    # True
print(4 <= 4)    # True
print(4 == '4')  # False  — no automatic type conversion!
print(4 != 5)    # True
```

**Do not confuse `=` with `==`:** `=` assigns a value; `==` asks 'are these equal?'.

**Why it matters:** comparisons are the engine of every condition — thresholds, filters, and anomaly checks all reduce to `True`/`False`.

**Run the code cell below** and read each output against its predicate.

````python
# Relational Operators
print(4>5)

print(4<5)

print(4>=4)

print(4<=4)

print(4==4)

print(4!=4)
````

## Logical Operators

**What it is:** combine truth values: `and`, `or`, `not`. They evaluate operands and **return the operand value** (short-circuit), not always a bool.

```python
print(1 and 0)    # 0      — first falsy wins
print(1 or 0)     # 1      — first truthy wins
print(not 1)      # False  — always a bool
```

**Truthiness:** falsy = `0, 0.0, '', [], {}, None, False`; everything else is truthy.

**Why it matters:** compound business rules ('category is X AND price > 50') are written with logical operators.

**Run the code cell below** — note that `and`/`or` print operand values.

````python
# Logical Operators
print(1 and 0)

print(1 or 0)

print(not 1)
````

## Bitwise Operators

**What it is:** operators that work on the **binary digits** of integers: `&` (AND), `|` (OR), `^` (XOR), `~` (NOT), `<<` (left shift), `>>` (right shift).

```python
print(2 & 3)     # 2   bits: 10 & 11 = 10
print(2 | 3)     # 3   bits: 10 | 11 = 11
print(2 ^ 3)     # 1   bits: 10 ^ 11 = 01
print(~3)        # -4  flips bits (two's complement)
print(4 >> 2)    # 1   shift right = divide by 2^n
print(5 << 2)    # 20  shift left  = multiply by 2^n
```

**Why it matters:** flags, permissions, and bit-mask tricks use these; they are also the fastest possible math on integers.

**Run the code cell below** and verify each binary result.

````python
# Bitwise Operators

# bitwise and
print(2 & 3)

# bitwise or
print(2 | 3)

# bitwise xor
print(2 ^ 3)

print(~3)

print(4 >> 2)

print(5 << 2)
````

## Assignment Operators

**What it is:** `=` stores a value; compound forms (`+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=`, `&=`, `|=`, `^=`, `>>=`, `<<=`) combine an operation with assignment.

```python
a = 2          # assignment
a = a + 1      # becomes 3
a += 1         # ALSO 3 — shorthand
a %= 2         # a = a % 2 → 3 % 2 = 1
```

**Important:** Python has **no** `a++` / `++a` — use `a += 1`.

**Run the code cell below** — the commented example shows `a %= 2` rewriting the variable in place.

````python
# Assignment Operators

# =
# a = 2

a = 2

# a = a % 2
a %= 2

# a++ ++a

print(a)
````

## Membership Operators

**What it is:** `in` / `not in` check whether a value exists inside a container (string, list, tuple, set, dict keys).

```python

print(1 in [2, 3, 4])       # False
print('x' in 'mumbai')      # False
```

**Details:** string membership is case-sensitive and checks substrings; dict membership checks **keys**.

**Run the code cell below** to watch `not in` and `in` in action.

````python
# Membership Operators

# in/not in

print('D' not in 'Delhi')

print(1 in [2,3,4,5,6])
````

## Program — sum of digits of a 3-digit number

**Problem:** read a 3-digit number like `123` and print the sum of its digits (`1+2+3 = 6`).

**The digit-extraction trick:**

```python
number = int(input('enter number'))   # 123
last  = number % 10                   # 3   remainder
rest  = number // 10                  # 12  chop the last digit
```

Repeat `% 10` to peel a digit, then `// 10` to shrink — until the number is 0. This single trick powers palindromes, reversals, and checksums.

**Run the code cell below** to see the full loop.

````python
# Program - Find the sum of a 3 digit number entered by the user

number = int(input('Enter a 3 digit number'))

# 345%10 -> 5
a = number%10

number = number//10

# 34%10 -> 4
b = number % 10

number = number//10
# 3 % 10 -> 3
c = number % 10

print(a + b + c)
````

## if-else — indentation defines the block

**What it is:**

```python
if condition:
    # indentation = block starts here
    # every indented line belongs to the if
else:
    # runs when condition is False
```

**Rules:**
1. Python has no `{}` — indentation (traditionally 4 spaces) groups statements
2. A condition can be any expression that evaluates to True/False
3. `else` is optional; you may chain `elif` for more branches
4. Nested `if` means an `if` inside another `if` body

**Why it matters:** a login check — email matches AND password matches — is a nested-if flow with distinct messages for each failure.

**Run the code cell below** (login program).

````python
# login program and indentation
# email -> nitish.campusx@gmail.com
# password -> 1234

email = input('enter email')
password = input('enter password')

if email == 'nitish.campusx@gmail.com' and password == '1234':
  print('Welcome')
elif email == 'nitish.campusx@gmail.com' and password != '1234':
  # tell the user
  print('Incorrect password')
  password = input('enter password again')
  if password == '1234':
    print('Welcome,finally!')
  else:
    print('beta tumse na ho paayega!')
else:
  print('Not correct')
````

## Loops in Python

## While loop — repeat while true

```python
i = 1
while i < 11:                 # runs 10 times
    print(number, 'x', i, '=', number * i)
    i += 1                    # MUST update the counter!
```

**Wait — is `while i < 11` enough?** No: the loop checks the condition, runs the body, then re-checks. Forgetting `i += 1` = infinite loop.

**Uses in this session:** printing a multiplication table and summing the digits of a number (peel `%10`, chop `//10`).

**Run the code cells below** (table + digit sum).

````python
# While loop example -> program to print the table
# Program -> Sum of all digits of a given number
# Program -> keep accepting numbers from users till he/she enters a 0 and then find the avg
````

````python
number = int(input('enter the number'))

i = 1

while i<11:
  print(number,'*',i,'=',number * i)
  i += 1
````

## while loop with `else`

A `while` loop may carry an `else` block that runs **once after the loop finishes normally** (the condition turned False — no `break`).

```python
x = 1
while x < 3:
    print(x)
    x += 1
else:
    print('loop ended')       # prints AFTER the loop
```

**Gotcha:** if a `break` exits the loop, the `else` is **skipped**.

**Run the code cell below** and notice the order of printed lines.

````python
# while loop with else

x = 1

while x < 3:
  print(x)
  x += 1

else:
  print('limit crossed')
````

## Guessing game — while in action

```python

target = random.randint(1, 100)   # secret number
guess = 0
```

**Why while and not for?** We do not know how many tries the user needs — the loop must keep going **until** a condition is met.

### Program — population problem

The current population of a town is **10000** and it grows 10% every year. Print the population at the end of each of the last 10 years.

## Population program

**Solution trick:** growth of 10% per year means *next year = current × 1.1*. To go **backwards** in time, undo that multiplication:

```python
curr_pop = 10000
for i in range(10, 0, -1):
    print(i, curr_pop)
    curr_pop /= 1.1          # reverse of *1.1
```

**Why `/= 1.1` and NOT `-= 10%`?** A 10% decrease is not the inverse of a 10% increase — percentages multiply; subtracting drifts further from the true value every year.

**Run the code cells below** — the first is the session code, the second the expected answer, and a short explanation follows.

````python
# Code Used in the session
curr_pop = 10000

for i in range(10,0,-1):
  print(i,curr_pop)
  curr_pop = curr_pop - 0.1*curr_pop
````

````python
# Correct Answer of above question:
curr_pop = 10000

for i in range(10,0,-1):
  print(i,curr_pop)
  curr_pop /= 1.1
````

#### Explanation

Each year the population = previous year × 1.1 (100% + 10%). Walking back 10 years, the population is divided by 1.1 at every step. Starting from 10000, the loop prints the population for years 10 down to 1. This 'reverse the multiplier' idea also shows up in compound interest and financial projections.

## Sequence sum

**What each term looks like:** term `i` = `i / i!`.

```python

for i in range(1, n+1):

```

**Performance trick:** do not recompute `i!` from scratch each time — update `fact *= i` (the previous factorial times `i`).

## For loop vs While

| For | While |
|---|---|
| when you know the count / iterable | when a condition decides how long |
| iterate a list/tuple/set/dict/str/range | keep running until something changes |
| impossible to forget the counter | you must update the counter yourself |
| e.g. `for x in data:` | e.g. `while guess != target:` |

**Rule of thumb:** if you can say 'do this for each item' → `for`. If you can only say 'do this while ...' → `while`.
