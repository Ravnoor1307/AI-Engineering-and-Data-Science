### Warm-up — The Population Problem

**Problem:** a town's population is 10000 and grows 10% per year. Print the population at the end of each of the last 10 years.

**The trick:** +10% per year means next year = current × 1.1. Going back in time must **undo** that growth:

```python
previous = current / 1.1
```

Why not subtract 10%? Percentages multiply, they do not add — subtracting accumulates error.

**Run the cell below** — it starts at 10000 and walks 10 years backwards, dividing by 1.1 each step.

````python
# Code here
curr_pop = 10000

for i in range(10,0,-1):
  print(i,curr_pop)
  curr_pop = curr_pop/1.1
````

### Warm-up — Sequence sum 1/1! + 2/2! + 3/3! + ...

**Problem:** compute the sum `1/1! + 2/2! + 3/3! + ...` up to `n` terms, where `!` is factorial (`4! = 4×3×2×1 = 24`).

**Efficient trick:** reuse the factorial instead of recomputing it — update `fact *= i` on every step, and accumulate `result += i / fact` inside a `range(1, n+1)` loop.

**Run the cell below** and try different values of `n`.

````python
# Code here

n = int(input('enter n'))

result = 0
fact = 1

for i in range(1,n+1):
  fact = fact * i
  result = result + i/fact

print(result)


````

## 1. Creating Strings

**What it is:** any text between quotes is a string. Python accepts single (`'...'`), double (`"..."`), and triple quotes (`'''...'''`), which is handy when the string itself contains quotes.

**How it works:**
- `'hello'`, `"hello"`, `'''hello'''` and `"""hello"""` are all equal
- triple quotes keep line breaks — perfect for multi-line text
- `'it\'s raining'` — backslash escapes the inner quote
- In a string like `"it's raining outside"`, the double quotes let you keep the apostrophe without escaping

**Rules:** quotes around a string only matter at the start and end — what matters inside is content. There is no 'correct' quote style; choose whichever avoids escaping.

**Run the cells below** — one creates strings every way, the other solves the apostrophe problem.

````python
s = 'hello'
s = "hello"
# multiline strings
s = '''hello'''
s = """hello"""
s = str('hello')
print(s)
````

# 2. Indexing & Slicing


## 2. Indexing & Slicing

**Indexing** — grab ONE character at a position. Python indexes from **0** from the left (`s[0]` is the first char) and from **-1** from the right (`s[-1]` is the last char).

```python
s = 'hello world'
s[0]   # 'h'
s[-3]  # 'r'  (three from the end)
```

**Slicing** — grab a SUBSTRING. The general form is
`s[start : stop : step]` — start included, **stop excluded**, step decides direction and stride.

```python
s[6:0:-2]   # start at index 6, go down to (not including) 0, step -2
s[::-1]     # entire string REVERSED
s[-1:-6:-1] # from the last char, backwards to the 6th-from-end
```

**Error:** an index outside the string raises `IndexError`. Slices never do — they just shrink.

**Run the cells below** — positive indexing, negative indexing, and three slicing recipes.

````python
# Positive Indexing
s = 'hello world'
print(s[41])
````

````python
# Negative Indexing
s = 'hello world'
print(s[-3])
````

````python
# Slicing
s = 'hello world'
print(s[6:0:-2])
````

````python
print(s[::-1])
````

````python
s = 'hello world'
print(s[-1:-6:-1])
````

# 3. Immutability & Editing


## 3. Strings are IMMUTABLE

**Immutability** means a string's content can never change once created. Python strings have no 'character slots' you can rewrite.

**What happens if you try?**

```python
s = 'hello world'
s[0] = 'H'   # TypeError: 'str' object does not support item assignment
```

**What you CAN do:** reassign the whole variable (`s = 'Hello'`) or build a new string from parts. `del s` deletes the variable itself, not 'parts' of it — deleting a slice (`del s[-1:-5:2]`) is also a `TypeError`, because strings do not support item or slice deletion.

**Why immutable?** Strings are hashed and shared everywhere (dict keys, interned values) — immutability makes that safe and fast.

**Run the cells below.** The first deliberately errors to show the rule; the others show deletion attempts.

````python
s = 'hello world'
s[0] = 'H'

# Python strings are immutable
````

````python
s = 'hello world'
del s
print(s)
````

````python
s = 'hello world'
del s[-1:-5:2]
print(s)
````

## 4. Operations on Strings

**Arithmetic:**
- `'delhi' + ' mumbai'` → concatenate (join) with `+`
- `'delhi' * 5` → repeat the string 5 times
- `'*' * 50` → a 50-star divider line (classic CLI trick)

**Relational** — strings compare **lexicographically** (dictionary order, by ASCII code):
- `'mumbai' > 'pune'` → False ('m' < 'p')
- `'Pune' > 'pune'` → False (uppercase letters come before lowercase in ASCII)

**Logical (truthiness):**
- `'hello' and 'world'` → 'world' (and returns the last truthy operand)
- `'hello' or 'world'` → 'hello' (or returns the first truthy operand)
- `'' and 'world'` → '' (empty string is **falsy**)
- `not 'hello'` → False

**Iteration & membership:**
- `for i in 'hello':` walks the string char by char
- `'D' in 'delhi'` → membership check

# 5. Common Functions


## 5. Common Functions — len / max / min / sorted

Built-in functions that take a string (or any iterable):

- `len('hello world')` → number of characters, INCLUDING spaces: 11
- `max('hello world')` → largest character by ASCII code: `'w'` (lowercase wins over most things)
- `min('hello world')` → smallest character: a **space** (ASCII 32 is smaller than letters)
- `sorted('hello world', reverse=True)` → returns a LIST of single characters sorted descending, e.g. `['w','r','o','o','l','l','h','e',' ','d']`

**Why sorted matters:** it is the backbone of frequency analysis and reorderable tokens in text pipelines.

**Note:** strings have no 'digits as numbers' here — everything compares by raw character code.

# 6. Case Methods


## 6. Case Methods — capitalize / title / upper / lower / swapcase

All four return a NEW string — the original is untouched (immutability).

- `s.capitalize()` → first char uppercase, rest lower: `'Hello world'`
- `s.title()` → first letter of every word uppercase: `'Hello World'`
- `s.upper()` → everything UPPERCASE
- `'Hello World'.lower()` → everything lowercase
- `'HeLlO WorLD'.swapcase()` → swap each char's case: `'hElLo wORld'`

**ML connection:** `lower()` is the single most-used step in text cleaning — model inputs must be case-normalized or 'Hello' and 'hello' look like different words.

**Remember:** `print(s.capitalize()); print(s)` shows the original still unchanged afterwards.

**Run the cells below** — each tests one method.

````python
s = 'hello world'
print(s.capitalize())
print(s)
````

# 7. Search


## 7. Search: count / find / index

- `s.count(sub)` → how many times a substring appears: `'my name is nitish'.count('i')` → 3
- `s.find(sub)` → index of the FIRST occurrence, or **-1** if missing (safe): `'...'.find('x')` → -1
- `s.index(sub)` → same as find, but raises `ValueError` if missing

**Rule of thumb:** use `find()` when you do not know if the pattern is present; use `index()` when its presence is guaranteed and a missing one should be treated as a bug.

**Run the cells below.** Notice the contrast: find returns -1 for an absent letter, index crashes with an error.

````python
'my name is nitish'.count('i')
````

````python
'my name is nitish'.find('x')
````

````python
'my name is nitish'.index('x')
````

# 8. endswith / startswith


## 8. endswith / startswith

- `s.endswith(sub)` → True if `s` ENDS with `sub`, else False
- `s.startswith(sub)` → True if `s` STARTS with `sub`, else False

```python
'my name is nitish'.endswith('sho')     # False — ends with 'ish'... actually 'sh'
'my name is nitish'.startswith('1my')   # False — starts with 'my', not '1my'
```

**Why useful:** routing logic — 'is this file a .py?', 'does this URL start with https?', 'is this message an error?'

## 9. format() — placeholders

`'...{0}...{1}...'.format(v0, v1)` — `{0}` gets replaced by the first argument, `{1}` by the second, in order of the numbers inside the braces. You may also use named fields `{name}` with keyword arguments.

```python
name = 'nitish'
gender = 'male'
'Hi my name is {1} and I am a {0}'.format(name, gender)
# -> 'Hi my name is male and I am a nitish'   (note the SWAP!)
```

**The trick:** the `{0}`/`{1}` numbers refer to argument position, so you can reorder, repeat, and reuse values freely.

**Modern alternative:** f-strings (`f'Hi {name}'`) are preferred in new code, but `format()` still appears in older codebases and templates.

**Run the cell below** to see the placeholder swap.

````python
name = 'nitish'
gender = 'male'

'Hi my name is {1} and I am a {0}'.format(gender,name)
````

# 11. split / join


## 11. split / join

- `s.split(sep)` → break a string at `sep` (default: any whitespace) → a LIST of words: `'hi my name is nitish'.split()` → `['hi','my','name','is','nitish']`
- `sep.join(list)` → glue a list of strings into ONE string: `' '.join(['hi', 'my', 'name', 'is', 'nitish'])` → `'hi my name is nitish'`

Together they are a **text parser's best friends**: split to tokens for NLP, join to rebuild output lines.

**Common mistake:** using `.split()` on an empty or single-whitespace string — results can surprise. Always verify the list length.

**Run the cells below** — one splits, one joins.

````python
'hi my name is nitish'.split()
````

````python
" ".join(['hi', 'my', 'name', 'is', 'nitish'])
````

# 13. strip


## 13. strip

`s.strip()` removes leading AND trailing whitespace, tabs, and newlines. It is the first line of defense for messy input.

- `s.strip()` — both ends
- `s.lstrip()` — leading (left) only
- `s.rstrip()` — trailing (right) only


**Real-world relevance:** CSV and log data always carries stray spaces — strip before comparing, splitting, or hashing.

**Program 1 — length without len()**

Loop over the string with a counter: for every character, add 1. Emulates `len()` manually.

**Program 2 — username from email**

Everything before the `@` is the username: `'nitish24sacademy@gmail.com'` → `nitish24sacademy`.

**Program 3 — character frequency**

Count how many times a specific character appears.

**Ideas:** `s.count(ch)`, or walk the string with a manual counter, or build a frequency dict.

**Cell below:** count the frequency of a given character in a string.

````python
# Count the frequency of a particular character in a provided string. 
# Eg 'hello how are you' is the string, the frequency of h in this string is 2.

s = input('enter the email')
term = input('what would like to search for')

counter = 0
for i in s:
  if i == term:
    counter += 1

print('frequency',counter)
````

**Program 4 — remove a character**

Delete every occurrence of a given character from a string.

**Idea:** `s.replace(ch, '')` replaces the char with nothing — clean and one line.

**Cell below:** remove a particular character from a string.

````python
# Write a program which can remove a particular character from a string.
s = input('enter the string')
term = input('what would like to remove')

result = ''

for i in s:
  if i != term:
    result = result + i

print(result)
````

**Program 5 — palindrome check**

A word reads the same forwards and backwards (`'madam'`).

**Idea:** compare `s == s[::-1]` (reverse slice), ignoring case/spaces for a stronger check.

**Cell below:** check whether a string is a palindrome.

````python
# Write a program that can check whether a given string is palindrome or not.
# abba
# malayalam

s = input('enter the string')
flag = True
for i in range(0,len(s)//2):
  if s[i] != s[len(s) - i -1]:
    flag = False
    print('Not a Palindrome')
    break

if flag:
  print('Palindrome')

````

**Program 6 — word count without split()**

Count words by counting spaces: roughly `words = spaces + 1`.

**Idea:** loop and count every space; handle edge cases like empty or double-spaced strings carefully.

**Cell below:** count words in a string WITHOUT `split()`.

````python
# Write a program to count the number of words in a string without split()

s = input('enter the string')
L = []
temp = ''
for i in s:

  if i != ' ':
    temp = temp + i
  else:
    L.append(temp)
    temp = ''

L.append(temp)
print(L)

````

**Program 7 — manual title case**

Uppercase the first letter of every word without using `title()`.

**Idea:** split into words, uppercase the first char of each, join back — or track spaces while walking the string.

**Cell below:** convert a string to title case manually.

````python
# Write a python program to convert a string to title case without using the title()
s = input('enter the string')

L = []
for i in s.split():
  L.append(i[0].upper() + i[1:].lower())

print(" ".join(L))
````

**Program 8 — int to string without str()**

Convert an integer into its string form without calling `str()`.

**Idea:** repeatedly `% 10` to peel digits and `// 10` to shrink the number, then build the string from the digit characters.

**Cell below:** convert an integer to a string.

````python
# Write a program that can convert an integer to string.

number = int(input('enter the number'))

digits = '0123456789'
result = ''
while number != 0:
  result = digits[number % 10] + result
  number = number//10

print(result)
print(type(result))
````
