## Printing with `sep` and `end`

The first question tests something every Python user needs: total control over how `print()` formats its output.

When you pass several values to `print`, the default behaviour is to separate them with a single space and finish each call with a newline:

```python
print('Data', 'Science')        # Data Science
```

Two optional parameters give you precise control:

- **`sep`** controls the separator placed *between* the values. `print('a', 'b', sep='-')` prints `a-b`.
- **`end`** controls the text printed *after* all the values. Normally it is a newline, but `print('a', end='')` leaves the cursor on the same line so the next `print` continues there.

Read the required output carefully:

```
Data-Science-Mentorship-Program-started-By-CampusX
```

Notice the words are joined with a single hyphen, and the word `started` appears in the middle even though it is not one of the six given strings. So the print call must receive the extra literal `'started'` in the right position, and use `sep='-'`.

Study the pattern before coding: the desired output is exactly six or seven values joined by hyphens. Identify values 1 to 6, figure out where `started` goes, then build the `print` statement with the correct `sep`. You do not need `end` here — the default newline is fine.

##  Celsius to Fahrenheit conversion

Temperature conversion is a classic first program because it mixes three core ideas: input, formula, and output.

The conversion rule you need is:

```
F = C * 9/5 + 32
```

where `C` is the temperature in Celsius and `F` the result in Fahrenheit. Take a moment to decode the formula: each Celsius degree is `9/5` (1.8) Fahrenheit degrees, and the scales are offset by 32.

To solve this in Python you need three steps, and each maps to a concept from the session:

1. **Take input.** `input(...)` always returns a **string**. You cannot multiply a string, so convert it first with `int()` or `float()`. Since temperatures can be fractional (`37.5`), `float()` is the safer choice.
2. **Compute.** Apply the formula with Python arithmetic. Watch the brackets — `C * 9/5 + 32` follows operator precedence naturally (`*` and `/` before `+`), but writing `C * (9/5) + 32` makes the intent explicit.
3. **Print the result.** `print()` can output a label and the number together.

The most common mistakes here are (a) forgetting the conversion and trying arithmetic on a string, which raises a `TypeError`, and (b) mixing up the direction of the formula.

Test your program with known values: `0` should give `32.0`, and `100` should give `212.0`. If both match, your formula is correct. This pattern — convert input, compute, print — is the skeleton of virtually every numeric program you will write, so getting comfortable with `float(input(...))` now will pay back constantly.

### Q3:- Take 2 numbers as input from the user.Write a program to swap the numbers without using any special python syntax.

##  Swapping two numbers

The task: given two numbers, swap their values using only elementary Python — no tuple unpacking (`a, b = b, a`), no `swap` function.

Think about the problem before typing. If you write `a = b` immediately, `a` has been overwritten and the old value of `a` is lost forever. You need a **temporary third variable** to park a value while you shuffle:

```python
temp = a

a = b

b = temp
```

Trace it: `temp` grabs the original `a`. Now `a` can safely take `b`'s value. Finally `b` takes the preserved original of `a` from `temp`. After the three lines, the two variables have traded places and nothing was destroyed.

Why does this matter beyond textbooks? Swapping is one of the most common operations in all of programming — sorting algorithms swap out-of-order neighbours, and many data structures need to exchange two values while preserving both. Learning the temp-variable dance teaches you to think about *order of operations*. If you move the lines the wrong way, data is lost and bugs appear.

For this question you also need `input()` again: read both numbers (converting with `int`), swap them, then print them in their new order so you can confirm the swap happened.

Verify with a mental test: `a = 5`, `b = 9`. After swapping, printing `a` should give 9 and printing `b` should give 5. Trace through your code once on paper with these values — that one careful trace is worth ten runs.

### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

##  Euclidean distance between two coordinates

This question connects arithmetic with geometry. Given two points `(x1, y1)` and `(x2, y2)`, you want the straight-line distance between them.

The formula comes from the Pythagorean theorem — the horizontal gap and the vertical gap form the two legs of a right triangle, and the distance is the hypotenuse:

```
distance = sqrt((x2 - x1)**2 + (y2 - y1)**2)
```

In Python, you can compute a square root with `** 0.5`, since raising to the power `0.5` is exactly a square root:

```
distance = ((x2 - x1)**2 + (y2 - y1)**2) ** 0.5
```

Now think about the input. Each coordinate is a whole `(x, y)` pair. You can read the four values separately — `x1 = float(input('enter x1'))` and so on — converting each with `float` so fractional coordinates are accepted. The goal is simply to end up with four numbers `x1`, `y1`, `x2`, `y2`.

Then the computation is pure arithmetic: subtract, square each difference with `** 2`, add the two squares, and take the square root with `** 0.5`. Finally print the distance.

Test it: the distance between `(0, 0)` and `(3, 4)` is exactly `5.0`. Why? Because `3**2 + 4**2 = 9 + 16 = 25`, and the square root of 25 is 5. Running that test tells you immediately whether your formula and inputs are right. Euclidean distance is everywhere — navigation, clustering algorithms, and image processing all use it — so this small problem generalises far beyond the classroom.


- `P` = principle amount
- `R` = rate of interest in percent
- `T` = time period (typically in years)

The division by 100 exists because the rate is given as a percentage. If the rate is `10` for 10%, then `R/100 = 0.1`. Thinking of it as `0.1` per year is the cleanest mental model.

As before, the coding pattern is the same trio:

3. `print()` the result.

## Q6. Dogs and chickens puzzle

This is a delightful word problem hiding a set of two linear equations. You know two facts:

- **total heads** = number of dogs + number of chickens
- **total legs** = 4 × dogs + 2 × chickens

Translating to algebra with `d` for dogs and `c` for chickens:

```
d + c = heads

4*d + 2*c = legs
```

You could solve these by hand every time, but the better programmer move is to solve them once, generally. From the first equation, `c = heads - d`. Substitute that into the second:

```
4*d + 2*(heads - d) = legs
4*d + 2*heads - 2*d = legs
2*d = legs - 2*heads
d = (legs - 2*heads) // 2
c = heads - d
```

That `//` is integer (floor) division: counts must be whole animals, so you want exact integers. Since real data always works out proportionally, integer division is the faithful choice.

Test against the given example: heads = 4, legs = 12.

```
d = (12 - 2*4) // 2 = (12 - 8) // 2 = 2
dogs = 2
chickens = 4 - 2 = 2
```

Your program should read `heads` and `legs` from the user (as integers — `int(input(...))`), apply the two formulas, and print the number of dogs and chickens.

### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

## Q7. Sum of squares of the first n natural numbers

The task: add `1**2 + 2**2 + 3**2 + ... + n**2` for a given `n`. For example, when `n = 3`, the sum is `1 + 4 + 9 = 14`.

There are two completely different ways to solve it, and each teaches something valuable.

**1. The loop way.** Walk from `1` to `n`, square each number, and add it to a running total:

```python
total = 0

for i in range(1, n + 1):
    total = total + i**2

print(total)
```

`range(1, n + 1)` is crucial — `range` stops one before the end, so `n + 1` includes `n` itself. This approach is intuitive and works for any similar sum: cubes, linear terms, anything.

**2. The formula way.** Mathematicians proved a closed form:

```
sum of squares = n*(n + 1)*(2*n + 1) // 6
```

Check it with `n = 3`: `3 * 4 * 7 / 6 = 84 / 6 = 14`. It matches, without a single loop. The formula is instant even for enormous `n`.

Both answer the question; the formula wins on speed, the loop wins on flexibility and insight into what is being computed. For this exercise, implement whichever you understand better, but make sure you can explain *why* it is correct.

Practical details: read `n` with `int(input(...))`. If you use the loop, remember to add the square, not just `i`. If you use the formula, integer division `//` keeps the result a whole number (the product is always divisible by 6). Verify with `n = 5`: `1 + 4 + 9 + 16 + 25 = 55`. Whichever path you take, this question cements the two fundamental approaches to summation that pervade all of programming and statistics.

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

## Q8. Nth term of an arithmetic series

An arithmetic series is a sequence of numbers where the difference between consecutive terms is constant — for example `2, 5, 8, 11, ...` grows by `3` each step. This question gives you the **first two terms** and asks for the **nth term**.

The first term is `a`, and the constant difference is `d`. From just the first two terms you can recover both:

```
d = (second term) - (first term)
```

Once you know `a` and `d`, the nth term has a famous closed form:

```
nth term = a + (n - 1) * d
```

Why `(n - 1)`? The first term is `a + 0*d`. The second is `a + 1*d`. Each term adds one more `d`, so the k-th term is `a + (k-1)*d`. The `(n - 1)` is not a mystery — it is simply one fewer than the position, because the first position contributes no differences at all.

For the example `2, 5, 8, ...`: `a = 2`, `d = 3`. The 4th term is `2 + 3*3 = 11`, which matches the sequence.

Coding it is a small input->compute->print program:

1. Read the first term `a` and the second term; compute `d` by subtraction.
2. Read `n`.
3. Print `a + (n - 1) * d`.

Watch the parentheses — `a + (n - 1) * d` is not the same as `(a + n - 1) * d`. Multiplication binds tighter than addition, so writing the formula exactly as above is enough.

Test with `a = 2`, `d = 3`: term 1 should be 2, term 2 should be 5, term 10 should be `2 + 9*3 = 29`. Arithmetic progressions show up all over — loan repayments, depreciation, and growth estimates — and the `a + (n-1)d` formula is one of the most reused in applied maths.

### Q9:- Given 2 fractions, find the sum of those 2 fractions.Take the numerator and denominator values of the fractions from the user.

## Q9. Sum of two fractions

Adding fractions might feel like school maths, but this question is really about a clean algorithmic rule. Given two fractions `a/b` and `c/d` (numerator and denominator of each), find their sum as a fraction.

The rule is simple and deterministic:

```
numerator   = a*d + c*b

denominator = b*d
```

Why does that work? To add fractions you need a common denominator. Multiplying the two denominators — `b*d` — always gives a valid common denominator. Cross-multiplying each numerator by the *other* denominator then expresses both fractions with that common denominator:

```
a/b = (a*d)/(b*d)

c/d = (c*b)/(d*b)
```

and adding the numerators gives `(a*d + c*b) / (b*d)`. The result may not be in lowest terms (e.g. `3/6` rather than `1/2`), and that is perfectly acceptable here — you are only asked for the sum.

For example, `1/2 + 1/3`: `a = 1, b = 2, c = 1, d = 3`.

```
numerator   = 1*3 + 1*2 = 5

denominator = 2*3 = 6
```

So the sum is `5/6` — correct, since 0.5 + 0.333... = 0.833...

Your program needs four inputs: the numerator and denominator of the first fraction, then of the second. Read them with `int(input(...))`, compute the two values with the formula, and print them nicely — for example `print(numerator, '/', denominator)` gives `5 / 6`.

## Q10. Glasses of milk from a tank

This is a volume problem with a story. A milk tank has given dimensions, and so does a glass. The question asks how many full glasses of milk the tank can hold.

The tank is rectangular: height `H`, length `L`, breadth `B`, so its volume is the product:

```
tank_volume = H * L * B
```

The glass is a cylinder: height `h` and radius `r`, so its volume uses the cylinder formula:

```
glass_volume = 3.14 * r * r * h
```

(pi approximated as 3.14 is fine unless told otherwise.)

Then the number of glasses is how many times the glass volume fits into the tank volume:

```
glasses = tank_volume // glass_volume
```

Use **floor division `//`** rather than ordinary `/` when the question asks for full glasses — you cannot serve a fraction of a glass. If it instead wanted rounded-up servings, you would use an opposite strategy; read the wording to decide.

Work the example you are given:

```
tank: H = 20, L = 20, B = 20   -> volume = 8000

glass: h = 3, r = 1            -> volume = 3.14 * 1 * 1 * 3 = 9.42

8000 // 9.42 = 849
```

So about 849 glasses.

Notes for your code: read all five values with `float(input(...))`; compute the two volumes; then `print(int(tank_volume // glass_volume))`. The `int(...)` simply makes the displayed number tidy (the floor is already integral).

This question trains the two skills that dominate real engineering maths: computing volumes from a formula, and deliberately choosing the right division for the semantics of the question. Predicting the answer by hand before running tells you in one line whether your code is right.
