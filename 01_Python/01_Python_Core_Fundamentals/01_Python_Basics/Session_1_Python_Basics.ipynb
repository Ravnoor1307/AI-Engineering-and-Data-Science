# 1. Python Output


## 1.1 print() — basics, quoting & case sensitivity

**What it is:** `print()` is Python's built-in function that displays values as text on the console.

**Why:** without output you cannot see what a program computed. Debugging, logging, and printing metrics in ML all start with `print()`.

**How it works:** `print(value1, value2, ..., sep=' ', end='\n')`. Every value is converted with an internal `str()` before being written. Strings must be written in quotes; without quotes `print(Ravnoor Singh)` raises a SyntaxError/NameError. Python is **case sensitive** — `print` works, `Print` does not.

**Key parameters:**
- `sep` — the character placed between values (default: space)
- `end` — the character placed after the last value (default: `\n`)

**Small example:** the code prints `'Hello World'` first, then intentionally runs `print(Ravnoor Singh)` without quotes to demonstrate the quoting rule.

````python
# Python is a case sensitive language
print('Hello World')
````

````python
print('Ravnoor Singh')
````

````python
print(Ravnoor Singh)
````

## 1.2 Multiple values & the sep parameter

**What it is:** `print()` can output several values in one call, and `sep` sets the character placed between those values.

**Why:** showing multiple data points together (like all columns of one row) is cleaner in a single call.

**How it works:** `print('Hello', 1, 4.5, True)` joins the values with spaces since the default is `sep=' '`. With `sep='/'` the output becomes `Hello/1/4.5/True`. Types can be mixed freely — Python converts each one to a string.

**Small example:** both calls below use the same values; only `sep` differs, so the output changes. Observe the difference.

````python
print(7)
````

````python
print(7.7)
````

````python
print(True)
````

````python
print('Hello',1,4.5,True)
````

## 1.3 The sep & end pair

`sep` joins multiple values; `end` decides what comes after the last value. Together they fully control the shape of your output line.

````python
print('Hello',1,4.5,True,sep='/')
````

## 1.3 The end parameter & newline behaviour

**What it is:** `end` decides what gets printed after the last value. The default is `\n` (newline), so the next print starts on a new line.

**Why needed:** sometimes we want two prints on the same line, for example when showing progress steps.

**How it works:** `print('hello', end='-')` prints `hello-` without a newline. The following `print('world')` continues on the same line, producing `hello-world`.

**Small example:** the code below shows `print('hello')` + `print('world')` on two lines, then the same with `end='-'` merging them.

````python
print('hello')
print('world')
````

````python
print('hello',end='-')
print('world')
````

# 2. Data Types


## 2.1 Numeric types — int & float

**What it is:** `int` stores whole numbers (positive, negative, zero, of any size), `float` stores decimal numbers like 8.55 or 1.5e2.

**Why:** numbers are the actual data of machine learning — features, weights, and loss values are all numeric.

**How it works:** `print(8)` writes an integer. A huge float literal like `1e309` overflows to `inf` (infinity) instead of crashing, because float is not size-limited like C integers.

**Small example:** the code prints an int (`8`), a decimal (`8.55`) and a huge float (`1e309`); the `inf` output is the key observation.

````python
# Integer
print(8)
# 1*10^308
print(1e309)
````

````python
# Decimal/Float
print(8.55)
print(1.7e309)
````

## 2.2 bool & str — truth values and text

**What it is:** `bool` stores only `True`/`False`; `str` stores any text.

**Why:** conditions (True/False) and textual data (names, labels) are everywhere in ML — classification labels are strings too.

**How it works:** `bool` is actually a subclass of `int` (`True==1`, `False==0`). Strings are written inside quotes and are immutable. Both display as text when printed.

**Key point:** everything in Python is an object, and `type(True)` reveals its class.

**Small example:** below we print `True`, `False`, and a string.

````python
# Boolean
print(True)
print(False)
````

**Why:** complex numbers appear in linear algebra and FFT; lists and dicts are the building blocks of Pandas and NumPy; `type()` is used for debugging and validation.

# 3. Variables


## 3.1 Static vs Dynamic Typing

**What it is:** typing decides when a variable's type is fixed. In static languages (C/C++) you must declare the type at compile time (`int a = 5`); in Python the type is decided at runtime (`a = 5`).

**Why:** static typing catches errors earlier but makes code longer. Python's dynamic typing keeps ML prototyping fast and clean.

**How it works:** `a = 5` creates an integer object and tags it `a`. Writing `int a = 5` raises a SyntaxError because Python does not support that C-style syntax.

**Small example:** the code below runs `a = 5` and `b = 5` fine; the static-style `int a = 5` produces an error.

````python
# Static Vs Dynamic Typing
# Static Vs Dynamic Binding
# stylish declaration techniques
````

````python
# C/C++
name = 'nitish'
print(name)

a = 5
b = 6

print(a + b)
````

````python
# Dynamic Typing
a = 5
# Static Typing
int a = 5
````

## 3.2 Dynamic Binding & assignment styles

**What it is:** binding is how a name attaches to a value. Dynamic binding means a variable can change its type while the program runs.

**Why:** Python allows `a = 5` followed by `a = 'hello'`; C/C++ rejects a type change.

**Assignment styles:**
- separate lines: `a = 1; b = 2; c = 3`
- parallel / tuple unpacking: `a, b, c = 1, 2, 3`
- chained: `a = b = c = 5` (all get 5)

**How it works:** a variable is a label pointing to an object; a new assignment moves the label to a new object.

**Small example:** the code demonstrates dynamic binding (a = 5 becomes a = 'nitish') and prints the results of all three assignment styles.

````python
# Dynamic Binding
a = 5
print(a)
a = 'nitish'
print(a)

# Static Binding
int a = 5
````

````python
a = 1
b = 2
c = 3
print(a,b,c)
````

````python
a,b,c = 1,2,3
print(a,b,c)
````

````python
a=b=c= 5
print(a,b,c)
````

# 4. Keywords & Identifiers


## 5.1 Keywords — reserved words

**What it is:** keywords are Python's **reserved words** (`if`, `for`, `class`, `return`, ...). You cannot use them as variable/identifier names.

**Why:** these words drive the language's syntax, so Python reserves their meaning.

**How it works:** Python has 35 keywords. To list them:


**Small example:** `keyword.kwlist` is run below and prints the full list of reserved words.

## 5.2 Identifiers — naming rules

**What it is:** identifiers are the names you create (variables, functions).

**Rules:**
1. Only letters, digits, and underscore `_` are allowed
2. Cannot start with a digit (`1name` is invalid)
3. Cannot be a keyword
4. Case sensitive (`name` != `Name`)
5. A lone `_` is a valid identifier (used for unused values)

**Conventions (PEP 8):** variables use `snake_case`; classes use `PascalCase`.

**Small example:** the code creates `name1` and `_` as valid identifiers.

````python
# Identifiers
# You can't start with a digit
name1 = 'Nitish'
print(name1)
# You can use special chars -> _
_ = 'ntiish'
print(_)
# identiers can not be keyword
````

# 6. Type Conversion


## 7.1 Implicit Type Conversion

**What it is:** Python converts a type **automatically** whenever it is safe.

**Why:** in `5 + 5.6` an int and a float are mixed, so Python widens the smaller type (int) to the larger one (float).

**How it works:** narrow → wide. `int + float → float`, done silently.

**IMPORTANT:** `4 + '4'` raises a **TypeError** — Python will NOT auto-convert a string to an int.

**Small example:** `5 + 5.6` works (implicit float), but `4 + '4'` errors — the golden rule: int + string never works until you cast.

````python
# Implicit Vs Explicit
print(5+5.6)
print(type(5),type(5.6))

print(4 + '4')
````

## 7.2 Explicit Type Conversion (Casting)

**What it is:** you force the conversion yourself using constructors: `int()`, `float()`, `str()`, `bool()`.

**Why:** input strings need to become numbers, and numbers need to become display-ready strings.

**How it works:**
- `int('1234')` → 1234 (digit strings are fine)
- `float(4)` → 4.0
- `str(5)` → '5'
- `int(4+5j)` → **TypeError** (a complex number cannot become an int)

**Small example:** the code shows str→int, int→str, and float conversions. Study the outputs to see each type change.

````python
# Explicit
# str -> int
#int(4+5j)

# int to str
str(5)

# float
float(4)
````

# 7. Literals


## 8.1 Numeric literals — binary / octal / hex / float

**What it is:** a literal is a value written directly in the source code.

**Numeric prefixes (exam favourite):**
- Binary: `0b1010` → 10
- Octal: `0o310` → 200
- Hex: `0x12c` → 300
- Float: `10.5`; scientific `1.5e2` → 150.0; `1.5e-3` → 0.0015

**Small example:** below, a, b, c, d hold the four number systems and are printed together — Python always displays them in decimal.

````python
a = 0b1010 #Binary Literals
b = 100 #Decimal Literal 
c = 0o310 #Octal Literal
d = 0x12c #Hexadecimal Literal

#Float Literal
float_1 = 10.5 
float_2 = 1.5e2 # 1.5 * 10^2
float_3 = 1.5e-3 # 1.5 * 10^-3

#Complex Literal 
x = 3.14j

print(a, b, c, d)
print(float_1, float_2,float_3)
print(x, x.imag, x.real)
````
