### **`2. What is a dynamically typed language?`**

Before we understand a dynamically typed language, we should learn about what typing is. **Typing** refers to type-checking in programming languages. In a **strongly-typed** language, such as Python, **"1" + 2** will result in a type error since these languages don't allow for "type-coercion" (implicit conversion of data types). On the other hand, a **weakly-typed** language, such as Javascript, will simply output **"12"** as result.

Type-checking can be done at two stages:
- **Static -** Data Types are checked before execution.
- **Dynamic -** Data Types are checked during execution.
Python is an interpreted language, executes each statement line by line and thus type-checking is done on the fly, during execution. Hence, Python is a Dynamically Typed Language.

### **`5. What are the common built-in data types in Python?`**

There are several built-in data types in Python. Although, Python doesn't require data types to be defined explicitly during variable declarations type errors are likely to occur if the knowledge of data types and their compatibility with each other are neglected. Python provides `type()` and `isinstance()` functions to check the type of these variables. These data types can be grouped into the following categories-

1. **None Type:** **None** keywork represents the null values in Python. Boolean equality operation can be performed using these NoneType objects.

2. **Numeric Type:** There are three distinct numeric types - `integers`, `floating-point numbers` and `complex numbers`. Additionally, `booleans` are a sub-type of integers.

3. **Sequence Types:** According to Python Docs, there are three basic Sequence Types - `lists`, `tuples`, and `range` objects. Sequence types have the `in` and `not in` operators defined for their traversing their elements. These operators share the same priority as the comparison operations.

4. **Mapping Types:** A mapping object can map hashable values to random objects in Python. Mappings objects are mutable and there is currently only one standard mapping type, the `dictionary`.

5. **Set Types:** Currently, Python has two built-in set types - `set` and `frozenset`. `set` type is mutable and supports methods like *add()* and *remove()*. `frozenset` type is immutable and can't be modified after creation.

6. **Callable Types:** Callable types are the types to which function call can be applied. They can be **user-defined functions, instance methods, generator functions**, and some other **built-in functions, methods** and **classes**.

## Q6. Operator precedence

Operator precedence decides the order in which Python evaluates the parts of an expression when more than one operator is present. Just as in school arithmetic, `2 + 3 * 4` equals `14`, not `20` — multiplication binds tighter than addition. Python's precedence table is a formal version of that idea, and a few of its rules appear in almost every program you write:

- **`**`** (exponentiation) binds tighter than `*`, `/`, `//`, `%`, and those bind tighter than `+` and `-`.
- **Comparison operators** (`<`, `>`, `==`, `!=`, `in`, `is`) bind looser than arithmetic, so `x + 1 == y - 2` groups as `(x + 1) == (y - 2)`.
- **`not`** binds looser than comparisons, **`and`** looser than `not`, and **`or`** loosest of all — which is why `a or b and c` means `a or (b and c)`.

Two things matter in practice:

1. **Do not memorise the whole table.** Memorise the top three levels above and, for anything more exotic, use **parentheses**. Extra brackets cost nothing and make your intention obvious to the reader — the Zen of Python itself says explicit is better than implicit.
2. **Be wary of mixed `and`/`or` chains** in conditions like `if x < 5 and y > 10 or flag`: many bugs come from assuming left-to-right grouping when precedence actually splits the expression differently.

A quick rule of thumb for the common tiers, weakest to strongest: `or` → `and` → `not` → comparisons → `+`/`-` → `*`/`/`/`%` → `**`. Write a few expressions yourself and predict the result before running them — precedence is one of those topics that feels trivial until you meet it inside a messy boolean condition.

## Q7. The ternary operator

Python has no `?:` from C++, but it ships the equivalent as a conditional expression:

```
[on true] if [expression] else [on false]
```

Read it in three parts: evaluate the **expression**; if it is `True`, the whole expression evaluates to the value written **before** `if`; otherwise it evaluates to the value after **else**. Exactly one of the two branches runs, and the thing it produces can be assigned, printed, or returned.

The canonical example chooses a minimum:

```
a, b = 2, 3
minimum = a if a < b else b
```

You can jump ahead to the demonstration cell below — it runs `print("Hi") if a < b else print("Bye")`, so when `a` is not less than `b`, the string `Bye` is printed.

Three rules keep ternaries clean:

- **The two outcomes are expressions, not statements.** You cannot write `x = ...` inside a branch — only an expression like `a`, `print(...)`, or a function call.
- **Keep them short.** A ternary is a readability tool, not a loop-replacement. If the branches are complex or the line grows long, an ordinary `if`/`else` is clearer.
- **Use them where a small value choice is happening** — picking a minimum, a sign, a default, or a message. That is precisely the 'pick one of two values' shape ternaries exist for.

Compare: the same logic as `a if a < b else b` written with a regular `if`/`else` takes four lines; the ternary does it in one. Interviewers ask this question because it tests whether you know Python's compact control-flow forms and when they are — and are not — appropriate.

## Q8. Why `is` behaves like this: identity and interned integers

The heading says *'What does the `is` operator do?'* — it checks whether two names refer to the **same object in memory**, i.e. whether their `id()` values are equal. `==` compares values; `is` compares identity.

The demonstration cells below show something surprising, and understanding it requires knowing about **integer interning**: CPython caches small integers in the range **-5 to 256** at startup. Every `1` you write anywhere in the program refers to that one cached object. That is why:

```
a = 1; b = 1; a is b   # True  → one shared object
```

but

```
a = 257; b = 257; a is b   # False → two equal but distinct objects
```

Each `257` allocates a new object; the values are equal (`a == b` is `True`), yet the identities differ — `id(a)` and `id(b)` print different numbers, as the cells below confirm. Even `-14` is outside the cache, so two `-14`s are distinct objects.

This is a classic interview trap for two reasons:

- It shows whether you understand **'equal' is not 'identical'** — the single most important object-model idea in Python.
- It shows whether you know that CPython's cache range is an *implementation detail* you must never depend on. Code that relies on `a is b` for small integers is fragile.

### `Q11 Identity operator (is) vs ==?`

->> Here’s the main difference between python “==” vs “is:”

Identity operators: The “is” and “is not” keywords are called identity operators that compare objects based on their identity.
Equality operator: The “==” and “!=” are called equality operators that compare the objects based on their values. 


## Q11. `==` vs `is` for lists: cloning demonstrates the difference

The question asks you to separate **equality** from **identity**, and this notebook demonstrates it most dramatically with lists.

Look at the code cells below: `s = [1, 2, 3]` creates one list object. Then `p = s` makes `p` *refer to the same object* — no copy happens, so `p is s` is `True` and their `id()` values match. By contrast, `q = s[:]` is a **slice-copy** (cloning): it builds a brand-new list with equal contents, so `s == q` is `True` but `s is q` is `False`, and `id(q)` is a different number.

The follow-up cell drives the point home: `a = [1, 2, 3]`, `b = a[:]`, then `a.append(4)`. Now `a` prints `[1, 2, 3, 4]` while `b` prints `[1, 2, 3]` — mutating the cloned list leaves the original untouched, proving they are separate objects. (Had we written `b = a`, `b` would have shown the 4 as well.)

Three takeaways to carry into the interview:

- **`is` asks 'are these the same object?', `==` asks 'do these have the same value?'**
- **Assignment never copies** — `p = s` duplicates the name, not the data.
- **Slicing copies** — `s[:]`, and the `s[::-1]` you used for reversing, both return new objects.

This is the same distinction behind the small-integer cache from Q8, but with lists the difference is visible: two lists with identical contents can clearly be two different things sitting at two different addresses.

````python
# Case 4:
# Here variable s is assigned a list, 
# and q assigned a list values same as s but on slicing of list a new list is generated
s=[1,2,3]
p=s
# cloning
q=s[:]
print("id of p", id(p))
print("Id of s", id(s))
print("id of q", id(q))
print("Comapare- s == q", s==q)
print("Identity- s is q", s is q)
print("Identity- s is p", s is p)
print("Comapare- s == p", s==p)
````

````python
a = [1,2,3]
b = a[:]

a.append(4)
print(a)
print(b)
````

## Q12. The throwaway `_` variable

The underscore `_` is Python's official 'I don't care about this value' placeholder. When a value is needed by syntax but not by your logic, you assign it to `_` instead of inventing a name.

The two canonical uses are exactly what the cells below demonstrate:

**1. Ignore a value while unpacking a tuple**

```
x, _, y = (1, 2, 3)
```

This unpacks 1 into `x`, 3 into `y`, and throws away the middle 2. The cell prints `x- 1` and `y- 3`; the `_` variable never appears in the output. Unpacking crops up wherever a function returns several values but you need only some of them.

**2. Ignore the loop index**

```
for _ in range(5):
    print('hello')
```

When the code inside the loop never uses the current index — we just want the body to run five times — `_` says so out loud. The cell prints `hello` five times and, again, the index plays no role.

A few things worth knowing for the interview:

- A plain `_` in the interactive interpreter also happens to hold the last expression's result, but the throwaway-variable reading above is the one the question is about.
- `_` is a perfectly normal variable name, so `_` does exist and holds the last assigned value; calling it a 'throwaway' is a convention about *intent*, not special language magic.
- The convention scales: many codebases use `_` and `__` to mean 'ignore one level deeper' when unpacking nested data.

The lesson generalises beyond `_` itself: good signal in variable names comes in two flavours — meaningful names for values you use, and a recognised placeholder for values you consciously discard.

````python
# Ignore a value when unpacking
x, _, y = (1, 2, 3)

print("x-",x)
print("y-", y)
````

````python
#Ignore the index

# Say we want to print hello 5 times, we don't need index value

for _ in range(5):
    print('hello')
````

### `Q13 Modules vs packages vs Library`



Python uses some terms that you may not be familiar with if you’re coming from a different language. Among these are modules, packages, and libraries.

* A **module** is a Python file that’s intended to be imported into scripts or other modules. It often defines members like classes, functions, and variables intended to be used in other files that import it.

* A **package** is a collection of related modules that work together to provide certain functionality. These modules are contained within a folder and can be imported just like any other modules. This folder will often contain a special `__init__` file that tells Python it’s a package, potentially containing more modules nested within subfolders

* A **library** is an umbrella term that loosely means “a bundle of code.” These can have tens or even hundreds of individual modules that can provide a wide range of functionality. Matplotlib is a plotting library. The Python Standard Library contains hundreds of modules for performing common tasks, like sending emails or reading JSON data. What’s special about the Standard Library is that it comes bundled with your installation of Python, so you can use its modules without having to download them from anywhere.

These are not strict definitions. Many people feel these terms are somewhat open to interpretation. Script and module are terms that you may hear used interchangeably.

### `Q14 Why 0.3 – 0.2 is not equal to 0.1 in Python?`

The reason behind it is called “*precision*”, and it’s due to the fact that computers do not compute in Decimal, but in Binary. Computers do not use a base 10 system, they use a base 2 system (also called Binary code).

https://www.geeksforgeeks.org/why-0-3-0-2-is-not-equal-to-0-1-in-python/

## Q14. Why 0.3 - 0.2 gives 0.09999999999999998

The headline is: `0.3 - 0.2` computes to `0.09999999999999998`, and `0.3 - 0.2 == 0.1` is `False`. The reason is that computers don't count in decimal — they count in **binary**, and binary simply cannot store most decimal fractions exactly.

Just as `1/3` has no clean decimal representation, `0.1` has **no clean binary representation**. Under the hood, 0.3 and 0.2 are stored as the nearest binary approximations, and the subtraction of two already-approximate values lands one tiny step off the decimal `0.1` you expected. Python prints the shortest decimal that round-trips to that binary value — `0.09999999999999998` — which is why you *see* the error.

Three frames for the same fact:

- It is not a Python bug and not specific to Python — any IEEE-754 floating point gives the same answer in JavaScript, C, and Java.
- It is a general consequence of base 2 vs base 10, the same class of phenomenon that Q8's integer cache highlights: what a representation can express is limited by the representation itself.
- It only bites **real numbers**. Integers, and `%`, `//`, and arithmetic on `int`s, are exact — the techniques you used to sum series and extract digits in Session 2 are all safe.

In practice, never compare floats for exact equality. Compare with a tolerance instead.

```
abs(0.3 - 0.2 - 0.1) < 1e-9   # True
```

Or use the `decimal` module when money matters or exact results are required. Run the cell below and stare at the two lines — the message to memorise is that the printed float is an approximation, and `==` on floats is a bet you usually lose.

````python
# code
print(0.3 - 0.2)
print(0.3 - 0.2 == 0.1)
````

## Q15. Docstrings: documentation that lives in the code

A **docstring** is the string written as the first statement of a module, function, class, or method, intended to document what it does. Unlike a `#` comment, which describes a single line, a docstring describes the *contract* of the whole unit, and Python makes it introspectable — `help(func)` and `print(func.__doc__)` reveal it at runtime.

The convention is triple quotes, `"""..."""`, with a one-line summary, then a blank line and optional details:

```
def greet(name):
    """Return a friendly greeting for name."""
    return "Hello, " + name
```

The cells below demonstrate the introspection side: `print(input.__doc__)` and `print(type.__doc__)` print the built-in documentation Python already ships for these functions — showing that even the interpreter's own tooling documents itself through docstrings. The final cells also showcase two further curiosities that interviewers sometimes pair with this question:

- `s = 'have'; s = s.capitalize()` — strings are immutable, so `s.capitalize()` returns a *new* string object, visible in the changed `id(s)` across the two prints.
- `a, b = print('hello'), print('world')` — a function call's return value is also just a value; `print` returns `None`, so both `a` and `b` hold `None` while the printing side-effects happen.

For the interview, the essentials are the triple-quote convention, that a docstring is the first statement of the unit, that `__doc__` exposes it, and that good docstrings say *what* and *why*, leaving the *how* to the code itself.
