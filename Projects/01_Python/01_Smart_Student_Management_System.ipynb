Scenario:

Your college wants a small Python application to manage student records. The application should store students, their marks, and their attendance.

Your tasks:

1. Take the names of 5 students as input.
2. Store their names and marks using an appropriate Python collection.
3. Calculate the average marks of each student.
4. Identify students who scored more than 80.
5. Find the student with the highest marks.
6. Remove duplicate student names.
7. Sort students according to their marks in descending order.
8. Display students who have attendance below 75%.
9. Create a function that accepts student information and returns their result status:
   - 90 or above: Excellent
   - 75–89: Very Good
   - 50–74: Pass
   - Below 50: Fail

Concepts being tested: Lists, dictionaries, sets, tuples, loops, conditions, functions, sorting, type conversion, and list comprehension.

```python

class Student:
    # Class variable shared by all Student objects
    __student_counter = 101

    @classmethod
    def generate_student_id(cls):
        student_id = f"S{cls.__student_counter}"
        cls.__student_counter += 1
        return student_id

    def __init__(self, name, marks, attendance):
        if not name.strip():
            raise ValueError("Student name cannot be empty.")

        if not marks:
            raise ValueError("Marks cannot be empty.")

        if any(type(mark) is not int or not 0 <= mark <= 100 for mark in marks):
            raise ValueError("Marks must be integers between 0 and 100.")

        if not 0 <= attendance <= 100:
            raise ValueError("Attendance must be between 0 and 100.")

        self.student_id = Student.generate_student_id()
        self.name = " ".join(name.split())
        self.marks = marks.copy()
        self.attendance = attendance

    def average_marks(self):
        return sum(self.marks) / len(self.marks)

    def get_result_status(self):
        average = self.average_marks()

        if average >= 90:
            return "Excellent"
        elif average >= 75:
            return "Very Good"
        elif average >= 50:
            return "Pass"
        else:
            return "Fail"

    def __str__(self):
        return (
            f"Student ID: {self.student_id}\n"
            f"Name: {self.name}\n"
            f"Marks: {self.marks}\n"
            f"Average Marks: {self.average_marks():.2f}\n"
            f"Attendance: {self.attendance}%"
        )


class StudentManagementSystem:

    def __init__(self):
        # Dictionary stores student objects
        self.students = {}

    # Normalize names for duplicate checking
    @staticmethod
    def normalize_name(name):
        return " ".join(name.split()).casefold()

    # Take marks as input
    def read_marks(self):
        while True:
            try:
                raw_marks = input(
                    "Enter marks separated by spaces: "
                ).strip()

                marks = list(map(int, raw_marks.split()))

                if not marks:
                    print("Please enter at least one mark.")
                    continue

                if any(mark < 0 or mark > 100 for mark in marks):
                    print("Marks must be between 0 and 100.")
                    continue

                return marks

            except ValueError:
                print("Invalid input. Enter integer marks only.")

    # Take attendance as input
    def read_attendance(self):
        while True:
            try:
                attendance = int(
                    input("Enter attendance percentage: ")
                )

                if 0 <= attendance <= 100:
                    return attendance

                print("Attendance must be between 0 and 100.")

            except ValueError:
                print("Enter a valid integer.")

    # 1. Add Student
    def add_student(self):
        name = input("Enter student name: ").strip()

        if not name:
            print("Student name cannot be empty.")
            return

        marks = self.read_marks()
        attendance = self.read_attendance()

        try:
            student = Student(name, marks, attendance)

            self.students[student.student_id] = student

            print("\nStudent added successfully!")
            print("Generated ID:", student.student_id)

        except ValueError as error:
            print(error)

    # 2. Display Students
    def display_students(self):
        if not self.students:
            print("No students exist.")
            return

        print("\n----- ALL STUDENTS -----")

        for student in self.students.values():
            print(student)
            print("-" * 40)

    # 3. Calculate average marks of all students
    def average_marks_of_students(self):
        if not self.students:
            print("No students exist.")
            return

        print("\n----- STUDENT AVERAGES -----")

        for student in self.students.values():
            print(
                f"{student.name}: "
                f"{student.average_marks():.2f}"
            )

    # 4. Students with average marks above 80
    def students_marks_above_80(self):
        if not self.students:
            print("No students exist.")
            return

        found = False

        print("\n----- STUDENTS ABOVE 80 -----")

        for student in self.students.values():

            if student.average_marks() > 80:
                print(
                    f"{student.name}: "
                    f"{student.average_marks():.2f}"
                )
                found = True

        if not found:
            print("No students scored above 80.")

    # 5. Student with highest average marks
    def student_with_highest_marks(self):
        if not self.students:
            print("No students exist.")
            return

        highest_average = max(
            student.average_marks()
            for student in self.students.values()
        )

        print("\n----- HIGHEST SCORING STUDENT(S) -----")

        # Display all students in case of a tie
        for student in self.students.values():

            if student.average_marks() == highest_average:
                print(
                    f"{student.name}: "
                    f"{highest_average:.2f}"
                )

    # 6. Remove duplicate student names
    def remove_duplicate_student_names(self):
        if not self.students:
            print("No students exist.")
            return

        unique_students = {}
        seen_names = set()
        removed_count = 0

        for student_id, student in self.students.items():

            normalized_name = self.normalize_name(student.name)

            if normalized_name not in seen_names:
                seen_names.add(normalized_name)
                unique_students[student_id] = student
            else:
                removed_count += 1

        self.students = unique_students

        print(f"Duplicate records removed: {removed_count}")

    # 7. Sort students by average marks
    def sort_students_by_marks(self):
        if not self.students:
            print("No students exist.")
            return

        sorted_students = sorted(
            self.students.values(),
            key=lambda student: student.average_marks(),
            reverse=True
        )

        print("\n----- SORTED STUDENTS -----")

        for student in sorted_students:
            print(
                f"{student.name}: "
                f"{student.average_marks():.2f}"
            )

    # 8. Students with attendance below 75%
    def students_with_low_attendance(self):
        if not self.students:
            print("No students exist.")
            return

        found = False

        print("\n----- LOW ATTENDANCE STUDENTS -----")

        for student in self.students.values():

            if student.attendance < 75:
                print(
                    f"{student.name}: "
                    f"Attendance = {student.attendance}%"
                )
                found = True

        if not found:
            print("No students have attendance below 75%.")

    # 9. Get result status of a student
    def get_result_status(self):
        if not self.students:
            print("No students exist.")
            return

        student_id = input(
            "Enter student ID: "
        ).strip().upper()

        student = self.students.get(student_id)

        if student is None:
            print("Student not found.")
            return

        print("\n----- STUDENT RESULT -----")
        print("Name:", student.name)
        print("Average:", f"{student.average_marks():.2f}")
        print("Result:", student.get_result_status())

    # Main menu
    def menu(self):

        while True:

            print("\n===== STUDENT MANAGEMENT SYSTEM =====")
            print("1. Add Student")
            print("2. Display Students")
            print("3. Calculate Average Marks")
            print("4. Identify Students with Marks > 80")
            print("5. Find Student with Highest Marks")
            print("6. Remove Duplicate Student Names")
            print("7. Sort Students by Marks (Descending)")
            print("8. Display Students with Attendance < 75%")
            print("9. Get Result Status of a Student")
            print("10. Exit")

            try:
                choice = int(input("\nEnter your choice: "))

            except ValueError:
                print("Invalid choice. Enter a number from 1 to 10.")
                continue

            if choice == 1:
                self.add_student()

            elif choice == 2:
                self.display_students()

            elif choice == 3:
                self.average_marks_of_students()

            elif choice == 4:
                self.students_marks_above_80()

            elif choice == 5:
                self.student_with_highest_marks()

            elif choice == 6:
                self.remove_duplicate_student_names()

            elif choice == 7:
                self.sort_students_by_marks()

            elif choice == 8:
                self.students_with_low_attendance()

            elif choice == 9:
                self.get_result_status()

            elif choice == 10:
                print("Exiting Student Management System...")
                break

            else:
                print("Invalid choice. Please select 1 to 10.")
```

```python
if __name__ == "__main__":

    sms = StudentManagementSystem()
    sms.menu()
```
