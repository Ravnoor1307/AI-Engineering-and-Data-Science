These five challenges are designed to test your understanding of Python fundamentals, data types, functions, OOP, and time complexity.

They are not simple syntax exercises. Each case represents a small real-world problem where you need to decide which Python concepts to use.

## Case 1 – Smart Student Management System

[Microsoft Power BI Consulting Services | Data-Driven Insights – Comfygen](https://images.openai.com/static-rsc-4/PpY5noc-f2Iynax7LKgQAfC3wKj3lLuGZs9STPAC6gnhHUgrE7Kp4G_Jib_XrYpNcvHiEwigjBD1KuhB3IMw9mhsXENRtq9661Zsc8eX3V6nIvm6u4lA-nfspYtvUAVxJFhyvvjm6YE3wXWiFcAILNoppWOsbbcawzxYV7yXedY?purpose=inline)

[comfygen.com](https://www.comfygen.com/data/power-bi-consulting-services?utm_source=chatgpt.com)

Scenario:

Your college wants a small Python application to manage student records. The application should store students, their marks, and their attendance.

Your tasks:

1. Take the names of 5 students as input.
2. Store their names and marks using an appropriate Python collection.
3. Calculate the average marks of each student.
4. Identify students who scored more than 80.
5. Find the student with the highest marks.
6. Remove duplicate student names.
7. Sort students according to their marks in descending order.
8. Display students who have attendance below 75%.
9. Create a function that accepts student information and returns their result status:
   - 90 or above: Excellent
   - 75–89: Very Good
   - 50–74: Pass
   - Below 50: Fail

Concepts being tested: Lists, dictionaries, sets, tuples, loops, conditions, functions, sorting, type conversion, and list comprehension.

## Case 2 – Smart ATM Banking System

[Unrecognizable Man Entering PIN Code on Keypad, Taking US Dollars from ATM Machine](https://images.openai.com/static-rsc-4/4erANfxqpPILxypmz0CFpsH_p8W1H95AhxrHxJ-DBgsjXALmL2LHA6_3sQKm0KdclboONQvvlGkA4szNZwjlLq9kd4Al8Fe5ifxSMXHOI-QQjXRLG7TbyK-KfOJxk272RmcF-mL1NrNjHwQjmriKtgFCUuwTP60OF3oS78XKkROvKanrG6Yb7DzvduC_3L42?purpose=inline)

Scenario:

You are developing a basic ATM application. A customer should be able to create an account, deposit money, withdraw money, and check their balance.

Your tasks:

Create an `Account` class with the following features:

1. Initialize account holder name, account number, PIN, and balance.
2. Keep the balance private.
3. Provide methods to:
   - Deposit money.
   - Withdraw money.
   - Check balance after PIN verification.
   - Change PIN after verifying the old PIN.
4. Do not allow withdrawals greater than the available balance.
5. Do not allow negative deposits or withdrawals.
6. Maintain a class variable that counts the total number of accounts created.
7. Create a static method to validate a PIN.
8. Create a class method to display the total number of accounts.
9. Override `__str__()` to display basic account information without exposing the PIN.

Bonus challenge: Create a `SavingsAccount` class that inherits from `Account` and adds an interest-rate attribute.

Concepts being tested: Classes, objects, constructors, `self`, encapsulation, getters/setters, static variables, class methods, static methods, inheritance, method overriding, and magic methods.

## Case 3 – E-Commerce Product and Shopping Cart

[E-commerce Platforms Built for Modern Retailers](https://images.openai.com/static-rsc-4/Idi4TQ3QHWELFLopXpC1Gz1w8k4d9u6KqkYN4iVrg66jqA9Ogq2Yvzp5ss4i8F3pSAe_Wcw5p989_Coa9cMgKryeMCL2cwclQ5HvdZjw1rSQueMub0huAyfaqIyye81ThfyyV8vQ-Pil_YHC9AR5muh7aDaxvwu0zLB86Dfb3fA?purpose=inline)
