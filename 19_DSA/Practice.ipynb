```python
# minimum elements
numbers = [45, 12, 89, 3, 67, 21]

if (not numbers):
    print("None")
    
minimum = numbers[0]

for i in range(1,len(numbers)):
    if(numbers[i] < minimum):
        minimum = numbers[i]
    else:
        pass
print(minimum)
```

```python
# sum of array elements
numbers = [10, 20, 30, 40, 50]
if(not numbers):
    print("None")
sum = 0
for i in range(len(numbers)):
    sum += numbers[i]
print(sum)
```

```python
# count occurences
numbers = [10, 20, 10, 30, 10, 40]
target = 10
freq = {}
for i in range(len(numbers)):
    if numbers[i] not in freq:
        freq[numbers[i]] = 1
    else:
        freq[numbers[i]] += 1
print(freq[target])
```

```python
# Leetcode 1480. Running Sum of 1d Array
nums = [1,2,3,4]
running_sum = []
su_m = 0
for i in range(len(nums)):
    su_m += nums[i] 
    running_sum.append(su_m)
running_sum
```

```python
# LeetCode 1431. Kids With the Greatest Number of Candies
candies = [4,2,1,1,2]
extraCandies = 3
maximum = max(candies)
result = []
for i in range(len(candies)):
    candies[i] += extraCandies
    if(candies[i] >= maximum):
        result.append(True)
    else:
        result.append(False)
result
```

```python
# LeetCode 1512. Number of Good Pairs
nums = [1,2,3,1,1,3]
good_pairs = 0
for i in range(len(nums)):
    for j in range(len(nums)):
        if(nums[i] == nums[j]) and (i < j):
            good_pairs += 1
        else:
            pass
good_pairs
```

```python
# LeetCode : 485. Max Consecutive Ones
nums = [1,0,1,1,0,1]
consicutive = 0
maximum = 0

for num in nums:
    if num == 1:
        consicutive += 1
    else:
        consicutive = 0
    maximum = max(maximum,consicutive)
maximum
```
