The enumerate function is a built-in Python function that allows you to loop over an iterable while keeping track of the index of each element. This is particularly useful when you need both the index and the value without manually maintaining a counter.

```python
# we will not use enumerate

students = ["Alice", "Sam", "Will", "Adam"]

# Alice --> 1
# Sam --> 2

# roll no.: , name:

roll_no = 1

for student in students:
    print(f"roll no.: {roll_no}, name: {student}")
    roll_no += 1
```

output --> tuple (index, value)

```python
# with enumerate
students = ["Alice", "Sam", "Will", "Adam"]

for t in enumerate(students):
    print(t, type(t))
```

```python
# with enumerate
students = ["Alice", "Sam", "Will", "Adam"]

for index, student in enumerate(students):
    print(f"roll no.: {index}, name: {student}")
```

Syntax

enumerate(iterable, start)

```python
# with enumerate and using the start parameter
students = ["Alice", "Sam", "Will", "Adam"]

for index, student in enumerate(students, start=1):
    print(f"roll no.: {index}, name: {student}")
```

```python
# list of numbers

numbers = [3, 2, 4, 5, 9, 6, 7, 8, 10, 6]
```

Steps:
1. Calculate the mean
2. distances from mean
3. absolute
4. sort --> ascending

which numbers are closest to the mean by distances

```python
# calculate the mean

mean = sum(numbers) / len(numbers)

mean
```

```python
# calculate the distances

distances = [abs(num - mean) for num in numbers]
print(distances)
```

```python
# sort the distances

sorted_list = sorted(distances)

print(sorted_list)
```

```python
numbers = [3, 2, 4, 5, 9, 6, 7, 8, 10, 6]
```

```python
# calculate the distances

distances = [abs(num - mean) for num in numbers]
print(distances)
```

```python
indexed_distances = list(enumerate(distances))

print(indexed_distances)
```

```python
# sort the distances

sorted_distances = sorted(indexed_distances, key=lambda x: x[1])
print(sorted_distances)
```

```python
# fetch out the indexes

indexes = [t[0] for t in sorted_distances]
print(indexes)
```

```python
for ind in indexes:
    print(numbers[ind])
```
