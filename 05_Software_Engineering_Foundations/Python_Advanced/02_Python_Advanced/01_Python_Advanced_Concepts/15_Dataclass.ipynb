```python
# default values through field

from dataclasses import field

@dataclass
class Person:
    name: str
    age: int
    city_name: str
    email: list[str] = []
```

```python
@dataclass
class Person:
    name: str
    age: int
    city_name: str
    email: list[str] = field(default=list)
```

```python
person_5 = Person("Ankit", 25, "Chennai", ["ankit05@gmail.com", "ankit100@yahoo.co.in"])

person_5
```

```python
person_5.name
```

```python
person_5.email
```

```python
# using a callable for default values
from random import randint

def generate_id():
    return randint(1, 1000)

@dataclass
class Person:
    name: str
    age: int
    city_name: str
    id: int = field(default_factory=generate_id)
    email: list[str] = field(default=list)
```

```python
person_6 = Person("Rahul", 25, "Pune")

print(person_6)
```

```python
person_6 = Person("Rahul", 25, "Pune", 10)

print(person_6)
```

```python
# post initialization

@dataclass
class Person:
    first_name: str
    last_name: str
    age: int
    city_name: str
    full_name: str = field(init=False)
    id: int = field(default_factory=generate_id, init=False)
    email: list[str] = field(default=list)

    def __post_init__(self):
        self.full_name = self.first_name + " " + self.last_name
```

```python
person_8 = Person("Rahul", "Kumar", 25, "Pune", ["rahulk12@gmail.com"])

print(person_8)
```

```python
person_8.first_name
```

```python
person_8.full_name
```

```python
# hide information in repr

@dataclass
class Person:
    first_name: str
    last_name: str
    age: int
    city_name: str
    full_name: str = field(init=False)
    id: int = field(default_factory=generate_id, init=False)
    email: list[str] = field(default=list)
    _status: str = field(default="Active", init=False, repr=False)

    def __post_init__(self):
        self.full_name = self.first_name + " " + self.last_name
```

```python
person_9 = Person("Rahul", "Kumar", 25, "Pune", ["rahulk12@gmail.com"])

print(person_9)
```

```python
person_9._status
```

```python
person_9._status = "Inactive"
```

```python
# create frozen class --> immutable

@dataclass(frozen=True)
class Person:
    first_name: str
    last_name: str
    age: int
    city_name: str
    id: int = field(default_factory=generate_id, init=False)
    email: list[str] = field(default=list)
    _status: str = field(default="Active", init=False, repr=False)
```

```python
person_10 = Person("Amit", "Sharma", 45, "Gurugram")

print(person_10)
```

```python
person_10._status
```

```python
person_10._status = "Inactive"
```

```python
# pass arguments as kw args only

@dataclass(frozen=False, kw_only=True)
class Person:
    first_name: str
    last_name: str
    age: int
    city_name: str
    id: int = field(default_factory=generate_id, init=False)
    email: list[str] = field(default=list)
    _status: str = field(default="Active", init=False, repr=False)
```

```python
person_11 = Person(first_name="Amit", last_name="Sharma", age=45, city_name="Gurugram")

print(person_11)
```

```python
person_11 = Person(age=45, first_name="Amit", last_name="Sharma", city_name="Gurugram")

print(person_11)
```
