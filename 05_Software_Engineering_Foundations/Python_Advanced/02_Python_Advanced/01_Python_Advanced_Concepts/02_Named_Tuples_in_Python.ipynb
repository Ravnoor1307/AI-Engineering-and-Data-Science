```python
# 3 ways to create fields
```

```python
# passing as list
Person1 = namedtuple("Person",["Name","Age","City"])
```

```python
# passing as str with whitespaces

Person2 = namedtuple("Person","Name Age City")
```

```python
# passing as str with commas
Person3 = namedtuple("Person","Name,Age,City")
```

```python
Person1('noor',23,'Ludhiana')
```

```python
Person2('noor',23,'Ludhiana')
```

```python
Person3('noor',23,'Ludhiana')
```

```python
# _make() --> passes values of iterable

input_list = ["noor",23,'Ludhiana']
Person1._make(input_list)
```

```python
input_dict = {'Name':"noor",'Age':23,'City':'Ludhiana'}
Person1._make(input_dict.values())
```

```python
# _asdict --> converts to dict
person = Person1('noor',23,'Ludhiana')

person
```

```python
person._asdict()
```

```python
Point(2,4)._asdict()
```

```python
# _replace --> replaces values of as field

point1 = Point(3,4)
print(id(point1))
```

```python
point2 = point1._replace(y_cod=6)
```

```python
print(id(point2))
```
