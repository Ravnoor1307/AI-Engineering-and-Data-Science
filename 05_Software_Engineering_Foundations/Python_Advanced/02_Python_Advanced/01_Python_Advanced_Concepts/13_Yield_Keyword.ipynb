```python
# iterables

list1 = [1, 2, 3, 4, 5]

for i in list1:
    print(i)
```

Yield

Function --> Iterator(Generator)

```python
# yield can turn functions --> iterators

def my_generator():
    print("Inside Generator")
    yield 1
    print("After yield")
```

```python
gen = my_generator()
```

```python
hasattr(gen, "__iter__")
```

```python
hasattr(gen, "__next__")
```

```python
next(gen)
```

```python
# yield vs return

def func_with_return():
    num = 5
    print("Before return")
    return num
    print("After return")
    num += 1
    return num


def func_with_yield():
    num = 5
    print("Before yield")
    yield num
    print("After yield")
    num += 1
    yield num
```

```python
func_with_return()
```

```python
func_with_yield()
```

```python
# multiple yield statements

def test_gen():
    yield "Start"
    yield "Middle"
    yield "Stop"

gen = test_gen()

print(next(gen))
print(next(gen))
print(next(gen))
```

```python
# counter using yield

def counter(limit):
    num = 1
    while num <= limit:
        yield num
        num += 1

counter_gen = counter(10)

print(next(counter_gen))
print(next(counter_gen))
print(next(counter_gen))
```

```python
for i in counter_gen:
    print(i)
```

```python
# next on generator

counter_gen = counter(20)

next(counter_gen)
```

```python
for i in range(20):
    print(next(counter_gen))
```

```python
# for loop on generator
```

```python
# infinite counter

def inf_counter(step):
    num = 1
    while True:
        yield num
        num += step
```

```python
inf_gen = inf_counter(9)

inf_gen
```

```python
for i in range(10):
    print(next(inf_gen))
```
