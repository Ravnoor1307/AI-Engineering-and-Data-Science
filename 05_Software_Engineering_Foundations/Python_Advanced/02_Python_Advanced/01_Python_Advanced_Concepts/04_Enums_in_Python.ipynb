```python
# enums (enumerations)



class RGB(Enum):
    RED = 1
    GREEN = 2
    BLUE = 3
```

```python
type(RGB)
```

Individual items in an enum, they are called as members of the enum

all the members of the enum, are separate instances of the class

Each instance of the enum class(member) has two attributes:
1. name
2. value

```python
# enum members
print(RGB.RED)
print(type(RGB.RED))

# name attribute
print(RGB.GREEN.name)

# value attribute
print(RGB.BLUE.value)
```

```python
dir(RGB.GREEN)
```

```python
# str vs repr

print(RGB.RED) # --> str method
```

```python
RGB.RED # --> repr method
```

```python
# converted to list

enum_list = list(RGB)

print(enum_list)
```

```python
# tuples

tuple(RGB)
```

```python
# looping through enums

for color in RGB:
    print(color)
```

Properties of enums:
1. immutable
2. hashable
3. iterable

```python
# enums are immutable

RGB.RED = 10
```

```python
# enum members are hashable

colors_dict = {}

for color in RGB:
    colors_dict[color.name] = color.value

print(colors_dict)
```

```python
# identity

for color in RGB:
    print(id(color))
```

```python
# auto
class Weekdays(Enum):
    MON = auto()
    TUE = auto()
    WED = auto()
    THU = auto()
    FRI = auto()
```

```python
for day in Weekdays:
    print(f"{day.name} ---> {day.value}")
```

```python
# unique --> unique value

@unique
class Weekdays(Enum):
    MON = 1
    TUE = 2
    WED = 3
    THU = 4
    FRI = 5
    SAT = 6
```

the unique decorator soes not allow duplicate values

```python
class Debug(Enum):
    INFO = auto()
    DEBUG = auto()
    WARNING = auto()
    ERROR = auto()
    CRITICAL = auto()
```

```python
for state in Debug:
    print(state.name, state.value)
```
