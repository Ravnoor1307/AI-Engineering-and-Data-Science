Monkey patching in Python refers to the dynamic modification of a class or module at runtime. This allows you to change or extend behavior without altering the original source code.



```python
class Car:

    def __init__(self, brand, fuel_type):
        self.brand = brand
        self.fuel_type = fuel_type

    def speed(self, speed_input):
        return f"The speed of the car is {speed_input} mph"
```

```python
car_1 = Car("Toyota", "Petrol")

car_1.speed(60)
```

```python
# new function

def new_speed(self, speed_input):
    return f"The speed of the car is {speed_input} km/h"
```

```python
car_1 = Car("Toyota", "Petrol")

car_1.speed(60)
```

```python
# patching

Car.speed = new_speed
```

```python
car_2 = Car("Suzuki", "Diesel")

car_2
```

```python
car_2.speed(40)
```

```python
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b

print(Calculator.add(2, 5))
```

```python
def new_add(a, b):
    result = a + b
    return f"The sum of numbers {a} and {b} is {result}"
```

```python
# patch

Calculator.add = new_add
```

```python
print(Calculator.add(2, 5))
```

# Mocking API calls

```python
from requests import get
```

```python
def my_get(url, *args, **kwargs):

    class MyResponse:
        def __init__(self):
            self.status_code = 200
            self.text = "HTML code"

        def json(self):
            return {"message": "This contains some data",
                    "data_type": "The data type is str"}

    return MyResponse()
```

```python
# patch

get = my_get
```

```python
# send a get request

response = get("https://www.wikipedia.org")

response
```
