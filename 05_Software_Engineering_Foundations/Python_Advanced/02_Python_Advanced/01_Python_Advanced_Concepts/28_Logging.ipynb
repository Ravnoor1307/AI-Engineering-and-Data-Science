```python
import logging
```

3 main comnponents:

1. Logger - Main component that does the logging.
2. Handler - it handles the output generated by logger -->
    - Stream(console)
    - File(log file)
3. Formatter - It formats the logs

Levels the logs in python
1. DEBUG - 10
2. INFO - 20
3. WARNING - 30
4. ERROR - 40
5. CRITICAL - 50

```python
# make a logger

logger = logging.getLogger("Logger")

# set level for logger
logger.setLevel(logging.DEBUG)

logger
```

```python
logging.NOTSET
```

```python
# handler

handler = logging.StreamHandler()

# attach handler to logger
logger.addHandler(handler)

# set level on handler
handler.setLevel(logging.DEBUG)
```

```python
logger.handlers
```

```python
# formatter

formatter = logging.Formatter(fmt="%(asctime)s - %(levelname)s - %(message)s")

# attach formatter to handler
handler.setFormatter(formatter)
```

```python
handler.formatter
```

```python
# log messages

logger.debug("This is a debug message")
logger.info("This is a info message")
logger.warning("This is a warning message")
logger.error("This is a error message")
logger.critical("This is a critical message")
```

```python
# new logger

logger = logging.getLogger("Test")

# set level
logger.setLevel(logging.INFO)

# file handler
handler = logging.FileHandler("log.log", mode="w")

# set level of handler
handler.setLevel(logging.WARNING)

# attach file handler to logger
logger.addHandler(handler)

# formatter
formatter = logging.Formatter(fmt="%(name)s - %(levelname)s - %(message)s")

# attach formatter to handler
handler.setFormatter(formatter)
```

```python
def add(a: int, b: int) -> int:
    try:
        if isinstance(a, int) and isinstance(b, int):
            logger.info("The inputs are integer inputs")
        else:
            raise TypeError("Inputs should be integer")
    except TypeError as e:
        logger.error(e, exc_info=True)
    else:
        return a + b
```
