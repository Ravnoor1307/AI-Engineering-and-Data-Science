## Functionalities

```python
# str representation --> as_posix

type(path.as_posix())
```

```python
# resolve --> absolute path

path.resolve()
```

```python
# exists

path.parent.parent.parent.parent.resolve().exists()
```

```python
# is dir

content_path = path.parent.parent.parent.parent.resolve()

content_path.is_dir()
```

```python
# is file

content_path.is_file()
```

```python
sample_file_path = Path("/content/test/sample.txt")

sample_file_path.is_file()
```

```python
# open file

with sample_file_path.open(mode="r") as f:
    print(f.read())
```

```python
# readtext

sample_file_path.read_text()
```
