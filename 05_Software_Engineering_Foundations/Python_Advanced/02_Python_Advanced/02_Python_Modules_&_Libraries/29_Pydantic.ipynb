```python
# data validation --> the manual way

def add_person(name: str, age: int):
    if not isinstance(name, str):
        raise TypeError("Name must be a str")
    elif not isinstance(age, int):
        raise TypeError("Age must be in int")
    print(name, type(name))
    print(age, type(age))

add_person("Mike", 25)
```

```python
# type validation using pydantic

from pydantic import BaseModel

class Person(BaseModel):
    name: str
    age: int

def add_person(person: Person):
    print(person.name, type(person.name))
    print(person.age, type(person.age))
```

```python
person_1 = Person(name="Mike", age=25)

person_1
```

```python
add_person(person_1)
```

```python
# data validation using field

from pydantic import Field
from typing import Annotated

# age should be more ge 18

class Person(BaseModel):
    name: str
    age: Annotated[int, Field(ge=18, description="Age should be an integer and greater than or equal to 18 years",
                              examples=[25, 26, 67])]

Person(name="Mike", age=25)
```

```python
Person(name="Mike", age=14)
```

```python
def add_person(name:str, age: int):
    if not isinstance(name, str):
        raise TypeError("Name must be a str")
    elif not isinstance(age, int):
        raise TypeError("Age must be in int")
    if age < 18:
        raise ValueError("Person should not be a minor")
    return {"name": name, "age": age}

add_person("Mike", 25)
```

```python
add_person("Mike", 15)
```

```python
# custom field validation --> field validator

# Mike  mike MIKE

from pydantic import field_validator


class Person(BaseModel):
    name: str
    age: int
    friends_list: list[str] | None = None


    @field_validator("name")
    @classmethod
    def validate_name(cls, value: str):
        if not value.istitle():
            raise ValueError("Name should be title cased")
        return value
```

```python
Person(name="Mike", age=23)
```

```python
Person(name="mike", age=23)
```

```python
Person(name="Mike", age="23", friends_list=["Mike", "Samantha"])
```

```python
Person(name="Mike", age="23", friends_list=["Mike", "Samantha", 43])
```

```python
# custom model validation

from pydantic import model_validator

class Person(BaseModel):

    name: str
    age: int


    @model_validator(mode="after")
    @classmethod
    def model_validation(cls, model):
        if not (model.age >= 18 and model.name.istitle()):
            raise ValueError("Age should be greater than or equal to 18 and name should be title cased")
        return model
```

```python
# computed fields --> after initialization calculations

from pydantic import computed_field

class Person(BaseModel):
    first_name: str
    last_name: str
    age: int

    @computed_field
    @property
    def full_name(self) -> str:
        return f"{self.first_name} {self.last_name}"
```

```python
Person(first_name="Ankit", last_name="Sharma", age=54)
```

```python
class Rectangle(BaseModel):

    length: Annotated[float, Field(gt=0)]
    width: Annotated[float, Field(gt=0)]

    @computed_field
    @property
    def area(self) -> float:
        return self.length * self.width
```

```python
Rectangle(length=5, width=10)
```

```python
# serialize models


# to a dict
Rectangle(length=5, width=10).model_dump(exclude=["area"])
```
