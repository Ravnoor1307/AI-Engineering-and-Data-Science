```python
def check_path(path):
    # check if the path exists
    if os.path.exists(path):
        if os.path.isdir(path):
            return f"Path {path} is a directory"
        elif os.path.isfile(path):
            return f"Path {path} is a file"
    else:
        raise FileNotFoundError("Path does not exist")
```

```python
check_path("./project/test.txt")
```

```python
check_path("./project/")
```

```python
path
```

```python
check_path(path)
```

```python
# os.walk() – Recursively generates file names in a directory tree


walk_gen = os.walk("./project/")
```

```python
tup_1 = next(walk_gen)
```

```python
print(tup_1)
```

```python
dir_path, dirs, filenames = tup_1
```

```python
print(dir_path)

print(dirs)

print(filenames)
```

```python
for dir_path, dirs, filenames in walk_gen:
    print(dir_path)
    print(dirs)
    print(filenames)
    print()
```
