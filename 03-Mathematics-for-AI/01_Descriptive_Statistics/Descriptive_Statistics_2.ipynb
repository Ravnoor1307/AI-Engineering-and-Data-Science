
# 1. Random Variables

## 1.1 Algebraic Variables

In algebra, a variable represents an unknown or changing quantity.

For example:

$$
x+5=10
$$

Therefore:

$$
x=5
$$

In this equation, the value of x is determined by solving the equation.

## 1.2 Random Variables

A random variable is a numerical function defined on the outcomes of a random experiment.

It assigns a numerical value to each possible outcome.

Mathematically:

$$
X:\Omega \rightarrow \mathbb{R}
$$

Where:

- $X$ is the random variable.
- $\Omega$ is the sample space.
- $\mathbb{R}$ represents real numbers.

### Example: Tossing a Coin

The possible outcomes are:

$$
\Omega=\{H,T\}
$$

Define the random variable:

$$
X=
\begin{cases}
1 & \text{if Heads}\\
0 & \text{if Tails}
\end{cases}
$$

Therefore:

$$
X\in\{0,1\}
$$

### Example: Rolling a Die

The sample space is:

$$
\Omega=\{1,2,3,4,5,6\}
$$

If $X$ represents the number appearing on the die:

$$
X\in\{1,2,3,4,5,6\}
$$

The actual value of $X$ depends on the random experiment.


````python

# Simulating a coin toss

outcomes = ["Heads", "Tails"]

results = random.choices(outcomes, k=1000)

coin_df = pd.Series(results)

print("First 10 outcomes:")
print(coin_df.head(10))

print("\nObserved frequencies:")
print(coin_df.value_counts())

print("\nObserved probabilities:")
print(coin_df.value_counts(normalize=True))
````


# 2. Types of Random Variables

Random variables are commonly classified into two categories.

## 2.1 Discrete Random Variables

A discrete random variable takes values from a finite or countably infinite set.

Examples:

- Number of heads in 10 coin tosses.
- Number appearing on a die.
- Number of students in a classroom.
- Number of customer purchases.

For a single die:

$$
X\in\{1,2,3,4,5,6\}

## 2.2 Continuous Random Variables

A continuous random variable can take values over an interval of real numbers.

Examples:

- Height
- Weight
- Temperature
- Time

For example, a person's height might be:

$$
X=172.43\text{ cm}

### Key Difference

For discrete random variables, we can assign probabilities to individual values.

For continuous random variables with a probability density, individual exact values have probability zero.

Instead, probabilities are calculated over intervals.


````python

rng = np.random.default_rng(42)

discrete = rng.integers(1, 7, size=1000)

continuous = rng.normal(
    loc=170,
    scale=10,
    size=1000
)

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].hist(
    discrete,
    bins=np.arange(0.5, 7.5, 1),
    edgecolor="black",
    color="skyblue"
)

axes[0].set_title("Discrete Random Variable")
axes[0].set_xlabel("Die Outcome")
axes[0].set_ylabel("Frequency")

axes[1].hist(
    continuous,
    bins=30,
    edgecolor="black",
    color="lightgreen"
)

axes[1].set_title("Continuous Random Variable")
axes[1].set_xlabel("Height")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
````


# 3. Probability Distributions

A probability distribution describes how probabilities are assigned to the possible values of a random variable.

For a discrete random variable, we can represent its distribution using a table.

### Example: Fair Die

| Outcome | Probability |
|---|---|
| 1 | 1/6 |
| 2 | 1/6 |
| 3 | 1/6 |
| 4 | 1/6 |
| 5 | 1/6 |
| 6 | 1/6 |

The probabilities sum to one:

$$
\sum_{x=1}^{6}P(X=x)=1
$$

## Why Use Probability Distribution Functions?

A table becomes inconvenient when:

- There are many possible outcomes.
- Outcomes form a countably infinite set.
- The random variable is continuous.

Instead, mathematical functions provide compact descriptions of probability distributions.

## Common Probability Distributions

### Discrete Distributions

- Bernoulli
- Binomial
- Poisson
- Geometric
- Negative Binomial

### Continuous Distributions

- Normal
- Uniform
- Exponential
- Gamma

### Why Are Probability Distributions Important?

They help us:

- Understand uncertainty.

# 4. Parameters of Probability Distributions

Parameters determine the shape, location, and scale of probability distributions.

Different distribution families use different parameters.

## Normal Distribution

A normal distribution is described by:

$$
X\sim N(\mu,\sigma^2)
$$

Where:

- $\mu$ is the population mean.
- $\sigma$ is the population standard deviation.
- $\sigma^2$ is the population variance.

The mean controls the center of the distribution.

The standard deviation controls its spread.

## Bernoulli Distribution

A Bernoulli distribution has one parameter:

$$
p=P(X=1)
$$

## Binomial Distribution

A binomial distribution has two parameters:

$$
X\sim Binomial(n,p)
$$

Where:

- $n$ is the number of independent trials.
- $p$ is the probability of success on each trial.

Understanding distribution parameters helps us estimate and interpret statistical models.


````python

x = np.linspace(-10, 10, 1000)

distributions = [
    (0, 1, "Mean=0, SD=1"),
    (0, 2, "Mean=0, SD=2"),
    (3, 1, "Mean=3, SD=1")
]

plt.figure(figsize=(10, 5))

for mu, sigma, label in distributions:
    plt.plot(
        x,
        norm.pdf(x, loc=mu, scale=sigma),
        label=label
    )

plt.title("Effect of Normal Distribution Parameters")
plt.xlabel("X")
plt.ylabel("Probability Density")
plt.legend()

plt.show()
````


# 5. Probability Mass Function (PMF)

A probability mass function describes the probability distribution of a discrete random variable.

The PMF is defined as:

$$
p_X(x)=P(X=x)
$$

For a valid PMF, two conditions must hold.

### Condition 1: Non-Negative Probabilities

$$
p_X(x)\geq0
$$

### Condition 2: Total Probability Equals One

$$
\sum_x p_X(x)=1
$$

## Example: Fair Die

For a fair six-sided die:

$$
p_X(x)=
\begin{cases}
\frac{1}{6} & x\in\{1,2,3,4,5,6\}\\
0 & \text{otherwise}
\end{cases}
$$

The probability of rolling a 4 is:

$$
P(X=4)=\frac{1}{6}
$$


````python

outcomes = np.arange(1, 7)

probabilities = np.repeat(1/6, 6)

pmf_df = pd.DataFrame({
    "Outcome": outcomes,
    "Probability": probabilities
})

display(pmf_df)

plt.figure(figsize=(9, 5))

plt.bar(
    outcomes,
    probabilities,
    color="steelblue",
    edgecolor="black"
)

plt.title("PMF of a Fair Die")
plt.xlabel("Outcome")
plt.ylabel("Probability")

plt.xticks(outcomes)
plt.ylim(0, 0.25)

plt.show()

print("Sum of probabilities:", probabilities.sum())
````


## 5.1 Bernoulli Distribution

A Bernoulli random variable represents a single experiment with two possible outcomes.

Examples:

- Success or failure.
- Heads or tails.
- Purchase or no purchase.

Let:

$$
X\sim Bernoulli(p)
$$

Its PMF is:

$$
P(X=x)=p^x(1-p)^{1-x}
$$

For:

$$
x\in\{0,1\}
$$

The expected value is:

$$
E[X]=p
$$

The variance is:

$$
Var(X)=p(1-p)
$$


````python

p = 0.7

x = np.array([0, 1])

probabilities = bernoulli.pmf(x, p)

plt.figure(figsize=(7, 4))

plt.bar(
    x,
    probabilities,
    color=["coral", "skyblue"],
    edgecolor="black"
)

plt.xticks([0, 1])
plt.title("Bernoulli Distribution")
plt.xlabel("Outcome")
plt.ylabel("Probability")

plt.show()

print("P(X=0):", probabilities[0])
print("P(X=1):", probabilities[1])
````


## 5.2 Binomial Distribution

A binomial random variable counts the number of successes in a fixed number of independent Bernoulli trials.

Assumptions:

1. A fixed number of trials.
2. Each trial has two possible outcomes.
3. Trials are independent.
4. The success probability remains constant.

The PMF is:

$$
P(X=k)=
\binom{n}{k}
p^k(1-p)^{n-k}
$$

Where:

- $n$ = Number of trials
- $k$ = Number of successes
- $p$ = Probability of success

Example:

How many heads occur in 10 independent coin tosses?


````python

n = 10
p = 0.5

x = np.arange(0, n + 1)

probabilities = binom.pmf(x, n, p)

plt.figure(figsize=(10, 5))

plt.bar(
    x,
    probabilities,
    color="mediumseagreen",
    edgecolor="black"
)

plt.title("Binomial Distribution: 10 Coin Tosses")
plt.xlabel("Number of Heads")
plt.ylabel("Probability")

plt.xticks(x)

plt.show()

print("P(X=5):", binom.pmf(5, n, p))
````


# 6. Cumulative Distribution Function (CDF)

The cumulative distribution function gives the probability that a random variable is less than or equal to a specified value.

$$
F_X(x)=P(X\leq x)
$$

For a discrete random variable:

$$
F_X(x)=\sum_{t\leq x}P(X=t)
$$

## Example: Fair Die

For a fair die:

$$
P(X\leq4)=P(1)+P(2)+P(3)+P(4)
$$

$$
P(X\leq4)=\frac{4}{6}
$$

$$
P(X\leq4)\approx0.6667
$$

### Properties of the CDF

1. It is non-decreasing.
2. Its values lie between zero and one.
3. It approaches zero toward negative infinity.
4. It approaches one toward positive infinity.
5. It is right-continuous.


````python

outcomes = np.arange(1, 7)

pmf = np.repeat(1/6, 6)

cdf = np.cumsum(pmf)

distribution_table = pd.DataFrame({
    "Outcome": outcomes,
    "PMF": pmf,
    "CDF": cdf
})

display(distribution_table)

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].bar(
    outcomes,
    pmf,
    color="skyblue",
    edgecolor="black"
)

axes[0].set_title("Probability Mass Function")
axes[0].set_xlabel("Outcome")
axes[0].set_ylabel("Probability")

axes[1].step(
    outcomes,
    cdf,
    where="post",
    color="red",
    linewidth=2
)

axes[1].scatter(outcomes, cdf, color="red")

axes[1].set_title("Cumulative Distribution Function")
axes[1].set_xlabel("Outcome")
axes[1].set_ylabel("Cumulative Probability")

plt.tight_layout()
plt.show()
````


# 7. Probability Distribution of Two Dice

Consider rolling two independent fair dice.

Each die has six possible outcomes.

The total number of equally likely ordered outcomes is:

$$
6\times6=36
$$

Let:

$$
X=D_1+D_2
$$

The possible sums are:

$$
X\in\{2,3,4,\ldots,12\}
$$

However, the sums are not equally likely.

For example:

$$
P(X=2)=\frac{1}{36}
$$

because only (1,1) produces a sum of 2.

But:

$$
P(X=7)=\frac{6}{36}
$$

because six ordered pairs produce a sum of 7.

We will simulate 10,000 rolls and compare the empirical probabilities with the theoretical distribution.


````python

L = []

for i in range(10000):

    a = random.randint(1, 6)
    b = random.randint(1, 6)

    L.append(a + b)

print("Total simulations:", len(L))
print("First five sums:", L[:5])
````

````python

dice_series = pd.Series(L)

empirical_pmf = (
    dice_series.value_counts(normalize=True)
    .sort_index()
)

empirical_cdf = empirical_pmf.cumsum()

result = pd.DataFrame({
    "Empirical PMF": empirical_pmf,
    "Empirical CDF": empirical_cdf
})

display(result)
````

````python

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

empirical_pmf.plot(
    kind="bar",
    ax=axes[0],
    color="steelblue"
)

axes[0].set_title("Empirical PMF of Two Dice")
axes[0].set_xlabel("Sum")
axes[0].set_ylabel("Probability")

empirical_cdf.plot(
    kind="bar",
    ax=axes[1],
    color="coral"
)

axes[1].set_title("Empirical CDF of Two Dice")
axes[1].set_xlabel("Sum")
axes[1].set_ylabel("Cumulative Probability")

plt.tight_layout()
plt.show()
````


## 7.1 Theoretical vs Empirical Probability

The theoretical probability of each sum can be calculated by enumerating all 36 ordered outcomes.

The empirical probability is estimated using the simulated observations.

As the number of simulations increases, empirical frequencies generally approach the theoretical probabilities.

This is an illustration of the Law of Large Numbers.


````python

from collections import Counter

all_sums = [
    a + b
    for a in range(1, 7)
    for b in range(1, 7)
]

theoretical_counts = Counter(all_sums)

theoretical_pmf = pd.Series({
    outcome: theoretical_counts[outcome] / 36
    for outcome in range(2, 13)
})

comparison = pd.DataFrame({
    "Theoretical": theoretical_pmf,
    "Empirical": empirical_pmf
})

display(comparison)

comparison.plot(
    kind="bar",
    figsize=(11, 5)
)

plt.title("Theoretical vs Empirical PMF")
plt.xlabel("Sum of Two Dice")
plt.ylabel("Probability")

plt.show()
````


# 8. Probability Density Function (PDF)

A probability density function describes the distribution of a continuous random variable.

For a continuous random variable with density $f(x)$:

$$
f(x)\geq0
$$

and:

$$
\int_{-\infty}^{\infty}f(x)\,dx=1
$$

## Probability vs Probability Density

For a continuous random variable:

$$
P(X=a)=0
$$

Probabilities are calculated over intervals.

For example:

$$
P(a\leq X\leq b)
=
\int_a^b f(x)\,dx
$$

### Important

The height of a PDF curve is not itself a probability.

Probability corresponds to the area under the density curve over an interval.

A density value can even exceed one, provided the total area remains one.

## Examples of Continuous Distributions

- Normal distribution
- Uniform distribution
- Exponential distribution
- Log-normal distribution

**Note:** The Poisson distribution is discrete and has a PMF, not a continuous PDF.



## 8.1 Normal Distribution

The normal distribution is one of the most widely used continuous probability distributions.

Its PDF is:

$$
f(x)=
\frac{1}{\sigma\sqrt{2\pi}}
\exp\left(
-\frac{(x-\mu)^2}{2\sigma^2}
\right)
$$

Where:

- $\mu$ is the mean.
- $\sigma$ is the standard deviation.

The normal distribution is symmetric around its mean.

Its mean, median, and mode are equal.

It is commonly used to model measurement errors and many approximately symmetric variables.


````python

mu = 50
sigma = 5

x = np.linspace(30, 70, 1000)

pdf = norm.pdf(x, loc=mu, scale=sigma)

plt.figure(figsize=(10, 5))

plt.plot(
    x,
    pdf,
    color="blue",
    linewidth=2
)

plt.fill_between(
    x,
    pdf,
    alpha=0.2,
    color="skyblue"
)

plt.title("Normal Probability Density Function")
plt.xlabel("X")
plt.ylabel("Probability Density")

plt.show()
````


## 8.2 Calculating Probability from a PDF

Suppose:

$$
X\sim N(50,5^2)
$$

We want:

$$
P(45\leq X\leq55)
$$

We can calculate this probability using:

$$
P(45\leq X\leq55)
=
F(55)-F(45)
$$

Where $F$ is the cumulative distribution function.

For a normal distribution, the interval from one standard deviation below the mean to one standard deviation above the mean contains approximately 68.27% of the probability.


````python

mu = 50
sigma = 5

lower = 45
upper = 55

probability = (
    norm.cdf(upper, loc=mu, scale=sigma)
    - norm.cdf(lower, loc=mu, scale=sigma)
)

print("P(45 <= X <= 55):", probability)

x = np.linspace(30, 70, 1000)

pdf = norm.pdf(x, loc=mu, scale=sigma)

plt.figure(figsize=(10, 5))

plt.plot(x, pdf, color="blue")

mask = (x >= lower) & (x <= upper)

plt.fill_between(
    x[mask],
    pdf[mask],
    color="orange",
    alpha=0.5
)

plt.title("Probability as Area Under the PDF")
plt.xlabel("X")
plt.ylabel("Density")

plt.show()
````


# 9. Density Estimation

Density estimation is the process of estimating an unknown probability density function from observed data.

It helps us understand the underlying distribution of numerical observations.

### Applications

- Data visualization
- Anomaly detection
- Probability estimation
- Generative modeling
- Exploratory data analysis
- Statistical inference

## Main Approaches

### Parametric Density Estimation

Assumes that the data follows a particular distribution family.

Examples:

- Normal distribution
- Exponential distribution
- Gamma distribution

The parameters of the chosen distribution are estimated from data.

### Non-Parametric Density Estimation

Does not assume a fixed finite-dimensional distribution family.

Instead, the density is estimated more flexibly from observations.

Examples:

- Histogram density estimation
- Kernel density estimation

Both approaches have advantages and limitations.



# 10. Parametric Density Estimation

Parametric density estimation assumes a particular family of probability distributions.

For example, suppose we believe that a dataset follows a normal distribution.

We estimate:

$$
\hat{\mu}=\frac{1}{n}\sum_{i=1}^{n}x_i
$$

and:

$$
\hat{\sigma}=
\sqrt{
\frac{1}{n}
\sum_{i=1}^{n}(x_i-\hat{\mu})^2
}
$$

These are maximum likelihood estimates for the mean and standard deviation of an independent normal sample.

Once the parameters are estimated, we construct the fitted PDF.

## Important

A parametric model is only useful if its distributional assumptions are reasonably appropriate.

A smooth fitted curve does not guarantee that the model accurately represents the data.


````python

from numpy.random import normal

sample = normal(
    loc=50,
    scale=5,
    size=1000
)

print("Sample mean:", sample.mean())
print("Sample standard deviation:", sample.std())

plt.figure(figsize=(9, 5))

plt.hist(
    sample,
    bins=20,
    edgecolor="black",
    color="skyblue"
)

plt.title("Histogram of Normally Distributed Data")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
````


## 10.1 Fitting a Normal Distribution

We estimate the distribution parameters from the observed sample.

Then we construct a normal distribution using those estimates.

The fitted PDF can be compared with a density-normalized histogram.

### Important Correction

In the original code, the sample was generated with:

$$
\mu=50,\quad \sigma=5
$$

but the PDF was created using:

```python
norm(60, 12)
```

That distribution does not use the estimated sample parameters.

Here we fit the distribution using the actual sample mean and standard deviation.


````python

sample_mean = sample.mean()
sample_std = sample.std(ddof=0)

dist = norm(
    loc=sample_mean,
    scale=sample_std
)

values = np.linspace(
    sample.min(),
    sample.max(),
    300
)

probabilities = dist.pdf(values)

plt.figure(figsize=(10, 5))

plt.hist(
    sample,
    bins=25,
    density=True,
    alpha=0.5,
    color="skyblue",
    edgecolor="black",
    label="Observed Data"
)

plt.plot(
    values,
    probabilities,
    color="red",
    linewidth=2,
    label="Fitted Normal PDF"
)

plt.title("Parametric Density Estimation")
plt.xlabel("Value")
plt.ylabel("Density")

plt.legend()
plt.show()

print("Estimated mean:", sample_mean)
print("Estimated standard deviation:", sample_std)
````


# 11. Non-Parametric Density Estimation

Sometimes the underlying distribution is unknown or cannot be adequately represented by a familiar parametric family.

In such situations, non-parametric density estimation can be useful.

It estimates the density without requiring a specific fixed distributional shape.

## Advantages

- Flexible distribution shapes.
- Can represent multimodal distributions.
- Does not require assuming normality.
- Useful for exploratory data analysis.

## Limitations

- Sensitive to smoothing parameters.
- Can require substantial data.
- May become computationally expensive.
- Density estimation becomes harder in high dimensions.

One of the most widely used non-parametric approaches is Kernel Density Estimation.


# 12. Kernel Density Estimation (KDE)

Kernel Density Estimation constructs a smooth estimate of the probability density function from observed data.

Instead of assuming a particular distribution family, KDE places a small kernel function around each observation.

The kernels are then combined to produce a smooth density estimate.

## KDE Formula

For one-dimensional data:

$$
\hat{f}_h(x)=
\frac{1}{nh}
\sum_{i=1}^{n}
K\left(
\frac{x-x_i}{h}
\right)
$$

Where:

- $n$ = Number of observations.
- $x_i$ = Individual observation.
- $K$ = Kernel function.
- $h$ = Bandwidth.
- $\hat{f}_h(x)$ = Estimated density.

## Gaussian Kernel

A commonly used kernel is the Gaussian kernel:

$$
K(u)=
\frac{1}{\sqrt{2\pi}}
e^{-u^2/2}
$$

## Bandwidth

Bandwidth controls the smoothness of the estimated density.

### Small Bandwidth

- Narrow kernels.
- More detailed density.
- Potentially noisy estimate.
- Risk of overfitting.

### Large Bandwidth

- Wider kernels.
- Smoother density.
- May hide important features.
- Risk of oversmoothing.

Selecting an appropriate bandwidth is essential.



## 12.1 KDE Using Scikit-Learn

Scikit-learn provides the `KernelDensity` estimator.

The general workflow is:

1. Generate or collect observations.
2. Convert observations into a two-dimensional array.
3. Create the KDE model.
4. Fit the model.
5. Evaluate the density on a grid.
6. Visualize the estimated density.

### Important

`score_samples()` returns log-density estimates.

Therefore, we use:

```python
np.exp(model.score_samples(values))
```

to convert the log-density estimates into density estimates.


````python

model = KernelDensity(
    bandwidth=5,
    kernel="gaussian"
)

sample_2d = bimodal_sample.reshape(-1, 1)

model.fit(sample_2d)

values = np.linspace(
    bimodal_sample.min(),
    bimodal_sample.max(),
    300
).reshape(-1, 1)

log_probabilities = model.score_samples(values)

probabilities = np.exp(log_probabilities)

plt.figure(figsize=(11, 5))

plt.hist(
    bimodal_sample,
    bins=50,
    density=True,
    alpha=0.4,
    color="skyblue",
    label="Histogram"
)

plt.plot(
    values.ravel(),
    probabilities,
    color="red",
    linewidth=2,
    label="KDE"
)

plt.title("Kernel Density Estimation")
plt.xlabel("Value")
plt.ylabel("Density")

plt.legend()
plt.show()
````


## 12.2 Understanding Bandwidth

Bandwidth is one of the most important hyperparameters in KDE.

A very small bandwidth can produce many artificial peaks.

A very large bandwidth can hide multiple modes.

We will compare KDE estimates using different bandwidth values.

In machine learning, bandwidth can be selected using:

- Cross-validation.
- Grid search.
- Statistical rules of thumb.
- Domain knowledge.


````python

bandwidths = [0.5, 2, 5, 10]

values = np.linspace(
    bimodal_sample.min() - 10,
    bimodal_sample.max() + 10,
    400
).reshape(-1, 1)

plt.figure(figsize=(12, 6))

for bandwidth in bandwidths:

    model = KernelDensity(
        bandwidth=bandwidth,
        kernel="gaussian"
    )

    model.fit(bimodal_sample.reshape(-1, 1))

    density = np.exp(
        model.score_samples(values)
    )

    plt.plot(
        values.ravel(),
        density,
        label=f"Bandwidth = {bandwidth}"
    )

plt.title("Effect of KDE Bandwidth")
plt.xlabel("Value")
plt.ylabel("Estimated Density")

plt.legend()
plt.show()
````


# 14. Empirical Cumulative Distribution Function (ECDF)

The empirical cumulative distribution function estimates the CDF directly from observed data.

For a sample of size $n$:

$$
\hat{F}_n(x)=
\frac{1}{n}
\sum_{i=1}^{n}
I(X_i\leq x)
$$

Where:

$$
I(X_i\leq x)
$$

equals one when the observation is less than or equal to x, and zero otherwise.

The ECDF shows the proportion of observed values that are less than or equal to a particular threshold.

### Difference Between KDE and ECDF

KDE estimates a smooth probability density.

ECDF estimates the cumulative distribution.

ECDF does not require a bandwidth parameter.


````python

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.kdeplot(
    data=iris,
    x="petal_width",
    hue="species",
    ax=axes[0]
)

axes[0].set_title("KDE of Petal Width")

sns.ecdfplot(
    data=iris,
    x="petal_width",
    hue="species",
    ax=axes[1]
)

axes[1].set_title("ECDF of Petal Width")

plt.tight_layout()
plt.show()
````


# 16. Standardization Using Z-Scores

Standardization transforms numerical observations so that the transformed sample has mean zero and sample standard deviation one.

The formula is:

$$
z_i=
\frac{x_i-\bar{x}}{s}
$$

Where:

- $x_i$ = Original observation
- $\bar{x}$ = Sample mean
- $s$ = Sample standard deviation

### Properties

For the standardized sample:

$$
\bar{z}=0
$$

and:

$$
s_z=1
$$

up to floating-point precision.

### Applications

Standardization is commonly used before:

- PCA
- K-Means clustering
- Support Vector Machines
- Regularized regression
- Distance-based algorithms

Standardization changes the center and scale but does not make a non-normal distribution normal.


````python

age = titanic["Age"].dropna()

z_scores = (
    age - age.mean()
) / age.std()

print("Standardized mean:", z_scores.mean())
print("Standardized sample SD:", z_scores.std())

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.kdeplot(
    x=age,
    ax=axes[0],
    fill=True
)

axes[0].set_title("Original Age Distribution")

sns.kdeplot(
    x=z_scores,
    ax=axes[1],
    fill=True,
    color="red"
)

axes[1].set_title("Standardized Age Distribution")

plt.tight_layout()
plt.show()
````


# 18. Three-Sigma Rule and Outlier Detection

For a normal distribution, approximately:

- 68.27% of observations lie within one standard deviation of the mean.
- 95.45% lie within two standard deviations.
- 99.73% lie within three standard deviations.

The three-sigma thresholds are:

$$
L=\mu-3\sigma
$$

$$
U=\mu+3\sigma
$$

Observations outside these thresholds may be considered unusual under a normal model.

### Important Limitation

The 68-95-99.7 rule is specific to normal distributions.

For skewed or heavy-tailed datasets, three-sigma thresholds may not accurately describe unusual observations.

Alternative approaches include:

- IQR fences
- Robust z-scores
- Quantile-based thresholds
- Domain-specific rules


````python

mean_age = age.mean()

std_age = age.std()

lower_threshold = mean_age - 3 * std_age
upper_threshold = mean_age + 3 * std_age

print("Mean:", mean_age)
print("Standard deviation:", std_age)

print("Lower threshold:", lower_threshold)
print("Upper threshold:", upper_threshold)

outliers = titanic[
    (titanic["Age"] < lower_threshold) |
    (titanic["Age"] > upper_threshold)
]

display(outliers)

print("Maximum age:", age.max())
````


# 19. CDF of a Continuous Random Variable

For a continuous random variable with PDF $f(x)$:

$$
F(x)=P(X\leq x)
$$

The CDF can be obtained by integrating the PDF:

$$
F(x)=
\int_{-\infty}^{x}
f(t)\,dt
$$

For an interval:

$$
P(a<X\leq b)=F(b)-F(a)
$$

For continuous random variables, using strict or non-strict inequalities at interval endpoints does not change the probability.

### Relationship Between PDF and CDF

Where the CDF is differentiable:

$$
f(x)=\frac{dF(x)}{dx}
$$

Thus:

- Integrating the PDF gives the CDF.
- Differentiating the CDF gives the PDF where the derivative exists.


````python

x = np.linspace(-4, 4, 1000)

pdf = norm.pdf(x)

cdf = norm.cdf(x)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(
    x,
    pdf,
    color="blue",
    linewidth=2
)

axes[0].set_title("Normal PDF")
axes[0].set_xlabel("X")
axes[0].set_ylabel("Density")

axes[1].plot(
    x,
    cdf,
    color="red",
    linewidth=2
)

axes[1].set_title("Normal CDF")
axes[1].set_xlabel("X")
axes[1].set_ylabel("Cumulative Probability")

plt.tight_layout()
plt.show()
````


# 20. Using PDF and CDF in Data Analysis

PDF and CDF provide different information.

## PDF

Useful for understanding:

- Distribution shape
- Concentration of observations
- Modes
- Relative density
- Possible anomalies

## CDF

Useful for calculating:

- Percentiles
- Probabilities below thresholds
- Probabilities above thresholds
- Probabilities between values

### Example

Suppose exam scores follow:

$$
X\sim N(70,10^2)
$$

We want to calculate:

1. Probability of scoring below 60.
2. Probability of scoring above 80.
3. Probability of scoring between 60 and 80.


````python

mu = 70
sigma = 10

p_below_60 = norm.cdf(
    60,
    loc=mu,
    scale=sigma
)

p_above_80 = 1 - norm.cdf(
    80,
    loc=mu,
    scale=sigma
)

p_between = (
    norm.cdf(80, loc=mu, scale=sigma)
    - norm.cdf(60, loc=mu, scale=sigma)
)

print("P(X < 60):", p_below_60)
print("P(X > 80):", p_above_80)
print("P(60 < X < 80):", p_between)
````

## 23.6 Gaussian Mixture Model Example

A Gaussian Mixture Model (GMM) represents a distribution as a weighted sum of Gaussian components.

$$
p(x)=
\sum_{k=1}^{K}
\pi_k
\mathcal{N}(x|\mu_k,\sigma_k^2)
$$

Where:

- $K$ is the number of components.
- $\pi_k$ is the mixing weight.
- $\mu_k$ is the mean of component k.
- $\sigma_k$ is its standard deviation.

The weights satisfy:

$$
\pi_k\geq0
$$

and:

$$
\sum_{k=1}^{K}\pi_k=1
$$

We will fit a two-component Gaussian mixture to our bimodal dataset.


````python

from sklearn.mixture import GaussianMixture

gmm = GaussianMixture(
    n_components=2,
    random_state=42
)

gmm.fit(bimodal_sample.reshape(-1, 1))

x = np.linspace(
    bimodal_sample.min() - 10,
    bimodal_sample.max() + 10,
    400
).reshape(-1, 1)

log_density = gmm.score_samples(x)

density = np.exp(log_density)

plt.figure(figsize=(11, 5))

plt.hist(
    bimodal_sample,
    bins=40,
    density=True,
    alpha=0.4,
    color="skyblue",
    label="Observed Data"
)

plt.plot(
    x.ravel(),
    density,
    color="red",
    linewidth=2,
    label="Gaussian Mixture Model"
)

plt.title("Gaussian Mixture Density Estimation")
plt.xlabel("Value")
plt.ylabel("Density")

plt.legend()
plt.show()

print("Estimated component means:")
print(gmm.means_.ravel())

print("\nEstimated mixing weights:")
print(gmm.weights_)
````
