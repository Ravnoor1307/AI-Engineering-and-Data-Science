In Python, the assert statement is a built-in construct that allows you to test assumptions about your code. It acts as a sanity check to ensure that certain conditions are met during the execution of a program.

```python
# The assert statement takes the following syntax:

assert condition, "message"
```
Here, the condition is the expression or condition that you want to test, and the message is an optional string that provides additional information about the assertion.

When the assert statement is encountered, it evaluates the given condition. If the condition is true, the program continues its execution without any interruptions. However, if the condition is false, the assert statement raises an AssertionError exception.

**Use Case**:
1. Checks.
2. Debugging
3. Testing --> unittest, pytest

> Assert only works in the development phase and is not implemented in production.

````python
# basic assert statement

print("hello")

assert 6 > 5, "Message"

print("world")
````

````python
# assertion exception

print("hello")

assert 6 < 5, "Number 1 should be greater than Number 2"

print("world")
````

Type assertions are used to verify the type of a variable or expression. They ensure that the expected type is matched, otherwise, an AssertionError is raised.

````python
# assertion used in python

# type check

def add_numbers(iterable):
    assert all(isinstance(x, int) for x in iterable), "All elements must be integers"
    total = sum(iterable)

    return total
````

````python
add_numbers([2, 4, 5, 6])
````

````python
add_numbers([2, 4, "5", 6])
````

Value assertions in Python are used to check whether a certain value meets specific conditions or criteria.

````python
# check on values

def divide(a, b):
    assert b != 0, "Cannot divide by 0"
    result = a / b

    return result
````

````python
divide(10, 2)
````

````python
divide(10, 0)
````

````python
def cast_vote(age):
    assert age >= 18, "You are a minor, Cannot cast your vote"
    return "Casted Vote"
````

````python
cast_vote(25)
````

````python
cast_vote(16)
````

Collection assertions are used to verify properties of collections, such as lists, tuples, sets, or dictionaries. These assertions allow you to check if a collection contains specific elements or satisfies certain conditions.

````python
# check on iterables

def calculate_average(numbers):
    assert len(numbers) > 0, "List must not be empty"
    total = sum(numbers)
    average = total / len(numbers)

    return average



data = [5, 10, 15, 20]

result = calculate_average(data)

result
````

````python
calculate_average([])
````

````python
# performance test in ML

class NakliLR:
    def __init__(self):
        self.weights = None
        self.bias = None

    def fit(self, X_train, y_train):
        return self

    def predict(self, X_test):
        return random.randint(1, 50)

    def score(self, X_test, y_test):
        return random.randrange(50,90)

def check_performance(X_test, y_test, model: NakliLR):
    model_prediction = model.predict(X_test)
    print(f"Prediction: {model_prediction}")
    score = model.score(X_test, y_test)
    print(f"score: {score}")

    threshold = 70

    assert score >= threshold, "Performance test failed, Model cannot pass the score of {}".format(threshold)

    print("Model has passed the performance test")
````
