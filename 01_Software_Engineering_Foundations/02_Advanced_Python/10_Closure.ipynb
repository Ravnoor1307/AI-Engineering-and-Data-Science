**A closure in Python is a function object that remembers values in enclosing scopes even if they are not present in memory.**

```python
def outer_func():
    x = 10
    print("Outer Function", f"x={x}")
    def inner_func():
        y = 5
        print("Inner Function", f"y={y}")
        print("Inner Function", f"x={x}")

    # call the inner function
    inner_func()
```

```python
outer_func()
```

```python
def adder(add_num):
    def add(number):
        return number + add_num

    # return the inner funtion
    return add
```

```python
add_by_2 = adder(2)
```

```python
add_by_2
```

```python
add_by_2(10)
```

```python
add_by_2(5)
```

Closures are inner functions, that maintains the variables of the enclosing scope even though the enclosing function has completed its execution.

```python
# use case --> factory functions
```

```python
def power_by(power):
    def execute(number):
        result = number ** power
        return result

    return execute
```

```python
square = power_by(2)

cube = power_by(3)

sqrt = power_by(0.5)
```

```python
square(2)
```

```python
square(square(2))
```

```python
cube(2)
```

```python
cube(4)
```

```python
sqrt(4)
```

```python
sqrt(25)
```
