TypedDict to make dictionaries in python but it also includes type hints and type checks.

```python
class Person(TypedDict):
    name: str
    age: int
    city_name: str
```

```python
person_dict: Person = {
    "name": "John",
    "age": 45,
    "city_name": "New york"
}

person_dict
```

```python
type(person_dict)
```

```python
person_dict_2: Person = {
    "full_name": "Neha Sharma",
    "age": 30,
    "city": "Indore"
}

person_dict_2
```

```python
Person(name="Vishal", age=45, city_name="Delhi")
```

```python
# another way to make typeddict

Animal = TypedDict("Animal", {"name": str, "sound": str, "age": int})
```

```python
Animal(name="Bruno", sound="bark", age=5)
```

```python
animal_dict: Animal = {
    "name": "Tuffy",
    "sound": "bark",
    "age": 3
}

animal_dict
```
