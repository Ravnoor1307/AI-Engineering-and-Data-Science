```python
import requests
```

The Requests library is the de facto standard for making HTTP requests in Python. It abstracts the complexities of making requests behind a beautiful, simple API so that you can focus on interacting with services and consuming data in your application.

```python
# urls

url_wiki = "https://en.wikipedia.org/wiki/Python_(programming_language)"

url_api = "https://api.github.com"
```

HTTP Methods:

1. GET - Retrieval
2. POST - Create
3. PUT - Update
4. DELETE - Delete

```python
# sending a get request

response = requests.get(url=url_wiki)
```

```python
# response object

response
```

```python
# response body --> content/payload

type(response.content)
```

```python
# as text

response.text[0:1000]
```

```python
# status code

response.status_code
```

```python
# status code used as conditionals

if response.status_code == 200:
    print("Success!")
elif response.status_code == 404:
    print("Not Found.")
```

```python
# raising errors

response.raise_for_status()
```

```python
# exception handling of errors

try:
    response = requests.get(url_wiki)
    response.raise_for_status()
except Exception as e:
    print(e)
else:
    print("Success")
```

```python
# response headers

response.headers["content-type"]
```

```python
# get request on api

response = requests.get(url=url_api)
```

```python
# responde headers of api

response.headers["content-type"]
```

```python
response.text
```

```python
# json response

json_content = response.json()

type(json_content)
```

```python
json_content["issues_url"]
```

```python
# query parameters

params= {"q": "language:python", "sort": "stars", "order": "desc"}

response = requests.get(
    url="https://api.github.com/search/repositories",
    params=params,
)

```

```python
response.headers["content-type"]
```

```python
json_response = response.json()

popular_repositories = json_response["items"]

for repo in popular_repositories[:3]:
    print(f"Name: {repo['name']}")
    print(f"Description: {repo['description']}")
    print(f"Stars: {repo['stargazers_count']}")
    print()
```

```python
# request body url

response.request.url
```

```python
# post url

url_post = "https://httpbin.org/post"
```

```python
# send post request with data

data = {"Name": "Amit", "Age": 42}

response = requests.post(url=url_post, data=data)

response
```

```python
# send data as text

response = requests.post(url=url_post, data=data)

response.request.headers["content-type"]
```

```python
# send data as json

response = requests.post(url=url_post, json=data)

response.request.headers["content-type"]
```

```python
# analysis of request

# request url

response.request.url
```

```python
# request headers
response.request.headers["content-type"]
```

```python
# request body

response.request.body
```

```python
# response in post

response.content
```
