🔧 What is a metaclass?  
A class defines how an object behaves.

A metaclass defines how a class behaves.    

----
Think of it like this:   

Objects are created from classes.

Classes themselves are created from metaclasses.

```python
class MyClass:

    def __init__(self, name):
        self.name = name

    def greet(self):
        return f"Hello my name is {self.name}"


p1 = MyClass("Ankit")
```

```python
p1
```

```python
p1.name
```

```python
p1.greet()
```

🧠 Simple Analogy   
- Object ← created by ← Class

- Class ← created by ← Metaclass



```python
# create a metaclass

type("MyClass1", (), {})
```

```python
def greet(self):
    return "Hello"

myclass = type("MyClass1", (), {"greet": greet})

myclass
```

```python
obj_1 = myclass()

obj_1.greet()
```

🎯 Why Use a Metaclass?

You use a metaclass when you want to control or customize class creation.

```python
class MyAttrValidation(type):
    def __new__(cls, name, bases, dct):
        new_attributes = {}
        for attr_name, attr_value in dct.items():
            if not attr_name.startswith("__"):
                new_attributes[attr_name.lower()] = attr_value
            else:
                new_attributes[attr_name] = attr_value

        return super().__new__(cls, name, bases, new_attributes)
```

```python
class Person(metaclass=MyAttrValidation):

    def __init__(self, name):
        self.name = name

    def greet(self):
        return f"Hello my name is {self.name}"

    def Lives_In(self, city):
        return f"Hi my name is {self.name} and i live in {city}"
```

```python
person_1 = Person("Ankit")
```

```python
person_1.name
```

```python
person_1.greet()
```

```python
person_1.Lives_In()
```

```python
person_1.lives_in("mumbai")
```
