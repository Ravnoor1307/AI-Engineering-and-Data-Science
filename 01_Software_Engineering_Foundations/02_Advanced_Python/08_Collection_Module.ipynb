```python
import collections
```

```python
# chainmap

fruits_dict = {"apple": "red", "banana": "yellow", "pear": "green", "orange": "orange"}

fruits_dict
```

```python
person_dict = {"John": "apple", "Alex": "Oranges", "Samantha": "pears"}

person_dict
```

```python
chain = collections.ChainMap(fruits_dict, person_dict)

chain
```

```python
chain["apple"]
```

```python
chain["orange"]
```

```python
chain["Samantha"]
```

```python
for k, v in chain.items():
    print(f"{k} --> {v}")
```

```python
vegetables_dict = {"tomato": "red", "cucumber": "green"}

vegetables_dict
```

```python
chain
```

```python
chain = chain.new_child(vegetables_dict)
```

```python
[k for k in chain.keys()]
```

```python
chain.parents
```

```python
# counter

list1 = ["ankit", "akash", "hemant", "neha", "akash", "akash", "ankit", "neha", "akash"]

list1
```

```python
count = collections.Counter(list1)

count
```

```python
for elem in count.elements():
    print(elem)
```

```python
# most common

count.most_common(n=2)
```
