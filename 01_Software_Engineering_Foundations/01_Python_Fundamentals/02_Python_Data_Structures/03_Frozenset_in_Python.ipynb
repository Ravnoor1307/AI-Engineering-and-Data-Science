frozenset --> immutable set

```python
# sets in python --> mutable

list_1 = [1, 2, 3, 4, 5, 2, 5, 3]

set_1 = set(list_1)

set_1
```

```python
id(set_1)
```

```python
# sets are mutable objects

set_1.add(6)

set_1
```

```python
id(set_1)
```

```python
# delete

set_1.discard(2)
```

```python
set_1
```

```python
id(set_1)
```

```python
# frozenset

f_1 = frozenset(list_1)

f_1
```

```python
# add element

f_1.add(10)
```

```python
id(f_1)
```

```python
list_2 = list_1 + [6]

list_2
```

```python
f_2 = frozenset(list_2)

f_2
```

```python
id(f_2)
```

```python
# frozenset are immutable
```

```python
# frozenset as dict keys


set_key = set(("first_name", "last_name"))

set_key
```

```python
dict_1 = {set_key: ["john", "smith"]}
```

```python
set_key = frozenset(("first_name", "last_name"))

set_key
```

```python
dict_1 = {set_key: ["john", "smith"]}

print(dict_1)
```

**Properties**:
1. immutable --> does not have add, discard methods
2. hashable --> used as keys in dictionaries

```python
# set operations on frozenset
set_1 = frozenset([1,2,3,4])
set_2 = frozenset([3,4,5,6])
# union
print(set_1 | set_2)
print(set_1.union(set_2))

# intersection
print(set_1.intersection(set_2))
print(set_1 & set_2)

# set difference
print(set_1.difference(set_2))
print(set_1 - set_2)

# symmetric difference
print(set_1.symmetric_difference(set_2))
print(set_1 ^ set_2)
```
